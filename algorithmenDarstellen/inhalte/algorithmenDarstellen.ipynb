### Algorithmen darstellen

Ein **Algorithmus** ist eine Schritt-für-Schritt-Anleitung zur Lösung eines Problems. Bevor wir ihn programmieren oder wenn wir
ihn anderen erklären, stellen wir ihn oft unabhängig von einer Programmiersprache dar. Wir lernen zwei Darstellungen kennen:

- **Pseudocode**: der Algorithmus als strukturierter Text
- **Nassi-Shneiderman-Diagramm** (auch *Struktogramm*): der Algorithmus als Diagramm


#### Pseudocode

**Pseudocode** ist eine vereinfachte Beschreibung eines Programms. Er verwendet eine leicht verständliche Sprache, die an
Programmiersprachen erinnert, aber nicht an die genaue Syntax einer bestimmten Programmiersprache gebunden ist.

Für Pseudocode gibt es keine feste Norm. Wichtig ist, dass jeder Schritt eindeutig ist und die Einrückung die Struktur zeigt.
Wir verwenden die folgenden Bausteine:

| Baustein | Pseudocode | Python |
|---|---|---|
| Zuweisung | `x = 0` oder `setze x auf 0` | `x = 0` |
| Eingabe | `lies x ein` | `x = int(input())` |
| Ausgabe | `gib x aus` | `print(x)` |
| Verzweigung | `wenn x > 0:` und `sonst:` | `if x > 0:` und `else:` |
| Solange-Schleife | `solange x > 0:` | `while x > 0:` |
| Zählschleife | `für i von 1 bis n:` | `for i in range(1, n+1):` |
| Schleife über eine Liste | `für jedes x in a:` | `for x in a:` |
| Funktion | `algorithmus maximum(a):` | `def maximum(a):` |
| Rückgabe | `gib m zurück` | `return m` |

**Achtung:** Bei `für i von 1 bis n` gehört `n` dazu. Bei `range` gehört die obere Grenze nicht dazu.

Bedingungen dürfen auch in Worten stehen, z. B. `wenn x gerade:` statt `wenn x % 2 == 0:`.

**Beispiel:** Das Maximum einer Liste im Pseudocode

    algorithmus maximum(a):
        m = a[0]
        für jedes x in a:
            wenn x > m:
                m = x
        gib m zurück

Die Übersetzung nach Python ist fast Zeile für Zeile möglich.

```python
def maximum(a):
    m = a[0]
    for x in a:
        if x > m:
            m = x
    return m

print(maximum([4, 7, 1, 9, 3]))
```

#### Das Nassi-Shneiderman-Diagramm

Ein **Nassi-Shneiderman-Diagramm** stellt einen Algorithmus als Rechteck dar, das aus einzelnen Blöcken besteht. Es wird von oben
nach unten gelesen. Jede Kontrollstruktur ist ein eigener Block, und in einem Block können wieder Blöcke stehen. Benannt ist es nach
Isaac Nassi und Ben Shneiderman.

**Folge von Anweisungen:** Die Blöcke stehen untereinander und werden der Reihe nach ausgeführt.

<img src='nsd_folge.png' width='248'>

```python
a = 3
b = 4
print(a + b)
```

**Verzweigung:** Im Dreieck steht die Bedingung. Ist sie wahr, wird der Teil unter „ja“ ausgeführt, sonst der Teil unter „nein“.
Danach geht es unterhalb des ganzen Blocks weiter. Gibt es kein `else`, bleibt die Seite „nein“ leer.

<img src='nsd_wenn.png' width='333'>

```python
x = 5
if x > 0:
    print('positiv')
else:
    print('nicht positiv')
```

**Mehrere Fälle:** Mit `elif` prüfen wir nacheinander mehrere Bedingungen. Im Pseudocode schreiben wir dafür `sonst wenn`:

    wenn punkte >= 90:
        gib "sehr gut" aus
    sonst wenn punkte >= 50:
        gib "bestanden" aus
    sonst:
        gib "nicht bestanden" aus

Im Diagramm schachteln wir die Verzweigungen: Jedes `elif` ist eine weitere Verzweigung im Teil „nein“ der vorherigen Verzweigung.

<img src='nsd_elif.png' width='560'>

```python
punkte = 72
if punkte >= 90:
    print('sehr gut')
elif punkte >= 50:
    print('bestanden')
else:
    print('nicht bestanden')
```

**Solange-Schleife:** Der Rumpf der Schleife ist eingerückt, links läuft die Leiste mit der Bedingung nach unten. Die Bedingung wird
**vor** jedem Durchlauf geprüft (*kopfgesteuerte* Schleife). Ist sie schon am Anfang falsch, wird der Rumpf gar nicht ausgeführt.

<img src='nsd_solange.png' width='214'>

```python
i = 3
while i > 0:
    print(i)
    i = i - 1
```

**Zählschleife:** Sie sieht genauso aus wie die Solange-Schleife, im Kopf steht aber der Zählbereich. Die Grenzen gehören dazu.
Auch `wiederhole für jedes x in a` ist möglich.

<img src='nsd_fuer.png' width='232'>

```python
s = 0
for i in range(1, 5):
    s = s + i
print(s)
```

**Wiederhole-bis-Schleife:** Die Bedingung steht unten und wird erst **nach** jedem Durchlauf geprüft (*fußgesteuerte* Schleife). Der
Rumpf läuft also mindestens einmal. Die Schleife endet, wenn die Bedingung **wahr** ist. Python hat dafür keinen eigenen
Befehl, wir schreiben `while True:` mit `break`.

<img src='nsd_bis.png' width='189'>

```python
i = 0
while True:
    i = i + 2
    print(i)
    if i >= 4:
        break
```

#### Beispiel: Collatz

Der Collatz-Algorithmus als Nassi-Shneiderman-Diagramm. Die Verzweigung steht im Rumpf der Schleife, die
Anweisung `z = z + 1` steht unter der Verzweigung und wird deshalb in jedem Durchlauf ausgeführt.

<img src='nsd_collatz.png' width='240'>

```python
x = 6
z = 0
while x != 1:
    if x % 2 == 0:
        x = x // 2
    else:
        x = 3 * x + 1
    z = z + 1
print(z)
```

#### Vom Programm zum Diagramm

So erstellen wir ein Nassi-Shneiderman-Diagramm aus einem Programm oder aus Pseudocode:

1. Wir gehen das Programm von oben nach unten durch. Jede einfache Anweisung wird ein Block.
2. `if`, `while` und `for` werden je ein Block über die volle Breite.
3. Was im Programm eingerückt ist, kommt in den Rumpf der Schleife oder unter „ja“ oder „nein“.

Umgekehrt entspricht beim Lesen jede Schachtelung im Diagramm einer Einrückung im Programm.
