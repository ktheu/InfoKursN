### Sort in Python

[Video](https://youtu.be/loBkBqOGKLY)

Der in Python implementierte Sortieralgorithmus ist eine
  Kombination von Insertionsort und Mergesort und wird nach
  seinem Erfinder **Tim Peters**
  <a href="https://de.wikipedia.org/wiki/Timsort">Timsort</a>
  genannt.

Timsort liegt in $O(n \cdot \log n)$. Ist die Liste schon (fast) sortiert, braucht es nur $O(n)$.

#### Eine Liste sortieren

```python
# a wird sortiert
a = [45, 97, 80, 23, 28, 66, 45, 37, 3]
a.sort()
a
```

```python
a.sort(reverse=True)
a
```

#### Eine neue sortierte Liste erstellen

```python
# a bleibt unverändert
a = [45, 97, 80, 23, 28, 66, 45, 37, 3]
b = sorted(a)
c = sorted(a,reverse=True)
a, b, c
```

#### Mit einer key-Funktion sortieren

```python
# Werte nach Modulo 5 sortieren
a = [45, 97, 80, 23, 28, 66, 45, 37, 3]
def f(x):
    return x % 5
a.sort(key = f)
a
```

Python ruft die key-Funktion für jedes Element einmal auf und sortiert dann nach den Ergebnissen.
Die Elemente selbst werden dabei nicht verändert.

Die key-Funktion kennen wir schon von `min` und `max`. Auch fertige Funktionen wie `len` oder `str.lower` können
wir als key-Funktion verwenden.

```python
woerter = ['Kiwi', 'apfel', 'Banane', 'feige']
print(sorted(woerter, key=len))
print(sorted(woerter, key=str.lower))
print(max(woerter, key=len))
```

#### Mit einer key-Funktion nach mehreren Kriterien sortieren

```python
# Werte erst nach Modulo 5, dann nach Größe sortieren
a = [45, 97, 80, 23, 28, 66, 45, 37, 3]
def f(x):
    return x % 5, x
a.sort(key = f)
a
```

Gibt die key-Funktion ein Tuple zurück, wird zuerst nach der ersten Komponente sortiert. Nur wenn diese gleich ist,
entscheidet die zweite Komponente, und so weiter.

#### Mit einer anonymen Funktion sortieren

Eine *anonyme Funktion* (lambda-Funktion) ist eine kurze Funktion ohne Namen. `lambda x: x % 5` bedeutet dasselbe wie
die Funktion `f` von oben: Sie bekommt `x` und gibt `x % 5` zurück. Eine lambda-Funktion besteht nur aus einem Ausdruck,
ein `return` wird nicht geschrieben.

```python
a = [45, 97, 80, 23, 28, 66, 45, 37, 3]
a.sort(key = lambda x: x % 5)
a
```

Soll nach einem Kriterium absteigend und nach einem anderen aufsteigend sortiert werden, hilft bei Zahlen ein
Minuszeichen: Hier wird nach `x % 5` aufsteigend und danach nach `x` absteigend sortiert.

```python
a = [45, 97, 80, 23, 28, 66, 45, 37, 3]
a.sort(key = lambda x: (x % 5 , -x))
a
```

```python
# Strings nach der Länge, dann alphabetisch sortieren
a = ['Thorben','Lena','Maike','Sören','Erik','Malte']
a.sort(key = lambda x: (len(x),x))
a
```

#### Wie werden Strings verglichen?

Strings werden Zeichen für Zeichen nach ihrem Unicode-Wert verglichen (siehe `ord`). Großbuchstaben haben kleinere
Werte als Kleinbuchstaben, deshalb stehen sie beim Sortieren vorne. Umlaute haben noch größere Werte.

```python
print(ord('A'), ord('Z'), ord('a'), ord('z'), ord('Ä'), ord('ä'))
print(sorted(['banane', 'Apfel', 'kiwi', 'Birne', 'Äpfel']))
print(sorted(['banane', 'Apfel', 'kiwi', 'Birne', 'Äpfel'], key=str.lower))
```

#### Sortieren ist stabil

Elemente mit gleichem Schlüssel behalten ihre ursprüngliche Reihenfolge. Man nennt das Sortierverfahren **stabil**.

```python
a = ['Ben', 'Anna', 'Tim', 'Eva', 'Lena']
print(sorted(a, key=len))     # Ben, Tim, Eva bleiben in der ursprünglichen Reihenfolge
```

Deshalb kann man auch in mehreren Schritten sortieren: erst nach dem weniger wichtigen Kriterium, dann nach dem
wichtigeren.

```python
a = ['Ben', 'Anna', 'Tim', 'Eva', 'Lena']
a.sort()                  # alphabetisch
a.sort(key=len)           # nach Länge, bei gleicher Länge bleibt die alphabetische Reihenfolge
print(a)
```

#### Datensätze sortieren

Bei Listen aus Tuples oder Dicts gibt die key-Funktion die Komponente zurück, nach der sortiert werden soll.

```python
schueler = [('Lena', 17, 2.1), ('Tom', 16, 1.7), ('Ida', 17, 1.3), ('Max', 16, 2.8)]
print(sorted(schueler, key=lambda s: s[2]))                # nach Note
print(sorted(schueler, key=lambda s: (-s[1], s[0])))       # nach Alter absteigend, dann nach Name
```

```python
haeufigkeit = {'der': 5, 'Hund': 2, 'bellt': 2, 'laut': 1}
print(sorted(haeufigkeit))                                         # die Schlüssel, alphabetisch
print(sorted(haeufigkeit, key=lambda w: haeufigkeit[w], reverse=True))   # nach Häufigkeit
print(sorted(haeufigkeit.items(), key=lambda p: p[1], reverse=True))     # Paare (Wort, Anzahl)
```

#### Anwendung: Das Rucksackproblem

Ein Rucksack kann ein Gewicht bis zur `kapazitaet` tragen. Jedes Item hat einen Wert und ein Gewicht. Gesucht ist eine
Auswahl von Items mit möglichst großem Gesamtwert, deren Gesamtgewicht die Kapazität nicht überschreitet.

Eine einfache Strategie: Wir sortieren die Items nach einem Kriterium und nehmen der Reihe nach jedes Item, das noch
passt. Ein solches Verfahren, das in jedem Schritt die gerade beste Wahl trifft, heißt **Greedy-Verfahren**
(greedy = gierig). Drei naheliegende Strategien:

- die wertvollsten zuerst
- die leichtesten zuerst
- die wertdichtesten zuerst (größter Wert pro Gewicht)

```python
def rucksack_greedy(wert, gewicht, kapazitaet, key):
    genommen = []
    rucksack_gewicht = 0
    rucksack_wert = 0
    for i in sorted(range(len(wert)), key=key):      # Indizes der Items in der Reihenfolge der Strategie
        if rucksack_gewicht + gewicht[i] <= kapazitaet:
            genommen.append(i)
            rucksack_gewicht += gewicht[i]
            rucksack_wert += wert[i]
    return sorted(genommen), rucksack_wert, rucksack_gewicht

wert = [1, 11, 13, 9]
gewicht = [2, 6, 8, 4]
kapazitaet = 10

print('wertvollste zuerst: ', rucksack_greedy(wert, gewicht, kapazitaet, lambda i: -wert[i]))
print('leichteste zuerst:  ', rucksack_greedy(wert, gewicht, kapazitaet, lambda i: gewicht[i]))
print('wertdichteste zuerst:', rucksack_greedy(wert, gewicht, kapazitaet, lambda i: -wert[i] / gewicht[i]))
```

Ein Greedy-Verfahren ist schnell, findet aber nicht immer die beste Lösung. Im folgenden Beispiel liefert keine der
drei Strategien das Optimum: Die Items 1 und 2 haben zusammen den Wert 14 und das Gewicht 10.

```python
wert = [10, 7, 7, 1]
gewicht = [6, 5, 5, 1]

print('wertvollste zuerst: ', rucksack_greedy(wert, gewicht, 10, lambda i: -wert[i]))
print('leichteste zuerst:  ', rucksack_greedy(wert, gewicht, 10, lambda i: gewicht[i]))
print('wertdichteste zuerst:', rucksack_greedy(wert, gewicht, 10, lambda i: -wert[i] / gewicht[i]))
```

#### Zusammenfassung

| | |
|---|---|
| `a.sort()` | sortiert die Liste `a` selbst, gibt `None` zurück |
| `sorted(a)` | erzeugt eine neue sortierte Liste, `a` bleibt unverändert |
| `reverse=True` | absteigend sortieren |
| `key=f` | nach den Werten `f(x)` sortieren, z.B. `key=len`, `key=str.lower` |
| `lambda x: ...` | kurze Funktion ohne Namen, oft als key-Funktion |
| Tuple als Schlüssel | nach mehreren Kriterien sortieren, `-x` für absteigend bei Zahlen |
| stabil | Elemente mit gleichem Schlüssel behalten ihre Reihenfolge |
| Laufzeit | $O(n \cdot \log n)$ (Timsort) |
