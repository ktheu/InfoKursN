## ASCII, Unicode, UTF-8

### 1. Zeichen als Zahlen

Ein Computer speichert nur binäre Daten. Um mit Zeichen umgehen zu können, wird über eine **Zeichensatztabelle** (Codepage) jedem Zeichen eine Zahl zugeordnet. Gespeichert wird dann die Binärdarstellung dieser Zahl.

```
Zeichen  ->  Zahl  ->  Bits
   A     ->   65   ->  01000001
```

Wichtig: Sender und Empfänger müssen **dieselbe Tabelle** benutzen. Sonst kommen beim Lesen falsche Zeichen heraus.

### 2. ASCII

Der **ASCII-Code** (American Standard Code for Information Interchange) sieht in seiner ursprünglichen Version 7 Bit zur Codierung von Zeichen vor. Damit lassen sich $2^7 = 128$ Zeichen darstellen.

<img src='./ascii.png' width='600'>

So liest man die Tabelle: Die **Zeile** gibt die erste Hexadezimalziffer an, die **Spalte** die zweite. `A` steht in Zeile `4…` und Spalte `…1`, hat also den Code `0x41` = 65.

##### Aufbau der ASCII-Tabelle

| Bereich (hex) | Bereich (dez) | Zeichen |
|:--|:--|:--|
| `00` bis `1F` | 0 bis 31 | **Steuerzeichen**, z. B. Tabulator `HT` (9), Zeilenumbruch `LF` (10) |
| `20` | 32 | Leerzeichen `SP` |
| `30` bis `39` | 48 bis 57 | Ziffern `0` bis `9` |
| `41` bis `5A` | 65 bis 90 | Großbuchstaben `A` bis `Z` |
| `61` bis `7A` | 97 bis 122 | Kleinbuchstaben `a` bis `z` |

Umlaute und `ß` kommen in ASCII nicht vor.

Achtung: Die Ziffer `'7'` ist ein Zeichen und hat den Code 55, nicht 7.

##### ord und chr

Die Python-Funktion `ord` liefert zu einem Zeichen die Zahl, `chr` liefert zu einer Zahl das Zeichen.

````python
print(ord('A'))
print(chr(65))
print(ord('7'))
````

````python
for zeichen in 'ABCabc':
    print(zeichen, ord(zeichen), hex(ord(zeichen)), format(ord(zeichen), '08b'))
````

##### Codierung der Buchstaben

```
A, B, C, ...  = 01000001, 01000010, 01000011, ...
a, b, c, ...  = 01100001, 01100010, 01100011, ...
```

Die Buchstaben stehen in alphabetischer Reihenfolge. Ein Kleinbuchstabe hat immer einen um **32** größeren Code als der zugehörige Großbuchstabe. In der Binärdarstellung unterscheiden sie sich nur in einem einzigen Bit (dem Bit mit dem Stellenwert 32).

````python
print(ord('a') - ord('A'))
print(chr(ord('g') - 32))    # Großbuchstabe zu g
````

##### Beispiel: Eine Bitfolge dekodieren

```
01000011 01101111 01100100 01100101
```

1. Jedes Byte in eine Zahl umrechnen, am einfachsten über Vierergruppen: `0100 0011` = `0x43` = 67.
2. Das Zeichen in der Tabelle nachsehen: Zeile `4…`, Spalte `…3` ist `C`.

| Byte | hex | dez | Zeichen |
|:--|:-:|:-:|:-:|
| `0100 0011` | 43 | 67 | C |
| `0110 1111` | 6F | 111 | o |
| `0110 0100` | 64 | 100 | d |
| `0110 0101` | 65 | 101 | e |

Ergebnis: `Code`

````python
bitfolge = '01000011 01101111 01100100 01100101'
text = ''
for byte in bitfolge.split():
    text += chr(int(byte, 2))
print(text)
````

### 3. ISO-8859-1 (ANSI)

**ISO-8859-1** ist eine Erweiterung des ASCII-Codes auf 8 Bit, also $2^8 = 256$ Zeichen. Die ersten 128 Zeichen sind genau die ASCII-Zeichen. In der oberen Hälfte (`A0` bis `FF`) stehen Umlaute, Akzentbuchstaben und Sonderzeichen. Das reicht für die meisten westeuropäischen Sprachen aus. Es fehlen aber zum Beispiel das Eurozeichen und einige französische Zeichen.

Gebräuchliche Namen für ähnliche Varianten sind *Windows-1252*, *Latin-1*, *Westeuropäisch* oder *ANSI*.

<img src='./iso.png' width='550'>

Beispiel: `ä` steht in Zeile `E…`, Spalte `…4`, hat also den Code `0xE4` = 228.

Das Problem: Für andere Sprachen (Griechisch, Russisch, Arabisch, ...) gibt es **andere** 8-Bit-Tabellen, in denen dieselben Codes andere Zeichen bedeuten. Ein Text mit Zeichen aus verschiedenen Sprachen lässt sich so nicht speichern.

### 4. Unicode

**Unicode** ist ein internationaler Standard, der jedem Schriftzeichen aller bekannten Sprachen eine eindeutige Zahl zuordnet, den **Code Point**. Dazu kommen viele Sonderzeichen und Emojis. [Unicode Charts](https://www.unicode.org/charts/)

Code Points schreibt man meist hexadezimal mit vorangestelltem `U+`:

| Zeichen | Code Point (dez) | Code Point (hex) | Schreibweise |
|:-:|--:|--:|:--|
| A | 65 | 41 | U+0041 |
| a | 97 | 61 | U+0061 |
| ß | 223 | DF | U+00DF |
| € | 8364 | 20AC | U+20AC |
| π | 960 | 3C0 | U+03C0 |
| 😀 | 128512 | 1F600 | U+1F600 |

Die ersten 256 Code Points stimmen mit ISO-8859-1 überein, die ersten 128 also auch mit ASCII. Der Unicode-Standard legt fest, dass Code Points nur von `0` bis `10FFFF` reichen dürfen. Das sind über eine Million mögliche Zeichen, belegt sind bisher rund 150.000.

Die Python-Funktionen `chr` und `ord` arbeiten mit Unicode Code Points:

````python
print(ord('€'))
print(chr(8364))
print(hex(ord('€')))
print(chr(0x1F600))
````

### 5. UTF-8

Unicode legt nur fest, welche **Zahl** zu einem Zeichen gehört. Wie diese Zahl als **Bytes** gespeichert wird, legt eine **Codierung** fest.

Der größte Code Point `10FFFF` braucht 21 Bit. Man könnte also jeden Code Point einfach mit 3 Byte speichern. Für Texte, die fast nur aus ASCII-Zeichen bestehen, wäre das aber dreimal so viel Speicher wie nötig. (Eine solche Codierung mit fester Länge gibt es tatsächlich: UTF-32. Sie benutzt sogar 4 Byte pro Zeichen, weil Computer besser mit Einheiten von 1, 2, 4 oder 8 Byte arbeiten als mit 3 Byte.)

**UTF-8** (1992) ist die am weitesten verbreitete Codierung für Unicode-Zeichen. [Numberphile](https://www.youtube.com/watch?v=MijmeoH9LT4)

- Zeichen bis 127 (ASCII) werden mit **1 Byte** codiert, genau wie in ASCII.
- Zeichen größer als 127 werden zu Byte-Ketten der Länge **2 bis 4** codiert.

##### Die UTF-8-Schablonen

| Code Point (hex) | Bytes | Schablone (binär) | freie Bits `x` |
|:--|:-:|:--|:-:|
| `0` bis `7F` | 1 | `0xxxxxxx` | 7 |
| `80` bis `7FF` | 2 | `110xxxxx 10xxxxxx` | 11 |
| `800` bis `FFFF` | 3 | `1110xxxx 10xxxxxx 10xxxxxx` | 16 |
| `10000` bis `10FFFF` | 4 | `11110xxx 10xxxxxx 10xxxxxx 10xxxxxx` | 21 |

- Das **erste Byte** zeigt an, wie viele Bytes zum Zeichen gehören: Die Anzahl der Einsen vor der ersten 0 ist die Anzahl der Bytes. Bei 1 Byte beginnt es mit `0`.
- Alle **weiteren Bytes** beginnen mit `10`.
- In die `x` werden die Bits des Code Points eingesetzt, **von rechts nach links**. Links wird mit Nullen aufgefüllt.

##### So wird ein Zeichen in UTF-8 codiert

1. Code Point hexadezimal bestimmen.
2. In der Tabelle die passende Schablone suchen.
3. Den Code Point binär schreiben, links mit Nullen auf die Anzahl der freien Bits auffüllen.
4. Die Bits von rechts nach links in die `x` der Schablone einsetzen.
5. Das Ergebnis in Vierergruppen hexadezimal schreiben.

Beispiel: Die UTF-8 Codierung von €

<img src='./unicode.png' width='800'>

```
Code Point          : 20AC  (liegt zwischen 800 und FFFF -> 3 Byte)
binär (16 Bit)      : 0010 0000 1010 1100
Schablone           : 1110xxxx 10xxxxxx 10xxxxxx
eingesetzt          : 1110 0010 1000 0010 1010 1100
UTF-8 Codierung     : E2 82 AC
```

Beispiel mit 2 Byte: das Zeichen `ö`

```
Code Point          : F6  (liegt zwischen 80 und 7FF -> 2 Byte)
binär (11 Bit)      : 000 1111 0110
Schablone           : 110xxxxx 10xxxxxx
eingesetzt          : 110 00011 10 110110  =  1100 0011 1011 0110
UTF-8 Codierung     : C3 B6
```

##### Zeichensalat

Wird ein UTF-8-Text so gelesen, als wäre er in ISO-8859-1 geschrieben, wird jedes Byte als eigenes Zeichen angezeigt. Aus `ö` (`C3 B6`) werden dann die zwei Zeichen `Ã` (`C3`) und `¶` (`B6`). Solchen Zeichensalat sieht man manchmal auf Webseiten oder in E-Mails.

````python
text = 'schön'
falsch = text.encode('utf-8').decode('latin-1')
print(falsch)
````
