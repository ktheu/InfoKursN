### SelectionSort

Idee: Das jeweils kleinste Element tauscht mit dem aktuellen Element die Plätze.

<img src='selectionSort.png' width='400'>

```python
def selection_sort(a):
    for i in range(len(a)-1):
        best = i                    # Index des kleinsten Elements
        best_val = a[i]
        for j in range(i+1,len(a)):
            if a[j] < best_val:
                best = j
                best_val = a[j]
        a[best],a[i] = a[i],a[best]
```

```python
a = [15, 23, 4, 42, 8, 16]
selection_sort(a)
print(a)
```

Mit einer Ausgabe nach jedem Durchgang der äußeren Schleife sehen wir, wie vorne der sortierte Teil wächst.

```python
def selection_sort_durchgaenge(a):
    for i in range(len(a)-1):
        best = i
        best_val = a[i]
        for j in range(i+1,len(a)):
            if a[j] < best_val:
                best = j
                best_val = a[j]
        a[best],a[i] = a[i],a[best]
        print(*a)

selection_sort_durchgaenge([15, 23, 4, 42, 8, 16])
```

#### Analyse von SelectionSort

Der Algorithmus wird nicht schneller, wenn die Daten schon sortiert sind. <br>
Anzahl Vergleiche bei den zwei ineinander geschachtelten for-Schleifen:   

$(n-1) +  (n-2) + (n-3) + ... + 1 =  \frac{n \cdot (n-1)}{2} =\frac{n^2 - n}{2} \in  O(n^2)$ 

worst case  = best case  = average case: $O(n^2)$   <br>
Zusätzlicher Platzbedarf: $O(1)$

### BubbleSort

Idee: Falls der Nachfolger kleiner ist, werden die Plätze getauscht. Oder vertikal: große Blasen steigen auf.

<img src='bubbleSort.png' width='501'>

```python
def bubble_sort(a):
    getauscht = True
    while getauscht:
        getauscht = False
        for i in range(len(a)-1):
            if a[i] > a[i+1]:
                a[i],a[i+1]=a[i+1],a[i]
                getauscht = True
```

```python
a = [15, 23, 4, 42, 8, 16]
bubble_sort(a)
print(a)
```

Ein Durchgang ist ein Durchlauf der for-Schleife. Nach jedem Durchgang steht mindestens ein weiteres großes Element
hinten an seinem endgültigen Platz.

```python
def bubble_sort_durchgaenge(a):
    getauscht = True
    while getauscht:
        getauscht = False
        for i in range(len(a)-1):
            if a[i] > a[i+1]:
                a[i],a[i+1]=a[i+1],a[i]
                getauscht = True
        print(*a)

bubble_sort_durchgaenge([15, 23, 4, 42, 8, 16])
```

Im letzten Durchgang wird nichts mehr getauscht, deshalb endet die while-Schleife.

#### Analyse von Bubble Sort

Best case:  $O(n)$  <br>
Worst case: (wenn die Folge umgekehrt sortiert ist) $O(n^2)$ <br>
Average case: $O(n^2)$

Weitere Verbesserungen möglich: Shaker Sort, aber es bleibt bei  $O(n^2)$. <br>
Grund: die Austauschpositionen liegen zu nahe beieinander.

Zusätzlicher Platzbedarf: $O(1)$ <br>

### InsertionSort

Idee: das aktuelle Element wird vorne richtig eingefügt.

<img src='insertionSort.png' width='401'>

```python
def insertion_sort(a):
    for i in range(1,len(a)):
        x = a[i]
        j = i
        while j > 0 and a[j-1] > x:
            a[j] = a[j-1]
            j = j - 1
        a[j] = x

a = [15, 23, 4, 42, 8, 16]
insertion_sort(a)
print(a)
```

Nach jedem Durchgang ist der vordere Teil bis zum Index `i` sortiert.

```python
def insertion_sort_durchgaenge(a):
    for i in range(1,len(a)):
        x = a[i]
        j = i
        while j > 0 and a[j-1] > x:
            a[j] = a[j-1]
            j = j - 1
        a[j] = x
        print(*a)

insertion_sort_durchgaenge([15, 23, 4, 42, 8, 16])
```

#### Analyse von InsertionSort

Best case:  $O(n)$ <br>
Worst case: (wenn die Folge umgekehrt sortiert ist) $O(n^2)$ <br>
Average case: $O(n^2)$ <br>
Zusätzlicher Platzbedarf: $O(1)$

Ist die Liste schon fast sortiert, muss jedes Element nur wenige Plätze nach vorne wandern.
Dann ist InsertionSort sehr schnell, fast $O(n)$.

### MergeSort
Idee (rekursiv formuliert): 

- Sortiere die vordere Hälfte der Liste  <br>
- Sortiere die hintere Hälfte der Liste <br>
- Mische die beiden sortierten Listen zu einer sortierten Liste

Diese Vorgehensweise wird *Divide and Conquer* genannt. Das ursprüngliche Problem wird in 
unabhängige Teilprobleme zerlegt, danach werden die Teillösungen zur Gesamtlösung zusammengefügt.

#### Der Merge-Algorithmus:

Zwei sortierte Listen werden zu einer sortierten Gesamtliste zusammengemischt. 

<img src='merge.png' width='800'>

```python
def merge(a, b):
    i = j = 0
    c=[]
    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            c.append(a[i])
            i+=1
        else:
            c.append(b[j])
            j+=1
    if i == len(a):
        c.extend(b[j:])
    else:
        c.extend(a[i:])
    return c
```

```python
a = [7,13,15,18]
b = [2,4,19,22]
merge(a,b)
```

Der Ablauf von MergeSort: Die zu sortierende Liste a wird in zwei Hälften b und c geteilt, die werden rekursiv mit mergeSort sortiert und dann mit merge zusammengefügt.

<img src='mergeSort.png' width='600'>

```python
def merge_sort(a):
    if len(a) <= 1: return a
    halb = len(a)//2
    b = a[:halb]
    c = a[halb:]
    return merge(merge_sort(b),merge_sort(c))
```

```python
a = [17,3,23,4,1,9,11]
merge_sort(a)
```

#### Übung

c .Bei jedem Aufruf von *merge* wird eine Liste zurückgegeben.

Mit Ausgaben und Zählern können wir die Lösung der Übung überprüfen.

```python
def merge_zaehlen(a, b):
    global anzahl_merge
    anzahl_merge += 1
    c = merge(a, b)
    print(*c)
    return c

def merge_sort_zaehlen(a):
    global anzahl_merge_sort
    anzahl_merge_sort += 1
    if len(a) <= 1: return a
    halb = len(a)//2
    b = a[:halb]
    c = a[halb:]
    return merge_zaehlen(merge_sort_zaehlen(b),merge_sort_zaehlen(c))

anzahl_merge = 0
anzahl_merge_sort = 0
merge_sort_zaehlen([17,3,23,4,1,9,11])
print('merge:', anzahl_merge, ' merge_sort:', anzahl_merge_sort)
```

Aufwand pro Ebene: $O(n)  \Rightarrow$ Laufzeit:  $O(n \cdot \log n)$

zusätzlicher Platzbedarf:  $O(n)$

### QuickSort

Idee: teile die Folge in eine elementweise kleinere und eine elementweise größere Hälfte und sortiere diese
nach demselben Verfahren. 

<img src='quicksort05.png' width='500'>

Beispiel:

<img src='quicksort.png' width='600'>

```python
def quick_sort(a, unten = 0, oben = None):
    if oben is None: oben = len(a)-1
    i , j = unten, oben
    mitte = (unten + oben) // 2
    pivot = a[mitte]
    while i <= j:
        while a[i] < pivot: i+=1
        while a[j] > pivot: j-=1
        if i <= j:
            a[i], a[j] = a[j], a[i]
            i+=1
            j-=1
    if unten < j: quick_sort(a,unten,j)
    if i < oben: quick_sort(a,i,oben)
```

```python
a = [15,26,22,18,16,28,9,38,8]
quick_sort(a)
a
```

#### Protokoll von QuickSort

Zu Beginn jedes Aufrufs schreiben wir eine Protokollzeile: die Liste, den Bereich von `unten` bis `oben` und das
Pivot-Element. Die nächste Zeile zeigt die Liste nach dem Aufteilen und den Bereich, der als nächstes sortiert wird.

```python
def quick_sort_protokoll(a, unten = 0, oben = None):
    if oben is None: oben = len(a)-1
    i , j = unten, oben
    mitte = (unten + oben) // 2
    pivot = a[mitte]
    print(*a, f'unten={unten} oben={oben} pivot={pivot}')
    while i <= j:
        while a[i] < pivot: i+=1
        while a[j] > pivot: j-=1
        if i <= j:
            a[i], a[j] = a[j], a[i]
            i+=1
            j-=1
    if unten < j: quick_sort_protokoll(a,unten,j)
    if i < oben: quick_sort_protokoll(a,i,oben)

quick_sort_protokoll([15,26,22,18,16,28,9,38,8])
```

Erst wird der linke Teil vollständig sortiert, der rechte Teil wartet so lange. Im Beispiel wartet der Bereich
von 4 bis 8, bis der Bereich von 0 bis 3 fertig ist.

#### Analyse von QuickSort

best case: Pivot-Element so, dass ungefähr gleich große Hälften entstehen.  Dann gibt es $\log n$ Rekursionsebenen. 
Pro Ebene muss einmal durch die Liste gelaufen werden,  also insgesamt:  $O(n \cdot \log n)$

worst case: Pivot-Element so, dass ein Teil immer nur aus einem Element besteht.  
$n + (n-1) + (n-2) + (n-3) + ... 1 \in O(n^2)$

average case: $O(n \cdot \log n)$.

Zusätzlicher Platz: $O(\log n)$ 
(nicht $O(1)$, denn in jeder der Rekursionsebenen benötigt man eine konstante Anzahl Variablen)

#### Vergleich

| | best case | average case | worst case | zusätzlicher Platz |
|---|---|---|---|---|
| SelectionSort | $O(n^2)$ | $O(n^2)$ | $O(n^2)$ | $O(1)$ |
| BubbleSort | $O(n)$ | $O(n^2)$ | $O(n^2)$ | $O(1)$ |
| InsertionSort | $O(n)$ | $O(n^2)$ | $O(n^2)$ | $O(1)$ |
| MergeSort | $O(n \cdot \log n)$ | $O(n \cdot \log n)$ | $O(n \cdot \log n)$ | $O(n)$ |
| QuickSort | $O(n \cdot \log n)$ | $O(n \cdot \log n)$ | $O(n^2)$ | $O(\log n)$ |



```python
import random
import time

a = [random.randint(0, 10**6) for _ in range(5000)]
for sortieren in [selection_sort, bubble_sort, insertion_sort, merge_sort, quick_sort]:
    b = a[:]
    start = time.perf_counter()
    sortieren(b)
    print(f'{sortieren.__name__:15} {time.perf_counter() - start:.3f} s')
```

In der Praxis sortiert man in Python mit `sorted(a)` oder `a.sort()`. Das Verfahren dahinter (Timsort) ist eine
Mischung aus MergeSort und InsertionSort und liegt in $O(n \cdot \log n)$.
