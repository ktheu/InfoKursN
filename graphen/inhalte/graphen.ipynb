## 1. Was ist ein Graph?

Ein **Graph** besteht aus **Knoten** (vertices) und **Kanten** (edges). Er kann gerichtet oder ungerichtet sein. Die Kanten
können **gewichtet** sein (Kosten).

<img src='graphen_01.png'>
<img src='graphen_02.png'>

Mit Graphen können binäre Beziehungen zwischen Objekten modelliert werden. Die Objekte werden
durch die Knoten, die Beziehungen durch die Kanten modelliert.

* Orte: Entfernungen, Dauer
* Personen: ist befreundet, verheiratet mit
* Ereignisse: a muss vor b geschehen

## 2. Begriffe

<img src='graphen40.png' width='600'>

* x ist zu y und z adjazent, x hat die Nachbarn y und z
* a hat die Nachbarn b, c, d
* der Grad von x ist 2 (Knotengrad: Anzahl der Kanten an einem Knoten)
* der Eingangsgrad von a ist 1, der Ausgangsgrad ist 3
* a ist Vorgänger von c, c ist Nachfolger von a
* Ein Weg ist eine Folge von adjazenten Knoten, z.B: y-x-z oder a-c-d-b
* Ein Kreis (Zyklus) ist ein Weg, der zurück zum Startknoten führt, z.B. a-c-d-b-a

Ein Graph, der einen Kreis enthält, heißt **zyklisch**. Ein Graph ohne Kreis heißt **azyklisch**. Der linke Graph ist azyklisch, der rechte zyklisch.

| Eigenschaft | Bedeutung |
|---|---|
| gerichtet / ungerichtet | Die Kanten haben eine Richtung / keine Richtung. |
| gewichtet / ungewichtet | Die Kanten haben Kosten / keine Kosten. |
| zyklisch / azyklisch | Der Graph enthält einen Kreis / keinen Kreis. |

## 3. Adjazenzmatrix

Eine Implementation sollte zwei Dinge zügig können:
* Die Frage klären, ob es eine Kante von Knoten a nach Knoten b gibt,
* Alle Nachbarn von a durchlaufen.

#### Adjazenzmatrix eines ungewichteten Graphen

Wir ordnen jedem Knoten einen Index zu. Die erste Reihe der Adjazenzmatrix gibt die Nachbarn des Knoten mit Index 0 an.

<img src='graphen_05.png'>

```python
V = ['a','b','c','d']    # Liste mit den Knoten

G = [[0, 1, 1, 0],
     [0, 0, 0, 1],
     [0, 1, 0, 1],
     [0, 1, 0, 0]]
```

```python
# Gibt es Kante von a nach b?
G[0][1] == 1
```

```python
# alle Nachbarn von a
[j for j in range(len(G)) if G[0][j] == 1]
```

Bei einem ungerichteten Graphen gehört zu jeder Kante von i nach j auch eine Kante von j nach i, die Matrix ist dann symmetrisch.

#### Adjazenzmatrix eines gewichteten Graphen

<img src='graphen_07.png'>

Nicht vorhandene Kanten haben unendliche Kosten.

```python
inf = float('inf')
G = [[0,   8,   2, inf],
     [inf, 0, inf,   4],
     [inf, 1,   0,   6],
     [inf, 2, inf,   0]]
```

```python
# Kosten von a nach b
G[0][1]
```

```python
# alle Nachbarn von a
[j for j in range(len(G)) if j!=0 and G[0][j] < inf]
```

## 4. Adjazenzlisten

Jedem Knoten wird eine Liste seiner Nachbarn zugeordnet. Wir werden statt Listen *sets* oder *dictionaries* verwenden.

Ein ungewichteter Graph wird durch ein dictionary implementiert, mit den Knoten als key und der Nachbarmenge als value.

<img src='graphen_10.png'>

```python
G = {'a':set('bc'),
     'b':set('d'),
     'c':set('bd'),
     'd':set('b')}
```

```python
# gibt es Kante von a nach b
'b' in G['a']
```

```python
# alle Nachbarn von a:
G['a']
```

```python
# alle Knoten von G durchlaufen:
for v in G:
    print(v)
```

Ein gewichteter Graph wird durch ein dictionary implementiert, mit den Knoten als key und einem dictionary, in dem die Kosten zu den Nachbarn verzeichnet sind, als value.

<img src='graphen_11.png'>

```python
G = {'a': {'b':8, 'c':2},
     'b': {'d':4},
     'c': {'b':1, 'd':6},
     'd': {'b':2}}
```

```python
# gibt es Kante von a nach b
'b' in G['a']
```

```python
# die Kosten der Kante von a nach b:
G['a']['b']
```

```python
# alle Nachbarn von a
[v for v in G['a']]
```

## 5. Vergleich der beiden Implementationen

$V$ = Knoten (vertices), $E$ = Kanten (edges). Ein Graph heißt **dünn besetzt**, wenn er viel weniger Kanten hat, als möglich wären.

| | Adjazenzmatrix | Adjazenzlisten |
|---|---|---|
| Platzbedarf | $O(\left|V\right|^2)$ | $O(\left|V\right|+\left|E\right|)$ |
| gibt es eine Kante von i nach j? | direkter Zugriff in konstanter Zeit | in konstanter Zeit nur mit sets oder dictionaries |
| alle Nachbarn eines Knotens | $O(\left|V\right|)$, bei dünn besetzten Graphen eher schlecht | nur so viele Schritte, wie der Knoten Nachbarn hat |
| sinnvoll | bei dicht besetzten Graphen | bei eher dünn besetzten Graphen und bei Algorithmen, die die Nachbarn eines Knotens verarbeiten |

## 6. Der Algorithmus von Dijkstra

Der Algorithmus von Dijkstra findet in einem gerichteten, mit nichtnegativen Kosten gewichteten
Graphen die kürzesten Wege von einem Startknoten zu allen anderen Knoten. Er löst das
**single source shortest path**-Problem. Das braucht zum Beispiel ein Navigationssystem.

<img src='graphen_26.png'>

Zu jedem Knoten merken wir uns die bisher beste Entfernung `dist` und den Vorgänger `prev` auf diesem Weg. Die Werte sind zunächst vorläufig.

    # Algorithmus von Dijkstra
    Setze dist des Startknotens s vorläufig auf 0,
        alle anderen vorläufig auf unendlich.
    Setze prev des Startknotens auf None
    Solange es noch vorläufige Knoten gibt
        setze dist des billigsten vorläufigen Knotens u auf endgültig
        Für jeden Nachbarn v von u:
            Falls (dist(u) + Kosten von u nach v) < dist(v):
                dist(v) =  dist(u) + Kosten von u nach v
                Setze Vorgänger von v auf u

#### Relaxieren

Die Kante (u,v) *relaxieren* bedeutet: Prüfen, ob man über diese Kante besser zum Ziel v kommt und ggf. updaten.

#### Beispiel

Wir starten bei a. In jeder Zeile steht `dist` nach dem Schritt, endgültige Werte sind fett.

| Schritt | endgültig | a | b | c | d | e | f |
|---|---|---|---|---|---|---|---|
| Start |  | 0 | ∞ | ∞ | ∞ | ∞ | ∞ |
| 1 | a | **0** | 2 | 9 | ∞ | ∞ | ∞ |
| 2 | b | **0** | **2** | 7 | 15 | ∞ | ∞ |
| 3 | c | **0** | **2** | **7** | 13 | 16 | ∞ |
| 4 | d | **0** | **2** | **7** | **13** | 14 | 18 |
| 5 | e | **0** | **2** | **7** | **13** | **14** | 16 |
| 6 | f | **0** | **2** | **7** | **13** | **14** | **16** |

In der Zeichnung sind die Werte an den Knoten notiert, überholte Werte sind durchgestrichen. Die roten Pfeile zeigen auf den Vorgänger `prev`.

<img src='graph51.png' width='500'>

Den kürzesten Weg lesen wir rückwärts ab: f hat den Vorgänger e, e hat den Vorgänger d, d hat den Vorgänger c, c hat den Vorgänger b und b den Vorgänger a.
Der kürzeste Weg von a nach f ist also a-b-c-d-e-f mit der Länge 16.

#### Warum funktioniert das?

Der billigste vorläufige Knoten u kann endgültig werden, weil jeder andere Weg zu u über einen vorläufigen Knoten führt, der mindestens genauso teuer
ist. Durch weitere Kanten wird der Weg nicht kürzer. Das gilt nur, wenn es keine negativen Kosten gibt.

Für alle Knoten wird einmal der billigste vorläufige Knoten gesucht. Suchen wir ihn in einer Liste, brauchen wir $O(\left|V\right|^2)$ Schritte.

#### Zusammenfassung

| | |
|---|---|
| Graph | besteht aus Knoten und Kanten, modelliert binäre Beziehungen |
| adjazent, Nachbar | durch eine Kante verbunden |
| Grad | Anzahl der Kanten an einem Knoten, im gerichteten Graphen Eingangsgrad und Ausgangsgrad |
| Weg | Folge von adjazenten Knoten |
| Kreis, Zyklus | Weg, der zurück zum Startknoten führt |
| zyklisch, azyklisch | Graph mit Kreis, Graph ohne Kreis |
| Adjazenzmatrix | `G[i][j]` gibt an, ob es eine Kante gibt oder welche Kosten sie hat |
| Adjazenzlisten | zu jedem Knoten seine Nachbarn, `{'a': set('bc')}` oder `{'a': {'b':8}}` |
| Dijkstra | kürzeste Wege von einem Startknoten bei nichtnegativen Kosten |
| relaxieren | prüfen, ob es über diese Kante billiger wird, und ggf. updaten |
