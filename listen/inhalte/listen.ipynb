## LI 1 · Datentypen und Typumwandlung

### Inhalte

Jeder Wert und damit jede Variable hat einen **Datentyp**. Wir kennen bisher:

| Datentyp | In Python | Beispiele |
|:--|:--|:--|
| Ganzzahl | `int` | `5`, `-3`, `0` |
| Gleitkommazahl | `float` | `2.5`, `-0.1`, `3.0` |
| Zeichenkette | `str` | `'Hallo'`, `'5'`, `''` |
| Wahrheitswert | `bool` | `True`, `False` |

Mit der Funktion `type()` können wir den Datentyp eines Wertes ausgeben.

````python
k = 5
x = 2.5
s = 'Hallo'
b = True
print(type(k))     # <class 'int'>
print(type(x))     # <class 'float'>
print(type(s))     # <class 'str'>
print(type(b))     # <class 'bool'>
````

##### Typumwandlung

Wenn es sinnvoll möglich ist, können wir Datentypen **umwandeln**. Das brauchen wir zum Beispiel bei `input()`, das immer einen String liefert.

| Funktion | wandelt um in | Beispiel | Ergebnis |
|:--|:--|:--|:--|
| `int()` | Ganzzahl | `int('5')` | `5` |
| `float()` | Gleitkommazahl | `float('3.14')` | `3.14` |
| `str()` | String | `str(42)` | `'42'` |

````python
s = '5'                   # str
x = int(s)                # str -> int
print(x, type(x))         # 5 <class 'int'>
````

````python
s = '3.14'                # str
x = float(s)              # str -> float
print(x, type(x))         # 3.14 <class 'float'>
````

````python
n = 42                    # int
s = str(n)                # int -> str
print(s + '!', type(s))   # 42! <class 'str'>
````

Ist die Umwandlung nicht möglich, bricht das Programm mit einem **ValueError** ab, zum Beispiel bei `int('Hallo')` oder `int('3.14')`. Wandelt man dagegen eine Gleitkommazahl mit `int()` um, werden die Nachkommastellen **abgeschnitten**: `int(3.9)` ergibt `3`.

## LI 2 · Listen

### Inhalte

In einer **Liste** speichern wir mehrere Werte unter einem Namen. Die Werte heißen **Elemente**. Wir schreiben sie in eckigen Klammern, durch Kommas getrennt. `len()` liefert die Anzahl der Elemente.

````python
a = [1, 2, 3]                                # Liste mit 3 ints
print(a, len(a))                             # [1, 2, 3] 3
````

````python
a = []                                       # leere Liste
print(a, len(a))                             # [] 0
````

````python
a = ['Montag', 'Dienstag', 'Mittwoch']       # Liste mit Strings
print(a, len(a))                             # ['Montag', 'Dienstag', 'Mittwoch'] 3
````

Die Liste ist ein eigener Datentyp:

````python
a = [3, 5, 8]
print(type(a))     # <class 'list'>
````

##### Indexing

Auf die Elemente greifen wir mit einem **Index** zu, genau wie bei den Zeichen eines Strings: von vorne gezählt beginnend bei `0`, von hinten gezählt beginnend bei `-1`.

**Sprechweise:** Zu `a[4]` sagen wir „**a an der Stelle 4**“. In der Liste im Bild ist `a[4]` das Element `0`.

<img src="./liste1.png" width="350" alt="Die Liste [7, 12, 1, 5, 0, 3, 5] mit den Indizes 0 bis 6 und -7 bis -1">

````python
a = [7, 12, 1, 5, 0, 3, 5]
x1 = a[0]          # erstes Element
x2 = a[1]          # zweites Element
x3 = a[len(a)-1]   # letztes Element
x4 = a[-1]         # letztes Element
x5 = a[-2]         # vorletztes Element
print(x1, x2, x3, x4, x5)    # 7 12 5 5 3
````

Auch **Slicing** funktioniert wie bei Strings: `a[1:3]` ergibt `[12, 1]`.

##### Elemente ändern

Anders als bei einem String können wir die Elemente einer Liste über den Index **verändern**:

````python
a = [7, 12, 1, 5, 0, 3, 5]
a[0] = 9
print(a)           # [9, 12, 1, 5, 0, 3, 5]
````

##### Elemente hinzufügen

Mit der Methode `append()` hängen wir ein Element **am Ende** der Liste an. Die Liste wird dadurch um ein Element länger.

````python
a = [3, 4, 5, 6, 7]
a.append(8)
print(a)           # [3, 4, 5, 6, 7, 8]
````

Häufig beginnen wir mit einer leeren Liste und füllen sie in einer Schleife:

````python
a = []
for i in range(5):
    a.append(i)
print(a)           # [0, 1, 2, 3, 4]
````

## LI 3 · Schleifen über Strings und Listen

### Inhalte

Es gibt zwei Arten, die Zeichen eines Strings oder die Elemente einer Liste nacheinander zu durchlaufen.

##### Laufvariable über den Index

Die Laufvariable `i` durchläuft die **Indizes** `0, 1, …, len(s)-1`. Das Zeichen bzw. Element erhalten wir dann mit `s[i]` bzw. `a[i]`.

````python
# Die Laufvariable i geht über die Indizes des Strings
s = 'Habicht'
for i in range(len(s)):
    print(i, s[i])
````

````python
# Die Laufvariable i geht über die Indizes der Liste
a = [3, 4, 7, 2, 8]
for i in range(len(a)):
    print(i, a[i])
````

##### Laufvariable über die Zeichen bzw. Elemente

Die Laufvariable nimmt direkt nacheinander die **Zeichen** bzw. **Elemente** an. Das ist kürzer, wir kennen dann aber den Index nicht.

````python
# Die Laufvariable c geht über die Zeichen des Strings
s = 'Habicht'
for c in s:
    print(c)
````

````python
# Die Laufvariable k geht über die Elemente der Liste
a = [3, 4, 7, 2, 8]
for k in a:
    print(k)
````

**Welche Art nehmen wir?** Wenn wir nur die Werte lesen wollen, ist die Schleife über die Elemente einfacher. Wenn wir den Index brauchen, etwa um **Elemente der Liste zu verändern**, nehmen wir die Schleife über den Index.

### Aufgabe 1

1. Berechne mit einer Schleife die Summe aller Elemente der Liste `zahlen = [3, 8, 2, 7]`.
2. Zähle mit einer Schleife, wie oft der Buchstabe `e` in `'Erdbeermarmelade'` vorkommt. Auch das große `E` soll mitzählen.

#### Lösung

**1.** Wir brauchen nur die Werte, also genügt die Schleife über die Elemente. `summe` wird **vor** der Schleife auf 0 gesetzt.

````python
zahlen = [3, 8, 2, 7]
summe = 0
for k in zahlen:
    summe += k
print(summe)       # 20
````

**2.** Mit `lower()` wird aus dem großen `E` ein kleines, dann reicht ein Vergleich:

````python
wort = 'Erdbeermarmelade'
anzahl = 0
for c in wort.lower():
    if c == 'e':
        anzahl += 1
print(anzahl)      # 5
````

### Aufgabe 2

Alle Elemente der Liste `a = [1, 5, 3]` sollen verdoppelt werden. Jemand schreibt:

````python
a = [1, 5, 3]
for k in a:
    k = 2 * k
print(a)
````

1. Was wird ausgegeben? Warum ändert sich die Liste nicht?
2. Korrigiere das Programm.

#### Lösung

**1.** Ausgabe: `[1, 5, 3]`. Die Laufvariable `k` erhält nacheinander eine **Kopie** der Werte 1, 5 und 3. Mit `k = 2 * k` wird nur `k` verändert, nicht das Element in der Liste.

**2.** Um ein Element zu verändern, brauchen wir seinen Index:

## LI 4 · Zufallszahlen

### Inhalte

Das Modul `random` erzeugt Zufallszahlen. Wir laden es einmal mit `import random`, danach rufen wir seine Funktionen mit `random.` davor auf.

| Funktion | liefert | Beispiel |
|:--|:--|:--|
| `random.random()` | Dezimalzahl x mit 0 ≤ x < 1 | `0.7664…` |
| `random.randint(a, b)` | ganze Zahl n mit a ≤ n ≤ b (**beide Grenzen** eingeschlossen) | `random.randint(1, 6)` |
| `random.uniform(a, b)` | Dezimalzahl x mit a ≤ x ≤ b | `random.uniform(2, 10)` |

Achtung: Anders als bei `range()` ist bei `randint()` die obere Grenze **eingeschlossen**.

````python
import random
for i in range(5):
    x = random.random()       # 0 <= x < 1
    print(x)
````

````python
print(random.randint(0, 10))       # ganze Zahl n:     0 <= n <= 10
print(random.random())             # Dezimalzahl x:    0 <= x < 1
print(random.uniform(2, 10))       # Dezimalzahl x:    2 <= x <= 10
````

##### Pseudozufall und seed

Die Zahlen sind **Pseudozufallszahlen**: Sie werden von einem Rechenverfahren erzeugt. Sie sehen zufällig aus, aber nach 2<sup>19937</sup> − 1 Zahlen wiederholt sich die Folge.

Mit der Funktion `seed()` legen wir den **Startwert** des Zufallsgenerators fest. Bei gleichem seed erzeugt er dann bei jedem Programmlauf **dieselbe Zahlenfolge**. Das ist praktisch beim Testen: Ein Fehler, der bei bestimmten Zufallszahlen auftritt, lässt sich so wiederholen.

````python
random.seed(12)
for i in range(5):
    x = random.random()
    print(x)           # bei jedem Lauf dieselben fünf Zahlen
````

### Aufgabe 1

Schreibe ein Programm, das 10-mal würfelt, jeden Wurf ausgibt und am Ende ausgibt, wie oft eine 6 gefallen ist.

#### Lösung

````python
import random
anzahl_sechsen = 0
for i in range(10):
    wurf = random.randint(1, 6)
    print(wurf)
    if wurf == 6:
        anzahl_sechsen += 1
print(f'Es wurde {anzahl_sechsen}-mal eine 6 gewürfelt.')
````

`random.randint(1, 6)` liefert die Zahlen 1 bis 6, **beide** Grenzen sind eingeschlossen. Mit `range(1, 6)` wäre es anders, dort wäre die 6 nicht dabei.

## LI 5 · Funktionen

### Inhalte

In **Funktionen** fassen wir Codeabschnitte zu einer Einheit zusammen. Eine Funktion schreiben wir einmal und können sie dann beliebig oft **aufrufen**.

<img src="./funktionen.png" width="400" alt="Die Funktion calc mit den Bezeichnungen Schlüsselwort, Name, Parameter, Kopf, Rumpf, Argumente und Aufruf der Funktion">

| Begriff | Bedeutung | im Beispiel |
|:--|:--|:--|
| **Schlüsselwort** | leitet die Funktionsdefinition ein | `def` |
| **Name** | unter diesem Namen wird die Funktion aufgerufen | `calc` |
| **Parameter** | Variablen, die beim Aufruf die übergebenen Werte erhalten | `x`, `y` |
| **Kopf** | die erste Zeile, sie endet mit einem Doppelpunkt | `def calc(x, y):` |
| **Rumpf** | die eingerückten Anweisungen der Funktion | die drei Zeilen darunter |
| **Rückgabewert** | der Wert nach `return`, er ist das Ergebnis des Aufrufs | `z` |
| **Argumente** | die konkreten Werte beim Aufruf | `2`, `4` |

Mit `return` gibt die Funktion ihr Ergebnis zurück, und die Funktion ist sofort beendet.

````python
def calc(x, y):
    z = x + y
    z = 3*z + 1
    return z
````

Die Definition allein bewirkt noch nichts. Erst beim **Aufruf** wird der Rumpf ausgeführt: Beim Aufruf `calc(3, 5)` erhält `x` den Wert 3 und `y` den Wert 5.

````python
a = calc(3, 5)
print(a)           # 25
print(calc(2, 4))  # 19
````

##### docstring

In einem **docstring** direkt unter dem Kopf beschreiben wir, wie die Funktion aufgerufen wird und was sie zurückgibt. Beispiele für Aufrufe mit ihren Ergebnissen machen die Beschreibung besonders verständlich.

````python
def div(x, y):
    '''
    x, y: zwei positive Zahlen
    returns: die kleinere dividiert durch die größere Zahl

    >>> div(5, 10)
    0.5

    >>> div(8, 2)
    0.25

    >>> div(4, 4)
    1.0
    '''
    if x < y:
        return x/y
    else:
        return y/x

print(div(5, 10), div(8, 2), div(4, 4))    # 0.5 0.25 1.0
````

Den docstring einer Funktion zeigt `help(div)` an.

## LI 6 · Dateien lesen

### Inhalte

Bevor wir aus einer Datei lesen oder in sie schreiben, **öffnen** wir sie mit `open()`. Nach der Verarbeitung **schließen** wir sie mit `close()`.

- Mit `readline()` lesen wir **eine Zeile**. Mit dem nächsten `readline()` wird die nächste Zeile gelesen.
- Der **Zeilenvorschub** am Ende der Zeile wird mitgelesen. Mit `strip()` entfernen wir ihn.
- `int()` wandelt die gelesene Zeile in eine ganze Zahl um und entfernt dabei den Zeilenvorschub automatisch.

Mit `%%writefile` in der ersten Zeile einer Zelle erzeugt Jupyter eine Datei mit dem Inhalt der Zelle. So legen wir uns Testdateien an.

````python
%%writefile input1.txt
Hallo
5
````

````python
f = open('input1.txt')
s = f.readline().strip()     # erste Zeile, ohne Zeilenvorschub
x = int(f.readline())        # zweite Zeile, als Zahl
f.close()
print(s)                     # Hallo
print(x + 1)                 # 6
````

##### Umlaute und encoding

Wenn wir bei Umlauten oder ß seltsame Zeichen sehen, hilft beim `open()` meist der Parameter `encoding='utf-8'`. Er gibt an, wie die Zeichen in der Datei gespeichert sind.

````python
%%writefile input2.txt
Daß üöä ÖÄÜ
````

````python
f = open('input2.txt', encoding='utf-8')
s = f.readline().strip()
f.close()
print(s)                     # Daß üöä ÖÄÜ
````

## LI 7 · Dateien schreiben

### Inhalte

Zum Schreiben öffnen wir die Datei mit dem Parameter `mode='w'` (*write*). Die Ausgaben schreiben wir mit `print()` und dem zusätzlichen Parameter `file=f` in die Datei statt auf den Bildschirm.

**Vorsicht:** Mit `mode='w'` wird eine vorhandene Datei **überschrieben**. Gibt es die Datei noch nicht, wird sie angelegt.

Erst mit `close()` ist sichergestellt, dass alles in der Datei gespeichert ist.

````python
'''
Die Datei output.txt bekommt folgenden Inhalt:
Die Summe von 5 und 10 ist 15.
Das Produkt von 5 und 10 ist 50.
'''
f = open('output.txt', encoding='utf-8', mode='w')
x = 5
y = 10
print(f'Die Summe von {x} und {y} ist {x+y}.', file=f)
print(f'Das Produkt von {x} und {y} ist {x*y}.', file=f)
f.close()
````

Zur Kontrolle lesen wir die Datei wieder ein:

````python
f = open('output.txt', encoding='utf-8')
print(f.readline().strip())
print(f.readline().strip())
f.close()
````

### Aufgabe 1

### Aufgabe 2

Schreibe ein Programm, das die drei Zahlen aus `zahlen.txt` (LI 6, Aufgabe 1) einliest und in die Datei `ergebnis.txt` die Zeile `Die Summe ist 28.` schreibt.

#### Lösung

````python
f = open('zahlen.txt')
summe = 0
for i in range(3):
    summe += int(f.readline())
f.close()

g = open('ergebnis.txt', encoding='utf-8', mode='w')
print(f'Die Summe ist {summe}.', file=g)
g.close()
````

Die Lesedatei und die Schreibdatei sind zwei verschiedene Dateien. Damit man sie nicht verwechselt, geben wir ihnen verschiedene Variablennamen, hier `f` und `g`.
