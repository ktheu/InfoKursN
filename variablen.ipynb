### Variablen, Datentypen, Deklaration, Initialisierung, Typumwandlung 


Eine **Variable** ist ein benannter Speicherplatz, in dem ein Wert abgelegt wird.
Wichtige Begriffe:

| Begriff | Bedeutung |
|---|---|
| **Bezeichner** | der Name der Variable (z. B. `alter`) |
| **Datentyp** | legt fest, welche Art von Werten gespeichert werden können (z. B. `int`, `str`) |
| **Wert** | der konkrete Inhalt der Variable |
| **Deklaration** | das "Ankündigen" einer Variable (in Python geschieht dies implizit bei der ersten Zuweisung) |
| **Initialisierung** | die erste Wertzuweisung an eine Variable |
| **Wertzuweisung** | das (erneute) Zuweisen eines Wertes mit `=` |
| **Gültigkeitsbereich (Scope)** | der Codebereich, in dem eine Variable bekannt und nutzbar ist |

Anders als z. B. Java benötigt Python **keine explizite Typangabe** – der Typ wird
zur Laufzeit aus dem zugewiesenen Wert abgeleitet ("dynamische Typisierung").
Man kann optional **Type Hints** verwenden, die von Python zur Laufzeit jedoch
nicht erzwungen werden.

```python
# Deklaration und Initialisierung in einem Schritt (typisch für Python)
alter = 17          # Bezeichner: alter, Datentyp: int (wird automatisch erkannt)
name = "Alex"        # Bezeichner: name, Datentyp: str

# Wertzuweisung (Änderung eines bereits initialisierten Wertes)
alter = 18

print(name, "ist", alter, "Jahre alt.")

# Optionale Type Hints (nur Dokumentation, keine Erzwingung durch Python)
groesse: float = 1.78
print("Groesse:", groesse)
```

#### Primitive Datentypen und ihr Wertebereich

Python kennt u. a. folgende primitive (einfache) Datentypen:

- `int` – ganzzahlige Werte (in Python **beliebig groß**, nur durch den Arbeitsspeicher begrenzt)
- `float` – Gleitkommazahlen (mit begrenzter Präzision)
- `bool` – Wahrheitswerte `True` / `False`

Im Unterschied zu Sprachen wie Java gibt es in Python **keine** festen Wertebereiche
wie `int8`, `int32` etc. – `int` wächst dynamisch. Bei `float` gibt es aber sehr wohl
Grenzen der **Genauigkeit**.

```python
# int: beliebig groß
grosse_zahl = 2 ** 100
print("2^100 =", grosse_zahl)
print("Typ:", type(grosse_zahl))

# float: Genauigkeitsprobleme durch die IEEE-754-Darstellung
ergebnis = 0.1 + 0.2
print("0.1 + 0.2 =", ergebnis)          # nicht exakt 0.3!
print("0.1 + 0.2 == 0.3 ->", ergebnis == 0.3)

# bool: Wahrheitswerte
ist_wach = True
print("Typ von ist_wach:", type(ist_wach))

# bool ist technisch eine Unterklasse von int (True == 1, False == 0)
print("True + True =", True + True)
```

#### Typumwandlungen (implizit und explizit)

- **Implizite Typumwandlung**: Python wandelt automatisch um, z. B. wenn ein `int`
  mit einem `float` verrechnet wird.
- **Explizite Typumwandlung (Type Cast)**: Wir wandeln den Typ bewusst selbst um,
  z. B. mit `int(...)`, `float(...)`, `str(...)`.

Bei expliziten Umwandlungen können **Probleme** auftreten, z. B. Informationsverlust
(`float` → `int` schneidet Nachkommastellen ab) oder Laufzeitfehler, wenn eine
Zeichenkette nicht in eine Zahl umgewandelt werden kann.

```python
# Implizite Typumwandlung
ganzzahl = 5
kommazahl = 2.5
summe = ganzzahl + kommazahl   # int wird automatisch zu float "erweitert"
print("Summe:", summe, "-> Typ:", type(summe))

# Explizite Typumwandlung (Type Cast)
text_zahl = "42"
zahl = int(text_zahl)          # str -> int
print("Umgewandelt:", zahl, type(zahl))

pi_gerundet = int(3.99)        # float -> int: Nachkommastellen werden ABGESCHNITTEN (nicht gerundet!)
print("int(3.99) =", pi_gerundet)

zahl_als_text = str(123)       # int -> str
print("Verkettet:", "Wert: " + zahl_als_text)
```
