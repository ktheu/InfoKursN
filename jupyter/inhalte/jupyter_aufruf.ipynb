## Jupyter Notebooks 

Hinweise zum Aufruf, zur Installation und zum Download von Notebooks zur Bearbeitung

#### Aufruf in der Schule

- Öffne mit Windows+E den Explorer und gehe auf dem H-Laufwerk in deinen Informatik Ordner.
- Im Explorer mit Strg-L in die Adresszeile springen und dort eingeben: jupyter lab
- Es öffnet sich ein Fenster, in dem der Jupyter-Server gestartet wird und der Browser zeigt die Jupyter-Oberfläche.
Das Fenster darf während der Sitzung nicht geschlossen werden.

#### Installation am heimischen PC

- Installiere Thonny (https://thonny.org/)
- Öffne Thonny, Ansicht: Dateien, navigiere zu deinem Informatik-Ordner.
- In Thonny: Werkzeuge, Verwalte Pakete, im Feld 'Suche im PyPI' eingeben: jupyterlab  (ohne Leerzeichen!). Dieses Paket installieren.

#### Aufruf am heimischen PC
- Thonny starten, Werkzeuge, System Terminal öffnen (das Terminal sollte in deinem Informatik-Ordner öffnen), dort eingeben: jupyter lab (mit Leerzeichen)
- Das System Terminal darf während der Jupyter Sitzung nicht geschlossen werden.

#### Download eines Jupyter-Notebooks zur Bearbeitung

- Auf der Github-Seite des Notebooks auf den Download-Button clicken
 
  <img src='bild1.png' width='400'>

- Auf den Download-Button des Browsers für den Download-Verlauf clicken und den Ordner des aktuellen Downloads anclicken.

  <img src='bild2.png' width='300'>

- In dem Download-Ordner das Notebook mit Strg-X ausschneiden und in den Informatik-Ordner mit Strg-C einfügen.
- Falls Jupyterlab schon im Informatik-Ordner geöffnet ist, auf den 'Refresh'-Button clicken. Dann mit Doppelklick das Notebook zur Bearbeitung öffnen.
