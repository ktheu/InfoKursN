### 1. Zellen ausführen

Ein Notebook besteht aus einzelnen **Zellen**. Es gibt zwei Arten:

- **Code-Zellen** enthalten Python-Code, den wir ausführen können. Sie sind grau hinterlegt.
- **Textzellen** (*Markdown-Zellen*) enthalten formatierten Text, wie diese Zelle hier.

Der blaue Balken links markiert die **aktive Zelle**. Um sie auszuführen, drückst du

- **Shift + Enter**: Die Zelle wird ausgeführt, und du springst zur nächsten Zelle.
- **Strg + Enter**: Die Zelle wird ausgeführt, und du bleibst in ihr stehen.

Die **Ausgabe** erscheint direkt unter der Code-Zelle.

**Probiere es aus:** Klicke in die erste Code-Zelle und führe beide Zellen nacheinander mit Shift + Enter aus.

````python
print('Montag')
````

### 2. Code ändern und Lücken füllen

Du kannst jede Code-Zelle ändern: hineinklicken, Code bearbeiten, erneut ausführen.

In Aufgaben findest du oft Code mit **Lücken**, die mit drei Unterstrichen `___` markiert sind. Dort setzt du deinen eigenen Code ein, der Rest ist schon vorgegeben.

**Probiere es aus:** Ersetze in der Zelle unten `___` durch deinen Namen in Hochkommata, z. B. `'Lena'`, und führe die Zelle aus.

**Kontrolle:** Die Ausgabe ist `Hallo Lena` (mit deinem Namen).

**Achtung:** Vergisst du eine Lücke, gibt es nicht immer eine Fehlermeldung. Manchmal kommt einfach ein falsches Ergebnis heraus. Prüfe deshalb vor dem Ausführen, ob du alle `___` ersetzt hast.

````python
name = 'Lena'
print('Hallo', name)
````

### 3. Fehlermeldungen lesen

Fehler passieren ständig, auch Profis. Dabei geht nichts kaputt. Wichtig ist, dass du die Fehlermeldung lesen kannst.

**Probiere es aus:** Die Zelle unten soll `Montag` ausgeben, enthält aber einen Fehler. Führe sie aus.

````python
print('Montag')
````

Fehlermeldungen sind oft lang, aber wichtig ist vor allem die **letzte Zeile**:

```
NameError: name 'Montag' is not defined
```

Sie sagt, **was** schiefgegangen ist: Ohne Hochkommata hält Python `Montag` für den Namen einer Variablen, und die gibt es nicht. Gemeint war der Text `'Montag'`.

**Jetzt du:** Korrigiere die Zelle oben und führe sie noch einmal aus.

> **Merke:** Lies bei einer Fehlermeldung zuerst die letzte Zeile. Dann korrigierst du den Code und führst die Zelle einfach erneut aus.

### 4. Die Ausführungsreihenfolge und der Kernel

Links neben einer Code-Zelle erscheint nach dem Ausführen eine Zahl in eckigen Klammern, z. B. `[3]`. Sie zeigt, in welcher **Reihenfolge du die Zellen ausgeführt hast**, nicht die Reihenfolge, in der sie im Notebook stehen. Solange eine Zelle noch läuft, steht dort `[*]`.

Variablen, die du in einer Zelle anlegst, bleiben gespeichert, auch für alle anderen Zellen. Gespeichert werden sie im **Kernel**, dem Programm, das im Hintergrund deinen Python-Code ausführt.

**Probiere es aus:**

1. Führe die beiden folgenden Zellen der Reihe nach aus. Die Ausgabe ist `Punkte: 10`.
2. Führe jetzt **nur die zweite Zelle** noch zweimal aus. Am Code ändert sich nichts, an der Ausgabe schon. Achte auch auf die Zahl in den eckigen Klammern.
3. Führe zum Schluss die erste Zelle aus und danach wieder die zweite. Überlege vorher, was herauskommt.

````python
punkte = 0
````

````python
punkte = punkte + 10
print('Punkte:', punkte)
````

Dieselbe Zelle liefert also je nach Vorgeschichte ein anderes Ergebnis.

**Kernel neu starten:** Manchmal möchte man ganz von vorne anfangen, zum Beispiel wenn sich etwas „verklemmt“ hat und auch nach dem Korrigieren nichts mehr funktioniert. Dann startest du den Kernel neu:

- Menü **Kernel → Restart Kernel and Clear Outputs of All Cells …** oder
- das **kreisförmige Pfeil-Symbol** in der Werkzeugleiste.

Dabei werden alle gespeicherten Variablen gelöscht. Dein Code im Notebook bleibt aber erhalten. Danach führst du die benötigten Zellen wieder von oben nach unten aus.

**Probiere es aus:** Starte den Kernel neu und führe dann **nur** die zweite Punkte-Zelle aus. Welche Fehlermeldung bekommst du, und warum?

<details>
<summary><b>Lösung</b></summary>

`NameError: name 'punkte' is not defined`. Durch den Neustart wurde die Variable `punkte` gelöscht. Die erste Zelle, die sie anlegt, wurde danach noch nicht wieder ausgeführt.
</details>

### 7. Aufgaben bearbeiten und Lösungen prüfen

In den Arbeits-Notebooks sind die Aufgaben immer gleich aufgebaut:

- **Aufgabe:** Was du tun sollst. Darunter steht eine Code-Zelle für deine Lösung, manchmal mit Lücken `___`.
- **Kontrolle:** Welches Ergebnis herauskommen muss. Vergleiche es genau mit deiner Ausgabe.
- **Tipp** und **Lösung** sind **eingeklappt**. Klicke auf das kleine Dreieck **▸** davor, um sie zu öffnen, und noch einmal, um sie zu schließen.

So gehst du vor:

1. Versuche die Aufgabe zuerst **allein**.
2. Vergleiche deine Ausgabe mit der **Kontrolle**.
3. Kommst du nicht weiter, öffne den **Tipp**.
4. Erst zum Schluss schaust du in die **Lösung**. Vergleiche sie mit deinem Code: Was hast du anders gemacht?

Manche Aufgaben heißen **Vorhersage**. Dort überlegst du zuerst, was passieren wird, und führst die Zelle erst danach aus.

**Probeaufgabe**

```
Setze p auf 6.
Setze q auf 9.
Erhöhe p um 3.
Erniedrige q um 4.
Gib in der ersten Zeile die Summe von p und q aus.
Gib in der nächsten Zeile das Produkt von p und q aus.
```

**Kontrolle:**
```
14
45
```

<details>
<summary><b>Tipp</b></summary>

„Erhöhe p um 3“ schreibt man in Python als `p = p + 3`.
</details>

<details>
<summary><b>Lösung</b></summary>

```python
p = 6
q = 9
p = p + 3
q = q - 4
print(p + q)
print(p * q)
```
</details>
