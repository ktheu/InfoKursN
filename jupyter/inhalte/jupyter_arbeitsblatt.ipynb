## Jupyter Notebooks

Jupyter Notebooks nutzen wir im Kurs zum präsentieren der Inhalte und als Arbeitsblätter für Übungen.  

Ein Jupyter Notebook besteht aus einzelnen **Zellen**. Es gibt zwei Arten von Zellen:

- **Code-Zellen**: enthalten Python-Code, den wir ausführen können.
- **Markdown-Zellen**: enthalten Text (wie diese Zelle hier), der formatiert
  dargestellt wird.

Den Zelltyp erkennen wir oben im Dropdown-Menü in der
Werkzeugleiste (dort steht 'Code' oder 'Markdown').

In den Arbeitsblättern schreibst du deine Lösungen in die Code-Zellen. 

### Zellen ausführen

Der blaue Balken links neben der Zelle markiert, welche Zelle gerade
aktiv ist. 
Um eine aktive Zelle auszuführen, klicken wir **Shift + Enter**
(wir springen danach automatisch zur nächsten Zelle) oder **Strg + Enter**
(wir bleiben in der Zelle stehen).

Probiere dies an den folgenden Code-Zelle aus:

````python
print('Montag')
````

````python
print('Dienstag')
````

### Die Ausführungsreihenfolge

Links neben einer Code-Zelle erscheint nach dem Ausführen eine Zahl in
eckigen Klammern, z.B. `[1]`. Sie zeigt, in welcher **Reihenfolge wir die
Zellen ausgeführt haben** — nicht die Reihenfolge, in der sie im Notebook
stehen!

Das ist wichtig: Variablen, die wir in einer Zelle definieren, bleiben so
lange gespeichert (im sogenannten *Kernel*), bis das Notebook neu
gestartet wird. Entscheidend ist also nicht, was im Notebook steht, sondern
was wir bereits ausgeführt haben.

**Probiere es aus:**

1. Führe die beiden folgenden Zellen der Reihe nach aus. Die Ausgabe
   ist `Punkte: 10`.
2. Führe jetzt **nur die zweite Zelle** noch dreimal aus. Am Code ändert
   sich nichts — an der Ausgabe schon. Achte dabei auch auf die Zahl in
   den eckigen Klammern.
3. Führe zum Schluss noch einmal die erste Zelle aus und danach wieder
   die zweite. Überlege dir vorher, was herauskommt.

````python
punkte = 0
````

````python
punkte = punkte + 10
print("Punkte:", punkte)
````

Dieselbe Zelle liefert also je nach Vorgeschichte ein anderes Ergebnis.
Wenn wir ein Notebook nochmal ganz von vorne starten wollen, dann wählen wir in der 
Werkzeugleiste  im Menü **Kernel → Restart Kernel and Clear Outputs of All Cells**

### Ein laufendes Programm abbrechen

Manchmal gerät ein Programm in eine **Endlosschleife** und hört nicht mehr
auf zu laufen (wir sehen dann links neben der Zelle ein `[*]` statt einer
Zahl — das bedeutet: die Zelle läuft noch). In dem Fall müssen wir die
Ausführung manuell abbrechen:

- Klicke auf das **■ (Stopp/Interrupt)**-Symbol in der Werkzeugleiste 

**Probiere es aus:** Führe die folgende Zelle aus (sie läuft absichtlich
endlos) und brich sie danach mit einer der beiden Methoden ab. Du bekommst
dann eine Fehlermeldung `KeyboardInterrupt` — das ist in diesem Fall
genau richtig und kein Problem.

````python
while True:
    pass
````

Wenn wir die Ausgabe einer selektierten Zelle löschen wollen, klicken wir die rechte Maustaste und wählen `clear cell output`.
Probiere auf diese Weise, die KeyboardInterrupt-Meldung der letzten Zelle zu löschen.

### Speichern

Ein Notebook wird automatisch zwischengespeichert, wir können aber
jederzeit auch manuell mit **Strg + S** speichern.

### Beispielaufgabe

#### Aufgabe A1
```
Setze p auf 6.
Setze q auf 9.
Erhöhe p um 3.
Erniedrige q um 4.
Gib in der ersten Zeile die Summe von p und q aus.
Gib in der nächsten Zeile das Produkt von p und q aus.
```
**Erwartete Ausgabe:**
```
14
45
```

````python
p = 6
q = 9
p = p + 3
q = q - 4
print(p + q)
print(p * q)
````
