### 1. Wo wird deine Arbeit gespeichert?

JupyterLite speichert deine Notebooks **in deinem Browser auf diesem Rechner**. Es gibt keinen Server, auf dem deine Arbeit liegt.
Das bedeutet:

- Am **selben Rechner im selben Browser** findest du deine Arbeit beim nächsten Mal wieder.
- An einem **anderen Rechner** oder in einem **anderen Browser** ist sie **nicht** da.
- In einem **privaten Fenster** (Inkognito-Modus) ist alles weg, sobald du das Fenster schließt.
- Wer die **Browserdaten löscht**, löscht auch seine Notebooks. An Schulrechnern passiert das oft automatisch.

Wenn du an wechselnden Rechnern arbeitest, nimmst du dein Notebook deshalb **als Datei** mit (Abschnitte 3 und 4).

### 2. Speichern und wieder öffnen am selben Rechner

**Speichern:** Drücke **Strg + S** oder wähle im Menü **File → Save Notebook**.
JupyterLite speichert zwar auch zwischendurch von selbst, aber verlass dich nicht darauf.

**Wieder öffnen:** Öffne denselben Link noch einmal, am selben Rechner und im selben Browser.
Du siehst dann deine gespeicherte Version, nicht das ursprüngliche Notebook.

**Wichtig:** Gespeichert werden der Text und der Code in den Zellen, nicht der Zustand von Python.
Nach dem Öffnen musst du die Zellen, die du brauchst, **noch einmal ausführen**, damit z. B. deine Variablen wieder da sind.

### 3. Herunterladen und hochladen

Damit du ein Notebook an einem anderen Rechner weiterbearbeiten kannst, machst du daraus eine **Datei** (Endung `.ipynb`).

**Herunterladen:**

1. Speichere das Notebook mit **Strg + S**.
2. Wähle im Menü **File → Download**.
3. Die Datei landet im **Download-Ordner**, z. B. `listen.ipynb`.

**Hochladen:**

1. Öffne die **JupyterLab-Ansicht** der passenden Seite (Links ganz oben). Dort siehst du links eine Liste mit allen Dateien.
2. Ziehe deine Datei mit der Maus in diese Liste. Oder klicke über der Liste auf das Symbol mit dem **Pfeil nach oben** (*Upload Files*) und wähle die Datei aus.
3. Gibt es schon eine Datei mit diesem Namen, fragt JupyterLite, ob sie ersetzt werden soll.
   Wähle **Overwrite** (Überschreiben), wenn deine hochgeladene Datei die neuere ist.
4. Öffne die Datei mit einem **Doppelklick**.

**Tipp:** Lade jedes Notebook auf **die Seite** hoch, zu der es gehört. Die Datenbank-Übungen brauchen z. B. die Datei
`sportverein.db`, die es nur auf der Seite mit den Übungen zu den Unterrichtseinheiten gibt.

### 4. Mit Moodle zwischen Schule und Zuhause wechseln

In Moodle hast du einen eigenen Speicherplatz, die **Privaten Dateien**. Nur du kannst sie sehen.
Im Kurs findest du sie im Block **Private Dateien**. Du erreichst sie auch über dein Profilbild oben rechts → **Private Dateien**.

#### Am Ende der Stunde: Notebook in Moodle ablegen

1. Speichere das Notebook in JupyterLite mit **Strg + S**.
2. Lade es herunter: **File → Download** (Abschnitt 3).
3. Öffne in Moodle den Block **Private Dateien** und klicke auf **Private Dateien verwalten …**
4. Ziehe die Datei aus dem Download-Ordner in das Feld. Oder klicke auf **Datei hinzufügen** (Symbol mit dem Blatt) und wähle sie aus.
5. Gibt es die Datei schon, fragt Moodle nach. Wähle **Überschreiben**.
6. Klicke auf **Änderungen speichern**. Ohne diesen Klick wird nichts gespeichert!

An Schulrechnern kannst du die Datei danach aus dem Download-Ordner löschen.

#### Beim nächsten Mal: Notebook aus Moodle holen

1. Öffne in Moodle den Block **Private Dateien** und klicke auf den Namen deiner Datei. Sie wird heruntergeladen.
2. Öffne die **JupyterLab-Ansicht** der passenden Seite (Links ganz oben).
3. Lade die Datei hoch (Abschnitt 3). Wähle bei Nachfrage **Overwrite**.
4. Öffne sie mit einem Doppelklick und führe die Zellen aus, die du brauchst.

#### Merke dir

**Zum Schluss immer: speichern → herunterladen → in Moodle hochladen → Änderungen speichern.**

Plane dafür am Ende der Stunde zwei Minuten ein. Was nur im Browser eines Schulrechners liegt, ist beim nächsten Mal meistens weg.

### 5. Wenn etwas nicht klappt

**Meine Änderungen sind weg.**
Du arbeitest an einem anderen Rechner, in einem anderen Browser oder in einem privaten Fenster, oder die Browserdaten wurden gelöscht.
Dann hilft nur die Datei, die du in Moodle abgelegt hast. Lade sie wie in Abschnitt 4 wieder hoch.

**Die heruntergeladene Datei heißt plötzlich `listen(1).ipynb`.**
Im Download-Ordner lag schon eine Datei mit diesem Namen, deshalb hat der Browser eine Nummer angehängt.
Lösche die alte Datei im Download-Ordner, bevor du herunterlädst. Oder benenne die neue Datei vor dem Hochladen in Moodle wieder um.
So überschreibst du in Moodle immer dieselbe Datei.

**Ich sehe eine alte Version eines Notebooks.**
Wenn du ein Notebook schon einmal geöffnet und gespeichert hast, zeigt JupyterLite dir immer deine gespeicherte Version,
auch wenn es inzwischen eine neue Version gibt. So kommst du an die neue Version:

- **Nur anschauen:** Öffne den Link in einem privaten Fenster.
- **Neu anfangen:** Öffne die JupyterLab-Ansicht der Seite (Links ganz oben) und wähle im Menü **Help → Clear Browser Data**.
  Danach siehst du wieder die aktuellen Notebooks.
  **Achtung:** Dabei werden **alle** Notebooks gelöscht, die du in diesem Browser in JupyterLite gespeichert hast.
  Lade deine eigenen Notebooks deshalb vorher herunter (**File → Download**) und lege sie in Moodle ab.

**Nach dem Hochladen gibt es Fehler wie `NameError`.**
Python wurde neu gestartet und kennt deine Variablen noch nicht. Führe die Zellen von oben nach unten noch einmal aus.
