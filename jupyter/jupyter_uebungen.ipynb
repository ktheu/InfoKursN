# Jupyter-Notebook selbst gestalten

In dieser Übung baust du selbst ein kleines Notebook. Erstelle dazu ein
neues Notebook und setze die Anleitung im Text unten Schritt für Schritt um — mit
den passenden Zellen, Überschriften, Formatierungen, einer Code-Zelle und
einem eingebetteten Bild.

Der Text ist absichtlich nicht formatiert. Lies ihn genau, er sagt dir an
jeder Stelle, was zu tun ist. 

Erstelle zunächst ein neues Notebook mit dem Namen 'mein_python_steckbrief.ipynb' im gleichen Ordner wie dieses Notebook hier. Setze darin die folgenden Anweisungen der Reihe nach in einzelnen Zellen um. Jeder neue Absatz hier im Text beschreibt eine neue Zelle, die du anlegen sollst.

Die erste Zelle ist eine Markdown-Zelle mit einer Überschrift 1. Ordnung. Der Text der Überschrift lautet 'Steckbrief Python'.

Die zweite Zelle ist wieder eine Markdown-Zelle, ganz ohne Überschrift. Schreibe hinein: 'Python ist eine weit verbreitete Programmiersprache, die wir in diesem Kurs benutzen. Der Name geht übrigens nicht auf die Schlange zurück, sondern auf die britische Comedy-Gruppe Monty Python'.

Die dritte Zelle ist eine neue Markdown-Zelle mit einer Überschrift 2. Ordnung mit dem Text 'Eigenschaften'. Darunter, in derselben Zelle, folgt eine Aufzählung mit drei Stichpunkten. Der erste Stichpunkt lautet 'Python wurde 1991 von Guido van Rossum veröffentlicht'. Der zweite Stichpunkt lautet 'Python gilt als gut lesbar und einfach zu lernen', wobei das Wort 'einfach' fett dargestellt werden soll. Der dritte Stichpunkt lautet 'Python wird unter anderem in der Datenanalyse, bei künstlicher Intelligenz und in der Webentwicklung eingesetzt', wobei die Wörter 'künstlicher Intelligenz' kursiv dargestellt werden sollen.

Die vierte Zelle ist wieder eine Markdown-Zelle mit einer Überschrift 2. Ordnung mit dem Text 'Ein Beispiel'. Schreibe darunter, in derselben Zelle, den Satz 'Schon mit einer einzigen Zeile Code kannst du in Python etwas ausgeben.'

Die fünfte Zelle ist eine Code-Zelle, keine Markdown-Zelle. Sie enthält genau eine Zeile Python-Code: print('Hallo, ich bin ein Python-Programm!')  Führe die Zelle aus, damit die Ausgabe sichtbar wird.

Die sechste Zelle ist wieder eine Markdown-Zelle mit einer Überschrift 2. Ordnung mit dem Text 'Das Python-Logo'. Füge darunter, in derselben Zelle, mit einem HTML img-Tag das Bild unter der Adresse https://www.python.org/static/community_logos/python-logo-master-v3-TM.png mit einer Breite von 150 Pixeln ein.

Die siebte und letzte Zelle ist eine Markdown-Zelle mit dem Satz: 'Damit ist der Steckbrief fertig', wobei das Wort fertig fett dargestellt werden soll.

Wenn du alle Zellen erstellt und ausgeführt hast, vergleiche dein Ergebnis noch einmal mit den Anweisungen oben. Stimmen alle Überschriften-Ebenen? Steht fett und kursiv an der richtigen Stelle? Ist der Code in einer eigenen, ausführbaren Zelle statt in einer Markdown-Zelle? Wird das Bild angezeigt?
