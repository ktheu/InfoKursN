## Was ist ein Jupyter Notebook?

Ein Jupyter Notebook besteht aus einzelnen **Zellen**, die du nacheinander
bearbeitest. Es gibt zwei Arten von Zellen:

- **Code-Zellen**: enthalten Python-Code, den du ausführen kannst.
- **Markdown-Zellen**: enthalten Text (wie diese Zelle hier), der formatiert
  dargestellt wird.

Den Zelltyp erkennst und änderst du über das Dropdown-Menü in der
Werkzeugleiste (dort steht 'Code' oder 'Markdown').

## Zellen ausführen

Um eine Zelle auszuführen, klickst du sie an und drückst **Shift + Enter**
(du springst danach automatisch zur nächsten Zelle) oder **Strg + Enter**
(du bleibst in der Zelle stehen).

Probiere es direkt an der folgenden Code-Zelle aus:

````python
print("Hallo, ich bin deine erste ausgeführte Zelle!")
````

## Bearbeitungsmodus und Kommandomodus

Der blaue Balken links neben der Zelle markiert, welche Zelle gerade
aktiv ist.

Eine Zelle hat zwei Zustände. In beiden ist die Umrandung blau — sie
unterscheiden sich aber darin, wie groß der blaue Rahmen ist:

- **Bearbeitungsmodus**: Der blaue Rahmen umschließt nur den
  Eingabebereich der Zelle. Du hast in den Text hineingeklickt und kannst
  ihn bearbeiten.
- **Kommandomodus**: Der blaue Rahmen reicht bis fast an den linken Rand
  und schließt den schmalen vertikalen **blauen Balken** links neben der
  Zelle mit ein. Die Zelle ist dann nur ausgewählt, aber nicht zum
  Bearbeiten geöffnet — einzelne Tasten lösen jetzt Aktionen auf die
  ganze Zelle aus, statt Buchstaben in sie hineinzutippen. Das werden wir weiter unten benötigen.

Wenn du im Bearbeitungsmodus bist, kannst du mit ESC in den Kommandomodus wechseln.

## Die Ausführungsreihenfolge

Links neben einer Code-Zelle erscheint nach dem Ausführen eine Zahl in
eckigen Klammern, z.B. `[1]`. Sie zeigt, in welcher **Reihenfolge du die
Zellen ausgeführt hast** — nicht die Reihenfolge, in der sie im Notebook
stehen!

Das ist wichtig: Variablen, die du in einer Zelle definierst, bleiben so
lange gespeichert (im sogenannten *Kernel*), bis du das Notebook neu
startest. Führst du Zellen mehrfach oder in der falschen Reihenfolge aus,
kann das zu überraschenden Ergebnissen führen.

**Probiere es aus:** Führe zuerst die nächste Zelle aus, danach die
übernächste. Verändere dann den Wert in der nächsten Zelle, führe nur
diese Zelle erneut aus (die übernächste NICHT) und überlege dir, was
passiert. Führe die übernächste Zelle danach trotzdem aus und vergleiche
mit deiner Vermutung.

````python
x = 6
````

````python
print(x)
````

## Ein laufendes Programm abbrechen

Manchmal gerät ein Programm in eine **Endlosschleife** und hört nicht mehr
auf zu laufen (du siehst dann links neben der Zelle ein `[*]` statt einer
Zahl — das bedeutet: die Zelle läuft noch). In dem Fall musst du die
Ausführung manuell abbrechen:

- Klicke auf das **■ (Stopp/Interrupt)**-Symbol in der Werkzeugleiste, oder
- wechsle in den Kommandomodus (Esc) und drücke zweimal **I** (also
  **I I**).

**Probiere es aus:** Führe die folgende Zelle aus (sie läuft absichtlich
endlos) und brich sie danach mit einer der beiden Methoden ab. Du bekommst
dann eine Fehlermeldung `KeyboardInterrupt` — das ist in diesem Fall
genau richtig und kein Problem.

````python
while True:
    pass
````

## Variablen im Speicher löschen (Kernel neu starten)

Wenn du alle bisher gespeicherten Variablen loswerden und komplett neu
anfangen willst, startest du den **Kernel** neu. Das löscht alle Variablen
aus dem Speicher — du musst die benötigten Zellen danach wieder von oben
nach unten neu ausführen.

- Klicke auf das **Neustart (kreisförmiger Pfeil)**-Symbol in der
  Werkzeugleiste bzw. wähle im Menü **Kernel → Restart Kernel...**, oder
- wechsle in den Kommandomodus (Esc) und drücke zweimal **0** (also
  **0 0**).

## Neue Zellen einfügen und Zelltyp ändern

Mit dem **+**-Symbol in der Werkzeugleiste fügst du eine neue Zelle ein.
Über das Dropdown-Menü (oder die rechte Maustaste) änderst du den Zelltyp
zwischen Code und Markdown.

**Schneller geht es mit Tastenkürzeln im Kommandomodus** (Esc, siehe oben):

- **M** → macht die Zelle zu einer Markdown-Zelle
- **Y** → macht die Zelle zu einer Code-Zelle
- **D D** (zweimal hintereinander **D** drücken) → löscht die Zelle

**Aufgabe:** Füge unterhalb dieser Zelle eine neue Code-Zelle ein und
lass sie deinen Namen ausgeben (mit print).

## Markdown-Grundlagen

Markdown-Zellen formatierst du mit einfachen Zeichen:

- `# Text` → Überschrift (mehr `#` = kleinere Überschrift, z.B. `##`, `###`)
- `**Text**` → **fett**
- `*Text*` → *kursiv*
- `- Text` → Aufzählungspunkt
- `` `Text` `` → `Code-Formatierung`

Um eine Markdown-Zelle zu bearbeiten, klickst du sie doppelt an. Zum
Rendern führst du sie aus (Shift+Enter) — genau wie eine Code-Zelle.

**Aufgabe:** Doppelklicke diese Zelle, um sie zu bearbeiten. Ersetze
diesen Text durch:
- eine Überschrift mit deinem Lieblingsfach
- einen fett gedruckten Satz, warum du es magst
- eine Liste mit 2 Stichpunkten dazu

Führe die Zelle danach aus (Shift+Enter), um das Ergebnis zu sehen.

## Bilder einfügen

Markdown-Zellen erlauben eingebettetes HTML. Mit dem `<img>`-Tag kannst
du ein Bild einfügen und zusätzlich seine Breite festlegen:

```html
<img src="https://www.python.org/static/community_logos/python-logo-master-v3-TM.png" width="200">
```

- `src` ist die Adresse des Bildes (eine Internet-URL oder ein Dateiname
  im Notebook-Ordner).
- `width` legt die Breite in Pixeln fest, die Höhe wird automatisch
  passend mitskaliert.
