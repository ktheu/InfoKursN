## 1. Klassen und Objekte

#### Objekte

Objekte können wir uns vorstellen als Container (Behälter) für Daten und Funktionalität. Welche Art von Daten und welche
Funktionalität zu einem Objekt gehören, wird durch seinen Typ bestimmt. Eine Liste ist ein solches Objekt.

```python
a = [1, 2, 3, 4]          # a ist ein Objekt vom Typ list
a.append(5)               # append ist eine Funktionalität des Objekts a
print(a)
print(type(a))
```

Die Daten eines Objekts werden in **Attributen** gespeichert, die Verarbeitungsfunktionen heißen **Methoden** des Objekts.

Ein Objekt befindet sich stets in einem bestimmten **Zustand**. Der aktuelle Objektzustand wird durch die aktuellen Werte
der Attribute festgelegt. Den Zustand eines Objekts verdeutlichen wir mit einem **Objektdiagramm**.

Hinweis: Häufig werden die Objektkarten mit runden Ecken gezeichnet.

#### Eigene Klassen

Wir wollen ein Objekt zu der folgenden Objektkarte erstellen.

<img src='klassen2b.png' width='400'>

Dafür brauchen wir einen Bauplan, der festlegt, wie ein Objekt vom Typ `Person` aussehen soll. Ein solcher Bauplan heißt
**Klasse**. Nach diesem Bauplan können wir beliebig viele Objekte erzeugen. Ein Objekt nennt man auch eine **Instanz** der Klasse.

```python
class Person:
    def __init__(self, name):
        self.name = name

p = Person('Lena')
print(type(p))
```

- `class Person:` beginnt die Klasse. Klassennamen schreiben wir mit großem Anfangsbuchstaben.
- Die Methode `__init__` (sprich: 'dunder init') heißt **Konstruktor**. Sie wird automatisch durchlaufen, wenn ein Objekt erzeugt wird.
- Ihr erster Parameter heißt `self`: Das ist das neue Objekt selbst. Mit `self.name = name` bekommt das Objekt das Attribut `name`.
- Die anderen Parameter geben wir beim Erzeugen an: `Person('Lena')`. Für `self` geben wir nichts an.

#### Attribute

Auf die Attribute eines Objekts greifen wir mit der Punktnotation zu. Wir können den Objektzustand ändern, indem wir den
Attributen andere Werte zuweisen.

```python
print(p.name)
p.name = 'Lisa'
print(p.name)
```

#### Beispiel: Datum

Wir wollen Objekte zu den folgenden Objektkarten erstellen.

<img src='klassen2a.png' width='701'>

```python
class Datum:
    def __init__(self, tag, monat, jahr):
        self.tag = tag
        self.monat = monat
        self.jahr = jahr

d1 = Datum(12, 3, 1990)
d2 = Datum(3, 9, 1995)
print(d1.monat, d2.monat)
```

Jedes Objekt hat **eigene** Attribute: Ändern wir `d1.monat`, bleibt `d2.monat` unverändert.

```python
d1.monat = 7
print(d1.monat, d2.monat)
```

#### Startwerte und berechnete Attribute

Der Konstruktor soll das Objekt in einen vernünftigen Anfangszustand versetzen. Nicht jedes Attribut muss dafür ein Parameter
sein: Der Konstruktor kann Attributen auch einen festen Startwert geben oder sie aus den Parametern berechnen. Parameter
können Standardwerte haben, wie bei Funktionen.

```python
class Konto:
    def __init__(self, inhaber, kontostand=0):
        self.inhaber = inhaber
        self.kontostand = kontostand
        self.buchungen = []

class Rechteck:
    def __init__(self, breite, hoehe):
        self.breite = breite
        self.hoehe = hoehe
        self.flaeche = breite * hoehe

k = Konto('Ali')
print(k.inhaber, k.kontostand, k.buchungen)
print(Konto('Bea', 50).kontostand)
print(Rechteck(3, 4).flaeche)
```

#### `isinstance()`

Die Klasse ist der Typ ihrer Objekte. `isinstance(objekt, Klasse)` prüft, ob ein Objekt eine Instanz der Klasse ist.

```python
print(isinstance(k, Konto))
print(isinstance(k, Rechteck))
```

## 2. Methoden

#### Methoden definieren und aufrufen

Eine **Methode** ist eine Funktion, die in der Klasse steht und zu einem Objekt gehört.

Wir möchten, dass unsere Datumsobjekte Auskunft darüber geben können, ob sie im Frühling liegen (d. h. in einem Monat von 3
bis 5). Dazu schreiben wir in die Klasse `Datum` eine Methode `im_fruehling`.

```python
class Datum:
    def __init__(self, tag, monat, jahr):
        self.tag = tag
        self.monat = monat
        self.jahr = jahr

    def im_fruehling(self):
        return 3 <= self.monat <= 5

d1 = Datum(12, 3, 1990)
d2 = Datum(3, 9, 1995)
print(d1.im_fruehling())
print(d2.im_fruehling())
```

Mit der Punktnotation rufen wir den Objekten ihre Methoden zu.

#### `self`

Eine Methode steht innerhalb des class-Blocks und erhält mit dem ersten Parameter `self` immer eine Referenz auf die Instanz,
über die sie aufgerufen wird. Dieser erste Parameter muss nur bei der Definition hingeschrieben werden und wird beim Aufruf
der Methode automatisch mit der entsprechenden Instanz verknüpft. Über `self` kommt die Methode an die Attribute des Objekts.

Die beiden folgenden Aufrufe bedeuten dasselbe:

```python
print(d1.im_fruehling())
print(Datum.im_fruehling(d1))
```

#### Methoden mit Parametern und Rückgabewert

Methoden können weitere Parameter haben und mit `return` einen Wert zurückgeben, wie Funktionen. Sie können auch Attribute
und damit den Zustand des Objekts ändern, z. B. mit `self.kontostand += betrag`.

```python
class Konto:
    def __init__(self, inhaber):
        self.inhaber = inhaber
        self.kontostand = 0

    def einzahlen(self, betrag):
        self.kontostand += betrag

    def abheben(self, betrag):
        if betrag <= self.kontostand:
            self.kontostand -= betrag
        else:
            print('Nicht genug Geld!')

    def ist_im_plus(self):
        return self.kontostand > 0

k = Konto('Ali')
print(k.ist_im_plus())
k.einzahlen(50)
k.abheben(20)
k.abheben(100)
print(k.kontostand, k.ist_im_plus())
```

#### Methoden rufen Methoden auf

Innerhalb der Klasse rufen wir eine andere Methode desselben Objekts mit `self.methode()` auf.

Bekommt eine Methode ein **zweites Objekt derselben Klasse** als Parameter, nennen wir es meist `other`: `self` ist das eigene
Objekt, `other` das andere. Mit `other.flaeche()` rufen wir dessen Methode auf.

```python
class Rechteck:
    def __init__(self, breite, hoehe):
        self.breite = breite
        self.hoehe = hoehe

    def flaeche(self):
        return self.breite * self.hoehe

    def ist_groesser_als(self, other):
        return self.flaeche() > other.flaeche()

a = Rechteck(3, 4)
b = Rechteck(2, 5)
print(a.flaeche(), b.flaeche())
print(a.ist_groesser_als(b))
```

## 4. Objekte ausgeben mit `__str__`

`print(d1)` zeigt normalerweise nur etwas wie `<__main__.Datum object at 0x...>`. Schreiben wir in die Klasse eine Methode
`__str__(self)`, die eine **Zeichenkette zurückgibt**, verwendet `print()` diese Zeichenkette. Auch `str(objekt)` und f-Strings
benutzen `__str__`.

`__str__` gibt den Text mit `return` zurück. Sie gibt nichts mit `print()` aus.

```python
class Datum:
    def __init__(self, tag, monat, jahr):
        self.tag = tag
        self.monat = monat
        self.jahr = jahr

    def im_fruehling(self):
        return 3 <= self.monat <= 5

    def __str__(self):
        return f'{self.tag}.{self.monat}.{self.jahr}'

d1 = Datum(12, 3, 1990)
print(d1)
print('Geboren am ' + str(d1))
print(f'Das Datum ist {d1}.')
```

## 5. Objekte in Listen und Dictionaries

#### Viele Objekte verwalten

Objekte können wir wie andere Werte in Listen speichern. Mit einer Schleife gehen wir sie durch und greifen auf ihre Attribute
und Methoden zu. Zum Sortieren und für `min()`/`max()` geben wir mit `key=` an, nach welchem Attribut verglichen wird.

```python
class Schueler:
    def __init__(self, name, note):
        self.name = name
        self.note = note

kurs = [Schueler('Ali', 2), Schueler('Bea', 1), Schueler('Cem', 2)]
for s in kurs:
    print(s.name, s.note)
beste = min(kurs, key=lambda s: s.note)
print('Beste Note:', beste.name)
for s in sorted(kurs, key=lambda s: s.name, reverse=True):
    print(s.name)
```

#### Objekte nachschlagen

In einem Dictionary legen wir Objekte unter einem Schlüssel ab, z. B. unter ihrem Namen. Zu einem Schlüssel kann auch eine
Liste von Objekten gehören, hier alle Schüler mit derselben Note.

```python
verzeichnis = {}
for s in kurs:
    verzeichnis[s.name] = s
print(verzeichnis['Bea'].note)

nach_note = {}
for s in kurs:
    if s.note not in nach_note:
        nach_note[s.note] = []
    nach_note[s.note].append(s)
for s in nach_note[2]:
    print(s.name)
```

## 6. Referenzen

#### Variablen verweisen auf Objekte

Eine Variable enthält nicht das Objekt selbst, sondern einen **Verweis** (eine **Referenz**) darauf.
Nach `b = a` verweisen `a` und `b` auf **dasselbe** Objekt. Ändern wir das Objekt über `b`, sehen wir die Änderung auch über `a`.

`a is b` prüft, ob zwei Variablen auf dasselbe Objekt verweisen.

```python
class Konto:
    def __init__(self, stand):
        self.stand = stand

a = Konto(100)
b = a
b.stand = 0
print(a.stand)
print(a is b)
c = Konto(0)
print(a is c)
```

#### `is` und `==`

Zwei **verschiedene** Objekte mit gleichen Attributen sind für Python zunächst **nicht** gleich: Ohne weitere Angaben bedeutet `==`
bei eigenen Objekten dasselbe wie `is`. Wollen wir die Attribute vergleichen, vergleichen wir sie einzeln.

```python
class Punkt:
    def __init__(self, x, y):
        self.x = x
        self.y = y

p = Punkt(1, 2)
q = Punkt(1, 2)
print(p == q, p is q)
print(p.x == q.x and p.y == q.y)
```

#### Zahlen und Zeichenketten verhalten sich anders

Zahlen und Zeichenketten können wir nicht verändern. `x += 1` erzeugt eine **neue** Zahl, auf die `x` dann verweist. Deshalb ändert
sich `y` nach `y = x` nicht mit. Bei Objekten (und Listen), die wir über ihre Attribute **verändern**, verweisen dagegen alle
Variablen auf dasselbe veränderte Objekt.

```python
x = 5
y = x
x += 1
print(x, y)

p = Punkt(5, 0)
q = p
p.x += 1
print(p.x, q.x)
```

#### Objekte als Parameter

Übergeben wir ein Objekt an eine Funktion oder Methode, bekommt sie einen Verweis auf **dasselbe** Objekt.
Ändert sie seine Attribute, ist die Änderung danach auch außerhalb sichtbar. Eine Zahl als Parameter kann die Funktion dagegen
nicht von außen sichtbar ändern.

```python
class Spieler:
    def __init__(self, name):
        self.name = name
        self.punkte = 0

def belohnen(spieler, punkte):
    spieler.punkte += punkte

def verdoppeln(zahl):
    zahl = zahl * 2

s = Spieler('Mia')
belohnen(s, 10)
print(s.punkte)
n = 7
verdoppeln(n)
print(n)
```

#### Eine Kopie anlegen und `None`

Wollen wir ein **unabhängiges** Objekt mit gleichen Werten, erzeugen wir ein neues mit dem Konstruktor, z. B. in einer Methode `kopie()`.

`None` bedeutet „kein Objekt“. Eine Variable kann `None` sein, wenn sie (noch) auf kein Objekt verweist. Wir prüfen das mit `is None`.

```python
class Punkt:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def kopie(self):
        return Punkt(self.x, self.y)

p = Punkt(1, 2)
q = p.kopie()
q.x = 10
print(p.x, q.x)
leer = None
print(leer is None)
```

## 7. Datenkapselung

#### Das Problem mit öffentlichen Attributen

Ist der Kontostand ein öffentliches Attribut, kann jeder Teil des Programms ihn beliebig ändern, auch auf unsinnige Werte.
Die Methode `abheben()` prüft zwar, ob genug Geld da ist, aber wir können sie einfach umgehen.

```python
class Konto:
    def __init__(self, inhaber):
        self.inhaber = inhaber
        self.kontostand = 0

    def abheben(self, betrag):
        if betrag <= self.kontostand:
            self.kontostand -= betrag

k = Konto('Ali')
k.abheben(500)
print(k.kontostand)
k.kontostand = -1000000
print(k.kontostand)
```

#### Datenkapselung und Geheimnisprinzip

In der objektorientierten Programmierung werden die Daten und die Methoden, die auf ihnen arbeiten, als eine Einheit gesehen
(**Datenkapselung**). Im Idealfall kann auf die Daten der Kapsel (die Werte der Attribute) nur durch wohldefinierte
Schnittstellenmethoden zugegriffen werden. Die Attribute sind nach außen hin nicht direkt sichtbar. Die Klasse entscheidet also
selbst, was mit ihren Daten passieren darf.

Wie die Klasse ihre Daten intern speichert, bleibt ihr Geheimnis (**Geheimnisprinzip**, Information Hiding). Allgemein bedeutet
das Prinzip, dass ein Teilsystem (hier ein Objekt) nichts von den Implementierungsentscheidungen eines anderen Teilsystems
wissen darf.

#### Zugriffsrechte

Der Entwickler einer Klasse hat die Möglichkeit, **Zugriffsrechte für Attribute und Methoden** zu vergeben.

| Zugriffsrecht | zugreifbar | Name in Python | im Klassendiagramm |
|---|---|---|---|
| public (öffentlich) | von überall | `name` | `+` |
| protected (geschützt) | innerhalb der Klasse und in abgeleiteten Klassen | `_name` | `#` |
| private (privat) | nur innerhalb der Klasse | `__name` | `-` |

In Python gibt es keine Möglichkeit, die Zugriffsrechte sicher durchzusetzen. Es gilt die Konvention: Ein privater Name beginnt
mit einem doppelten Unterstrich und endet ohne Unterstrich, ein geschützter Name beginnt mit einem einfachen Unterstrich.

#### Private Attribute

Ein privates Attribut benutzen wir innerhalb der Klasse ganz normal, von außen gibt `k.__kontostand` einen `AttributeError`.

<img src='uml_konto_privat.png' width='293'>

```python
class Konto:
    def __init__(self, inhaber):
        self.__inhaber = inhaber
        self.__kontostand = 0

    def einzahlen(self, betrag):
        if betrag > 0:
            self.__kontostand += betrag

    def get_kontostand(self):
        return self.__kontostand

k = Konto('Ali')
k.einzahlen(50)
print(k.get_kontostand())
print(k.__kontostand)
```

#### Achtung: Zuweisen von außen gibt keinen Fehler

`k.__kontostand = 5` von außen gibt **keinen** Fehler. Python legt dann aber ein **neues**, öffentliches Attribut an, das mit dem
privaten Attribut nichts zu tun hat. Das private Attribut bleibt unverändert.

Python versteckt private Attribute nur, indem es ihren Namen intern ändert (aus `__kontostand` wird `_Konto__kontostand`).
Wer es darauf anlegt, kommt also doch heran. Datenkapselung ist in Python eine **Abmachung**: Von außen benutzen wir nur die Methoden.

```python
k = Konto('Ali')
k.einzahlen(100)
k.__kontostand = 5
print(k.__kontostand)
print(k.get_kontostand())
```

#### Getter und Setter

Damit wir private Attribute trotzdem lesen und ändern können, schreiben wir **Zugriffsmethoden**:

- Ein **Getter** `get_name()` gibt den Wert zurück.
- Ein **Setter** `set_name(wert)` setzt einen neuen Wert. Der Setter kann dabei **prüfen**, ob der Wert sinnvoll ist, und ihn sonst ablehnen.

Nicht jedes Attribut braucht beides: Gibt es nur einen Getter, können wir den Wert von außen lesen, aber nicht ändern.

Im Beispiel benutzt auch der Konstruktor den Setter. So wird die Prüfung nicht doppelt geschrieben.

<img src='uml_schueler.png' width='297'>

```python
class Schueler:
    def __init__(self, name, alter):
        self.__name = name
        self.__alter = 0
        self.set_alter(alter)

    def get_name(self):
        return self.__name

    def get_alter(self):
        return self.__alter

    def set_alter(self, alter):
        if 0 <= alter <= 120:
            self.__alter = alter

s = Schueler('Ali', 15)
s.set_alter(16)
s.set_alter(-3)
print(s.get_name(), s.get_alter())
```

#### Private Methoden

Auch Methoden können privat sein: `def __hat_ziffer(self)`. Sie sind **Hilfsmethoden**, die nur innerhalb der Klasse gebraucht
werden. Von außen können wir sie nicht aufrufen.

<img src='uml_passwortpruefer.png' width='306'>

```python
class Passwortpruefer:
    def __init__(self, passwort):
        self.__passwort = passwort

    def __hat_ziffer(self):
        for z in self.__passwort:
            if z.isdigit():
                return True
        return False

    def ist_sicher(self):
        return len(self.__passwort) >= 8 and self.__hat_ziffer()

print(Passwortpruefer('geheim42!').ist_sicher())
print(Passwortpruefer('geheimnis').ist_sicher())
```

## 8. Beziehungen zwischen Klassen

#### Assoziation

Eine **Assoziation** ist eine Beziehung „kennt“ oder „hat“ zwischen zwei Klassen: Eine Person hat ein Geburtsdatum.
In Python ist das einfach ein **Attribut, das auf ein anderes Objekt verweist**. Der Typ des Attributs ist die andere Klasse.

Die Klasse `Person` nutzt also die Klasse `Datum`. Wir machen das im Klassendiagramm mit einem Assoziationspfeil zwischen den
Klassen deutlich: Eine Person kennt ihr Datum, aber das Datum kennt die Person nicht.

<img src='klassen6a.png' width='700'>

```python
import datetime

class Person:
    def __init__(self, vorname, nachname, tag, monat, jahr):
        self.vorname = vorname
        self.nachname = nachname
        self.gebdatum = Datum(tag, monat, jahr)

    def jahrgang(self):
        return self.gebdatum.jahr

    def alter(self):
        heute = datetime.date.today()
        akt_jahr = heute.year
        return akt_jahr - self.jahrgang()

    def __str__(self):
        return f'{self.vorname} {self.nachname}, geboren am {self.gebdatum}'

p = Person('Willi', 'Wacker', 9, 8, 2005)
print(p)
print(p.jahrgang())
print(p.alter())
print(p.gebdatum.im_fruehling())
```

Im Objektdiagramm sehen wir, dass das Attribut `gebdatum` eine Referenz auf ein Datum-Objekt enthält.

<img src='klassen5a.png' width='600'>

#### Eine Klasse kennt viele Objekte

Kennt ein Objekt **viele** andere, speichert es sie in einer **Liste**. Der Typ des Attributs ist dann `list`. Typische Methoden
einer solchen Klasse sind `hinzufuegen()`, `anzahl()` sowie Suchen und Auswerten über alle Objekte der Liste.

<img src='uml_kurs.png' width='691'>

```python
class Schueler:
    def __init__(self, name, note):
        self.name = name
        self.note = note

class Kurs:
    def __init__(self, name):
        self.name = name
        self.teilnehmer = []

    def hinzufuegen(self, s):
        self.teilnehmer.append(s)

    def anzahl(self):
        return len(self.teilnehmer)

    def schnitt(self):
        summe = 0
        for s in self.teilnehmer:
            summe += s.note
        return summe / len(self.teilnehmer)

k = Kurs('Informatik')
k.hinzufuegen(Schueler('Ali', 2))
k.hinzufuegen(Schueler('Bea', 1))
k.hinzufuegen(Schueler('Cem', 3))
print(k.anzahl(), k.schnitt())
```

#### Rekursive Assoziation

Bei einer **rekursiven Assoziation** verweist ein Objekt auf ein Objekt **derselben** Klasse: Ein Mitarbeiter hat einen Chef, der
selbst ein Mitarbeiter ist. Im Diagramm geht der Pfeil von der Klasse zu sich selbst zurück.

Die Chefin ganz oben hat keinen Chef: Bei ihr steht `None`. Mit einer `while`-Schleife hangeln wir uns von
Chef zu Chef nach oben, bis wir bei `None` ankommen.

<img src='uml_mitarbeiter.png' width='451'>

```python
class Mitarbeiter:
    def __init__(self, name, chef=None):
        self.name = name
        self.chef = chef

boss = Mitarbeiter('Frau König')
leiter = Mitarbeiter('Herr Graf', boss)
azubi = Mitarbeiter('Ali', leiter)
print(azubi.chef.name)
m = azubi
while m is not None:
    print(m.name)
    m = m.chef
```

## 9. Vererbung

#### Erben

Aus bestehenden Klassen können wir neue Klassen ableiten. Eine abgeleitete Klasse erbt von der Oberklasse ihre Attribute und
Methoden, fügt ggf. eigene hinzu und kann ihnen durch Überschreiben eine neue Bedeutung geben.

`class Hund(Tier):` bedeutet: Die Klasse `Hund` **erbt** von der Klasse `Tier`. `Tier` ist die **Oberklasse**, `Hund` die **Unterklasse**.
Vererbung passt, wenn wir sagen können: „Ein Hund **ist ein** Tier.“ (Bei einer Assoziation sagen wir „hat ein“.)

Im Klassendiagramm zeigt ein Pfeil mit **leerem Dreieck** von der Unterklasse zur Oberklasse. In der Unterklasse stehen nur die
**neuen** Attribute und Methoden.

<img src='uml_tier.png' width='401'>

```python
class Tier:
    def __init__(self, name):
        self.name = name

    def fressen(self):
        return self.name + ' frisst.'

class Hund(Tier):
    def bellen(self):
        return self.name + ' bellt: Wuff!'

class Katze(Tier):
    def miauen(self):
        return self.name + ' miaut.'

bello = Hund('Bello')
minka = Katze('Minka')
print(bello.fressen())
print(bello.bellen())
print(minka.fressen())
print(minka.miauen())
```

Ein Hund ist auch ein Tier: `isinstance(bello, Tier)` ist `True`. Aber nicht jedes Tier ist ein Hund.

```python
tier = Tier('irgendwas')
print(isinstance(bello, Hund), isinstance(bello, Tier))
print(isinstance(tier, Hund), isinstance(tier, Tier))
```

#### Der Konstruktor der Unterklasse: `super().__init__()`

Hat die Unterklasse **eigene Attribute**, braucht sie einen eigenen Konstruktor. Damit die Attribute der Oberklasse trotzdem angelegt
werden, ruft er zuerst den Konstruktor der Oberklasse auf: `super().__init__(name)`. `super()` steht für „die Oberklasse“.

Im Diagramm steht der Konstruktor der Unterklasse mit allen Parametern, auch denen, die an die Oberklasse weitergegeben werden.

<img src='uml_hund_rasse.png' width='277'>

```python
class Tier:
    def __init__(self, name):
        self.name = name

class Hund(Tier):
    def __init__(self, name, rasse):
        super().__init__(name)
        self.rasse = rasse

bello = Hund('Bello', 'Dackel')
print(bello.name, bello.rasse)
```

Vergessen wir den Aufruf, wird der Konstruktor der Oberklasse **nicht** ausgeführt. Die Attribute der Oberklasse fehlen dann, und
wir bekommen einen `AttributeError`.

```python
class Hund(Tier):
    def __init__(self, name, rasse):
        self.rasse = rasse

bello = Hund('Bello', 'Dackel')
print(bello.name)
```

#### Methoden überschreiben

Hat die Unterklasse eine Methode mit **demselben Namen** wie die Oberklasse, wird für Objekte der Unterklasse **ihre** Methode benutzt.
Wir sagen: Sie **überschreibt** die Methode der Oberklasse. Objekte der Oberklasse benutzen weiter die alte Methode.

```python
class Tier:
    def __init__(self, name):
        self.name = name

    def laut(self):
        return '...'

class Hund(Tier):
    def laut(self):
        return 'Wuff'

class Katze(Tier):
    def laut(self):
        return 'Miau'

class Fisch(Tier):
    pass

print(Hund('Bello').laut())
print(Katze('Minka').laut())
print(Fisch('Nemo').laut())
```

#### Methoden erweitern mit `super()`

Wollen wir die Methode der Oberklasse nicht ersetzen, sondern **ergänzen**, rufen wir sie in der neuen Methode mit `super().methode()`
auf und machen dann noch etwas dazu. Das geht auch bei `__str__`.

```python
class Tier:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name

class Hund(Tier):
    def __init__(self, name, rasse):
        super().__init__(name)
        self.rasse = rasse

    def __str__(self):
        return super().__str__() + ' (' + self.rasse + ')'

print(Tier('Nemo'))
print(Hund('Bello', 'Dackel'))
```

#### Beispiel: Student

Die Klasse `Student` erbt von der Klasse `Person` aus Abschnitt 8. Die Methode `jahrgang` der Oberklasse wird überschrieben:
Sie soll das Attribut `jsb` (Jahr des Studienbeginns) zurückgeben.

Die geerbte Methode `alter` ruft `self.jahrgang()` auf. Bei einem `Student` ist das die überschriebene Methode, `alter` liefert
hier also die Jahre seit dem Studienbeginn.

<img src='klassen7.png' width='750'>

```python
class Student(Person):
    def __init__(self, vorname, nachname, tag, monat, jahr, fach, jsb):
        super().__init__(vorname, nachname, tag, monat, jahr)
        self.fach = fach
        self.jsb = jsb

    def jahrgang(self):
        return self.jsb

s = Student('Peter', 'Döhlmann', 10, 4, 1992, 'Bio', 2011)
print(s)
print(s.jahrgang())
print(s.alter())
```

#### Geschützt und privat bei Vererbung

- **Öffentlich** (`name`, `+`): alle dürfen zugreifen.
- **Geschützt** (`_name`, `#`): die Klasse selbst und ihre **Unterklassen** (eine Abmachung, Python prüft das nicht).
- **Privat** (`__name`, `-`): **nur** die Klasse selbst. Auch eine Unterklasse kommt nicht direkt heran.

Braucht eine Unterklasse ein privates Attribut der Oberklasse, benutzt sie die Zugriffsmethoden der Oberklasse (`self.get_stand()`).
Deshalb machen wir Attribute, die Unterklassen brauchen, meist **geschützt**.

<img src='uml_sparkonto.png' width='213'>

```python
class Konto:
    def __init__(self, stand):
        self.__stand = stand

    def get_stand(self):
        return self.__stand

class Sparkonto(Konto):
    def zinsen(self):
        return self.get_stand() * 0.02

    def zinsen_falsch(self):
        return self.__stand * 0.02

s = Sparkonto(500)
print(s.zinsen())
print(s.zinsen_falsch())
```

#### Mehrstufige Vererbung

Eine Unterklasse kann selbst wieder Oberklasse sein: `Dackel` erbt von `Hund`, `Hund` erbt von `Tier`. Ein Dackel hat dann alles von
`Hund` **und** von `Tier`. Sucht Python eine Methode, schaut es zuerst in der Klasse des Objekts, dann in ihrer Oberklasse, dann in
deren Oberklasse usw.

Jede Klasse erbt am Ende von der eingebauten Klasse `object`. Von dort stammen zum Beispiel das Standard-`__str__` und das Standard-`==`.

<img src='uml_dackel.png' width='176'>

```python
class Tier:
    def atmen(self):
        return 'atmet'

class Hund(Tier):
    def bellen(self):
        return 'bellt'

class Dackel(Hund):
    def buddeln(self):
        return 'buddelt'

d = Dackel()
print(d.atmen(), d.bellen(), d.buddeln())
print(isinstance(d, Tier), type(d) == Hund)
```

#### Zusammenfassung

| | |
|---|---|
| Klasse | Bauplan für Objekte, `class Hund:` |
| Objekt, Instanz | wird nach dem Bauplan erzeugt, `bello = Hund('Bello')` |
| Attribut | Variable eines Objekts, `self.name` |
| Methode | Funktion eines Objekts, Aufruf mit der Punktnotation |
| Objektzustand | die aktuellen Werte der Attribute, dargestellt im Objektdiagramm |
| Konstruktor | `__init__`, wird beim Erzeugen eines Objekts durchlaufen |
| `self` | das Objekt, über das die Methode aufgerufen wird |
| `__str__` | gibt den Text zurück, den `print(objekt)` zeigt |
| Referenz | Verweis einer Variablen auf ein Objekt, `is` prüft auf dasselbe Objekt |
| Datenkapselung | Daten und Methoden bilden eine Einheit, Zugriff nur über Methoden |
| Geheimnisprinzip | die Implementierung einer Klasse bleibt nach außen verborgen |
| Zugriffsrechte | öffentlich `name` (+), geschützt `_name` (#), privat `__name` (-) |
| Getter, Setter | Zugriffsmethoden für private Attribute, `get_name()`, `set_name(wert)` |
| Assoziation | „hat ein“: ein Attribut verweist auf ein anderes Objekt |
| Vererbung | „ist ein“: `class Hund(Tier):`, die Unterklasse übernimmt alles von der Oberklasse |
| `super()` | die Oberklasse, z. B. `super().__init__(name)` |
| Überschreiben | die Unterklasse legt eine Methode der Oberklasse neu fest |
