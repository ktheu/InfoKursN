## Veränderbare und unveränderbare Objekte in Python

[Video](https://youtu.be/JcNHEZA38qU) - [AufgabenVideo](https://youtu.be/-CyCSCVe44U)

Jedes Objekt in Python hat einen *Typ* und die binären Daten, die zu dem Objekt gehören, sind an einer bestimmten Adresse gespeichert. Wir nennen diese Daten des Objekts den *internal state* des Objekts. Wenn wir den internal state eines Objekts ändern können, dann nennen wir das Objekt *veränderbar (mutable)*, sonst *unveränderbar (immutable)*.

#### Lists sind mutable

Eine Liste ist ein veränderbares Objekt. Mit der Methode *append* verändern wir den internal state des Objekts.
Im abgebildeten Beispiel ändert sich die Variable b, obwohl wir nur a ändern. Das nennt man einen *Seiteneffekt*. Unerwünschte Seiteneffekte können unangenehme Fehlerquellen sein.

![](veraenderbar1.png)

```python
a = [1,2,3]
b = a
a.append(7)
print(a)
print(b)
```

#### Strings sind immutable

Strings sind unveränderbar (immutable). Wenn wir einen String durch Anfügen eines Zeichens ändern, wird ein neuer String an anderer Stelle erstellt.
Bei Strings können keine Seiteneffekte auftreten.

![](veraenderbar2.png)

```python
a = '123'
b = a
a = a + '4'
print(a)
print(b)
```

Wird eine Liste nicht mit append, sondern mit + erweitert, dann wird, wie in dem String-Beispiel, eine neue Liste erzeugt.

```python
a = [1,2,3]
b = a
a = a + [7]
print(a)
print(b)
```

**Achtung bei `+=`.** Bei Zahlen und Strings bedeutet `a += x` dasselbe wie `a = a + x`. Bei Listen ist das anders:
`a += [7]` erzeugt keine neue Liste, sondern verändert die Liste wie `append`. Der Seiteneffekt tritt also auf.

```python
a = [1,2,3]
b = a
a += [7]
print(a)
print(b)
```

##### Verändern oder neu erzeugen?

**Neu:** Bei Listen muss man also wissen, ob eine Operation die Liste verändert oder eine neue Liste erzeugt.

| verändert die Liste `a` | erzeugt eine neue Liste |
|---|---|
| `a[i] = x`, `del a[i]` | `a + b`, `a * 3` |
| `a.append(x)`, `a.insert(i, x)` | `a[1:3]`, `a[::-1]` (Slicing) |
| `a.pop()`, `a.remove(x)` | `sorted(a)` |
| `a.sort()` | `a.copy()`, `list(a)`, `a[:]` |
| `a += b` | |

Methoden, die eine Liste verändern, geben `None` zurück. Ein häufiger Fehler ist deshalb `a = a.sort()`:
Die Liste wird zwar sortiert, aber danach zeigt `a` auf `None`.

```python
a = [3, 1, 2]
b = sorted(a)       # neue sortierte Liste, a bleibt unverändert
print(a, b)

a.sort()            # a wird sortiert
print(a)

a = a.sort()        # Fehler: sort() gibt None zurück
print(a)
```

##### Kopien erstellen

Wenn eine Liste verändert werden soll, ohne dass sich andere Variablen mitändern, braucht man eine Kopie.
`a.copy()`, `list(a)` und `a[:]` erzeugen jeweils eine neue Liste mit denselben Elementen.

```python
a = [1, 2, 3]
b = a.copy()
c = list(a)
d = a[:]
print(a == b, a is b)    # gleicher Inhalt, aber verschiedene Objekte
a.append(7)
print(a, b, c, d)
```

#### Tuples sind immutable

Tuples sind unveränderbar. Alles, was bei Listen den internal state ändert, ist bei Tuples nicht möglich.

```python
t = (1,2,3)
t.append(7)      # nicht möglich
```

```python
t = (1,2,3)
t = t + (7,)     # möglich
print(t)
```

```python
t = (1,2,3)
t[1] = 7         # nicht möglich
```

Aber wenn die Elemente des Tuples aus veränderbaren Objekten bestehen, dann kann sich doch von außen gesehen der Inhalt des Tuples ändern.

![](veraenderbar3.png)

```python
a = [1,2]
b = [3,4]
t = (a,b)
print(t)
```

```python
a.append(7)
print(t)
```

#### Dicts und Sets sind mutable

Dicts und Sets sind ebenfalls veränderbar. Es gelten dieselben Regeln wie bei Listen: `e = d` erzeugt keine
Kopie, eine Kopie erhält man mit `d.copy()` bzw. `s.copy()`.

Hier liegt auch der Grund, warum Schlüssel eines Dicts und Elemente eines Sets unveränderbar sein müssen: Python legt
ein Paar an der Stelle ab, die sich aus dem Schlüssel berechnet. Würde sich der Schlüssel später ändern, fände Python
das Paar nicht mehr.

```python
d = {'Anna': 2}
e = d
f = d.copy()
e['Ben'] = 1
print(d)
print(e)
print(f)

s = {1, 2}
t = s
t.add(3)
print(s)
```

#### Mutable und Immutable Datentypen

| veränderbar (mutable) | unveränderbar (immutable) |
|---|---|
| Listen, Dicts, Sets | Ints, Floats, Bools, Strings, Tuples, None |

#### Funktionsaufrufe

Bei einem Funktionsaufruf wird der Parameter zu einem Zeiger auf die Adresse, auf die das Argument zeigt.

![](aufruf1.png)

Die Übergabe des Arguments an den Parameter können wir uns wie eine Zuweisung vorstellen. Diesen Mechanismus bezeichnet man mit
*call by assignment* oder *call by object reference*.

```python
def func(x):
    print(hex(id(x)))

k = 500
print(hex(id(k)))
func(k)
```

Wenn wir beispielsweise einen langen String an eine Funktion übergeben, wird nicht der lange String (aufwendig) kopiert (das wäre *call by value*), sondern nur die Adresse, an der der String steht.

```python
s = 'Dies ist ein langer Text'
print(hex(id(s)))
func(s)
```

#### Funktionsaufrufe und Mutability

Bei immutable Objekten sind wir (einigermaßen) sicher vor unerwünschten Seiteneffekten. Bei mutable Objekten kann es zu Seiteneffekten kommen.
In dem Beispiel wird die Liste a in der Funktion verändert. Wenn wir das verhindern wollen, müssen wir eine Kopie der Liste entweder beim Funktionsaufruf oder in der Funktion erstellen.

![](aufruf2.png)

```python
def doit(x):
    x.append(7)
    return x

a = [1,4]
print('a',a)
b = doit(a)
print('a',a)
print('b',b)
```

```python
def doit(x):
    x.append(7)
    return x

a = [1,4]
print('a',a)
b = doit(a.copy())        # Kopie beim Aufruf um Seiteneffekte zu verhindern
print('a',a)
print('b',b)
```

```python
def doit(x):
    x = x.copy()          # Kopie vor der Veränderung um Seiteneffekte zu verhindern
    x.append(7)
    return x

a = [1,4]
print('a',a)
b = doit(a)
print('a',a)
print('b',b)
```

Wird dem Parameter in der Funktion eine **neue** Liste zugewiesen, dann zeigt der Parameter auf ein anderes Objekt.
Die Liste des Aufrufers bleibt dann unverändert.

```python
def doit(x):
    x = [0]               # x zeigt jetzt auf eine neue Liste
    x.append(7)
    return x

a = [1,4]
b = doit(a)
print('a',a)
print('b',b)
```

Manchmal ist es gewollt, dass eine Funktion die übergebene Liste verändert, z.B. bei `a.sort()` oder bei einem
Sortierverfahren, das „an Ort und Stelle“ sortiert. Solche Funktionen geben dann meistens nichts zurück. Wichtig ist,
dass man aus dem Namen und dem Docstring erkennt, ob eine Funktion ihr Argument verändert.

```python
def verdoppeln(a):
    '''
    a: Liste mit Zahlen
    returns: neue Liste mit den doppelten Werten, a bleibt unverändert
    '''
    ergebnis = []
    for x in a:
        ergebnis.append(2 * x)
    return ergebnis

def verdoppeln_in_liste(a):
    '''
    a: Liste mit Zahlen, wird verändert: jeder Wert wird verdoppelt
    returns: None
    '''
    for i in range(len(a)):
        a[i] = 2 * a[i]

zahlen = [1, 2, 3]
print(verdoppeln(zahlen), zahlen)
verdoppeln_in_liste(zahlen)
print(zahlen)
```

#### Flache und tiefe Kopie

Bei der flachen Kopie einer Liste werden die Inhalte der Liste kopiert. Aber wenn die Elemente der Liste Zeiger zu veränderbaren Objekten sind, kann dies trotzdem zu Seiteneffekten führen. Bei einer tiefen Kopie werden alle veränderbaren Objekte, die irgendwie zur Liste gehören, kopiert.

![](deepcopy.png)

```python
from copy import deepcopy

a = [1,2]
b = [3,4]
c = [a,b]
d = c.copy()         # flache Kopie von c
e = deepcopy(c)      # tiefe Kopie von c

a.append(7)
print(d)
print(e)
```

##### Achtung bei zweidimensionalen Listen

Eine Liste mit `*` zu vervielfachen kopiert nur die Zeiger. Bei einer Liste von Listen zeigen dann alle Zeilen
auf **dieselbe** Liste. Ändert man ein Element, ändert es sich in allen Zeilen.

```python
feld = [[0] * 3] * 3          # drei Zeiger auf dieselbe Zeile
feld[0][0] = 1
print(feld)
print(feld[0] is feld[1])
```

Richtig ist es, für jede Zeile eine eigene Liste zu erzeugen. (Innerhalb einer Zeile ist `[0] * 3` kein Problem,
denn Zahlen sind unveränderbar.)

```python
feld = []
for i in range(3):
    feld.append([0] * 3)      # jedes Mal eine neue Liste
feld[0][0] = 1
print(feld)
print(feld[0] is feld[1])
```

#### Zusammenfassung

- Listen, Dicts und Sets sind veränderbar, Zahlen, Strings, Tuples und None sind unveränderbar.
- `b = a` erzeugt keine Kopie: a und b zeigen auf dasselbe Objekt. Wird es verändert (z.B. mit `append`), sieht man die Änderung über beide Variablen (Seiteneffekt).
- Eine neue Zuweisung (`a = a + [7]`) lässt a auf ein neues Objekt zeigen, b bleibt unverändert. Achtung: `a += [7]` verändert die Liste.
- Methoden, die eine Liste verändern (`append`, `sort`, ...), geben `None` zurück.
- Kopien erhält man mit `a.copy()`, `list(a)` oder `a[:]`. Das sind flache Kopien. Enthält die Liste selbst veränderbare Objekte, braucht man eine tiefe Kopie mit `deepcopy`.
- Beim Funktionsaufruf zeigt der Parameter auf dasselbe Objekt wie das Argument (*call by assignment*). Verändert die Funktion das Objekt, sieht der Aufrufer die Änderung.
- Eine zweidimensionale Liste nicht mit `[[0] * n] * m` erzeugen, sondern jede Zeile einzeln.
