## Tuples, Dicts und Sets

Neben Strings und Listen hat Python drei weitere eingebaute Datenstrukturen, die man ständig braucht:

| Datenstruktur | Beispiel | wofür? |
|---|---|---|
| **Tuple** | `(3, 4)` | mehrere Werte fest zusammenfassen, z.B. einen Punkt oder ein Datum |
| **Dict** (Dictionary) | `{'Anna': 2, 'Ben': 1}` | zu einem **Schlüssel** schnell den zugehörigen **Wert** finden |
| **Set** (Menge) | `{2, 3, 5, 7}` | Elemente ohne Duplikate sammeln, schnell prüfen, ob etwas enthalten ist |

### Tuples

Ein Tuple ist eine Folge von Werten in runden Klammern. Wie bei einer Liste gibt es `len()`, Indexing, Slicing, `in`
und die `for`-Schleife. Der Unterschied: **Ein Tuple kann man nicht verändern.** Es gibt kein `append`, kein `pop`
und keine Zuweisung `t[i] = x`.

Ein Tuple nimmt man, wenn mehrere Werte **zusammengehören** und sich die Anzahl nicht ändert: ein Punkt `(x, y)`,
ein Datum `(tag, monat, jahr)`, eine Farbe `(rot, grün, blau)`.

```python
p = (3, 4)
datum = (29, 10, 2026)
print(p[0], p[1])
print(len(datum), datum[-1], datum[:2])
print(10 in datum)
```

```python
p = (3, 4)
p[0] = 5          # TypeError: ein Tuple kann man nicht verändern
```

Achtung bei Tuples mit nur einem Element: `(7)` ist einfach die Zahl 7 in Klammern. Für ein Tuple braucht man ein Komma: `(7,)`.
Die Klammern darf man oft weglassen, entscheidend ist das Komma.

```python
a = (7)
b = (7,)
c = 1, 2, 3        # auch ohne Klammern ein Tuple
leer = ()
print(type(a), type(b), type(c), len(leer))
```

##### Entpacken

Die Werte eines Tuples kann man in einem Schritt auf mehrere Variablen verteilen. Links müssen genauso viele Variablen
stehen, wie das Tuple Elemente hat. Damit lassen sich auch zwei Variablen ohne Hilfsvariable tauschen.

```python
p = (3, 4)
x, y = p
print(x, y)

tag, monat, jahr = (29, 10, 2026)
print(f'{tag}.{monat}.{jahr}')

a = 1
b = 2
a, b = b, a        # rechts entsteht das Tuple (2, 1), das wird entpackt
print(a, b)
```

##### Listen von Tuples

Sehr häufig ist eine **Liste von Tuples**: Die Liste ist veränderbar und kann wachsen, jedes Tuple ist ein Datensatz.
In der `for`-Schleife kann man jedes Tuple direkt entpacken.

`sorted()` sortiert Tuples zuerst nach dem ersten Element, bei Gleichheit nach dem zweiten usw.

```python
ergebnisse = [('Ben', 12), ('Anna', 15), ('Carla', 9)]
ergebnisse.append(('Dana', 15))

for name, punkte in ergebnisse:
    print(f'{name:6} {punkte:3}')

print(sorted(ergebnisse))
```

### Dicts

Ein Dict (Dictionary, Wörterbuch) speichert **Schlüssel-Wert-Paare**. Über den Schlüssel greift man auf den Wert zu,
so wie man in einem Wörterbuch zu einem Wort die Übersetzung nachschlägt. Die Schreibweise ist
`{schlüssel: wert, schlüssel: wert, ...}`.

Statt eines Index `0, 1, 2, ...` wie bei der Liste verwendet man den Schlüssel, meistens einen String oder eine Zahl.
Jeder Schlüssel kommt **höchstens einmal** vor.

```python
telefon = {'Anna': '0711 2345', 'Ben': '0176 98765', 'Carla': '07031 555'}
print(telefon['Ben'])
print(len(telefon))
print('Anna' in telefon, 'Dana' in telefon)    # in prüft die Schlüssel
```

Ein Zugriff mit einem Schlüssel, den es nicht gibt, führt zu einem `KeyError`. Mit `in` kann man vorher prüfen,
oder man nimmt `get`: `d.get(schlüssel, standard)` liefert den Wert oder, falls der Schlüssel fehlt, den Standardwert.

```python
telefon = {'Anna': '0711 2345', 'Ben': '0176 98765'}
print(telefon.get('Anna', 'unbekannt'))
print(telefon.get('Dana', 'unbekannt'))
print(telefon['Dana'])                         # KeyError
```

##### Dicts verändern

| Anweisung | Wirkung |
|---|---|
| `d[k] = w` | neues Paar hinzufügen, oder den Wert ändern, falls `k` schon Schlüssel ist |
| `del d[k]` | Paar mit dem Schlüssel `k` löschen |
| `d.pop(k)` | Paar löschen und den Wert zurückgeben |
| `{}` | leeres Dict |

```python
noten = {}                 # leeres Dict
noten['Anna'] = 2
noten['Ben'] = 3
noten['Ben'] = 1           # Wert ändern
noten['Carla'] = 4
del noten['Carla']
print(noten)
```

##### Dicts durchlaufen

Die `for`-Schleife durchläuft die **Schlüssel**. Mit `d.items()` erhält man die Paare als Tuples `(schlüssel, wert)`,
die man in der Schleife gleich entpacken kann. `d.keys()` liefert nur die Schlüssel, `d.values()` nur die Werte.

Die Reihenfolge ist die Reihenfolge, in der die Paare eingefügt wurden.

```python
preise = {'Brot': 2.49, 'Milch': 1.19, 'Käse': 3.79}

for ware in preise:
    print(ware, preise[ware])

for ware, preis in preise.items():
    print(f'{ware:6} {preis:5.2f} €')

print(list(preise.keys()))
print(round(sum(preise.values()), 2))
```

##### Typische Anwendung: Zählen

Wie oft kommt jedes Zeichen in einem Text vor? Die Zeichen sind die Schlüssel, die Anzahlen die Werte.
Kommt ein Zeichen zum ersten Mal vor, gibt es den Schlüssel noch nicht. Mit `get(c, 0)` erhält man in diesem Fall `0`.

```python
text = 'abrakadabra'
anzahl = {}
for c in text:
    anzahl[c] = anzahl.get(c, 0) + 1
print(anzahl)

# häufigstes Zeichen suchen
best = ''
for c, n in anzahl.items():
    if best == '' or n > anzahl[best]:
        best = c
print(best, anzahl[best])
```

##### Typische Anwendung: Nachschlagen statt `if`-Kette

Statt einer langen `if`-`elif`-Kette kann man die Zuordnung in einem Dict ablegen.

```python
tage = {1: 'Montag', 2: 'Dienstag', 3: 'Mittwoch', 4: 'Donnerstag',
        5: 'Freitag', 6: 'Samstag', 7: 'Sonntag'}
nr = 4
print(tage[nr])

englisch = {'Hund': 'dog', 'Katze': 'cat', 'Maus': 'mouse'}
satz = 'Hund Katze Maus Pferd'
woerter = []
for w in satz.split():
    woerter.append(englisch.get(w, '?'))
print(' '.join(woerter))
```

##### Datensätze

Ein Dict eignet sich auch als **Datensatz**, die Schlüssel sind dann die Attribute, so wie die Spalten einer Tabelle
in einer Datenbank. Eine Liste solcher Dicts entspricht einer ganzen Tabelle.

```python
schueler = [
    {'name': 'Anna', 'klasse': '11a', 'punkte': 13},
    {'name': 'Ben', 'klasse': '11b', 'punkte': 9},
    {'name': 'Carla', 'klasse': '11a', 'punkte': 11},
]

# entspricht SELECT name FROM schueler WHERE klasse = '11a'
for s in schueler:
    if s['klasse'] == '11a':
        print(s['name'])
```

##### Welche Schlüssel sind erlaubt?

Schlüssel müssen **unveränderbar** sein: Zahlen, Strings und Tuples sind erlaubt, Listen nicht.
Ein Tuple als Schlüssel ist z.B. praktisch für Koordinaten auf einem Spielfeld. Die Werte dürfen beliebig sein,
auch Listen oder andere Dicts.

```python
feld = {(0, 0): 'Turm', (0, 4): 'König', (1, 3): 'Bauer'}
print(feld[(0, 4)])
print(feld.get((5, 5), 'leer'))

kurse = {'Informatik': ['Anna', 'Ben'], 'Physik': ['Carla']}
kurse['Physik'].append('Dana')
print(kurse)
```

### Sets

Ein Set ist eine **Menge** im mathematischen Sinn: Jedes Element kommt höchstens einmal vor, und die Elemente haben
**keine Reihenfolge**. Deshalb gibt es auch kein Indexing und kein Slicing.

Ein Set schreibt man mit geschweiften Klammern `{1, 2, 3}`. Achtung: `{}` ist ein leeres **Dict**,
ein leeres Set erzeugt man mit `set()`.

```python
s = {3, 1, 4, 1, 5, 9, 2, 6, 5}
print(s, len(s))               # Duplikate verschwinden
print(4 in s, 7 in s)

leer = set()
print(type(leer), type({}))
```

##### Sets verändern

| Anweisung | Wirkung |
|---|---|
| `s.add(x)` | `x` hinzufügen (ist `x` schon drin, passiert nichts) |
| `s.remove(x)` | `x` entfernen, `KeyError`, falls `x` nicht enthalten ist |
| `s.discard(x)` | `x` entfernen, falls enthalten, sonst passiert nichts |
| `set(folge)` | Set aus einer Liste, einem String, ... erzeugen |

```python
farben = set()
farben.add('rot')
farben.add('blau')
farben.add('rot')
farben.discard('gelb')
print(farben)

print(set('Mississippi'))
```

##### Typische Anwendung: Duplikate entfernen und zählen

`set(liste)` entfernt alle Duplikate, `len(set(liste))` ist die Anzahl **verschiedener** Elemente.
Braucht man wieder eine Liste, wandelt man mit `list()` oder `sorted()` zurück.

```python
wuerfe = [3, 6, 1, 3, 3, 6, 2]
print(len(set(wuerfe)), 'verschiedene Augenzahlen')
print(sorted(set(wuerfe)))

wort = 'Habicht'
if len(set(wort.lower())) == len(wort):
    print(wort, 'hat keinen Buchstaben doppelt')
```

##### Typische Anwendung: Schon gesehen?

Ein Set merkt sich, was schon vorgekommen ist. Hier wird die erste Zahl gesucht, die doppelt vorkommt.

```python
zahlen = [4, 8, 15, 16, 8, 42, 4]
gesehen = set()
for k in zahlen:
    if k in gesehen:
        print('erste Wiederholung:', k)
        break
    gesehen.add(k)
```

##### Mengenoperationen

| Operator | Bedeutung | Mathematik |
|---|---|---|
| `a \| b` | Vereinigung: in `a` oder in `b` | $A \cup B$ |
| `a & b` | Schnittmenge: in `a` und in `b` | $A \cap B$ |
| `a - b` | Differenz: in `a`, aber nicht in `b` | $A \setminus B$ |
| `a <= b` | ist `a` Teilmenge von `b`? | $A \subseteq B$ |

```python
informatik = {'Anna', 'Ben', 'Carla', 'Dana'}
physik = {'Ben', 'Dana', 'Emil'}

print(informatik | physik)     # in mindestens einem der Kurse
print(informatik & physik)     # in beiden Kursen
print(informatik - physik)     # nur in Informatik
print({'Ben', 'Emil'} <= physik)
```

##### `in` ist bei Sets und Dicts schnell

Bei einer Liste muss `x in a` im schlimmsten Fall jedes Element ansehen (lineare Suche). Ein Set oder ein Dict
berechnet dagegen aus dem Element direkt, wo es gespeichert sein müsste (Hashing). Die Suche dauert deshalb fast
gleich lang, egal wie groß das Set ist. Bei vielen Suchen in großen Datenmengen ist das ein riesiger Unterschied.

```python
import time

n = 1_000_000
liste = list(range(n))
menge = set(liste)

start = time.perf_counter()
for i in range(100):
    n - 1 in liste
print(f'Liste: {time.perf_counter() - start:.4f} s')

start = time.perf_counter()
for i in range(100):
    n - 1 in menge
print(f'Set:   {time.perf_counter() - start:.4f} s')
```

### Überblick

| | Liste | Tuple | Dict | Set |
|---|---|---|---|---|
| Beispiel | `[7, 12, 7]` | `(3, 4)` | `{'Anna': 2}` | `{2, 3, 5}` |
| leer | `[]` | `()` | `{}` | `set()` |
| Zugriff | Index `a[0]` | Index `t[0]` | Schlüssel `d['Anna']` | kein Zugriff, nur `in` |
| Reihenfolge | ja | ja | Einfügereihenfolge | nein |
| Duplikate | erlaubt | erlaubt | Schlüssel eindeutig | nein |
| veränderbar | ja | **nein** | ja | ja |
| `in` prüft | Elemente (langsam) | Elemente (langsam) | Schlüssel (schnell) | Elemente (schnell) |
| durchlaufen | `for k in a:` | `for k in t:` | `for k, w in d.items():` | `for k in s:` |

Faustregeln:
- eine Folge von Werten, die sich ändern kann → **Liste**
- wenige zusammengehörende Werte, die fest bleiben → **Tuple**
- zu einem Schlüssel einen Wert nachschlagen oder etwas zählen → **Dict**
- Duplikate vermeiden, schnell prüfen, ob etwas enthalten ist, Mengenoperationen → **Set**
