## Bedingungen, boolesche Ausdrücke, Schleifen

### Bedingungen 

Mit bedingten Anweisungen können wir Verzweigungen im Programmfluss erreichen. Die Einrückung bestimmt, welche Anweisungen
in den Rumpf der if-Anweisung gehören.

##### if

````python
x = 3
if x > 1:
  print('A')
  print('B')
print('Ende')
````

##### if - else

````python
x = 0
if x > 1:
    print('A')
else:
    print('B')
````

##### if - elif - else
Diese Konstruktion nutzen wir für eine Fallunterscheidung: Nur der erste zutreffende Fall wird durchlaufen. 

````python
x = 2.5
if x > 3:
    print('A')
elif x > 2:
    print('B')
elif x > 1:
    print('C')
else:
    print('D')
print('Ende')
````

### Boolesche Ausdrücke 

Die Bedingungen in den if-Anweisungen sind *boolesche Ausdrücke*. Boolsche Ausdrücke werten sich immer zu einem der beiden Wahrheitswerte True oder False aus. 

Mit den Vergleichsoperatoren für Zahlen können wir boolesche Ausdrücke erstellen. 

````python
x = 2
print(x > 1)
print(x > 3)
````

Die Gleichheit zweier Zahlen prüfen wir mit ```==```, die Ungleichheit mit ```!=```

````python
i = 5
j = 7
print(i > j, i >= j, i < j, i <= j, i == j, i != j)
````

#### Beispiel  

Eine Zahl ist genau dann gerade, wenn der Rest bei Division durch 2 Null ist.

````python
x = int(input('Bitte eine ganze Zahl eingeben'))
if x % 2 == 0:
    print(f'Die Zahl {x} ist gerade.')
else:
    print(f'Die Zahl {x} ist ungerade.')
````

### Boolesche Operatoren

Mit den booleschen Operatoren ```and```, ```or``` und ```not``` können wir aus booleschen Ausdrücken kompliziertere boolesche Ausdrücke erstellen. Die Wahrheitstafeln für die Operatoren zeigen alle Fälle für die verschiedenen Wahrheitswerte.

Die Auswertungsreihenfolge ist ```not```, ```and```, ```or```. Im Zweifelsfall setzen wir Klammern.

### while-Schleife

Der Rumpf der while-Schleife wird solange durchlaufen, bis die while-Bedingung sich zu False auswertet. 

````python
x = 10
while x >= 4:
    print(x)
    x -= 2
print("End")
````

### for-Schleife mit range()

Die Funktion range() liefert Aufzählungen von ganzen Zahlen und kann ein, zwei oder drei Parameter erhalten.

````python
# der Bereich beginnt bei 0, endet bei 5 (5 ist ausgeschlossen)
for k in range(5):
    print(k)
````

````python
# der Bereich beginnt bei 2, endet bei 7 (7 ist ausgeschlossen)
for k in range(2,5):
    print(k)
````

````python
# der Bereich beginnt bei 3, endet bei 9 (9 ist ausgeschlossen) mit Schrittweite 2
for i in range(3, 9, 2):
    print(i)
````

````python
# der Bereich beginnt bei 5, endet bei -1 (-1 ist ausgeschlossen) mit Schrittweite -1
for i in range(5, -1, -1):
    print(i)
````

````python
# Berechne die Summe aller ungeraden Zahlen von 1 bis 100:
summe = 0
for i in range(1, 100, 2):
    summe += i
print(f"Die Summe aller ungeraden Zahlen von 1 bis 100 ist: {summe}")
````

**Faustregel:** For-Schleifen nutzen wir meist dann, wenn die Anzahl der Schleifendurchgänge bekannt ist, andernfalls nutzen wir while-Schleifen.

### break und continue

Mit *break* verlassen wir eine Schleife und setzen den Ablauf hinter der Schleife fort. Mit *continue* hören mit dem laufenden Schleifendurchgang auf und beginnen den nächsten Schleifendurchgang. *break* und *continue* können nur in Schleifen verwendete werden.

````python
x = 2
while True:
    print(x)
    if x == 4: 
        break
    x+=1
print('End')
````

````python
for i in range(8):
    if i % 2 == 0: 
        continue
    print(i)
````

````python
# Die Schleife läuft, bis der Benutzer 0 eingibt
print('Gib nacheinander Zahlen ein (0 für Ende)')
summe = 0
while True:
    zahl = int(input())
    if zahl == 0:
        break
    summe = summe + zahl
print(f'Die Summe aller eingegebenen Zahlen ist {summe}')
````
