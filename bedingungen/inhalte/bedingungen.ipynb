## BE 1 · Bedingte Anweisungen

### Inhalte

Mit **bedingten Anweisungen** verzweigen wir den Programmablauf: Ein Teil des Programms wird nur ausgeführt, wenn eine Bedingung erfüllt ist.

##### if

Nach der Bedingung steht ein **Doppelpunkt**. Die **Einrückung** bestimmt, welche Anweisungen in den Rumpf der if-Anweisung gehören. Üblich sind vier Leerzeichen.

````python
x = 3
if x > 1:
    print('A')     # gehört zum Rumpf
    print('B')     # gehört zum Rumpf
print('Ende')      # nicht eingerückt: wird immer ausgeführt
````

##### if – else

Der else-Zweig wird ausgeführt, wenn die Bedingung **nicht** erfüllt ist. Es wird also immer genau einer der beiden Zweige durchlaufen.

````python
x = 0
if x > 1:
    print('A')
else:
    print('B')     # B
````

##### if – elif – else

Diese Konstruktion nutzen wir für eine **Fallunterscheidung**. Die Bedingungen werden von oben nach unten geprüft. **Nur der erste zutreffende Fall wird durchlaufen**, alle weiteren werden übersprungen. Trifft kein Fall zu, wird der else-Zweig ausgeführt.

````python
x = 2.5
if x > 3:
    print('A')
elif x > 2:
    print('B')     # B – der erste zutreffende Fall
elif x > 1:
    print('C')     # würde auch zutreffen, wird aber übersprungen
else:
    print('D')
print('Ende')
````

In Python ist die Einrückung also **Teil der Sprache** und nicht nur eine Lesehilfe.

## BE 2 · Boolesche Ausdrücke und Vergleiche

### Inhalte

Die Bedingungen in den if-Anweisungen sind **boolesche Ausdrücke**. Boolesche Ausdrücke werten sich immer zu einem der beiden **Wahrheitswerte** `True` oder `False` aus. Der Datentyp heißt `bool`.

Mit den **Vergleichsoperatoren** erstellen wir boolesche Ausdrücke:

| Operator | Bedeutung | Beispiel mit `i = 5`, `j = 7` | Ergebnis |
|:--|:--|:--|:--|
| `>` | größer | `i > j` | `False` |
| `>=` | größer oder gleich | `i >= j` | `False` |
| `<` | kleiner | `i < j` | `True` |
| `<=` | kleiner oder gleich | `i <= j` | `True` |
| `==` | gleich | `i == j` | `False` |
| `!=` | ungleich | `i != j` | `True` |

````python
x = 2
print(x > 1)       # True
print(x > 3)       # False
````

````python
i = 5
j = 7
print(i > j, i >= j, i < j, i <= j, i == j, i != j)
# False False True True False True
````

**Vorsicht:** `=` und `==` sind verschieden. Mit `=` **weisen** wir einen Wert **zu**, mit `==` **vergleichen** wir zwei Werte. `if x = 5:` ist ein Syntaxfehler.

Einen booleschen Ausdruck können wir auch in einer Variablen speichern:

````python
alter = 16
volljaehrig = alter >= 18
print(volljaehrig)     # False
````

##### Beispiel

Eine Zahl ist genau dann gerade, wenn der Rest bei der Division durch 2 null ist.

````python
x = int(input('Bitte eine ganze Zahl eingeben: '))
if x % 2 == 0:
    print(f'Die Zahl {x} ist gerade.')
else:
    print(f'Die Zahl {x} ist ungerade.')
````

## BE 3 · Boolesche Operatoren

### Inhalte

Mit den **booleschen Operatoren** `and`, `or` und `not` setzen wir aus booleschen Ausdrücken kompliziertere boolesche Ausdrücke zusammen.

- `a and b` ist nur dann `True`, wenn **beide** Teile `True` sind.
- `a or b` ist `True`, wenn **mindestens einer** der beiden Teile `True` ist.
- `not a` kehrt den Wahrheitswert um.

Die **Wahrheitstafeln** zeigen alle Fälle (0 steht für `False`, 1 für `True`):

<img src="./wahrheitstafeln.png" width="500" alt="Wahrheitstafeln für and, or und not">

Die **Auswertungsreihenfolge** ist `not` vor `and` vor `or`. Im Zweifelsfall setzen wir Klammern.

````python
alter = 15
print(alter >= 13 and alter <= 17)    # True   Jugendlicher: beide Bedingungen erfüllt
print(alter < 6 or alter >= 65)       # False  keine der beiden Bedingungen erfüllt
print(not alter >= 18)                # True   nicht volljährig
````

````python
tag = 'Sonntag'
if tag == 'Samstag' or tag == 'Sonntag':
    print('Wochenende')
else:
    print('Werktag')
````

#### Lösung

| | Ausdruck | Rechnung | Ergebnis |

| G | `b and c or a` | (False and True) or True = False or True | `True` |
| H | `b and (c or a)` | False and (True or True) = False and True | `False` |

**G und H zeigen, warum die Reihenfolge wichtig ist:** Ohne Klammern wird `and` vor `or` ausgewertet. Mit Klammern ändert sich das Ergebnis.

````python
# Zum Nachprüfen
a = True
b = False
c = True
print(not a, a and b, a or b, not a or c)
print(a and not b, not (a and b), b and c or a, b and (c or a))
````

## BE 4 · while-Schleife

### Inhalte

Der Rumpf der **while-Schleife** wird so lange wiederholt, wie die Bedingung `True` ist. Die Bedingung wird **vor jedem Durchlauf** geprüft. Sobald sie `False` ergibt, geht es hinter der Schleife weiter.

````python
x = 10
while x >= 4:
    print(x)       # 10 8 6 4
    x -= 2
print('Ende')
````

Im Rumpf muss sich etwas ändern, das die Bedingung beeinflusst. Sonst bleibt die Bedingung immer `True`, und wir erhalten eine **Endlosschleife**. Eine Endlosschleife brechen wir in Jupyter mit dem Stopp-Knopf ■ ab (oder im Menü **Kernel → Interrupt Kernel**).

Ein typisches Beispiel: Wir verdoppeln eine Zahl, bis sie größer als 100 ist. Wie viele Durchläufe nötig sind, wissen wir vorher nicht.

````python
zahl = 3
while zahl <= 100:
    zahl = zahl * 2
print(zahl)        # 192
````

### Aufgabe 1

Was gibt das Programm aus? Lege eine Tabelle an, die nach jedem Durchlauf den Wert von `x` festhält.

````python
x = 1
while x < 50:
    x = x * 3
    print(x)
````

#### Lösung

| Durchlauf | Bedingung `x < 50` vorher | `x` danach | Ausgabe |
|:--|:--|:--|:--|
| 1 | 1 < 50 ✓ | 3 | 3 |
| 2 | 3 < 50 ✓ | 9 | 9 |
| 3 | 9 < 50 ✓ | 27 | 27 |
| 4 | 27 < 50 ✓ | 81 | 81 |
| – | 81 < 50 ✗ | | Schleife endet |

**Beachte:** Es wird auch 81 ausgegeben, obwohl 81 größer als 50 ist. Die Bedingung wird nur vor jedem Durchlauf geprüft, nicht mitten im Rumpf.

### Aufgabe 2

Ein Guthaben von 1000 € wird jedes Jahr mit 5 % verzinst. Schreibe ein Programm, das berechnet, nach wie vielen Jahren sich das Guthaben mindestens verdoppelt hat. Gib die Anzahl der Jahre und das Guthaben mit zwei Nachkommastellen aus.

#### Lösung

Wie viele Jahre es dauert, wissen wir vorher nicht, deshalb ist eine while-Schleife richtig:

````python
guthaben = 1000
jahre = 0
while guthaben < 2000:
    guthaben = guthaben * 1.05
    jahre += 1
print(f'Nach {jahre} Jahren beträgt das Guthaben {guthaben:.2f} €.')
# Nach 15 Jahren beträgt das Guthaben 2078.93 €.
````

`jahre` muss **vor** der Schleife auf 0 gesetzt werden. In jedem Durchlauf wird das Guthaben verzinst und `jahre` um 1 erhöht.

## BE 5 · for-Schleife mit range()

### Inhalte

Die **for-Schleife** durchläuft nacheinander alle Werte einer Aufzählung. Die Funktion `range()` liefert Aufzählungen von ganzen Zahlen und kann einen, zwei oder drei Parameter erhalten:

| Aufruf | Bedeutung | liefert |
|:--|:--|:--|
| `range(5)` | von 0 bis 5 (ausschließlich) | 0, 1, 2, 3, 4 |
| `range(2, 5)` | von 2 bis 5 (ausschließlich) | 2, 3, 4 |
| `range(3, 9, 2)` | von 3 bis 9 (ausschließlich), Schrittweite 2 | 3, 5, 7 |
| `range(5, -1, -1)` | von 5 bis -1 (ausschließlich), Schrittweite -1 | 5, 4, 3, 2, 1, 0 |

Wie beim Slicing gilt: **Die bis-Zahl wird nicht mitgenommen.**

````python
# der Bereich beginnt bei 0, endet bei 5 (5 ist ausgeschlossen)
for k in range(5):
    print(k)
````

````python
# der Bereich beginnt bei 2, endet bei 5 (5 ist ausgeschlossen)
for k in range(2, 5):
    print(k)
````

````python
# der Bereich beginnt bei 3, endet bei 9 (9 ist ausgeschlossen), Schrittweite 2
for i in range(3, 9, 2):
    print(i)
````

````python
# der Bereich beginnt bei 5, endet bei -1 (-1 ist ausgeschlossen), Schrittweite -1
for i in range(5, -1, -1):
    print(i)
````

Ein typisches Beispiel ist das Aufsummieren. Die Variable `summe` wird **vor** der Schleife auf 0 gesetzt:

````python
# Summe aller ungeraden Zahlen von 1 bis 100
summe = 0
for i in range(1, 100, 2):
    summe += i
print(f'Die Summe aller ungeraden Zahlen von 1 bis 100 ist {summe}.')    # 2500
````

**Faustregel:** for-Schleifen nutzen wir meist dann, wenn die Anzahl der Schleifendurchgänge bekannt ist. Andernfalls nutzen wir while-Schleifen.

## BE 6 · break und continue

### Inhalte

- Mit **`break`** verlassen wir eine Schleife sofort und setzen den Ablauf hinter der Schleife fort.
- Mit **`continue`** beenden wir nur den laufenden Schleifendurchgang und beginnen sofort den nächsten.

`break` und `continue` können nur in Schleifen verwendet werden.

Mit `while True` schreiben wir eine Schleife, deren Bedingung immer erfüllt ist. Sie endet erst durch ein `break`:

````python
x = 2
while True:
    print(x)       # 2 3 4
    if x == 4:
        break
    x += 1
print('Ende')
````

Mit `continue` überspringen wir hier alle geraden Zahlen:

````python
for i in range(8):
    if i % 2 == 0:
        continue
    print(i)       # 1 3 5 7
````

`while True` mit `break` ist praktisch, wenn wir Eingaben lesen, bis eine bestimmte Eingabe kommt:

````python
# Die Schleife läuft, bis der Benutzer 0 eingibt
print('Gib nacheinander Zahlen ein (0 für Ende)')
summe = 0
while True:
    zahl = int(input())
    if zahl == 0:
        break
    summe = summe + zahl
print(f'Die Summe aller eingegebenen Zahlen ist {summe}.')
````

### Aufgabe 2

Schreibe eine Passwortabfrage: Das Programm fragt so lange nach dem Passwort, bis `geheim` eingegeben wird. Danach gibt es aus, wie viele Versuche nötig waren.

Beispielablauf:
```
Passwort: hallo
Falsch, bitte noch einmal.
Passwort: geheim
Richtig nach 2 Versuchen.
```

#### Lösung

````python
versuche = 0
while True:
    eingabe = input('Passwort: ')
    versuche += 1
    if eingabe == 'geheim':
        break
    print('Falsch, bitte noch einmal.')
print(f'Richtig nach {versuche} Versuchen.')
````

Die Anzahl der Versuche ist vorher nicht bekannt, deshalb eine while-Schleife. `versuche` wird erhöht, **bevor** geprüft wird, damit auch der richtige Versuch mitgezählt wird.

*Ohne `break` geht es auch:*

````python
versuche = 1
eingabe = input('Passwort: ')
while eingabe != 'geheim':
    print('Falsch, bitte noch einmal.')
    eingabe = input('Passwort: ')
    versuche += 1
print(f'Richtig nach {versuche} Versuchen.')
````
