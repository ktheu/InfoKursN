### Bedingungen 

Mit **bedingten Anweisungen** verzweigen wir den Programmablauf: Ein Teil des Programms wird nur ausgeführt, wenn eine Bedingung erfüllt ist.

##### if

Nach der Bedingung steht ein **Doppelpunkt**. Die **Einrückung** bestimmt, welche Anweisungen in den Rumpf der if-Anweisung gehören. Üblich sind vier Leerzeichen.

````python
x = 0
if x > 1:
    print('A')
    print('B')
print('Ende')
````

##### if - else

Der else-Zweig wird ausgeführt, wenn die Bedingung **nicht** erfüllt ist. Es wird also immer genau einer der beiden Zweige durchlaufen.

````python
x = 0
if x > 1:
    print('A')
else:
    print('B')
````

##### if - elif - else

Diese Konstruktion nutzen wir für eine **Fallunterscheidung**. Die Bedingungen werden von oben nach unten geprüft. **Nur der erste zutreffende Fall wird durchlaufen**, alle weiteren werden übersprungen. Trifft kein Fall zu, wird der else-Zweig ausgeführt.

````python
x = 2.5
if x > 3:
    print('A')
elif x > 2:
    print('B')
elif x > 1:
    print('C')
else:
    print('D')
print('Ende')
````

### Boolesche Ausdrücke 

Die Bedingungen in den if-Anweisungen sind **boolesche Ausdrücke**. Boolesche Ausdrücke werten sich immer zu einem der beiden **Wahrheitswerte** `True` oder `False` aus. Der Datentyp heißt `bool`.

Mit den Vergleichsoperatoren für Zahlen können wir boolesche Ausdrücke erstellen. 

| Operator | Bedeutung | Beispiel mit `i = 5`, `j = 7` | Ergebnis |
|:--|:--|:--|:--|
| `>` | größer | `i > j` | `False` |
| `>=` | größer oder gleich | `i >= j` | `False` |
| `<` | kleiner | `i < j` | `True` |
| `<=` | kleiner oder gleich | `i <= j` | `True` |
| `==` | gleich | `i == j` | `False` |
| `!=` | ungleich | `i != j` | `True` |

````python
x = 2
print(x > 1)
print(x > 3)
````

**Vorsicht:** `=` und `==` sind verschieden. Mit `=` **weisen** wir einen Wert **zu**, mit `==` **vergleichen** wir zwei Werte. `if x = 5:` ist ein Syntaxfehler.

Einen booleschen Ausdruck können wir auch in einer Variablen speichern:

````python
alter = 16
volljaehrig = alter >= 18
print(volljaehrig)      
````

#### Beispiel  

Eine Zahl ist genau dann gerade, wenn der Rest bei Division durch 2 Null ist.

````python
x = int(input('Bitte eine ganze Zahl eingeben'))
if x % 2 == 0:
    print(f'Die Zahl {x} ist gerade.')
else:
    print(f'Die Zahl {x} ist ungerade.')
````

### Boolesche Operatoren

Mit den **booleschen Operatoren** `and`, `or` und `not` setzen wir aus booleschen Ausdrücken kompliziertere boolesche Ausdrücke zusammen.

- `a and b` ist nur dann `True`, wenn **beide** Teile `True` sind.
- `a or b` ist `True`, wenn **mindestens einer** der beiden Teile `True` ist.
- `not a` kehrt den Wahrheitswert um.

Die Auswertungsreihenfolge ist ```not```, ```and```, ```or```. Im Zweifelsfall setzen wir Klammern.
<img src='./wahrheitstafeln.png' width='500'>

Beispiele:

````python
alter = 15
print(alter >= 13 and alter <= 17)    
````

````python
print(alter < 6 or alter >= 65)  
````

````python
print(not alter >= 18)       
````


### while-Schleife

Der Rumpf der **while-Schleife** wird so lange wiederholt, wie die Bedingung `True` ist. Die Bedingung wird **vor jedem Durchlauf** geprüft. Sobald sie `False` ergibt, geht es hinter der Schleife weiter.

````python
x = 10
while x >= 4:
    print(x)
    x -= 2
print("End")
````

Im Rumpf muss sich etwas ändern, das die Bedingung beeinflusst. Sonst bleibt die Bedingung immer `True`, und wir erhalten eine **Endlosschleife**. Eine Endlosschleife brechen wir in Jupyter mit dem Stopp-Knopf ■ ab (oder im Menü **Kernel → Interrupt Kernel**).

Ein typisches Beispiel: Wir verdoppeln eine Zahl, bis sie größer als 100 ist. Wie viele Durchläufe nötig sind, wissen wir vorher nicht.

````python
zahl = 3
while zahl <= 100:
    zahl = zahl * 2
print(zahl)        
````

### for-Schleife mit range()

Die **for-Schleife** durchläuft nacheinander alle Werte einer Aufzählung. Die Funktion `range()` liefert Aufzählungen von ganzen Zahlen und kann einen, zwei oder drei Parameter erhalten:

| Aufruf | Bedeutung | liefert |
|:--|:--|:--|
| `range(5)` | von 0 bis 5 (ausschließlich) | 0, 1, 2, 3, 4 |
| `range(2, 5)` | von 2 bis 5 (ausschließlich) | 2, 3, 4 |
| `range(3, 9, 2)` | von 3 bis 9 (ausschließlich), Schrittweite 2 | 3, 5, 7 |
| `range(5, -1, -1)` | von 5 bis -1 (ausschließlich), Schrittweite -1 | 5, 4, 3, 2, 1, 0 |

Wie beim Slicing gilt: **Die bis-Zahl wird nicht mitgenommen.**

````python
# der Bereich beginnt bei 0, endet bei 5 (5 ist ausgeschlossen)
for k in range(5):
    print(k)
````

````python
# der Bereich beginnt bei 2, endet bei 5 (5 ist ausgeschlossen)
for k in range(2,5):
    print(k)
````

````python
# der Bereich beginnt bei 3, endet bei 9 (9 ist ausgeschlossen) mit Schrittweite 2
for i in range(3, 9, 2):
    print(i)
````

````python
# der Bereich beginnt bei 5, endet bei -1 (-1 ist ausgeschlossen) mit Schrittweite -1
for i in range(5, -1, -1):
    print(i)
````

Ein typisches Beispiel ist das Aufsummieren. Die Variable `summe` wird **vor** der Schleife auf 0 gesetzt:

````python
# Berechne die Summe aller ungeraden Zahlen von 1 bis 100:
summe = 0
for i in range(1, 100, 2):
    summe += i
print(f"Die Summe aller ungeraden Zahlen von 1 bis 100 ist: {summe}")
````

**Faustregel:** For-Schleifen nutzen wir meist dann, wenn die Anzahl der Schleifendurchgänge bekannt ist, andernfalls nutzen wir while-Schleifen.

### break und continue

- Mit **`break`** verlassen wir eine Schleife sofort und setzen den Ablauf hinter der Schleife fort.
- Mit **`continue`** beenden wir nur den laufenden Schleifendurchgang und beginnen sofort den nächsten.

`break` und `continue` können nur in Schleifen verwendet werden.

Mit `while True` schreiben wir eine Schleife, deren Bedingung immer erfüllt ist. Sie endet erst durch ein `break`:

````python
x = 2
while True:
    print(x)
    if x == 4: 
        break
    x+=1
print('End')
````

Mit `continue` überspringen wir hier alle geraden Zahlen:

````python
for i in range(8):
    if i % 2 == 0: 
        continue
    print(i)
````

`while True` mit `break` ist praktisch, wenn wir Eingaben lesen, bis eine bestimmte Eingabe kommt:

````python
# Die Schleife läuft, bis der Benutzer 0 eingibt
print('Gib nacheinander Zahlen ein (0 für Ende)')
summe = 0
while True:
    zahl = int(input())
    if zahl == 0:
        break
    summe = summe + zahl
print(f'Die Summe aller eingegebenen Zahlen ist {summe}')
````
