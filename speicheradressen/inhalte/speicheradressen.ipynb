## Speicheradressen (memory address)

[Video](https://youtu.be/YLPw6zjepsM) - [AufgabenVideo](https://youtu.be/OfBHBAJy5kA)

![](speicher0.png)

Den Computerspeicher stellen wir uns vor als eine riesige Sammlung von Bits (0 und 1). 8 Bits werden zu einem Byte zusammengefasst. In der Abbildung ist der Speicher als eine Kette von Bytes dargestellt. Jedes Byte im Speicher hat eine Adresse. Über die Adressen können wir an bestimmte Stellen des Speichers springen.
Adressen sind selbst Zahlen, die üblicherweise hexadezimal angegeben werden. In einem 32-Bit Betriebssystem darf eine Zahl, die eine Speicheradresse sein soll, höchstens 32 Bit lang sein.

In einem 32-Bit Betriebssystem gibt es $2^{32}$ Adressen, von 0 bis $2^{32}-1$ (0x00000000 - 0xFFFFFFFF).

```python
bin(2**32-1)
```

```python
int('FFFFFFFF',16)
```

```python
2**32
```

```python
# Hier wird Kilo mit 1024 benutzt, wie in diesem Zusammenhang üblich
bdict = {1:'KiloByte', 2:'MegaByte', 3:'GigaByte', 4:'TeraByte', 5:'PetaByte', 6:'ExaByte'}
x = 2**32
for k in bdict:
    print(f'{x} Byte = {x//(1024**k)} {bdict[k]}')
    if k == 3: break
```

Ein 32-Bit Betriebssystem kann einen Speicherbereich bis 4 GB Speicher verwalten. Diese Begrenzung war einer der Hauptgründe für die Einführung von 64-Bit Betriebssystemen.

```python
x = 2**64
for k in bdict:
    print(f'{x} Byte = {x//(1024**k)} {bdict[k]}')
```

Ein 64-Bit Betriebssystem kann einen Speicherbereich bis 16 ExaByte Speicher verwalten.

Mit der Abschätzung $2^{10} \approx 10^3$ aus dem Notebook zu Variablen geht das auch im Kopf:
$2^{32} = 2^2 \cdot 2^{30} = 4 \cdot 1024^3$ Byte, also 4 GigaByte, und $2^{64} = 2^4 \cdot 2^{60} = 16 \cdot 1024^6$ Byte, also 16 ExaByte.

#### Zuweisungen

Im Speicherbereich können wir die binäre Darstellung von Objekten speichern. In der Abbildung benötigt Objekt $x_1$ nur 1 Byte. Das Objekt $x_2$ beginnt bei Speicheradresse 0x1001 und benötigt 3 Byte. Der nächste freie Speicherplatz ist also 0x1004.

![](speicher.png)

Bei der Zuweisung `a = 2` erzeugt der Python Memory Manager in einem freien Speicherbereich ein Objekt mit dem Inhalt 2 und a ist ein Name für die Adresse, an der dieses Objekt im Speicher beginnt. Genau genommen ist a also nicht identisch mit der 2, sondern mit der Startadresse, an der die 2 gespeichert wurde. Man sagt dazu auch: *a ist ein Zeiger auf die 2*. **Alle Variablen in Python sind Zeiger auf Speicheradressen.** Die Adresse, auf die eine Variable zeigt, kann man sich mit der Funktion `id()` ausgeben lassen.

In den meisten Fällen sagen wir weiterhin: *In der Variablen a speichern wir die 2*, auch wenn damit eigentlich gemeint ist: *Die Adresse, an der die 2 im Speicher steht, nennen wir a*.

![](speicher2.png)

Anmerkung: Manchmal wird gesagt, dass Variablen *labels* oder *tags* für die Objekte sind und keine *Zeiger*. Der Begriff *Zeiger* ist dann für solche Fälle reserviert, wo die Sprache gezielte Änderungen an den Adressen zulässt. In Python gibt es keine *pointer-arithmetic* wie in C oder C++, aber ansonsten ist das Bild vom Zeiger zutreffend.

```python
a = 2
hex(id(a))
```

Die Anweisung `b = a` bewirkt, dass b auf dieselbe Adresse zeigt, auf die a zeigt. Die Gleichheit der Adressen können wir mit `is` bzw. `is not` überprüfen.

```python
b = a
hex(id(b))
```

```python
a is b
```

Bekommt `a` danach einen neuen Wert, dann zeigt `a` auf ein anderes Objekt. `b` zeigt weiter auf das alte Objekt.
Eine Zuweisung verändert also nie ein Objekt, sie lässt nur die Variable auf ein anderes Objekt zeigen.

```python
a = 2
b = a
a = a + 1
print(a, b)
print(hex(id(a)), hex(id(b)))
print(a is b)
```

#### `==` und `is`

Python kennt zwei Arten von Gleichheit:

| Ausdruck | prüft | Beispiel |
|---|---|---|
| `a == b` | Haben die Objekte den **gleichen Wert**? | `2 == 2.0` ist `True` |
| `a is b` | Zeigen a und b auf **dasselbe Objekt**, also auf dieselbe Adresse? | `2 is 2.0` ist `False` |

`a is b` ist dasselbe wie `id(a) == id(b)`. Wenn `a is b` gilt, dann gilt immer auch `a == b`, umgekehrt nicht.

Bei Listen sieht man den Unterschied gut: Jedes `[...]` erzeugt eine neue Liste, auch wenn sie den gleichen Inhalt hat.

```python
a = [1, 2]
b = [1, 2]
c = a
print(a == b, a is b)
print(a == c, a is c)
print(hex(id(a)), hex(id(b)), hex(id(c)))
```

#### Interning

*Interning* verwendet Python als eine Maßnahme zur Speicheroptimierung. Dabei werden Objekte, die schon existieren, wiederverwendet.
Dazu gehören alle Zahlen von -5 bis 256 und Strings, die so aussehen, als könnten sie Variablennamen sein.

![](interning.png)

```python
# Das Objekt 5 wird nur einmal erzeugt.
a = 5
b = 5
print(hex(id(a)))
print(hex(id(b)))
```

```python
print(a is b)
print(a == b)
```

```python
# Das Objekt 500 wird zweimal erzeugt.
a = 500
b = 500
print(hex(id(a)))
print(hex(id(b)))
```

```python
print(a is b)
print(a == b)
```

```python
# Das Objekt 'Hallo' wird nur einmal erzeugt
a = 'Hallo'
b = 'Hallo'
print(hex(id(a)))
print(hex(id(b)))
```

```python
print(a is b)
print(a == b)
```

```python
# Das Objekt 'Hallo Welt' wird zweimal erzeugt
a = 'Hallo Welt'
b = 'Hallo Welt'
print(hex(id(a)))
print(hex(id(b)))
```

```python
print(a is b)
print(a == b)
```

Interning gilt auch für Werte, die erst beim Rechnen entstehen: Ergibt eine Rechnung 100, bekommt man das schon
vorhandene Objekt 100. Ergibt sie 500, wird ein neues Objekt erzeugt.

```python
h = 100
a = 100
b = 500
print(h + h - a is a)          # 100 liegt im Bereich -5 bis 256
print(h * 5 is b, h * 5 == b)  # 500 wird neu erzeugt
```

**Achtung, das Ergebnis hängt davon ab, wie der Code ausgeführt wird.** Jupyter übersetzt jede Anweisung einer Zelle
einzeln. Wird dagegen ein ganzes Programm auf einmal übersetzt, z.B. in Thonny oder mit `python datei.py`, dann bemerkt
Python, dass im Programmtext zweimal derselbe Wert `500` steht, und legt das Objekt nur einmal an. Dann ist `a is b`
plötzlich `True`. Mit `exec` können wir das hier nachstellen: `exec` übersetzt den ganzen String auf einmal.

```python
exec('a = 500\nb = 500\nprint(a is b)')
exec("a = 'Hallo Welt'\nb = 'Hallo Welt'\nprint(a is b)")
```

Das Interning ist eine Implementationsentscheidung der jeweiligen Python-Version (CPython), und wie wir gesehen haben,
hängt das Ergebnis sogar von der Art der Ausführung ab. Deshalb sollte der `is`-Operator nicht zum Abfragen der
Gleichheit von Objekten benutzt werden, sondern der `==`-Operator. Ausnahme: Das Objekt `None`.

#### None

Auch None ist ein Objekt. Es hat einen eigenen Typ.

```python
type(None)
```

Das Objekt None ist ein *Singleton*. Es gibt nur ein Objekt vom Typ NoneType. Ein zweites lässt sich nicht erzeugen. Alle Variablen, die auf None gesetzt werden, zeigen auf die gleiche Speicheradresse. Deswegen kann die Abfrage auf None mit dem `is`-Operator bzw. `is not`-Operator durchgeführt werden.

```python
a = None
b = None

print(hex(id(a)))
print(hex(id(b)))
```

```python
a is None
```

```python
d = 5
d is not None
```

`None` steht für „kein Wert“. Eine Funktion ohne `return` gibt `None` zurück. Das nutzt man z.B. bei einer Suche:
Wird nichts gefunden, ist das Ergebnis `None`. Auch als Startwert für „noch kein Wert vorhanden“ ist `None` praktisch.

```python
def erstes_langes_wort(woerter, laenge):
    for w in woerter:
        if len(w) >= laenge:
            return w
    # kein return erreicht: die Funktion gibt None zurück

ergebnis = erstes_langes_wort(['Hund', 'Katze', 'Maus'], 7)
if ergebnis is None:
    print('nicht gefunden')
else:
    print(ergebnis)

best = None                    # noch kein bester Wert
for x in [4, 9, 2]:
    if best is None or x > best:
        best = x
print(best)
```

#### Zusammenfassung

- Jedes Byte im Speicher hat eine Adresse. Mit 32-Bit Adressen kann man $2^{32}$ Byte = 4 GB adressieren, mit 64-Bit Adressen 16 ExaByte.
- Variablen in Python sind Zeiger auf Objekte im Speicher. `id(a)` liefert die Adresse.
- `b = a` lässt b auf dasselbe Objekt zeigen wie a. Eine neue Zuweisung an a ändert daran nichts für b.
- `==` vergleicht die Werte, `is` prüft, ob es dasselbe Objekt ist.
- Wegen Interning zeigen gleiche kleine Zahlen und kurze Strings oft auf dasselbe Objekt. Darauf darf man sich nicht verlassen: Gleichheit immer mit `==` prüfen.
- Nur bei `None` verwendet man `is None` bzw. `is not None`.
