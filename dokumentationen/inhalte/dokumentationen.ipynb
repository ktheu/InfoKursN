### Dokumentationen nutzen

Niemand kennt alle Funktionen von Python auswendig. Wichtiger ist, dass wir schnell nachsehen können, was eine Funktion macht.
Eine **Dokumentation** beantwortet drei Fragen:

1. Was macht die Funktion oder Klasse?
2. Was müssen wir ihr übergeben?
3. Was bekommen wir zurück?

Wir lernen, die Hilfe von Python zu lesen, die Dokumentation von Bibliotheken zu nutzen und mit fremdem Code zu arbeiten, ohne ihn
vollständig zu lesen. Die meisten Dokumentationen sind auf Englisch.

#### `help()`

`help(name)` zeigt den Docstring einer Funktion, Methode, Klasse oder eines Moduls an. Die erste Zeile ist die **Signatur**: der Name
mit den Parametern. Darunter steht die Beschreibung.

```python
help(len)
```

Bei Methoden schreiben wir den Typ davor. Der Parameter `self` steht für das Objekt selbst, wir übergeben ihn nicht.
Funktionen aus einem Modul können wir erst nach dem `import` nachsehen.

```python
help(str.replace)
print('banane'.replace('a', 'o', 1))
```

```python
import random

help(random.randint)
```

#### Die Signatur lesen

Die Signatur sagt uns genau, wie wir eine Funktion aufrufen dürfen.

| Schreibweise | Bedeutung | Beispiel |
|---|---|---|
| `name` | Pflichtargument | `random.randint(a, b)` |
| `name=wert` | optional, ohne Angabe gilt der Standardwert | `round(number, ndigits=None)` |
| `/` | die Argumente davor werden nur über ihre Position übergeben | `len(obj, /)` |
| `*` | die Argumente danach werden nur mit Namen übergeben | `sorted(iterable, /, *, key=None, reverse=False)` |
| `*args` | beliebig viele Argumente | `print(*args, sep=' ', end='\n')` |
| `**kwargs` | beliebig viele Argumente mit Namen | `S.format(*args, **kwargs)` |
| `[...]` | optional | `S.find(sub[, start[, end]])` |

Bei `sorted` steht `reverse` hinter dem `*`. Deshalb müssen wir `reverse=True` schreiben, sonst gibt es einen `TypeError`.

```python
help(sorted)
```

```python
a = [3, 1, 2]
print(sorted(a, reverse=True))
print(1, 2, 3, sep='-', end='!\n')
print(sorted(a, True))
```

#### Wichtige Wörter in Dokumentationen

| Englisch | Bedeutung |
|---|---|
| return, returns | gibt zurück |
| optional | darf weggelassen werden |
| default | Standardwert |
| raises | löst einen Fehler aus, z. B. `Raises ValueError` |
| in place | verändert das Objekt selbst |
| iterable | etwas, das wir mit `for` durchlaufen können: Liste, String, `range`, ... |
| sequence | Folge mit Indizes: Liste, String, Tuple, ... |
| separator | Trennzeichen |
| substring | Teilstring |
| including both end points | beide Grenzen gehören dazu |

#### Was wird zurückgegeben?

Die Dokumentation sagt auch, ob eine Funktion ein neues Objekt erzeugt oder das vorhandene verändert. Das ist eine häufige
Fehlerquelle: Bei `sorted` steht *Return a new list*, bei `list.sort` dagegen *Sort the list in ascending order and return None*.

```python
a = [3, 1, 2]
b = sorted(a)
print(a, b)
c = a.sort()
print(a, c)
```

#### Die Dokumentation einer Bibliothek nutzen

Die offizielle Dokumentation von Python steht unter [docs.python.org/3/library](https://docs.python.org/3/library/). Dort ist jedes
Modul mit allen Funktionen und vielen Beispielen beschrieben, zum Beispiel
[random](https://docs.python.org/3/library/random.html) und [statistics](https://docs.python.org/3/library/statistics.html).
Andere Bibliotheken wie matplotlib haben eigene Seiten, die ähnlich aufgebaut sind.

So gehen wir vor:

1. Wir formulieren, was wir brauchen: „Ich suche den Median einer Liste.“
2. Wir suchen das passende Modul, in der Liste der Module oder mit einer Suchmaschine.
3. Wir lesen Signatur und Beschreibung der Funktion.
4. Wir probieren sie an einem kleinen Beispiel aus, das wir von Hand überprüfen können.
5. Wir testen Sonderfälle, z. B. eine Liste mit gerader Länge.

`dir(modul)` zeigt alle Namen in einem Modul, `dir(typ)` alle Methoden eines Typs, z. B. `dir(list)`. Namen mit Unterstrichen am Anfang können wir dabei weglassen.

```python
import statistics
dir(statistics)
```

```python
help(statistics.median)
```

```python
print(statistics.median([5, 1, 3]))
print(statistics.median([5, 1, 3, 7]))
```

**Beispiel:** Wir suchen die häufigsten Wörter in einem Text. Im Modul `collections` finden wir die Klasse `Counter`, die Dinge zählt, und
ihre Methode `most_common`.

```python
from collections import Counter

help(Counter.most_common)
woerter = 'der Hund und der Fuchs und der Hase'.split()
print(Counter(woerter).most_common(2))
```

**Beispiel:** Das Modul `datetime` kennt Kalenderdaten. Die Dokumentation von `weekday` sagt uns, welche Zahl zu welchem Tag gehört.

```python
import datetime

help(datetime.date.weekday)
d = datetime.date(2026, 10, 30)
print(d.weekday())
```

#### Dokumentation von gegebenem Quellcode

Auch fremden Code müssen wir nicht Zeile für Zeile lesen. Bekommen wir eine Klasse oder ein Modul, reichen die Docstrings, um sie zu
benutzen. Das ist der Sinn des **Geheimnisprinzips**: Wir kennen die Schnittstelle, nicht die Umsetzung. Ein Klassendiagramm liefert
dieselben Angaben in anderer Form. Beispiele im Docstring zeigen die Benutzung oft am schnellsten.

Hier stellen wir uns vor, dass jemand anders die Klasse `Einkaufsliste` geschrieben hat.

```python
class Einkaufsliste:
    '''Eine Einkaufsliste mit Preisen.'''

    def __init__(self):
        self.__artikel = {}

    def hinzufuegen(self, name, preis):
        '''
        Nimmt einen Artikel auf. Einen vorhandenen Artikel ersetzt die Methode.
        name: String
        preis: Zahl in Euro
        '''
        self.__artikel[name] = preis

    def summe(self):
        '''
        returns: Summe aller Preise, bei leerer Liste 0
        '''
        return sum(self.__artikel.values())

    def teuerster(self):
        '''
        returns: Name des teuersten Artikels, bei leerer Liste None
        '''
        if len(self.__artikel) == 0:
            return None
        return max(self.__artikel, key=lambda name: self.__artikel[name])
```

Um die Klasse zu benutzen, schauen wir nur in die Hilfe.

```python
help(Einkaufsliste)
```

```python
liste = Einkaufsliste()
liste.hinzufuegen('Brot', 3.20)
liste.hinzufuegen('Käse', 4.50)
print(liste.summe())
print(liste.teuerster())
print(Einkaufsliste().teuerster())
```

Bei einer eigenen Datei `name.py` geht das genauso: Nach `import name` zeigt `help(name)` den Docstring der Datei und aller
Funktionen und Klassen darin.

#### Hilfsmittel der Entwicklungsumgebung

| Was | Thonny | Jupyter |
|---|---|---|
| Hilfe zu einem Namen | `help(name)` in der Kommandozeile | `help(name)`, `name?` oder **Shift + Tab** |
| Methoden eines Objekts sehen | Punkt tippen, dann **Strg + Leertaste** | Punkt tippen, dann **Tab** |
| Parameter einer Funktion sehen | **Strg + Umschalt + Leertaste** | **Shift + Tab** |

#### Zusammenfassung

- `help(name)` zeigt Signatur und Beschreibung, `dir(objekt)` die vorhandenen Namen.
- In der Signatur sind Parameter ohne Standardwert Pflicht, vor `/` stehen reine Positionsargumente, hinter `*` Argumente mit Namen.
- Wir achten darauf, ob eine Funktion etwas Neues zurückgibt oder das Objekt selbst verändert (*in place*).
- Die Dokumentation aller Module steht unter docs.python.org/3/library.
- Eine neue Funktion probieren wir an einem kleinen Beispiel aus, das wir von Hand überprüfen können.
- Fremden Code benutzen wir über seine Docstrings, ohne ihn vollständig zu lesen.
