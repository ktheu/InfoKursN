### Binärer Suchbaum

[Video](https://youtu.be/4rNUM-6UJKA)

In einer sortierten Liste können wir mit der binären Suche sehr schnell suchen. Das Einfügen einer neuen Zahl ist aber
aufwendig, weil wir dafür im Schnitt die Hälfte der Liste verschieben müssen. Ein binärer Suchbaum verbindet beides:
Suchen, Einfügen und Löschen gehen im Durchschnitt so schnell wie die binäre Suche.

Wir verwenden die Klasse `Node` und die Funktionen `baumString` und `ebenen` aus der Einheit *Binärer Baum*.

```python
class Node:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def baumString(node):
    def dfs(node, tiefe=0):
        s = ""
        punkte = "." * tiefe
        if node.right:
            s = s + dfs(node.right, tiefe + 1)
        s = s + punkte + str(node.val) + "\n"
        if node.left:
            s = s + dfs(node.left, tiefe + 1)
        return s

    if node is None: return ""      # leerer Baum
    return dfs(node)[:-1]

def ebenen(node):
    '''
    node: Wurzel eines Baums oder None
    returns: Anzahl der Ebenen
    '''
    if node is None:
        return 0
    return 1 + max(ebenen(node.left), ebenen(node.right))
```

Ein *binärer Suchbaum (BST - binary search tree)* ist ein binärer Baum, bei dem alle Einträge im linken
Teilbaum eines Knotens x kleiner sind als der Eintrag im Knoten x und bei
dem alle Einträge im rechten Teilbaum eines Knotens x größer sind als der
Eintrag im Knoten x. Das gilt für jeden Knoten, also auch für alle Teilbäume.
Jeder Eintrag kommt höchstens einmal vor.

#### Methoden des Suchbaums

Ein Suchbaum hat üblicherweise drei Methoden: `lookup` (suchen), `insert` (einfügen) und `delete` (löschen).
Alle drei beginnen an der Wurzel und gehen bei jedem Knoten nach links, wenn die gesuchte Zahl kleiner ist, und nach
rechts, wenn sie größer ist.

- lookup

lookup 20, 19, 85

<img src="suchbaum3.png">

Bei der 20 gehen wir von der 42 nach links zur 10, nach rechts zur 19, nach rechts zur 31 und dann nach links. Dort ist
aber kein Knoten mehr, die 20 ist also nicht im Suchbaum. Die 19 finden wir auf demselben Weg. Bei der 85 gehen wir nach
rechts, links, rechts und müssten dann bei der 87 nach links. Dort ist nichts, die 85 ist nicht im Suchbaum.

Da der Suchbaum eine rekursive Datenstruktur ist, schreiben wir `lookup` rekursiv. Basisfall ist der leere Baum.

```python
def lookup(node, x):
    '''
    node: Wurzel eines Suchbaums oder None
    returns: True, falls x im Suchbaum vorkommt, sonst False
    '''
    if node is None: return False
    if x == node.val: return True
    if x < node.val:
        return lookup(node.left, x)
    return lookup(node.right, x)
```

- insert

insert 5, 18, 70, 71, 20

<img src="suchbaum4.png">

Wir gehen den Weg wie bei `lookup`. Wo wir auf einen leeren Teilbaum stoßen, hängen wir die neue Zahl als Blatt an.
Ist die Zahl schon vorhanden, passiert nichts.

`insert` gibt die Wurzel des Baums zurück. So funktioniert die Funktion auch für einen leeren Baum: Dann ist der neue
Knoten die Wurzel.

```python
def insert(node, x):
    '''
    node: Wurzel eines Suchbaums oder None
    returns: Wurzel des Suchbaums, in den x eingefügt wurde
    '''
    if node is None: return Node(x)
    if x < node.val:
        node.left = insert(node.left, x)
    elif x > node.val:
        node.right = insert(node.right, x)
    return node                 # x war schon vorhanden: nichts tun
```

Wir bauen den Suchbaum aus dem Beispiel auf, suchen 20, 19, 85 und fügen dann 5, 18, 70, 71, 20 ein.

```python
baum = None
for x in [42, 10, 88, 7, 19, 72, 14, 31, 87]:
    baum = insert(baum, x)
print(baumString(baum))
print(lookup(baum, 20), lookup(baum, 19), lookup(baum, 85))
```

```python
for x in [5, 18, 70, 71, 20]:
    baum = insert(baum, x)
print(baumString(baum))
```

Die Form des Suchbaums hängt von der Reihenfolge ab, in der die Zahlen eingefügt werden. Die erste Zahl wird die Wurzel.
Fügen wir die Zahlen 1 bis 7 in der Reihenfolge 4, 2, 6, 1, 3, 5, 7 ein, entsteht ein vollkommen ausgeglichener Baum
mit 3 Ebenen. Fügen wir sie sortiert ein, hängt jede Zahl rechts an der vorigen, und es entsteht eine Kette mit 7 Ebenen.

```python
ausgeglichen = None
for x in [4, 2, 6, 1, 3, 5, 7]:
    ausgeglichen = insert(ausgeglichen, x)
kette = None
for x in [1, 2, 3, 4, 5, 6, 7]:
    kette = insert(kette, x)
print(baumString(ausgeglichen))
print()
print(baumString(kette))
```

- delete

Zuerst suchen wir den Knoten mit der Zahl, die gelöscht werden soll. Dann gibt es drei Fälle.

delete 42 - Fall 1: Blatt löschen

<img src = 'suchbaum8.png' width='200'> &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; <img src = 'suchbaum9.png' width='120'>

Der Zeiger auf das Blatt wird auf `None` gesetzt.

delete 42 - Fall 2: ein Kind

<img src = 'suchbaum10.png'> &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; <img src = 'suchbaum11.png'>

Der Zeiger auf die 42 zeigt jetzt auf ihr Kind, die 50. Das ist wieder ein Suchbaum: Die 50 und alles darunter hing
schon vorher links von der 78, ist also kleiner als 78.

delete 42 - Fall 3: zwei Kinder

<img src = 'suchbaum12.png'>

Die 42 kann nicht einfach durch eines ihrer Kinder ersetzt werden, denn das hat vielleicht selbst schon zwei Kinder.
Stattdessen suchen wir die **größte Zahl im linken Teilbaum**: Wir gehen einmal nach links und dann immer nach rechts,
bis es nicht mehr weitergeht. Hier ist das die 35. Sie ist größer als alle anderen Zahlen im linken Teilbaum und kleiner
als alle Zahlen im rechten Teilbaum. Deshalb kann sie den Platz der 42 einnehmen.

<img src='delete01.png' width='600'>

Die 35 hat kein rechtes Kind, sonst wären wir weiter nach rechts gegangen. An ihrer alten Stelle löschen wir sie
deshalb nach Fall 1 oder Fall 2.

<img src = 'suchbaum13.png'>

```python
def groesstes(node):
    '''
    node: Wurzel eines nicht leeren Suchbaums
    returns: größte Zahl im Suchbaum
    '''
    while node.right is not None:
        node = node.right
    return node.val

def delete(node, x):
    '''
    node: Wurzel eines Suchbaums oder None
    returns: Wurzel des Suchbaums, aus dem x gelöscht wurde
    '''
    if node is None: return None              # x kommt nicht vor
    if x < node.val:
        node.left = delete(node.left, x)
    elif x > node.val:
        node.right = delete(node.right, x)
    else:
        if node.left is None: return node.right   # Fall 1 und Fall 2
        if node.right is None: return node.left   # Fall 2
        m = groesstes(node.left)                  # Fall 3
        node.val = m
        node.left = delete(node.left, m)
    return node
```

Wir löschen die 42 aus dem Baum im Bild oben.

```python
baum3 = None
for x in [78, 42, 10, 50, 8, 17, 43, 55, 4, 9, 35, 30]:
    baum3 = insert(baum3, x)
print(baumString(baum3))
baum3 = delete(baum3, 42)
print()
print(baumString(baum3))
```

#### Inorder-Durchlauf

In der Einheit *Binärer Baum* haben wir die Knoten in der Reihenfolge Knoten, linker Teilbaum, rechter Teilbaum
besucht (*preorder*). Besuchen wir sie in der Reihenfolge linker Teilbaum, Knoten, rechter Teilbaum, heißt das
*inorder*. Bei einem Suchbaum kommen dabei die Zahlen sortiert heraus: Alles Kleinere steht links, alles Größere rechts.
Ein inorder-Durchlauf durch einen binären Suchbaum ist also wie das Wandern durch eine sortierte Liste.

```python
def inorder(node):
    if node is None: return
    inorder(node.left)
    print(node.val, end=' ')
    inorder(node.right)

inorder(baum)
```

#### Komplexität der Methoden im Suchbaum: Anzahl Ebenen

`lookup`, `insert` und `delete` gehen von der Wurzel aus einen Weg nach unten. Die Laufzeit hängt deshalb von der Anzahl
der Ebenen ab.

worst case: $O(n)$

Werden die Zahlen sortiert eingefügt, entsteht eine Kette. Dann hat der Baum so viele Ebenen wie Zahlen.

<img src = 'suchbaum6.png'>

best case, average case: $O(\log(n))$

Ein vollkommen ausgeglichener Baum mit $k$ Ebenen hat $2^k - 1$ Knoten. Für $n$ Zahlen braucht er also nur etwa
$\log_2(n)$ Ebenen, bei einer Million Zahlen rund 20. Auch wenn die Zahlen in zufälliger Reihenfolge eingefügt werden,
bleibt die Anzahl der Ebenen im Durchschnitt in der Größenordnung von $\log_2(n)$.

<img src = 'suchbaum7.png'>

```python
import random
random.seed(1)
zahlen = list(range(500))
baum_sortiert = None
for x in zahlen:
    baum_sortiert = insert(baum_sortiert, x)
random.shuffle(zahlen)
baum_zufall = None
for x in zahlen:
    baum_zufall = insert(baum_zufall, x)
print(ebenen(baum_sortiert), ebenen(baum_zufall))
```

Es gibt Suchbäume, die sich beim Einfügen und Löschen selbst ausgleichen, z.B. AVL-Bäume oder Rot-Schwarz-Bäume. Bei
ihnen ist die Laufzeit immer $O(\log(n))$.

#### Anwendungen

Mit Suchbäumen lassen sich sets und dictionaries implementieren: Bei einem dictionary speichert jeder Knoten zusätzlich
zum Schlüssel einen Wert. Python selbst verwendet für `set` und `dict` ein anderes Verfahren, das Hashing. In Java gibt es
neben den Hash-Varianten aber z.B. `TreeSet` und `TreeMap`, die mit ausgeglichenen Suchbäumen arbeiten. Ein Vorteil des
Suchbaums: Die Einträge lassen sich mit einem inorder-Durchlauf sortiert ausgeben, und die kleinste und die größte Zahl
findet man schnell.

#### Zusammenfassung

| | |
|---|---|
| binärer Suchbaum | links von jedem Knoten nur kleinere, rechts nur größere Einträge, jeder Eintrag höchstens einmal |
| `lookup` | bei jedem Knoten nach links oder rechts gehen, bis die Zahl gefunden oder der Teilbaum leer ist |
| `insert` | wie `lookup`, am leeren Teilbaum die Zahl als Blatt anhängen |
| `delete` | Blatt: entfernen, ein Kind: Kind rückt nach, zwei Kinder: größte Zahl des linken Teilbaums rückt nach |
| inorder | linker Teilbaum, Knoten, rechter Teilbaum: liefert die Einträge sortiert |
| Laufzeit | Anzahl der Ebenen: im worst case $O(n)$, im best und average case $O(\log(n))$ |
