#### House Robber

https://leetcode.com/problems/house-robber/ - [Video](https://youtu.be/J-a8OX_Cts0)

Ein Räuber will in die Häuser einer Straße einbrechen. Die Werte in den Häusern sind in einer Liste gegeben. Er darf
keine zwei benachbarten Häuser ausrauben. Welchen maximalen Wert kann er erreichen?

<img src='robber1.png' width='501'>

dp-guess: Wird das Haus i ausgeraubt oder nicht? Wenn ja, darf Haus i + 1 nicht ausgeraubt werden.

```python
def dp(i):
    '''
    returns: maximaler Wert für a[i:], i = 0..n
    dp-guess: wird a[i] ausgeraubt oder nicht?
    '''
    if i in memo:
        return memo[i]
    if i == n:
        return 0
    if i == n - 1:
        return a[n - 1]
    res = max(dp(i + 1), a[i] + dp(i + 2))
    memo[i] = res
    return res

a = [2, 7, 9, 6, 1]
n = len(a)
memo = {}
print(dp(0))
```

Am besten raubt er die Häuser mit 7 und 6 aus, das ergibt 13. Mit 2, 9 und 1 wären es nur 12. Laufzeit $O(n)$.

#### Climbing Stairs

https://leetcode.com/problems/climbing-stairs/ - [Video](https://youtu.be/jeGr99fSkW8)

Eine Treppe hat n Stufen. Wir stehen vor der ersten Stufe und können jeweils 1 oder 2 Stufen hochsteigen. Wie viele
verschiedene Möglichkeiten gibt es, bis nach ganz oben zu kommen?

<img src='stairs2.png' width='402'>

<img src='stairs.png' width='600'>

```python
def dp(i):
    '''
    returns: die Anzahl der Möglichkeiten, nach oben zu kommen, wenn ich vor Stufe i stehe
    dp-guess: ein Schritt oder zwei Schritte?
    '''
    if i in memo:
        return memo[i]
    if i == n - 1:
        return 1
    if i == n - 2:
        return 2
    res = dp(i + 1) + dp(i + 2)
    memo[i] = res
    return res

for n in [2, 3, 4, 5, 10]:
    memo = {}
    print(n, dp(0))
```

Hier zählen wir Möglichkeiten, deshalb werden die Ergebnisse der beiden Fälle addiert und nicht das Beste gewählt. Es
entstehen die Fibonacci-Zahlen.

#### Minimizing Coins

https://cses.fi/problemset/task/1634 - [Video](https://youtu.be/3jz-3jhh2GA)

Gegeben sind verschiedene Münzwerte und eine Zahl k. Was ist die kleinste Anzahl von Münzen, deren Werte zusammen genau
k ergeben? Jeder Münzwert darf beliebig oft benutzt werden. Gib -1 zurück, wenn es nicht geht.

<img src='coins.png' width='401'>

Hier ist die kleinere Problemgröße eine kleinere Summe. dp-guess: Was ist die erste Münze einer optimalen Lösung?

<img src='coins2.png' width='600'>

```python
def dp(i):
    '''
    returns: die minimale Anzahl von Münzen für die Summe i, i = 0..k
    dp-guess: welches ist die erste Münze einer optimalen Lösung?
    '''
    if i in memo:
        return memo[i]
    if i == 0:
        return 0                          # für die Summe 0 braucht man 0 Münzen
    res = float('inf')                    # noch keine Lösung gefunden
    for c in coins:
        if i - c >= 0 and 1 + dp(i - c) < res:
            res = 1 + dp(i - c)
    memo[i] = res
    return res

coins = [1, 5, 7]
for k in [3, 10, 11]:
    memo = {}
    ans = dp(k)
    print(k, -1 if ans == float('inf') else ans)
```

Für 10 braucht man 2 Münzen (5 + 5). Wer gierig zuerst die größte Münze nimmt, kommt auf 7 + 1 + 1 + 1, also 4 Münzen.
Laufzeit: k Teilprobleme mit je so viel Arbeit, wie es Münzwerte gibt.

#### Edit Distance

https://leetcode.com/problems/edit-distance/ - [Video](https://youtu.be/DCzT0BKFue0)

Gegeben sind zwei Strings w1 und w2. Wie viele Schritte braucht man mindestens, um w1 in w2 zu verwandeln? Ein Schritt
ist: ein Zeichen in w1 einfügen (insert), löschen (delete) oder durch ein anderes ersetzen (replace).

<img src='editdistance.png' width='450'>

Wenn das Problem zwei Listen oder Strings enthält, probieren wir es mit zwei Suffixen: `dp(i, j)` für `w1[i:]` und
`w2[j:]`.

dp-guess: Was machen wir mit dem ersten Zeichen von `w1[i:]`? Sind die ersten Zeichen gleich, ist nichts zu tun.
Sonst probieren wir alle drei Operationen aus:

- insert: Wir fügen `w2[j]` vorne ein, weiter mit `dp(i, j + 1)`.
- delete: Wir löschen `w1[i]`, weiter mit `dp(i + 1, j)`.
- replace: Wir ersetzen `w1[i]` durch `w2[j]`, weiter mit `dp(i + 1, j + 1)`.

```python
def dp(i, j):
    '''
    returns: minimale Anzahl von Operationen, um w1[i:] in w2[j:] zu überführen
    dp-guess: was machen wir mit dem Zeichen w1[i]?
    '''
    if (i, j) in memo:
        return memo[(i, j)]
    if i == n1:
        return n2 - j                     # der Rest von w2 muss eingefügt werden
    if j == n2:
        return n1 - i                     # der Rest von w1 muss gelöscht werden
    if w1[i] == w2[j]:
        res = dp(i + 1, j + 1)
    else:
        res = 1 + min(dp(i, j + 1), dp(i + 1, j), dp(i + 1, j + 1))
    memo[(i, j)] = res
    return res

w1 = 'horse'
w2 = 'ros'
n1 = len(w1)
n2 = len(w2)
memo = {}
print(dp(0, 0))
```

horse → rorse (h durch r ersetzen) → rose (r löschen) → ros (e löschen). Laufzeit $O(n_1 \cdot n_2)$.

#### Longest Increasing Subsequence (LIS)

[Video](https://youtu.be/7E0xaqMenT0)

Gegeben eine Liste mit Zahlen. Finde die Länge der längsten Teilfolge, bei der jedes Element größer als sein
Vorgänger ist. Eine Teilfolge muss nicht zusammenhängend sein, die Reihenfolge bleibt aber erhalten.

<img src='LIS.png' width='600'>

Erster Versuch:

    dp(i) = Länge der längsten Teilfolge in a[i:]
    dp-guess: nehmen wir a[i] in die Teilfolge auf oder nicht?

Wenn wir `a[i]` nehmen, wissen wir, dass die nächste Zahl größer sein muss. Wenn wir `a[i]` aber nicht nehmen, kennen
wir die Bedingung für die nächste Zahl nicht, weil wir nicht wissen, welches Element vorher genommen wurde. Deshalb
formulieren wir das Teilproblem um:

    dp(i) = Länge der längsten Teilfolge in a[i:], die mit a[i] beginnt
    dp-guess: welches Element nehmen wir als nächstes auf?

Das Ausgangsproblem ist dann das Maximum aller `dp(i)`.

```python
def dp(i):
    '''
    returns: Länge der längsten aufsteigenden Teilfolge in a[i:], die mit a[i] beginnt, i = 0..n-1
    '''
    if i in memo:
        return memo[i]
    res = 1 + max([dp(j) for j in range(i + 1, n) if a[i] < a[j]], default=0)
    memo[i] = res
    return res

a = [10, 9, 2, 5, 3, 7, 101, 18]
n = len(a)
memo = {}
print(max(dp(i) for i in range(n)))
```

Eine längste Teilfolge ist 2, 5, 7, 101. `max(..., default=0)` liefert 0, wenn es keine größere Zahl mehr gibt. Laufzeit
$O(n^2)$: n Teilprobleme mit je $O(n)$ nicht-rekursiver Arbeit.

#### Longest Common Subsequence (LCS)

[Video](https://youtu.be/M-uTCi7xLdw)

Gegeben zwei Worte. Gesucht ist die Länge der längsten gemeinsamen Teilfolge.

<img src='LCS.png' width='600'>

    dp(i, j) = Länge der längsten gemeinsamen Teilfolge von w1[i:] und w2[j:]

- Wenn `w1[i] == w2[j]`, nehmen wir beide Buchstaben in die Teilfolge auf: `1 + dp(i + 1, j + 1)`.
- Sonst ist einer der beiden Buchstaben sicher nicht in der Teilfolge. Wir wissen nicht welcher, also probieren wir
  beide Möglichkeiten: `max(dp(i, j + 1), dp(i + 1, j))`.

```python
def dp(i, j):
    '''
    returns: Länge der längsten gemeinsamen Teilfolge von w1[i:] und w2[j:], i = 0..n1, j = 0..n2
    '''
    if (i, j) in memo:
        return memo[(i, j)]
    if i == n1 or j == n2:
        return 0
    if w1[i] == w2[j]:
        res = 1 + dp(i + 1, j + 1)
    else:
        res = max(dp(i, j + 1), dp(i + 1, j))
    memo[(i, j)] = res
    return res

w1 = 'HIEROGLYPHOLOGY'
w2 = 'MICHELANGELO'
n1 = len(w1)
n2 = len(w2)
memo = {}
print(dp(0, 0))
```

Eine längste gemeinsame Teilfolge ist z.B. HELLO.

#### Predict the Winner

https://leetcode.com/problems/predict-the-winner/ - [Video](https://youtu.be/UAtgufz6Ijk)

Zwei Spieler nehmen abwechselnd von einer Liste vorne oder hinten eine Zahl weg. Gewonnen hat, wer am Schluss die
größere Summe hat. Bei Gleichstand gewinnt Spieler 1. Kann der Spieler, der beginnt, sicher gewinnen?

<img src='predict_winner.png' width='500'>

Hier wird die Liste von beiden Seiten kürzer. Die Teilprobleme sind deshalb Teillisten `a[i:j+1]` von Index i bis j.
Bei einem Spiel für zwei Personen bekommt `dp` außerdem einen Parameter für den Spieler, der am Zug ist. Spieler 1
nimmt das Beste für sich, Spieler 2 das, was für Spieler 1 am schlechtesten ist, wie beim Spielbaum.

```python
def dp(i, j, p):
    '''
    returns: maximale Summe für Spieler 1 bei a[i:j+1], wenn Spieler p am Zug ist
             p = 0: Spieler 1, p = 1: Spieler 2
    dp-guess: welche Zahl nimmt Spieler p?
    '''
    if (i, j, p) in memo:
        return memo[(i, j, p)]
    if i == j:
        return a[i] if p == 0 else 0
    if p == 0:
        res = max(a[i] + dp(i + 1, j, 1), a[j] + dp(i, j - 1, 1))
    else:
        res = min(dp(i + 1, j, 0), dp(i, j - 1, 0))
    memo[(i, j, p)] = res
    return res

a = [1, 5, 233, 7]
memo = {}
me = dp(0, len(a) - 1, 0)
print(me, sum(a) - me, me >= sum(a) - me)
```

Spieler 1 nimmt zuerst die 1. Danach kann er immer die 233 erreichen, egal ob Spieler 2 die 5 oder die 7 nimmt.

Als Tabelle müssen wir die Teillisten nach ihrer Länge ordnen: Zuerst alle Teillisten mit einem Element, dann mit
zweien und so weiter. Dann sind die kürzeren Teillisten immer schon berechnet.

```python
n = len(a)
dp_tab = [[[0] * 2 for j in range(n)] for i in range(n + 1)]
for laenge in range(1, n + 1):
    for i in range(n - laenge + 1):
        j = i + laenge - 1
        if i == j:
            dp_tab[i][j][0] = a[i]
            dp_tab[i][j][1] = 0
        else:
            dp_tab[i][j][0] = max(a[i] + dp_tab[i + 1][j][1], a[j] + dp_tab[i][j - 1][1])
            dp_tab[i][j][1] = min(dp_tab[i + 1][j][0], dp_tab[i][j - 1][0])
print(dp_tab[0][n - 1][0])
```

#### Rod Cutting

Eine Stange der Länge n soll in Stücke zersägt werden. Die Liste `a` gibt den Preis für Stücke der Länge 1 bis n an:

    a = [1, 10, 13, 18, 20, 31, 32]       # Preise für die Längen 1..7

Wie zersägt man die Stange, damit der Gesamtpreis möglichst hoch ist?

Die kleinere Problemgröße ist eine kürzere Stange. dp-guess: Wie lang ist das erste Stück?

```python
def dp(i):
    '''
    returns: maximaler Preis für eine Stange der Länge i, i = 0..n
    dp-guess: wie lang ist das erste Stück?
    '''
    if i in memo:
        return memo[i]
    if i == 0:
        return 0
    res = max(a[laenge - 1] + dp(i - laenge) for laenge in range(1, i + 1))
    memo[i] = res
    return res

a = [1, 10, 13, 18, 20, 31, 32]
memo = {}
print(dp(len(a)))
```

Am besten zersägt man die Stange in Stücke der Längen 2, 2 und 3: $10 + 10 + 13 = 33$. Die Stücke dürfen höchstens
so lang sein wie die restliche Stange, deshalb läuft `laenge` nur bis `i`.

#### Subset Sum

Gegeben sind eine Liste `a` von positiven Zahlen und eine Zahl T. Gibt es eine Auswahl von Zahlen aus `a`, deren Summe
genau T ist?

    a = [2, 5, 7, 8, 9], T = 21

    dp(i, t) = gibt es eine Auswahl aus a[i:] mit der Summe t?   i = 0..n, t = 0..T
    dp-guess: nehmen wir a[i] in die Auswahl auf oder nicht?

```python
def dp(i, t):
    '''
    returns: True, falls es eine Auswahl aus a[i:] mit der Summe t gibt
    '''
    if (i, t) in memo:
        return memo[(i, t)]
    if i == n:
        return t == 0
    res = dp(i + 1, t)
    if t - a[i] >= 0:
        res = res or dp(i + 1, t - a[i])
    memo[(i, t)] = res
    return res

a = [2, 5, 7, 8, 9]
n = len(a)
for T in [21, 6]:
    memo = {}
    print(T, dp(0, T))
```

Für 21 geht es mit 5 + 7 + 9, für 6 nicht.

Die Laufzeit ist $O(n \cdot T)$. Das sieht polynomiell aus, ist es aber nicht unbedingt: T kann im Vergleich zu n sehr
groß sein, z.B. $2^n$. Dann ist die Laufzeit exponentiell. Solange T nicht sehr groß ist, ist der Algorithmus gut. Man
nennt das **pseudopolynomiell**.

#### Two Sets II

https://cses.fi/problemset/task/1093

Auf wie viele Arten kann man die Zahlen 1 bis n in zwei Mengen mit gleicher Summe aufteilen?

Die Gesamtsumme ist $S = n(n+1)/2$. Ist S ungerade, geht es nicht. Sonst zählen wir die Teilmengen mit der Summe
$S/2$. Jede Aufteilung wird dabei zweimal gezählt, einmal für jede der beiden Mengen.

    dp(i, k) = Anzahl der Teilmengen von {1, ..., i} mit der Summe k
    dp-guess: ist die Zahl i in der Teilmenge oder nicht?

```python
def dp(i, k):
    '''
    returns: Anzahl der Teilmengen von {1, ..., i} mit der Summe k
    '''
    if (i, k) in memo:
        return memo[(i, k)]
    if k < 0:
        return 0
    if i == 0:
        return 1 if k == 0 else 0
    res = dp(i - 1, k) + dp(i - 1, k - i)
    memo[(i, k)] = res
    return res

for n in [3, 4, 7, 8]:
    memo = {}
    S = n * (n + 1) // 2
    if S % 2 != 0:
        print(n, 0)
    else:
        print(n, dp(n, S // 2) // 2)
```

Für n = 7: {1, 6, 7} und {2, 3, 4, 5} ist eine der 4 Aufteilungen. Bei CSES ist n bis 500 groß, dort wird das Ergebnis
modulo $10^9 + 7$ verlangt.

#### Das Rucksackproblem

[Video 1](https://youtu.be/tAC2o0XAuPw) - [Video 2](https://youtu.be/VwgC0IAEyDY) - [Video 3](https://youtu.be/-KHkWcMogog)

Gegeben sind Gegenstände mit einem Wert und einem Gewicht und ein Rucksack mit einer Kapazität. Welche Gegenstände
packt man ein, damit der Gesamtwert möglichst groß ist, ohne dass das Gewicht die Kapazität übersteigt? Jeder
Gegenstand kann nur einmal eingepackt werden.

    wert = [6, 10, 12]
    gewicht = [1, 2, 3]
    kapazitaet = 5

Hier nehmen wir als Teilprobleme die ersten Gegenstände 0 bis j und eine kleinere Kapazität k:

    opt(k, j) = maximaler Wert bei Kapazität k, wenn nur die Gegenstände 0..j benutzt werden
    dp-guess: wird Gegenstand j eingepackt oder nicht?

Passt Gegenstand j nicht mehr, wird er nicht eingepackt. Sonst probieren wir beides aus: Ohne ihn bleibt `opt(k, j-1)`,
mit ihm bekommen wir seinen Wert und haben für die übrigen Gegenstände nur noch die Kapazität `k - gewicht[j]`.

Zuerst rekursiv ohne Memoization. Die Funktion gibt jeden berechneten Wert aus:

```python
def opt(k, j):
    '''
    returns: Wert des optimalen Rucksacks mit Kapazität k, der nur Gegenstände 0..j enthält
    dp-guess: wird Gegenstand j eingepackt oder nicht?
    '''
    if j == 0:
        res = wert[0] if gewicht[0] <= k else 0
    elif gewicht[j] <= k:
        res = max(opt(k, j - 1), wert[j] + opt(k - gewicht[j], j - 1))
    else:
        res = opt(k, j - 1)
    print(f'opt({k},{j})={res}')
    return res

wert = [6, 10, 12]
gewicht = [1, 2, 3]
kapazitaet = 5
opt(kapazitaet, len(wert) - 1)
```

Die Werte erscheinen in der Reihenfolge, in der die Aufrufe fertig werden: Zuerst `opt(k, j - 1)` mit allen seinen
Unteraufrufen, dann `opt(k - gewicht[j], j - 1)`, zum Schluss der Aufruf selbst.

Als Tabelle: Zeile j gehört zu den Gegenständen 0..j, Spalte k zur Kapazität k. Jede Zeile wird aus der Zeile darüber
berechnet. Rechts unten steht der Wert des optimalen Rucksacks.

```python
def rucksack_tabelle(wert, gewicht, kapazitaet):
    m = [[0] * (kapazitaet + 1) for i in range(len(wert))]
    for k in range(kapazitaet + 1):
        m[0][k] = wert[0] if gewicht[0] <= k else 0
    for j in range(1, len(wert)):
        for k in range(kapazitaet + 1):
            if gewicht[j] <= k:
                m[j][k] = max(m[j - 1][k], wert[j] + m[j - 1][k - gewicht[j]])
            else:
                m[j][k] = m[j - 1][k]
    return m

def zeige(m):
    print('    ' + ''.join(f'{k:4}' for k in range(len(m[0]))))
    for j in range(len(m)):
        print(f'{j:2} !' + ''.join(f'{x:4}' for x in m[j]))

m = rucksack_tabelle(wert, gewicht, kapazitaet)
zeige(m)
```

Welche Gegenstände sind im Rucksack? Wir gehen von rechts unten nach oben. Ist der Wert in Zeile j größer als darüber,
dann wurde Gegenstand j eingepackt, und wir gehen in der Zeile darüber um sein Gewicht nach links. Sonst gehen wir nur
nach oben. In Zeile 0 ist Gegenstand 0 dabei, wenn der Wert nicht 0 ist.

```python
def gewaehlte(m, gewicht):
    j = len(m) - 1
    k = len(m[0]) - 1
    items = []
    while j > 0:
        if m[j][k] != m[j - 1][k]:
            items.append(j)
            k -= gewicht[j]
        j -= 1
    if m[0][k] != 0:
        items.append(0)
    return sorted(items)

items = gewaehlte(m, gewicht)
print('Items =', *items)
print('Wert =', m[-1][-1], ' Gewicht =', sum(gewicht[i] for i in items))
```

Laufzeit $O(n \cdot K)$ für n Gegenstände und die Kapazität K. Wie bei Subset Sum ist das pseudopolynomiell.
