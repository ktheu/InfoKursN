### Variablen, Datentypen, Deklaration, Initialisierung, Typumwandlung

#### Begriffe

Eine **Variable** ist ein benannter Speicherplatz, in dem ein Wert abgelegt wird.

| Begriff | Bedeutung | Beispiel |
|---|---|---|
| **Bezeichner** | der Name der Variable | `alter` |
| **Wert** | der konkrete Inhalt der Variable | `17` |
| **Datentyp** | legt fest, welche Art von Werten gespeichert werden kann und welche Operationen damit möglich sind | `int` |
| **Deklaration** | das Bekanntmachen einer Variable mit Bezeichner (und in vielen Sprachen mit Datentyp) | Java: `int alter;` |
| **Initialisierung** | die erste Wertzuweisung an eine Variable | `alter = 17` |
| **Wertzuweisung** | das Zuweisen eines (neuen) Wertes mit `=` | `alter = 18` |
| **Gültigkeitsbereich** (Scope) | der Teil des Programms, in dem eine Variable bekannt ist und benutzt werden kann | siehe unten |

Für Bezeichner gelten die Regeln aus den Grundlagen: Buchstaben, Ziffern und Unterstrich, keine Ziffer am Anfang,
keine Schlüsselwörter, und wir verwenden sprechende Namen, die mit einem kleinen Buchstaben beginnen.

#### Deklaration und Initialisierung in Java und Python

In Java sind Deklaration und Initialisierung zwei verschiedene Schritte, die man auch zusammen schreiben kann.
Bei der Deklaration wird der Datentyp festgelegt und kann danach nicht mehr geändert werden (*statische Typisierung*):

```java
int alter;          // Deklaration: Bezeichner alter, Datentyp int
alter = 17;         // Initialisierung
alter = 18;         // Wertzuweisung
int anzahl = 5;     // Deklaration und Initialisierung in einem Schritt
alter = "Alex";     // Fehler beim Übersetzen: ein String passt nicht in eine int-Variable
```

In Python gibt es keine eigene Deklaration. Eine Variable entsteht bei ihrer ersten Wertzuweisung,
Deklaration und Initialisierung fallen also zusammen. Der Datentyp gehört zum Wert, nicht zur Variable.
Deshalb kann eine Variable später einen Wert eines anderen Typs bekommen (*dynamische Typisierung*).

````python
alter = 17           # Deklaration und Initialisierung: alter ist jetzt eine int-Variable
name = 'Alex'        # Bezeichner: name, Datentyp: str
print(name, 'ist', alter, 'Jahre alt.')

alter = 18           # Wertzuweisung: der alte Wert 17 ist verloren
print(type(alter))

alter = 'achtzehn'   # in Python erlaubt: jetzt ist der Wert ein str
print(type(alter))
````

Eine Wertzuweisung wird **von rechts nach links** gelesen: Zuerst wird der Ausdruck auf der rechten Seite ausgewertet,
dann wird das Ergebnis der Variable auf der linken Seite zugewiesen. Deshalb ist `x = x + 1` sinnvoll, obwohl es als
Gleichung keinen Sinn ergibt.

Benutzt man eine Variable, bevor sie initialisiert wurde, gibt es einen Fehler:

```python
print(punkte)    # NameError: name 'punkte' is not defined
```

Man kann in Python den Datentyp freiwillig mit einem **Type Hint** angeben. Er dient der Dokumentation und hilft der
Entwicklungsumgebung, Python selbst prüft ihn aber nicht.

````python
x = 5
x = x + 1            # rechts: 5 + 1 = 6, dann wird 6 an x zugewiesen
print(x)

groesse: float = 1.78     # Type Hint: groesse soll ein float sein
print(groesse)
````

#### Gültigkeitsbereich

Eine Variable, die innerhalb einer Funktion eine Wertzuweisung bekommt, ist eine **lokale Variable**.
Sie ist nur innerhalb dieser Funktion gültig und existiert nur, solange die Funktion ausgeführt wird.
Eine Variable, die außerhalb aller Funktionen eine Wertzuweisung bekommt, ist eine **globale Variable**.
Sie ist im ganzen Programm bekannt und kann in Funktionen gelesen werden.

````python
mwst = 0.19                    # globale Variable


def brutto(netto):
    steuer = netto * mwst      # steuer und netto sind lokale Variablen, mwst wird nur gelesen
    return netto + steuer


print(brutto(100))
print(mwst)
````

Außerhalb der Funktion sind `steuer` und `netto` nicht bekannt:

```python
print(steuer)    # NameError: name 'steuer' is not defined
```

Bekommt eine Variable in einer Funktion einen Wert, ist sie dort lokal, auch wenn es eine globale Variable mit
demselben Bezeichner gibt. Die globale Variable bleibt dann unverändert:

````python
zaehler = 0


def erhoehen():
    zaehler = 10       # neue lokale Variable, die globale Variable zaehler wird nicht verändert
    print('in der Funktion:', zaehler)


erhoehen()
print('außerhalb:', zaehler)
````

Gültigkeitsbereiche sorgen dafür, dass sich Funktionen nicht gegenseitig ihre Variablen verändern.
Man sollte deshalb möglichst wenige globale Variablen verwenden und Werte lieber als Parameter übergeben
und mit `return` zurückgeben.

In Java ist eine Variable sogar nur in dem Block `{ ... }` gültig, in dem sie deklariert wurde, z.B. nur innerhalb
einer Schleife. In Python gibt es diese Block-Gültigkeit nicht: Eine Variable, die in einer Schleife oder in einem
`if`-Zweig einen Wert bekommt, ist danach auch außerhalb bekannt.

````python
for i in range(3):
    quadrat = i * i

print(i, quadrat)      # in Python bekannt, in Java wären i und quadrat hier nicht gültig
````

#### Primitive Datentypen und ihr Wertebereich

Für Zahlen und Wahrheitswerte gibt es einfache (*primitive*) Datentypen:

| Datentyp | Python | Java | Beispiele |
|---|---|---|---|
| ganze Zahl | `int` | `byte`, `short`, `int`, `long` | `17`, `-3`, `0` |
| Gleitkommazahl | `float` | `float`, `double` | `3.14`, `-0.5`, `2.0`, `1e-3` |
| Wahrheitswert | `bool` | `boolean` | `True`, `False` |

Der **Wertebereich** eines Datentyps ist die Menge der Werte, die eine Variable dieses Typs annehmen kann.
Er hängt davon ab, wie viele Bits für einen Wert zur Verfügung stehen.

(In Python ist streng genommen jeder Wert ein Objekt. `int`, `float` und `bool` entsprechen aber den primitiven Datentypen anderer Sprachen.)

#### Ganze Zahlen

In Java und vielen anderen Sprachen haben ganze Zahlen eine **feste Anzahl von Bits** und werden im Zweierkomplement
gespeichert. Mit $n$ Bits lassen sich die Zahlen von $-2^{n-1}$ bis $2^{n-1}-1$ darstellen:

| Java-Typ | Bits | kleinster Wert | größter Wert |
|---|---:|---:|---:|
| `byte` | 8 | -128 | 127 |
| `short` | 16 | -32 768 | 32 767 |
| `int` | 32 | -2 147 483 648 | 2 147 483 647 |
| `long` | 64 | ca. $-9{,}2 \cdot 10^{18}$ | ca. $9{,}2 \cdot 10^{18}$ |

Wird der größte Wert überschritten, gibt es einen **Überlauf** (*Overflow*): Das Ergebnis springt ohne Fehlermeldung
in den negativen Bereich. In Java ergibt z.B. `2147483647 + 1` den Wert `-2147483648`.

In Python ist `int` **nicht** auf eine feste Anzahl von Bits beschränkt. Eine ganze Zahl kann beliebig groß werden,
solange der Arbeitsspeicher reicht. Ein Überlauf kann also nicht passieren.
Mit dem Modul `ctypes` kann man die Typen fester Länge aber nachbilden und den Überlauf sichtbar machen:

````python
import ctypes

print(2 ** 100)                         # Python-int: beliebig groß

print(ctypes.c_int8(127).value)         # 8 Bit: größter Wert
print(ctypes.c_int8(127 + 1).value)     # Überlauf: -128
print(ctypes.c_int32(2147483647 + 1).value)
````

#### Zweierpotenzen abschätzen

Um Wertebereiche einzuschätzen, muss man wissen, wie groß eine Zweierpotenz $2^n$ ungefähr ist.
Dafür reicht eine einzige Näherung:

$$2^{10} = 1024 \approx 1000 = 10^3$$

**10 Bit entsprechen also ungefähr 3 Dezimalstellen.** Man zerlegt den Exponenten in Zehner und einen Rest
und ersetzt jedes $2^{10}$ durch $10^3$:

$$2^{n} = 2^{\text{Rest}} \cdot 2^{10} \cdot 2^{10} \cdots \approx 2^{\text{Rest}} \cdot 10^3 \cdot 10^3 \cdots$$

| Zweierpotenz | Zerlegung | Schätzung | genauer Wert |
|---|---|---|---|
| $2^{31}$ (Grenze von Java-`int`, 32 Bit) | $2^1 \cdot (2^{10})^3$ | $2 \cdot 10^9$, also ca. 2 Milliarden | 2 147 483 648 |
| $2^{53}$ | $2^3 \cdot (2^{10})^5$ | $8 \cdot 10^{15}$, also 16 Stellen | 9 007 199 254 740 992 |
| $2^{63}$ (Grenze von Java-`long`, 64 Bit) | $2^3 \cdot (2^{10})^6$ | $8 \cdot 10^{18}$ | ca. $9{,}2 \cdot 10^{18}$ |

Weil $1024$ etwas größer als $1000$ ist, ist die Schätzung immer etwas zu klein, und zwar umso mehr, je mehr Zehner
im Exponenten stecken. Für die Größenordnung und die Anzahl der Stellen reicht sie aber aus.

**Faustformel für die Anzahl der Stellen:** Jedes Bit bringt ungefähr $0{,}3$ Dezimalstellen. Eine Zahl in der Größenordnung
$10^k$ hat $k + 1$ Stellen. Daraus ergibt sich:

$$2^n \text{ hat ungefähr } \frac{3 \cdot n}{10} \text{ (abgerundet)} + 1 \text{ Stellen,}$$

in Python `3 * n // 10 + 1`. Beispiel: $2^{53}$ hat $3 \cdot 53 // 10 + 1 = 15 + 1 = 16$ Stellen.
Bis $n = 100$ stimmt die Formel genau, für größere $n$ ist sie manchmal um 1 zu klein.

````python
for n in [10, 31, 53, 63]:
    wert = 2 ** n
    geschaetzt = 3 * n // 10 + 1
    print(f'2^{n} = {wert}  ({len(str(wert))} Stellen, Faustformel: {geschaetzt})')
````

#### Gleitkommazahlen

Ein `float` in Python entspricht dem Typ `double` in Java: 64 Bit nach dem Standard IEEE 754.
Ein Teil der Bits speichert die Ziffern (Mantisse), ein Teil den Exponenten. Daraus ergeben sich zwei Grenzen:

- **Größe:** Der größte Wert ist ungefähr $1{,}8 \cdot 10^{308}$. Darüber ergibt sich `inf` (unendlich).
- **Genauigkeit:** Es werden nur etwa 15 bis 16 Dezimalstellen gespeichert. Viele Dezimalzahlen wie $0{,}1$ lassen sich
  im Binärsystem nicht exakt darstellen (so wie $1/3$ im Dezimalsystem). Es entstehen **Rundungsfehler**.

Deshalb vergleicht man Gleitkommazahlen nicht mit `==`, sondern prüft, ob sie *fast* gleich sind, z.B. mit `math.isclose`.

````python
import sys
import math

print(sys.float_info.max)           # größter float-Wert
print(sys.float_info.max * 10)      # zu groß: inf

print(0.1 + 0.2)                    # nicht exakt 0.3
print(0.1 + 0.2 == 0.3)
print(math.isclose(0.1 + 0.2, 0.3))

print(1e16 + 1 == 1e16)             # die 1 geht verloren: zu viele Stellen
````

#### Wahrheitswerte

Der Datentyp `bool` hat nur die zwei Werte `True` und `False`. Vergleiche liefern einen Wahrheitswert,
und Wahrheitswerte werden mit `and`, `or` und `not` verknüpft. Man braucht sie vor allem für Bedingungen und Schleifen.

````python
alter = 17
volljaehrig = alter >= 18
print(volljaehrig, type(volljaehrig))

hat_erlaubnis = True
darf_mitfahren = volljaehrig or hat_erlaubnis
print(darf_mitfahren)
print(not darf_mitfahren)
````

#### Typumwandlungen

Oft muss ein Wert eines Datentyps in einen anderen Datentyp umgewandelt werden.

- **Implizite Typumwandlung:** Die Umwandlung geschieht automatisch. Python wandelt z.B. einen `int` in einen `float` um,
  wenn mit einem `float` gerechnet wird. Das ist unproblematisch, weil dabei (fast) keine Information verloren geht.
- **Explizite Typumwandlung (Type Cast):** Man wandelt den Wert selbst mit einer Funktion um:
  `int(...)`, `float(...)`, `str(...)`, `bool(...)`. Das ist nötig, wenn Information verloren gehen kann
  oder wenn die Umwandlung fehlschlagen kann.

In Java schreibt man einen Type Cast so: `int n = (int) 3.99;`

````python
# implizite Typumwandlungen
summe = 5 + 2.5            # int + float: 5 wird zu 5.0
print(summe, type(summe))

print(7 / 2, 6 / 2)        # / liefert immer einen float, auch bei 6 / 2
print(7 // 2)              # // liefert bei zwei int einen int
print(True + 1)            # True wird als 1 gerechnet, False als 0
````

````python
# explizite Typumwandlungen
zahl = int('42')           # str -> int
print(zahl, type(zahl))

wert = float('3.5')        # str -> float
print(wert, type(wert))

text = str(123)            # int -> str
print('Wert: ' + text)

print(int(3.99))           # float -> int
print(float(7))            # int -> float
````

#### Probleme bei Typumwandlungen

**1. Informationsverlust bei `float` → `int`:** `int()` schneidet die Nachkommastellen ab, es rundet nicht.
Bei negativen Zahlen wird in Richtung 0 abgeschnitten. Zum Runden nimmt man `round()`.

````python
print(int(3.99), int(-3.99))       # abgeschnitten: 3 und -3
print(round(3.99), round(-3.99))   # gerundet: 4 und -4
````

**2. Umwandlung eines Strings in eine Zahl kann fehlschlagen.** Der String muss genau eine passende Zahl enthalten.
Leerzeichen am Anfang und am Ende sind erlaubt, ein Dezimalkomma aber nicht.

```python
int('3.5')       # ValueError: invalid literal for int() with base 10: '3.5'
int('abc')       # ValueError: invalid literal for int() with base 10: 'abc'
float('3,5')     # ValueError: could not convert string to float: '3,5'
```

Das ist wichtig bei `input()`, denn `input()` liefert immer einen String. Gibt der Benutzer etwas Falsches ein,
bricht das Programm mit einem `ValueError` ab.

````python
print(int(' 42 '))         # Leerzeichen werden ignoriert
print(float('3.5'))        # mit Punkt funktioniert es
print(int(float('3.5')))   # erst in float, dann in int umwandeln
````

**3. Keine implizite Umwandlung in einen String.** Java wandelt bei `"Alter: " + 15` die Zahl automatisch in einen
String um. Python tut das nicht:

```python
'Alter: ' + 15   # TypeError: can only concatenate str (not "int") to str
```

Man muss explizit umwandeln oder einen f-String verwenden.

````python
alter = 15
print('Alter: ' + str(alter))
print(f'Alter: {alter}')
````

**4. Überlauf bei der Umwandlung in einen kleineren Typ.** Passt ein Wert nicht in den Wertebereich des Zieltyps,
werden in Java einfach die überzähligen Bits weggelassen. Aus `(byte) 200` wird so `-56`.
Python-`int` hat keinen festen Wertebereich, mit `ctypes` lässt sich das aber wieder nachbilden.

**5. Genauigkeitsverlust bei großen ganzen Zahlen → `float`:** Ein `float` speichert nur etwa 16 Stellen.
Sehr große ganze Zahlen werden bei der Umwandlung gerundet.

````python
import ctypes

print(ctypes.c_int8(200).value)     # wie (byte) 200 in Java: -56

n = 2 ** 53 + 1
print(n)
print(float(n))                     # die letzte Stelle stimmt nicht mehr
print(int(float(n)) == n)
````

**6. Umwandlung in `bool`:** `bool()` liefert für `0`, `0.0` und den leeren String `''` den Wert `False`,
für alle anderen Werte `True`. Deshalb ist auch `bool('False')` gleich `True`, denn der String ist nicht leer.

````python
print(bool(0), bool(0.0), bool(''))
print(bool(5), bool('0'), bool('False'))
````
