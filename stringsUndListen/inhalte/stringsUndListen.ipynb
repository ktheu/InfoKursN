## Strings und Listen – Wiederholung

Strings und Listen kennen wir aus dem Brückenkurs. Hier ist das Wichtigste kompakt zusammengefasst, mit ein paar Ergänzungen.

Beide sind **Folgen** (Sequenzen): Ein String ist eine Folge von Zeichen, eine Liste eine Folge von beliebigen Werten.
Vieles funktioniert deshalb bei beiden gleich: `len()`, Indexing, Slicing, `in`, `+`, `*` und die `for`-Schleife.
Der wichtigste Unterschied: **Listen kann man verändern, Strings nicht.**

### Strings

Ein String steht in einfachen oder doppelten Hochkommas. `len()` liefert die Anzahl der Zeichen.
Mit `+` werden Strings **verkettet**, mit `*` **vervielfacht**. Eine Zahl muss vorher mit `str()` in einen String
umgewandelt werden, oder man nimmt einen f-String.

```python
s = 'Habicht'
leer = ''
print(len(s), len(leer))

gruss = 'Hallo' + ' ' + 'Welt'      # Verketten
lachen = 'Ha' * 3                   # Vervielfachen
print(gruss, lachen)

alter = 17
print('Alter: ' + str(alter))
print(f'{s} hat {len(s)} Buchstaben')
```

##### Indexing

Jedes Zeichen hat zwei Indizes: von vorne gezählt beginnend bei `0`, von hinten gezählt beginnend bei `-1`.
Ein Index außerhalb des Strings führt zu einem `IndexError`, z.B. `'Habicht'[7]`.

<img src='habicht.png' width='301'>

```python
s = 'Habicht'
print(s[0], s[1], s[-1], s[-2], s[len(s) - 1])
```

##### Slicing

`s[von:bis]` liefert den Teilstring von Index `von` bis **ausschließlich** `bis`. Die Länge des Ergebnisses ist `bis - von`.
Fehlt `von`, beginnt der Teilstring am Anfang, fehlt `bis`, geht er bis zum Ende.

**Neu:** Mit einer dritten Zahl gibt man die **Schrittweite** an: `s[::2]` nimmt jedes zweite Zeichen, `s[::-1]` liefert den String rückwärts.

```python
s = 'Habicht'
print(s[1:5])      # Index 1 bis 4
print(s[:3])       # die ersten 3 Zeichen
print(s[-3:])      # die letzten 3 Zeichen
print(s[2:])       # alles außer den ersten beiden Zeichen
print(s[:-2])      # alles außer den letzten beiden Zeichen
print(s[::2])      # jedes zweite Zeichen
print(s[::-1])     # rückwärts
```

##### Strings sind unveränderlich

Einzelne Zeichen eines Strings kann man nicht ändern: `s[0] = 'K'` führt zu einem `TypeError`.
Stattdessen erzeugt man einen **neuen** String und weist ihn der Variable zu. Auch String-Methoden wie `upper()` oder `replace()`
verändern den String nicht, sondern liefern einen neuen.

```python
s = 'Habicht'
s = 'K' + s[1:]          # neuer String aus 'K' und dem Rest
print(s)

t = 'abc'
t.upper()                # liefert 'ABC', aber t bleibt unverändert
print(t)
t = t.upper()            # erst durch die Zuweisung ändert sich t
print(t)
```

##### String-Methoden

Methoden werden mit einem Punkt hinter dem String aufgerufen. Die wichtigsten (neue sind markiert):

| Methode | Ergebnis | Beispiel mit `s = 'Abiabiabi'` |
|---|---|---|
| `s.upper()`, `s.lower()` | alles groß bzw. klein | `'ABIABIABI'`, `'abiabiabi'` |
| `s.count(t)` | wie oft `t` vorkommt | `s.count('abi')` ergibt `2` |
| `s.replace(alt, neu)` | jedes `alt` durch `neu` ersetzt | `s.replace('abi', 'oha')` ergibt `'Abiohaoha'` |
| `s.find(t)` | erster Index von `t`, `-1` wenn nicht enthalten | `s.find('bi')` ergibt `1` |
| `s.strip()` | Leerzeichen und Zeilenvorschub am Anfang und Ende entfernt | `' ja\n'.strip()` ergibt `'ja'` |
| `s.startswith(t)`, `s.endswith(t)` (neu) | beginnt bzw. endet `s` mit `t`? | `s.startswith('Abi')` ergibt `True` |
| `s.isdigit()` (neu) | besteht `s` nur aus Ziffern? | `'2025'.isdigit()` ergibt `True` |
| `s.split()` (neu) | Liste der Wörter (getrennt an Leerzeichen) | `'a b c'.split()` ergibt `['a', 'b', 'c']` |
| `s.split(t)` (neu) | Liste der Teile, getrennt an `t` | `'1;2;3'.split(';')` ergibt `['1', '2', '3']` |
| `t.join(liste)` (neu) | die Strings der Liste, verbunden mit `t` | `'-'.join(['a', 'b'])` ergibt `'a-b'` |

Mit `in` und `not in` prüft man, ob ein Teilstring vorkommt. Eine vollständige Übersicht gibt es bei
[w3schools](https://www.w3schools.com/python/python_ref_string.asp).

```python
s = 'Abiabiabi'
print(s.count('abi'), s.find('bi'), s.find('xy'))
print(s.replace('abi', 'oha'))
print('abi' in s, 'aba' not in s)

eingabe = '  2025 \n'
jahr = eingabe.strip()
print(jahr.isdigit(), int(jahr) + 1)

satz = 'Strings und Listen sind Folgen'
woerter = satz.split()
print(woerter)
print(' / '.join(woerter))
```

Methoden kann man **verketten**. Sie werden von links nach rechts ausgewertet, jede Methode arbeitet mit dem Ergebnis der vorherigen:

```python
s = 'Düsäne'
s1 = s.replace('ü', 'ue')     # 'Duesäne'
s2 = s1.replace('ä', 'ae')    # 'Duesaene'
s3 = s2.upper()               # 'DUESAENE'
s4 = s3[2:6]                  # 'ESAE'
print(s4.count('E'))

# dasselbe in einer Zeile
print(s.replace('ü', 'ue').replace('ä', 'ae').upper()[2:6].count('E'))
```

##### Zeichen durchlaufen und neue Strings bauen

Wie bei Listen gibt es zwei Arten von Schleifen: über die Zeichen oder über die Indizes.
Weil Strings unveränderlich sind, baut man ein Ergebnis oft in einer Variable auf, die mit dem leeren String `''` beginnt.

```python
s = 'Habicht'

ohne_vokale = ''
for c in s:                       # Laufvariable über die Zeichen
    if c.lower() not in 'aeiou':
        ohne_vokale = ohne_vokale + c
print(ohne_vokale)

for i in range(len(s)):           # Laufvariable über die Indizes
    if s[i] == 'i':
        print('i an Index', i)
```

##### Strings vergleichen

Jedem Zeichen ist mit `ord()` eine Zahl zugeordnet (der Unicode-Wert), `chr()` liefert zu einer Zahl das Zeichen.
Strings werden mit `<`, `>`, `==` Zeichen für Zeichen nach diesen Zahlen verglichen. Ohne Umlaute entspricht das
der alphabetischen Ordnung, aber: **Großbuchstaben kommen vor allen Kleinbuchstaben**, und Umlaute kommen nach `z`.
Für einen alphabetischen Vergleich wandelt man deshalb beide Strings vorher in Kleinbuchstaben um.

```python
print(ord('A'), ord('Z'), ord('a'), ord('z'), ord('ä'))
print(chr(65), chr(8364))

print('abc' < 'abd')               # True: das dritte Zeichen entscheidet
print('abc' < 'Abd')               # False: 'a' (97) ist größer als 'A' (65)
print('abc'.lower() < 'Abd'.lower())
print('Mähne' < 'Mazda')           # False: 'ä' (228) kommt nach 'a'
```

### Listen

Eine Liste speichert mehrere Werte (**Elemente**) in eckigen Klammern. Die Elemente können verschiedene Datentypen haben.
`len()`, Indexing, Slicing, `in`, `+` und `*` funktionieren wie bei Strings.

```python
a = [7, 12, 1, 5, 0, 3, 5]
print(len(a), a[0], a[-1], a[1:3], a[::-1])
print(5 in a, 9 in a)
print([1, 2] + [3], [0] * 4)

gemischt = ['Ada', 17, 1.65, True]
leer = []
print(gemischt, len(leer))

zahlen = list(range(1, 6))       # Liste aus einem range
print(zahlen)
```

##### Listen verändern

Anders als Strings sind Listen **veränderlich**. Die Methoden verändern die Liste selbst (und liefern meistens nichts zurück):

| Anweisung | Wirkung |
|---|---|
| `a[i] = x` | Element an Index `i` ersetzen |
| `a.append(x)` | `x` am Ende anhängen |
| `a.insert(i, x)` (neu) | `x` an Index `i` einfügen, die folgenden Elemente rücken nach hinten |
| `a.pop()` (neu) | letztes Element entfernen und zurückgeben |
| `a.pop(i)` (neu) | Element an Index `i` entfernen und zurückgeben |
| `a.remove(x)` (neu) | das erste Element mit dem Wert `x` entfernen |
| `a.sort()` (neu) | die Liste aufsteigend sortieren |

```python
a = [7, 12, 1, 5]
a[0] = 9
a.append(8)
print(a)

a.insert(1, 4)          # an Index 1 einfügen
print(a)

letztes = a.pop()       # letztes Element entfernen
print(letztes, a)

a.remove(12)            # den Wert 12 entfernen
print(a)

a.sort()
print(a)
```

##### Nützliche Funktionen für Listen

| Funktion / Methode | Ergebnis |
|---|---|
| `sum(a)` | Summe der Elemente (bei Zahlen) |
| `min(a)`, `max(a)` | kleinstes bzw. größtes Element |
| `sorted(a)` (neu) | eine **neue**, sortierte Liste, `a` bleibt unverändert |
| `a.count(x)` | wie oft `x` vorkommt |
| `a.index(x)` (neu) | Index des ersten Vorkommens von `x` (Fehler, wenn nicht enthalten) |

`sorted()` und `.sort()` sortieren auch Listen von Strings, und zwar nach der Ordnung von oben (Großbuchstaben zuerst).

```python
noten = [2, 1, 3, 2, 4, 2]
print(sum(noten), min(noten), max(noten))
print('Durchschnitt:', sum(noten) / len(noten))
print(noten.count(2), noten.index(3))

print(sorted(noten))    # neue Liste
print(noten)            # unverändert

namen = ['ben', 'Ada', 'carla']
print(sorted(namen))
```

##### Listen durchlaufen

Die Schleife über die Elemente ist kürzer. Die Schleife über die Indizes braucht man, wenn man den Index kennen
oder **Elemente der Liste verändern** will. Neue Listen füllt man oft, indem man mit der leeren Liste beginnt und
in einer Schleife `append()` aufruft.

```python
a = [3, 4, 7, 2, 8]

for k in a:                      # über die Elemente
    print(k, end=' ')
print()

for i in range(len(a)):          # über die Indizes: Elemente verändern
    a[i] = a[i] * 10
print(a)

gross = []                       # neue Liste füllen
for k in a:
    if k > 50:
        gross.append(k)
print(gross)
```

##### Typische Algorithmen

Viele Aufgaben lassen sich mit einem Durchlauf durch die Liste lösen. Man benutzt eine Variable, in der das Ergebnis
gesammelt wird (siehe *Einfache Algorithmen* im Brückenkurs):

```python
temperaturen = [12, 15, 9, 21, 18, 7]

anzahl_warm = 0                  # zählen
for t in temperaturen:
    if t > 14:
        anzahl_warm += 1
print('warme Tage:', anzahl_warm)

best = temperaturen[0]           # Maximum suchen (lineare Suche)
best_index = 0
for i in range(len(temperaturen)):
    if temperaturen[i] > best:
        best = temperaturen[i]
        best_index = i
print('Maximum', best, 'an Index', best_index)
```

### Zwischen Strings und Listen umwandeln

| von | nach | wie | Beispiel |
|---|---|---|---|
| String | Liste der Zeichen | `list(s)` | `list('abc')` ergibt `['a', 'b', 'c']` |
| String | Liste der Teile | `s.split()`, `s.split(t)` | `'3 7 1'.split()` ergibt `['3', '7', '1']` |
| Liste von Strings | String | `t.join(liste)` | `''.join(['a', 'b'])` ergibt `'ab'` |

Nach `split()` sind die Teile Strings. Zum Rechnen muss man sie noch mit `int()` umwandeln.
Umgekehrt kann `join()` nur Strings verbinden, Zahlen muss man vorher mit `str()` umwandeln.

```python
zeile = '3 7 1 9'
teile = zeile.split()
print(teile)

zahlen = []
for t in teile:
    zahlen.append(int(t))
print(zahlen, sum(zahlen))

texte = []
for k in zahlen:
    texte.append(str(k))
print('+'.join(texte), '=', sum(zahlen))

print(list('Habicht'))
```
