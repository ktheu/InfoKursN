#### Aufgabe A13
```
Der folgende Code soll das größte Element einer Liste finden, liefert
bei dieser Liste aber ein falsches Ergebnis. Finde den Fehler und
korrigiere ihn.
```
**Erwartete Ausgabe:**
```
-2
```

````python
def find_maximum(a):
    best = 0
    for x in a:
        if x > best:
            best = x
    return best

numbers = [-8, -2, -15, -6]
result = find_maximum(numbers)
print(result)
````
