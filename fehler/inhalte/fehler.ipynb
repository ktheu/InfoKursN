### Fehler

Wir unterscheiden drei Arten von Fehlern:

| Fehlerart | Was passiert? | Beispiel |
|:--|:--|:--|
| **Syntaxfehler** | Das Programm kann gar nicht erst gestartet werden. | Klammer vergessen |
| **Laufzeitfehler** | Das Programm startet, bricht dann aber mit einer Fehlermeldung ab. | Division durch 0 |
| **Logischer Fehler** | Das Programm läuft bis zum Ende, liefert aber ein falsches Ergebnis. | falsche Formel |

##### Syntaxfehler

Ein **Syntaxfehler** ist ein Verstoß gegen die Regeln der Sprache. Python prüft das Programm, bevor es irgendetwas ausführt. Deshalb wird bei einem Syntaxfehler **keine einzige Zeile** ausgeführt.

- **SyntaxError:** zum Beispiel fehlt eine Klammer, ein Hochkomma oder ein Doppelpunkt.
- **IndentationError:** die Einrückung ist falsch.

Wir bekommen einen Hinweis auf die Zeile und eine (hoffentlich) aussagekräftige Beschreibung des Fehlers. Beispiele:

**Tipp:** Die angegebene Zeile ist nicht immer die Zeile mit dem Fehler. Fehlt eine Klammer, bemerkt Python das oft erst in der **nächsten** Zeile. Schau deshalb auch in die Zeile davor.

#### Laufzeitfehler


Bei einem **Laufzeitfehler** startet das Programm, bricht aber an der fehlerhaften Stelle ab. Alle Anweisungen **davor** wurden schon ausgeführt.

Als Fehlermeldung bekommen wir einen **Traceback**. Er zeigt die Abfolge der letzten Anweisungen vor dem Fehler und gibt uns eine Zeilennummer als Hinweis. Die **letzte Zeile** des Tracebacks nennt die Art des Fehlers und eine Beschreibung. Bei der Fehlersuche sollte man auch die unmittelbar davor liegenden Zeilen im Blick haben.

| **NameError** | Variable oder Funktion ist nicht definiert | Tippfehler im Namen |
| **ValueError** | Typ stimmt, aber der Wert ist ungültig | `int("abc")` |
| **ZeroDivisionError** | Division durch 0 | `5 / 0` |
| **KeyError** | Schlüssel nicht vorhanden | bei Dictionaries, die wir später kennenlernen |

#### Logische Fehler


Bei einem **logischen Fehler** meldet Python nichts: Das Programm läuft bis zum Ende, liefert aber ein falsches Ergebnis. Solche Fehler sind manchmal schwierig zu finden.

So gehen wir vor:

1. **Erwartetes Ergebnis überlegen:** Wir testen das Programm mit Werten, bei denen wir das richtige Ergebnis kennen.
2. **print-Anweisungen einbauen:** An geeigneten Stellen geben wir Zwischenwerte aus und vergleichen sie mit unserer Erwartung.
3. **Debugger benutzen:** Manchmal müssen wir den Ablauf Schritt für Schritt verfolgen (siehe später).

Beispiel: Das Programm soll die Zahlen von 1 bis 4 addieren, liefert aber 6 statt 10. Mit einer print-Anweisung in der Schleife sehen wir sofort, dass die 4 nie addiert wird:

```python
summe = 0
for i in range(1, 4):
    summe += i
    print(f'{i=} {summe=}')     # Zwischenwerte zur Fehlersuche
print(summe)
```

In dem Beispiel oben sehen wir einen weiteren Vorteil des f-Strings. mit {i=} sehen wir den Namen und Wert der Variablen i.

Auch in dem folgenden Beispiel hat sich ein logischer Fehler eingeschlichen

```python
# Durchschnitt von 10 + 20 berechnen
durchschnitt = (10 + 20) / 2 
durchschnitt
```

### Debugger
Mit einem **Debugger** lassen wir ein Programm **Schritt für Schritt** ablaufen und verfolgen dabei, wie sich die Werte der Variablen verändern. Die Bilder zeigen den Debugger der Entwicklungsumgebung **Thonny**.

Den Debugger starten wir mit einem Klick auf das **Käfer-Symbol**. Unter *Ansicht* öffnen wir das Fenster *Variablen*.

##### Einzelschritt

Mit **Einzelschritt** (F6) wird eine Anweisung ausgeführt, und im Fenster *Variablen* sehen wir die aktuellen Werte. Die **gelb markierte Zeile** ist die **nächste** auszuführende Anweisung. Sie ist also **noch nicht** ausgeführt: Im Bild hat `x` schon den Wert 2, `y` gibt es noch nicht.

Wollen wir den gelben Bereich nicht in einem Schritt auswerten, klicken wir auf **Eintreten** (F7). Dann wird die Anweisung in kleinere Teile zerlegt, zum Beispiel erst der Ausdruck `x + 3*y`, und wir können die Auswertung genauer verfolgen.

##### Breakpoints

Durch einen Klick in den grauen Bereich neben der Zeilennummer setzen wir einen **Breakpoint** (roter Punkt). Beim Klick auf den Käfer läuft das Programm dann bis zum ersten Breakpoint und hält dort an. Mit **Fortfahren** (F8, grünes Dreieck links neben dem Stoppschild) läuft das Programm bis zum nächsten Breakpoint bzw. bis zum Ende.

Breakpoints sind praktisch, wenn uns nur ein bestimmter Teil eines längeren Programms interessiert.

### Einfache Algorithmen

#### Durchlaufen einer Liste

Viele einfache Algorithmen durchlaufen eine Liste einmal von vorne bis hinten. Dabei benutzen wir eine Variable, in der das Ergebnis **gesammelt** wird. Sie wird **vor** der Schleife mit einem passenden Startwert belegt.

##### Beispiel 1


Wir berechnen die Summe aller Zahlen, ohne die eingebaute Funktion `sum()` zu nutzen. Der Startwert ist 0.

```python
a = [15, 23, 4, 16, 8, 42]
summe = 0
for x in a:
    summe += x
print(summe)
```

##### Beispiel 2: Zählen

Wir zählen, wie viele gerade Zahlen in der Liste vorkommen. Der Zähler wird nur erhöht, wenn die Bedingung erfüllt ist.

```python
a = [15, 23, 4, 16, 8, 42]
zaehl = 0
for x in a:
    if x % 2 == 0:
       zaehl += 1
print(zaehl)
```

##### Beispiel 3: Teilsummen

Wir bilden eine neue Liste, in der an jeder Stelle die Summe aller Zahlen bis dorthin steht.

```python
a = [2, 4, 9, 12, 1, 5]
b = []
teilsumme = 0
for x in a:
    teilsumme += x
    b.append(teilsumme)
print(b)
```

##### Beispiel 4: Vergleich mit dem Vorgänger

Wir zählen, wie viele Zahlen einen kleineren Vorgänger haben. Bei `[2, 4, 9, 12, 1, 5]` sind das 4, 9, 12 und 5, also 4 Zahlen.

Weil wir jedes Element mit seinem **Vorgänger** vergleichen, brauchen wir den **Index**. Die Schleife beginnt bei Index 1, denn das erste Element hat keinen Vorgänger.

```python
a = [2, 4, 9, 12, 1, 5]
zaehl = 0
for i in range(1,len(a)):
    if a[i-1] < a[i]:
        zaehl += 1
print(zaehl)
```

#### Lineare Suche

Die **lineare Suche** durchsucht eine Liste der Reihe nach nach einem besonderen Element, zum Beispiel dem größten.

Die Idee: Das erste Element ist **vorläufig** das beste. Dann prüfen wir jedes weitere Element, ob es besser ist als das bisher beste. Wenn ja, merken wir es uns.

##### Beispiel 1

Wir suchen nach dem größten Element in einer Liste.

```python
# wir suchen das größte Element in der Liste
a = [3, 4, 3, 8, 12, 52, 35, 17, 33]

best = a[0]                    # das erste ist das beste - vorläufig
for x in a:      
    if x > best:               # was besseres gefunden ?
        best = x               
print(best)
```

Wir beginnen mit `a[0]` und nicht mit 0. Sonst geht die Suche bei Listen mit lauter negativen Zahlen schief.

##### Beispiel 2: das beste Element nach einem anderen Merkmal

Wenn das, was ein Element zum besten macht, **nicht sein eigener Wert** ist, nutzen wir zwei Variablen: `best` für das Element selbst und `best_val` für den Wert des Merkmals.

Wir suchen das Element der Liste, das den größten Rest bei der Division durch 7 hat.

```python
a = [3, 4, 3, 8, 12, 52, 35, 17, 33]

best = a[0]                    # das Erste ist das Beste - vorläufig!
best_val = a[0] % 7            # der Wert des Besten 
for x in a:                   
    if x % 7 > best_val:       # was besseres gefunden ?
        best_val = x % 7       
        best = x
print(best)
```

#### BubbleSort

**Bubblesort** ist ein einfacher Sortieralgorithmus.

Die Idee: Wir gehen die Liste von vorne nach hinten durch und vergleichen jedes Element mit seinem Nachfolger. **Ist der Nachfolger kleiner, werden die beiden vertauscht.** So wandert das größte Element bei jedem Durchgang ans Ende – wie eine große Blase, die nach oben steigt.

Das wiederholen wir so lange, bis bei einem Durchgang **nichts mehr getauscht** wird. Dann ist die Liste sortiert.

<img src='./bubbleSort.png' width='400'>



```python
def bubble_sort(a):
    getauscht = True
    while getauscht:                   # so lange noch getauscht wurde
        getauscht = False
        for i in range(len(a)-1):      # bis zum vorletzten Element, denn a[i+1] muss existieren
            if a[i] > a[i+1]:
                a[i], a[i+1] = a[i+1], a[i]    # die beiden Nachbarn tauschen
                getauscht = True
```

```python
a = [15, 23, 4, 42, 8, 16]
bubble_sort(a)
print(a)
```
