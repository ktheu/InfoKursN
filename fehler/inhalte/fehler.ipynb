## FE 1 · Fehlerarten und Syntaxfehler

### Inhalte

Wir unterscheiden drei Arten von Fehlern:

| Fehlerart | Was passiert? | Beispiel |
|:--|:--|:--|
| **Syntaxfehler** | Das Programm kann gar nicht erst gestartet werden. | Klammer vergessen |
| **Laufzeitfehler** | Das Programm startet, bricht dann aber mit einer Fehlermeldung ab. | Division durch 0 |
| **Logischer Fehler** | Das Programm läuft bis zum Ende, liefert aber ein falsches Ergebnis. | falsche Formel |

##### Syntaxfehler

Ein **Syntaxfehler** ist ein Verstoß gegen die Regeln der Sprache. Python prüft das Programm, bevor es irgendetwas ausführt. Deshalb wird bei einem Syntaxfehler **keine einzige Zeile** ausgeführt.

- **SyntaxError:** zum Beispiel fehlt eine Klammer, ein Hochkomma oder ein Doppelpunkt.
- **IndentationError:** die Einrückung ist falsch.

Wir bekommen einen Hinweis auf die Zeile und eine (hoffentlich) aussagekräftige Beschreibung des Fehlers. Führe die folgenden Zellen aus und lies die Fehlermeldungen.

````python
print("Hello World"         # die schließende Klammer fehlt
````

````python
s = 'Montag                 # das schließende Hochkomma fehlt
````

````python
def greet():
print("Hello")              # der Rumpf ist nicht eingerückt
````

````python
1x = 'Hallo'                # ein Variablenname darf nicht mit einer Ziffer beginnen
````

**Tipp:** Die angegebene Zeile ist nicht immer die Zeile mit dem Fehler. Fehlt eine Klammer, bemerkt Python das oft erst in der **nächsten** Zeile. Schau deshalb auch in die Zeile davor.

## FE 2 · Laufzeitfehler

Als Fehlermeldung bekommen wir einen **Traceback**. Er zeigt die Abfolge der letzten Anweisungen vor dem Fehler und gibt uns eine Zeilennummer als Hinweis. Die **letzte Zeile** des Tracebacks nennt die Art des Fehlers und eine Beschreibung. Bei der Fehlersuche sollte man auch die unmittelbar davor liegenden Zeilen im Blick haben.

| **NameError** | Variable oder Funktion ist nicht definiert | Tippfehler im Namen |

| **KeyError** | Schlüssel nicht vorhanden | bei Dictionaries, die wir später kennenlernen |

## FE 3 · Logische Fehler

### Inhalte

Bei einem **logischen Fehler** meldet Python nichts: Das Programm läuft bis zum Ende, liefert aber ein falsches Ergebnis. Solche Fehler sind manchmal schwierig zu finden.

So gehen wir vor:

1. **Erwartetes Ergebnis überlegen:** Wir testen das Programm mit Werten, bei denen wir das richtige Ergebnis kennen.
2. **print-Anweisungen einbauen:** An geeigneten Stellen geben wir Zwischenwerte aus und vergleichen sie mit unserer Erwartung.
3. **Debugger benutzen:** Manchmal müssen wir den Ablauf Schritt für Schritt verfolgen (siehe FE 4).

Beispiel: Das Programm soll die Zahlen von 1 bis 4 addieren, liefert aber 6 statt 10. Mit einer print-Anweisung in der Schleife sehen wir sofort, dass die 4 nie addiert wird:

````python
summe = 0
for i in range(1, 4):
    summe += i
    print('i =', i, ' summe =', summe)     # Zwischenwerte zur Fehlersuche
print(summe)
````

Die Ausgabe zeigt: `i` läuft nur bis 3. Richtig ist `range(1, 5)`. Wenn der Fehler gefunden ist, entfernen wir die zusätzliche print-Anweisung wieder.

### Aufgabe 2

Die Funktion soll das Maximum einer Liste finden. Für `[3, 9, 2]` liefert sie richtig 9. Teste sie mit der Liste unten. Was geht schief, und warum? Korrigiere die Funktion.

````python
def find_maximum(a):
    best = 0
    for x in a:
        if x > best:
            best = x
    return best

numbers = [-10, -3, -20, -5]
result = find_maximum(numbers)
print('Maximum:', result)
````

#### Lösung

Ausgegeben wird `Maximum: 0`, richtig wäre `-3`. Der Startwert `best = 0` ist größer als alle Zahlen der Liste. Keine Zahl ist also größer als `best`, und `best` behält den Wert 0 – eine Zahl, die in der Liste gar nicht vorkommt.

Der Fehler fällt nur bei Listen mit ausschließlich negativen Zahlen auf. Deshalb ist es wichtig, auch **Sonderfälle** zu testen.

Richtig ist es, mit dem ersten Element der Liste zu beginnen:

````python
def find_maximum(a):
    best = a[0]                # das erste Element ist vorläufig das größte
    for x in a:
        if x > best:
            best = x
    return best

print('Maximum:', find_maximum([-10, -3, -20, -5]))    # -3
print('Maximum:', find_maximum([3, 9, 2]))             # 9
````

## FE 5 · Durchlaufen einer Liste

### Inhalte

Viele einfache Algorithmen durchlaufen eine Liste einmal von vorne bis hinten. Dabei benutzen wir eine Variable, in der das Ergebnis **gesammelt** wird. Sie wird **vor** der Schleife mit einem passenden Startwert belegt.

Wir berechnen die Summe aller Zahlen, ohne die eingebaute Funktion `sum()` zu nutzen. Der Startwert ist 0.

````python
a = [15, 23, 4, 16, 8, 42]
summe = 0
for x in a:
    summe += x
print(summe)       # 108
````

````python
a = [15, 23, 4, 16, 8, 42]
zaehl = 0
for x in a:
    if x % 2 == 0:
        zaehl += 1
print(zaehl)       # 4
````

Wir bilden eine neue Liste, in der an jeder Stelle die Summe aller Zahlen bis dorthin steht.

````python
a = [2, 4, 9, 12, 1, 5]
zaehl = 0
for i in range(1, len(a)):
    if a[i-1] < a[i]:
        zaehl += 1
print(zaehl)       # 4
````

## FE 6 · Lineare Suche

### Inhalte

Die **lineare Suche** durchsucht eine Liste der Reihe nach nach einem besonderen Element, zum Beispiel dem größten.

Die Idee: Das erste Element ist **vorläufig** das beste. Dann prüfen wir jedes weitere Element, ob es besser ist als das bisher beste. Wenn ja, merken wir es uns.

##### Beispiel 1: das größte Element

````python
a = [3, 4, 3, 8, 12, 52, 35, 17, 33]

best = a[0]                    # das erste ist das beste – vorläufig
for x in a:
    if x > best:               # etwas Besseres gefunden?
        best = x
print(best)        # 52
````

Wir beginnen mit `a[0]` und nicht mit 0. Sonst geht die Suche bei Listen mit lauter negativen Zahlen schief (siehe FE 3, Aufgabe 2).

##### Beispiel 2: das beste Element nach einem anderen Merkmal

Wenn das, was ein Element zum besten macht, **nicht sein eigener Wert** ist, nutzen wir zwei Variablen: `best` für das Element selbst und `best_val` für den Wert des Merkmals.

Wir suchen das Element der Liste, das den größten Rest bei der Division durch 7 hat.

````python
a = [3, 4, 3, 8, 12, 52, 35, 17, 33]

best = a[0]                    # das erste ist das beste – vorläufig
best_val = a[0] % 7            # der Wert des besten
for x in a:
    if x % 7 > best_val:       # etwas Besseres gefunden?
        best_val = x % 7
        best = x
print(best, best_val)          # 12 5
````

**2.** Für den Index brauchen wir die Schleife über den Index:

### Aufgabe 2

## FE 7 · Bubblesort

### Inhalte

**Bubblesort** ist ein einfacher Sortieralgorithmus.

Die Idee: Wir gehen die Liste von vorne nach hinten durch und vergleichen jedes Element mit seinem Nachfolger. **Ist der Nachfolger kleiner, werden die beiden vertauscht.** So wandert das größte Element bei jedem Durchgang ans Ende – wie eine große Blase, die nach oben steigt.

Das wiederholen wir so lange, bis bei einem Durchgang **nichts mehr getauscht** wird. Dann ist die Liste sortiert.

<img src="./bubbleSort.png" width="400" alt="Die Liste 15, 23, 4, 42, 8, 16 nach jedem Durchgang von Bubblesort, bis sie sortiert ist">

````python
def bubble_sort(a):
    getauscht = True
    while getauscht:                   # so lange noch getauscht wurde
        getauscht = False
        for i in range(len(a)-1):      # bis zum vorletzten Element, denn a[i+1] muss existieren
            if a[i] > a[i+1]:
                a[i], a[i+1] = a[i+1], a[i]    # die beiden Nachbarn tauschen
                getauscht = True
````

````python
a = [15, 23, 4, 42, 8, 16]
bubble_sort(a)
print(a)           # [4, 8, 15, 16, 23, 42]
````

Drei Dinge sind neu:

- **Tauschen:** Mit `a[i], a[i+1] = a[i+1], a[i]` werden zwei Elemente in einer Zeile vertauscht. Python wertet erst die rechte Seite vollständig aus und weist dann zu.
- **Die Variable `getauscht`** merkt sich, ob im aktuellen Durchgang getauscht wurde. Sie steuert die while-Schleife.
- **Kein return:** Die Funktion verändert die übergebene Liste direkt. Deshalb ist `a` nach dem Aufruf sortiert, obwohl die Funktion nichts zurückgibt.

### Aufgabe 1

Sortiere die Liste `[5, 1, 4, 2]` von Hand mit Bubblesort. Notiere die Liste nach jedem Tausch. Wie viele Durchgänge der while-Schleife sind nötig?

#### Lösung

| Durchgang | Vergleich | Tausch? | Liste danach |
|:--|:--|:--|:--|
| 1 | 5 > 1 | ja | `[1, 5, 4, 2]` |
| 1 | 5 > 4 | ja | `[1, 4, 5, 2]` |
| 1 | 5 > 2 | ja | `[1, 4, 2, 5]` |
| 2 | 1 > 4 | nein | `[1, 4, 2, 5]` |
| 2 | 4 > 2 | ja | `[1, 2, 4, 5]` |
| 2 | 4 > 5 | nein | `[1, 2, 4, 5]` |
| 3 | 1 > 2, 2 > 4, 4 > 5 | nein | `[1, 2, 4, 5]` |

Es sind **3 Durchgänge** nötig. Im dritten Durchgang wird nichts mehr getauscht, erst daran erkennt der Algorithmus, dass die Liste sortiert ist.

````python
# Zum Nachprüfen: print-Anweisung zeigt die Liste nach jedem Durchgang
a = [5, 1, 4, 2]
getauscht = True
while getauscht:
    getauscht = False
    for i in range(len(a)-1):
        if a[i] > a[i+1]:
            a[i], a[i+1] = a[i+1], a[i]
            getauscht = True
    print(a)
````

### Aufgabe 2

1. Ändere die Funktion `bubble_sort` so, dass sie **absteigend** sortiert.
2. Was passiert, wenn man in der for-Schleife `range(len(a))` statt `range(len(a)-1)` schreibt?

#### Lösung

**1.** Es muss nur der Vergleich umgedreht werden: Getauscht wird, wenn der Nachfolger **größer** ist.

````python
def bubble_sort_absteigend(a):
    getauscht = True
    while getauscht:
        getauscht = False
        for i in range(len(a)-1):
            if a[i] < a[i+1]:
                a[i], a[i+1] = a[i+1], a[i]
                getauscht = True

a = [15, 23, 4, 42, 8, 16]
bubble_sort_absteigend(a)
print(a)           # [42, 23, 16, 15, 8, 4]
````

**2.** Es gibt einen **IndexError**. Beim letzten Index `i = len(a)-1` greift `a[i+1]` auf ein Element hinter dem Ende der Liste zu.
