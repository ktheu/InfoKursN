### Einfache Algorithmen

#### Durchlaufen einer Liste

##### Beispiel 1

Wir wollen die Summe aller Zahlen berechnen (ohne die eingebaute Funktion sum zu nutzen).

```python
a = [15, 23, 4, 16, 8, 42]
summe = 0
for x in a:
    summe += x
print(summe)
```

##### Beispiel 2

Wir wollen zählen, wieviel gerade Zahlen in der Liste a vorkommen.

```python
a = [15, 23, 4, 16, 8, 42]
zaehl = 0
for x in a:
    if x % 2 == 0:
       zaehl += 1
print(zaehl)
```

#### Lineare Suche

Die lineare Suche sucht in einer Liste oder einer Folge von Werten ein ausgezeichnetes Element. 

##### Beispiel 1

Wir suchen nach dem größten Element in einer Liste.

```python
# wir suchen das größte Element in der Liste
a = [3, 4, 3, 8, 12, 52, 35, 17, 33]

best = a[0]                    # das erste ist das beste - vorläufig
for x in a:      
    if x > best:               # was besseres gefunden ?
        best = x               
print(best)
```

##### Beispiel 2

Wenn das, was das Element als bestes auszeichnet, nicht sein eigener Wert ist, nutzen wir zwei Variablen: *best* und *best_val*. 

Wir suchen nach dem Element in der Liste, das den größten Rest bei der Division durch 7 hat.

```python
a = [3, 4, 3, 8, 12, 52, 35, 17, 33]

best = a[0]                    # das Erste ist das Beste - vorläufig!
best_val = a[0] % 7            # der Wert des Besten 
for x in a:                   
    if x % 7 > best_val:       # was besseres gefunden ?
        best_val = x % 7       
        best = x
print(best)
```

#### BubbleSort

BubbleSort ist ein einfacher Sortieralgorithmus.
Idee: Falls der Nachfolger kleiner ist, werden die Plätze getauscht. Oder vertikal: große Blasen steigen auf.
Wird bei einem Durchgang nichts mehr getauscht, ist der Algorithmus fertig.

<img src='./bubbleSort.png' width='400'>



```python
def bubble_sort(a):
    getauscht = True
    while getauscht:
        getauscht = False
        for i in range(len(a)-1):
            if a[i] > a[i+1]:
                a[i],a[i+1]=a[i+1],a[i]
                getauscht = True
```

```python
a = [15, 23, 4, 42, 8, 16]
bubble_sort(a)
print(a)
```
