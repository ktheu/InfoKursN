##### Vorbemerkung: Feste vs. beliebige Bitlänge

In vielen Programmiersprachen (z.B. in Java oder C) werden Zahlen mit fester Bitlänge gespeichert (typischerweise 8, 16, 32 oder 64 Bit). Das ist auch der Grund, warum dort Overflow-Probleme auftreten können.

Python ist hier eine Ausnahme: Der Datentyp *int* in Python hat eine beliebige Genauigkeit. Python vergrößert den benötigten Speicher automatisch, wenn eine Zahl zu groß wird. Das folgende Beispiel zeigt das:

````python
import sys
print(sys.getsizeof(2**1000))       # in bytes
print(sys.getsizeof(2**10000)) 
````

### Negative Zahlen: Einerkomplement und Zweierkomplement

Für vorzeichenbehaftete ganze Zahlen mit fester Bitlänge gibt es verschiedene Darstellungen:

#### 1 - Vorzeichen-Betrag-Darstellung
Das höchstwertige Bit (MSB, most significant bit) gibt das Vorzeichen an (0 = positiv, 1 = negativ), der Rest den Betrag.

Beispiel (8 Bit): `+5 = 00000101`, `−5 = 10000101`

Nachteile:
- Es gibt zwei Darstellungen für die Null (`00000000` = +0 und `10000000` = −0).
- Addition und Subtraktion benötigen unterschiedliche Algorithmen 

#### 2 - Einerkomplement (= bitweise Negation)
Negative Zahlen erhält man, indem man alle Bits der positiven Zahl invertiert (0↔1)  

Beispiel (8 Bit): `+5 = 00000101` → `−5 = 11111010`

Auch hier gibt es zwei Nullen: `00000000` (+0) und `11111111` (−0).
Außerdem ist ggf. ein Korrekturschritt notwendig: Entsteht beim Addieren ein Übertrag aus dem höchstwertigen Bit (MSB) heraus, darf dieser nicht
einfach verworfen werden, sondern muss auf das niedrigstwertige Bit addiert werden.

Beispiel: 5 + (−3) im 8-Bit-Einerkomplement

```
   00000101   ( 5)
 + 11111100   (-3, Einerkomplement von 00000011)
 -----------
  100000001   (9-Bit-Zwischenergebnis, Übertrag = 1)

   00000001
 +        1   <- Übertrag wieder addieren (End-Around-Carry)
 -----------
   00000010   ( 2)  richtig
```

#### 3 - Zweierkomplement  
Negative Zahlen erhält man, indem man alle Bits invertiert und danach 1 addiert
(= Einerkomplement + 1).

Beispiel (8 Bit): `+5 = 00000101` → invertieren: `11111010` → +1 → `−5 = 11111011`

Vorteile des Zweierkomplements (deshalb wird es in der Praxis fast ausschließlich verwendet):
- Es gibt nur eine Darstellung der Null.
- Addition und Subtraktion funktionieren mit demselben Algorithmus wie bei
  vorzeichenlosen Zahlen – man addiert einfach bitweise mit Übertrag. Der Übertrag aus dem MSB wird einfach verworfen.

#####  Wertebereich im Zweierkomplement

              Wertebereich   größte Zahl   kleinste Zahl
    4-Bit       -8...7              0111          1000
    8-Bit     -128...127        01111111      10000000

Bei $n$ Bit ist der Wertebereich: 
$-2^{n-1} \; \text{bis} \; 2^{n-1}-1$

##### Darstellung im Zahlenkreis

Die Darstellung des Zweierkomplements im Zahlenkreis zeigt den Zusammenhang zwischen den positiven und negativen Zahlen.

<img src='../bilder/zahlenkreis2.png' width='500'>

Wir sehen, dass folgende Operationen äquivalent sind:

1. subtrahiere 1 (gehe 1 Schritt linksrum), dann bitweise Negation
2. bitweise Negation, dann addiere 1 (gehe 1 Schritt rechtsrum)

Insbesondere: Von der Darstellung einer negativen Zahl kommt man zur entsprechenden positiven Zahl, 
indem man alle Bits invertiert und danach 1 addiert.

##### Effekte beim Überlauf
In der Programmiersprache Java werden ganze Zahlen vom Typ *int* intern im 32-Bit Zweierkomplement gespeichert.
Wie lässt sich die Ausgabe erklären?  

````python
int k = 1;
while (k > 0) {
  k = k+1;
}
System.out.println(k);

Ausgabe: -2147483648
````

### Genauigkeit bei endlicher Stellenzahl

Neben dem Overflow bei ganzen Zahlen gibt es bei Gleitkommazahlen zusätzlich das Problem der
**begrenzten Genauigkeit** (Rundungsfehler), da nicht jede reelle Zahl exakt mit endlich vielen
Bits dargestellt werden kann.

````python
x = 0.3
print(3*x)
````
