### ADT Keller und Schlange

Keller und Schlange sind Abstrakte Datentypen. Wie bei der Liste legen wir fest, welche Operationen es gibt und was sie leisten.
Beide speichern eine Folge von Elementen, aber der Zugriff ist eingeschränkt: Wir kommen nur an ein bestimmtes
Element heran. Gerade das macht sie so nützlich, denn viele Abläufe in Rechnern und im Alltag folgen genau diesem Muster.

## 1. Der ADT Keller

Ein Keller heißt auch Stapel oder **stack**. Wir stellen uns einen Stapel Teller vor: Ein neuer Teller kommt oben drauf, und
genommen wird immer der oberste.

Prinzip **LIFO**: Last in, First out.

Ein *Keller* ist eine (ggf. leere) Folge von Elementen zusammen mit einem so genannten (ggf. undefinierten) *Top-Element*.

#### Schnittstelle des ADT Keller

| Operation | Bedeutung |
|---|---|
| `empty` | liefert True, falls der Keller leer ist |
| `push(x)` | legt das Element x auf den Keller |
| `top` | liefert das oberste Kellerelement |
| `pop` | entfernt das oberste Kellerelement |

`top` und `pop` setzen voraus, dass der Keller nicht leer ist. In anderen Quellen heißt `empty` auch `isEmpty`.

<img src='keller_ablauf.png' width='876'>

#### Beispiel

Wir notieren den Keller horizontal mit dem top-Element auf der linken Seite.

| Operation | Keller | liefert |
|---|---|---|
| `k = Keller()` | leer | |
| `k.push(3)` | 3 | |
| `k.push(8)` | 8 3 | |
| `k.push(5)` | 5 8 3 | |
| `k.top()` | 5 8 3 | 5 |
| `k.pop()` | 8 3 | |
| `k.push(1)` | 1 8 3 | |
| `k.pop()` | 8 3 | |
| `k.empty()` | 8 3 | False |

## 2. Keller in Python

Für den Keller nutzen wir die Python-Liste. Das Ende der Liste ist das obere Ende des Kellers. Dort kostet Einfügen und Entfernen
nur $O(1)$, weil keine Elemente verschoben werden müssen.

| ADT Keller | Python-Liste `a` |
|---|---|
| `empty` | `not a` oder `len(a) == 0` |
| `push(x)` | `a.append(x)` |
| `top` | `a[-1]` |
| `pop` | `a.pop()` |

Achtung: `a.pop()` entfernt das Element und gibt es außerdem zurück. Die Operation `pop` des ADT entfernt es nur.

```python
a = []
a.append(3)
a.append(8)
a.append(5)
print(a[-1])
a.pop()
a.append(1)
x = a.pop()
print(x, a, not a)
```

Da das zuletzt Hineingelegte zuerst wieder herauskommt, kehrt ein Keller die Reihenfolge um.

```python
k = []
for c in 'Keller':
    k.append(c)

s = ''
while k:
    s += k.pop()
print(s)
```

## 3. Anwendungen des Kellers

#### Rückgängig

Vor jeder Änderung legen wir den alten Zustand auf einen Keller. Die letzte Änderung lässt sich dann als erste zurücknehmen.
Genauso funktioniert der Zurück-Knopf im Browser.

```python
text = ''
verlauf = []
for teil in ['Hallo', ' Welt', '!']:
    verlauf.append(text)
    text += teil
print(text)

text = verlauf.pop()
print(text)
text = verlauf.pop()
print(text)
```

#### Der Aufrufstapel

Der Rechner verwaltet Funktionsaufrufe mit einem Keller, dem **Aufrufstapel** (call stack). Ruft eine Funktion eine andere auf, kommt der neue
Aufruf oben auf den Keller. Ist eine Funktion fertig, wird ihr Aufruf entfernt und die Funktion darunter läuft weiter. Die zuletzt
aufgerufene Funktion wird also als erste beendet.

<img src='aufrufstapel.png' width='876'>

```python
def f(x):
    return g(x) + 1

def g(x):
    return h(x) * 2

def h(x):
    return 10 // x

print(f(5))
```

Bei einem Fehler zeigt Python den Aufrufstapel im Traceback: von `f` über `g` bis `h`, wo der Fehler entstanden ist.

```python
print(f(0))
```

#### Rekursion auflösen

Bei einer Rekursion liegen alle noch offenen Aufrufe auf dem Aufrufstapel. Wir können uns die offene Arbeit auch selbst in einem Keller
merken und kommen dann ohne Rekursion aus.

```python
def fak(n):
    if n <= 1:
        return 1
    return n * fak(n-1)

def fak_keller(n):
    k = []
    while n > 1:
        k.append(n)
        n -= 1
    ergebnis = 1
    while k:
        ergebnis *= k.pop()
    return ergebnis

print(fak(5), fak_keller(5))
```

## 4. Der ADT Schlange

Eine Schlange heißt auch Warteschlange oder **queue**. Wer zuerst kommt, ist zuerst dran. Neue Elemente stellen sich hinten an.

Prinzip **FIFO**: First in, First out.

Eine *Schlange* ist eine (ggf. leere) Folge von Elementen zusammen mit einem so genannten (ggf. undefinierten) *Front-Element*.

#### Schnittstelle des ADT Schlange

| Operation | Bedeutung |
|---|---|
| `empty` | liefert True, falls die Schlange leer ist |
| `enq(x)` | fügt das Element x hinten ein |
| `front` | liefert das vorderste Element |
| `deq` | entfernt das vorderste Element |

`front` und `deq` setzen voraus, dass die Schlange nicht leer ist. In anderen Quellen heißen die Operationen auch `isEmpty`,
`enqueue` und `dequeue`.

<img src='schlange_ablauf.png' width='444'>

#### Beispiel

Wir notieren die Schlange horizontal mit dem Kopf der Schlange auf der linken Seite.

| Operation | Schlange | liefert |
|---|---|---|
| `s = Schlange()` | leer | |
| `s.enq(3)` | 3 | |
| `s.enq(8)` | 3 8 | |
| `s.enq(5)` | 3 8 5 | |
| `s.front()` | 3 8 5 | 3 |
| `s.deq()` | 8 5 | |
| `s.enq(1)` | 8 5 1 | |
| `s.deq()` | 5 1 | |
| `s.empty()` | 5 1 | False |

## 5. Schlange in Python

Für die Schlange nutzen wir die Funktion `deque` aus dem Modul `collections` mit den Methoden `append` und `popleft`.

| ADT Schlange | `q = deque()` |
|---|---|
| `empty` | `not q` oder `len(q) == 0` |
| `enq(x)` | `q.append(x)` |
| `front` | `q[0]` |
| `deq` | `q.popleft()` |

Auch `popleft` gibt das entfernte Element zurück.

```python
from collections import deque

a = [5, 4, 6]
q = deque(a)    # macht aus der Liste eine Schlange
q.append(7)
x = q.popleft()
print(x, q[0], q, not q)
```

Eine Python-Liste wäre hier schlecht geeignet: `a.pop(0)` muss alle übrigen Elemente verschieben und braucht $O(n)$, `popleft` nur $O(1)$.
Das zeigt die Messung. Wir entfernen $n$ Elemente von vorn.

```python
import time

def laufzeit(a, entferne, n):
    start = time.perf_counter()
    for i in range(n):
        entferne(a)
    return time.perf_counter() - start

print('n', 'pop(0)', 'popleft', sep='\t')
for n in [25000, 50000, 100000]:
    t1 = laufzeit(list(range(n)), lambda a: a.pop(0), n)
    t2 = laufzeit(deque(range(n)), lambda q: q.popleft(), n)
    print(n, round(t1, 3), round(t2, 3), sep='\t')
```

## 6. Anwendungen der Schlange

Mit einer Schlange modellieren wir Situationen, in denen etwas der Reihe nach abgearbeitet wird: die Kasse im Supermarkt, die Warteschleife
einer Hotline oder die Aufträge an einen Drucker.

#### Beispiel: Drucker

Die Klasse `Drucker` merkt sich ihre Aufträge in einer Schlange. Wer `Drucker` benutzt, muss davon nichts wissen.

```python
class Drucker:
    def __init__(self):
        self.auftraege = deque()

    def neuer_auftrag(self, datei):
        self.auftraege.append(datei)

    def drucke(self):
        if not self.auftraege:
            print('Keine Aufträge')
        else:
            print('Drucke', self.auftraege.popleft())

d = Drucker()
d.neuer_auftrag('brief.pdf')
d.neuer_auftrag('foto.png')
d.drucke()
d.neuer_auftrag('liste.txt')
d.drucke()
d.drucke()
d.drucke()
```

Auch in der Informatik kommt die Schlange oft vor, zum Beispiel bei der Breitensuche auf Bäumen.

## 7. Keller und Schlange im Vergleich

| | Keller | Schlange |
|---|---|---|
| Prinzip | LIFO | FIFO |
| Einfügen | oben, `push` | hinten, `enq` |
| Zugriff und Entfernen | oben, `top` und `pop` | vorn, `front` und `deq` |
| in Python | Liste: `append`, `pop` | `deque`: `append`, `popleft` |
| Reihenfolge beim Entnehmen | umgekehrt | wie beim Einfügen |
| Anwendungen | Rückgängig, Aufrufstapel | Warteschlangen, Breitensuche |

#### Zusammenfassung

| | |
|---|---|
| Keller, stack | LIFO: Das zuletzt hinzugefügte Element wird als erstes entfernt. |
| `push`, `top`, `pop` | oben einfügen, oberstes Element ansehen, oberstes Element entfernen |
| Keller in Python | `a.append(x)`, `a[-1]`, `a.pop()`, Test auf leer mit `not a` |
| Aufrufstapel | Keller, mit dem der Rechner die Funktionsaufrufe verwaltet |
| Schlange, queue | FIFO: Das zuerst hinzugefügte Element wird als erstes entfernt. |
| `enq`, `front`, `deq` | hinten einfügen, vorderstes Element ansehen, vorderstes Element entfernen |
| Schlange in Python | `q = deque()`, `q.append(x)`, `q[0]`, `q.popleft()`, Test auf leer mit `not q` |
