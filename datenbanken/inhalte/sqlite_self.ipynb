## SQ 1 · Datenbank anlegen

**Ziel:** Du kannst mit Python eine SQLite-Datenbank anlegen und darin Tabellen mit Datentypen, Primär- und Fremdschlüsseln erstellen.

### 1.1 Wer macht was?

Wenn wir mit einer Datenbank arbeiten, sind vier Dinge beteiligt:

- **SQLite** ist das **Datenbankmanagementsystem (DBMS)**. Es verwaltet die Daten, prüft die Regeln (z. B. Primärschlüssel) und führt unsere Anweisungen aus.
- Das Python-Modul **`sqlite3`** ist die **Datenbankschnittstelle**. Darüber schickt unser Python-Programm Anweisungen an das DBMS.
- **SQL** (*Structured Query Language*) ist die **Sprache**, in der wir die Anweisungen formulieren.
- Die **Datenbank** selbst steht bei SQLite in einer einzigen Datei, hier `lkwahl.db`.

```
Python-Programm  →  sqlite3 (Schnittstelle)  →  SQLite (DBMS)  →  lkwahl.db (Datei)
```

**Aufgabe 1.1** Im Python-Code findest du später Zeilen wie diese:

```python
cur.execute("SELECT * FROM Lehrer")
```

Welcher Teil ist Python, welcher ist SQL? Wer führt die SQL-Anweisung eigentlich aus?


<details>
<summary><b>Lösung</b></summary>

`cur.execute(...)` ist Python. Die Methode gehört zur Schnittstelle `sqlite3` und schickt eine Anweisung an das DBMS. Der String `"SELECT * FROM Lehrer"` ist die SQL-Anweisung. Python versteht sie nicht selbst, für Python ist sie nur ein String. Ausgeführt wird sie von SQLite, das dazu die Datei `lkwahl.db` liest.

</details>

### 1.2 Das Grundmuster: Connection und Cursor

Jede Arbeit mit der Datenbank folgt demselben Muster:

1. `sqlite3.connect(...)` baut eine **Verbindung** (*Connection*) zur Datenbank auf. Gibt es die Datei noch nicht, wird sie angelegt.
2. Über einen **Cursor** schicken wir SQL-Anweisungen mit `execute()` an das DBMS.
3. `commit()` **speichert** Änderungen dauerhaft. Wir brauchen es nur, wenn wir die Datenbank verändern.
4. `close()` **schließt** die Verbindung.

Führe die Zelle aus. Scheinbar passiert nichts, aber im Ordner gibt es jetzt die Datei `lkwahl.db` (falls sie nicht schon da war).

````python
import sqlite3
con = sqlite3.connect("lkwahl.db")    # wird neu angelegt, falls sie noch nicht existiert
cur = con.cursor()

# hier kommen unsere SQL-Anweisungen hin

con.commit()        # falls Änderungen in der Datenbank gespeichert werden sollen
con.close()
````

### 1.3 Tabellen anlegen: CREATE TABLE

Mit `CREATE TABLE` legen wir eine Tabelle an. Für jede Spalte geben wir einen Namen und einen Datentyp an:

| SQL-Datentyp | entspricht in Python | Beispiel |
|:--|:--|:--|
| `INTEGER` | `int` | `120` |
| `REAL` | `float` | `12.99` |
| `TEXT` | `str` | `'Schulte'` |

Zusätzlich können wir angeben:

- `PRIMARY KEY`: Die Spalte ist der Primärschlüssel. Besteht er aus mehreren Spalten, schreiben wir `PRIMARY KEY (Spalte1, Spalte2)` ans Ende.
- `NOT NULL`: Die Spalte darf nicht leer bleiben.
- `FOREIGN KEY (Spalte) REFERENCES Tabelle(Spalte)`: Die Spalte ist ein Fremdschlüssel.

Dabei gilt eine feste **Reihenfolge**:

- Innerhalb einer Spalte steht zuerst der **Name**, dann der **Datentyp**, dann die Bedingungen wie `PRIMARY KEY` oder `NOT NULL`.
- Zuerst kommen **alle Spalten**, danach die Bedingungen in eigenen Zeilen, also `PRIMARY KEY (…)` und `FOREIGN KEY …`. Eine Spalte nach einem `FOREIGN KEY` führt zu einem `syntax error`.

```
CREATE TABLE Kurse (
    KID TEXT PRIMARY KEY,                              -- Name, Datentyp, Bedingung
    Kuerzel TEXT NOT NULL,
    Stunden INTEGER,
    FOREIGN KEY (Kuerzel) REFERENCES Lehrer(Kuerzel)   -- erst nach allen Spalten
)
```

**Schreibweise:** SQL unterscheidet bei Schlüsselwörtern wie `CREATE TABLE`, `SELECT` oder `WHERE` nicht zwischen Groß- und Kleinschreibung, `select` funktioniert genauso wie `SELECT`. Üblich ist aber, Schlüsselwörter **groß** zu schreiben. Dann sieht man auf einen Blick, was zu SQL gehört und was Tabellen- oder Spaltennamen sind. Tabellen- und Spaltennamen schreiben wir so, wie sie angelegt wurden.

Mit `DROP TABLE IF EXISTS` löschen wir eine Tabelle, falls es sie schon gibt. So können wir die Zelle beliebig oft ausführen.

**Beispiel:** Die Tabelle *Lehrer(<u>Kuerzel</u>, Lehrer)*

````python
import sqlite3
con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("DROP TABLE IF EXISTS Lehrer")
cur.execute("""
CREATE TABLE Lehrer (
    Kuerzel TEXT PRIMARY KEY,
    Lehrer TEXT NOT NULL
)
""")

con.commit()
con.close()
````

**Vorhersage 1.2** Die nächste Zelle legt die Tabelle *Lehrer* noch einmal an, diesmal aber **ohne** `DROP TABLE IF EXISTS`. Was wird passieren? Überlege zuerst, dann führe die Zelle aus.

````python
import sqlite3
con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
CREATE TABLE Lehrer (
    Kuerzel TEXT PRIMARY KEY,
    Lehrer TEXT NOT NULL
)
""")

con.commit()
con.close()
````

<details>
<summary><b>Erklärung</b></summary>

SQLite meldet `OperationalError: table Lehrer already exists`. Eine Tabelle kann nicht zweimal angelegt werden. Deshalb steht in unseren Zellen vorher `DROP TABLE IF EXISTS`.

Lies Fehlermeldungen immer bis zur **letzten Zeile**. Dort steht, was schiefgegangen ist.

</details>

**Wichtig:** Die Zelle ist mit einem Fehler abgebrochen, bevor `con.close()` ausgeführt wurde. Die Verbindung ist also noch offen, und unter Windows bleibt die Datei dann gesperrt. Führe deshalb die nächste Zelle aus.

> **Merke:** Nach einer Fehlermeldung `con.close()` ausführen. Hilft das nicht, starte den Kernel neu (*Kernel → Restart Kernel*).
>
> Meldet SQLite später `database is locked`, ist meist noch eine Verbindung aus einer abgebrochenen Zelle offen. Auch dann hilft `con.close()` oder ein Neustart des Kernels.

````python
# Nach einer Fehlermeldung: offene Verbindung schließen
con.close()
````

### 1.4 Die Datenbank LK-Wahl aufbauen

Die folgende Zelle legt alle vier Tabellen an und füllt sie mit Daten. `execute()` kann immer nur **eine** SQL-Anweisung ausführen. Mit `executescript()` übergeben wir mehrere Anweisungen auf einmal, getrennt durch einen **Strichpunkt**. Wie `INSERT INTO` genau funktioniert, lernst du in SQ 2.

**Ausgangszustand:** Diese Zelle stellt die Datenbank jederzeit wieder her. Führe sie jetzt aus und später immer dann, wenn du dich vertan hast oder eine Aufgabe dich dazu auffordert.

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.executescript("""
DROP TABLE IF EXISTS LKWahl;
DROP TABLE IF EXISTS Schueler;
DROP TABLE IF EXISTS Kurse;
DROP TABLE IF EXISTS Lehrer;

CREATE TABLE Lehrer (
    Kuerzel TEXT PRIMARY KEY,
    Lehrer TEXT NOT NULL
);

CREATE TABLE Kurse (
    KID TEXT PRIMARY KEY,
    Kuerzel TEXT NOT NULL,
    Stunden INTEGER,
    FOREIGN KEY (Kuerzel) REFERENCES Lehrer(Kuerzel)
);

CREATE TABLE Schueler (
    SNr INTEGER PRIMARY KEY,
    Name TEXT NOT NULL,
    Vorname TEXT NOT NULL
);

CREATE TABLE LKWahl (
    SNr INTEGER,
    KID TEXT,
    PRIMARY KEY (SNr, KID),
    FOREIGN KEY (SNr) REFERENCES Schueler(SNr),
    FOREIGN KEY (KID) REFERENCES Kurse(KID)
);

INSERT INTO Lehrer VALUES ('SC', 'Schulte');
INSERT INTO Lehrer VALUES ('DA', 'Dallmeyer');
INSERT INTO Lehrer VALUES ('AD', 'Adam');
INSERT INTO Lehrer VALUES ('GE', 'Gebauer');

INSERT INTO Kurse VALUES ('M1', 'SC', 5);
INSERT INTO Kurse VALUES ('Ph1', 'SC', 5);
INSERT INTO Kurse VALUES ('Bio1', 'DA', 5);
INSERT INTO Kurse VALUES ('M2', 'AD', 5);
INSERT INTO Kurse VALUES ('Ge2', 'GE', 5);

INSERT INTO Schueler VALUES (120, 'Riedberg', 'Malte');
INSERT INTO Schueler VALUES (121, 'Dreis', 'Lena');
INSERT INTO Schueler VALUES (122, 'Rödding', 'Maike');

INSERT INTO LKWahl VALUES (120, 'M1');
INSERT INTO LKWahl VALUES (120, 'Ph1');
INSERT INTO LKWahl VALUES (121, 'Bio1');
INSERT INTO LKWahl VALUES (121, 'Ph1');
INSERT INTO LKWahl VALUES (122, 'M2');
INSERT INTO LKWahl VALUES (122, 'Ge2');
""")

con.commit()
con.close()
print('Datenbank lkwahl.db im Ausgangszustand.')
````

**Aufgabe 1.2** Sieh dir die `CREATE TABLE`-Anweisung für *LKWahl* in der Zelle oben an.

1. Warum steht `PRIMARY KEY` hier nicht hinter einer Spalte, sondern in einer eigenen Zeile?
2. Auf welche Tabellen verweisen die beiden Fremdschlüssel?


<details>
<summary><b>Lösung</b></summary>

1. Der Primärschlüssel besteht aus **zwei** Spalten, (*SNr*, *KID*). Eine Schülerin darf mehrere Kurse wählen, und ein Kurs hat mehrere Schüler. Eindeutig ist nur die Kombination. Ein zusammengesetzter Schlüssel wird deshalb am Ende mit `PRIMARY KEY (SNr, KID)` angegeben.
2. *SNr* verweist auf *Schueler*, *KID* verweist auf *Kurse*. Das entspricht den ↑ im Schema.

</details>

### 1.5 Eine Tabelle ansehen

Um zu prüfen, was in einer Tabelle steht, schicken wir eine **Abfrage** an das DBMS. `SELECT * FROM Lehrer` bedeutet: Gib alle Spalten (`*`) aller Datensätze der Tabelle *Lehrer* zurück.

Mit `cur.fetchall()` holen wir das Ergebnis ab. Es ist eine **Liste**, in der jeder Datensatz ein **Tupel** ist. Mit einer Schleife geben wir die Datensätze nacheinander aus.

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT *
    FROM Lehrer
""")

for datensatz in cur.fetchall():
    print(datensatz)

con.close()
````

## SQ 2 · Daten einfügen, ändern und löschen

**Ziel:** Du kannst Datensätze einfügen, ändern und löschen und weißt, wie das DBMS die Schlüssel überwacht.

### 2.1 Drei Anweisungen

| Anweisung | Wirkung | Beispiel |
|:--|:--|:--|
| `INSERT INTO` | Datensatz einfügen | `INSERT INTO Lehrer VALUES ('MU', 'Mueller')` |
| `UPDATE … SET … WHERE` | Datenfelder ändern | `UPDATE Lehrer SET Lehrer = 'Müller' WHERE Kuerzel = 'MU'` |
| `DELETE FROM … WHERE` | Datensätze löschen | `DELETE FROM Lehrer WHERE Kuerzel = 'MU'` |

- Bei `INSERT INTO` stehen die Werte in der **Reihenfolge der Spalten**. Strings stehen in **einfachen Hochkommata**.
- Nach jeder Änderung brauchen wir `commit()`.
- **Vorsicht:** Bei `UPDATE` und `DELETE` legt `WHERE` fest, **welche** Datensätze betroffen sind. Ohne `WHERE` werden **alle** Datensätze geändert bzw. gelöscht!

**Beispiel:** Wir fügen den Lehrer Mueller ein und sehen uns danach die Tabelle an.

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("INSERT INTO Lehrer VALUES ('MU', 'Mueller')")
con.commit()

cur.execute("SELECT * FROM Lehrer")
for datensatz in cur.fetchall():
    print(datensatz)

con.close()
````

**Vorhersage 2.1** Was passiert, wenn du die Zelle oben **ein zweites Mal** ausführst? Überlege zuerst, dann probiere es aus.


<details>
<summary><b>Erklärung</b></summary>

SQLite meldet `IntegrityError: UNIQUE constraint failed: Lehrer.Kuerzel`. *Kuerzel* ist der Primärschlüssel, und jeder Wert darf darin nur einmal vorkommen. Das DBMS lehnt den zweiten Datensatz mit dem Kürzel MU deshalb ab. Der Primärschlüssel schützt also vor doppelten Einträgen.

Denk daran, nach dem Fehler die Verbindung zu schließen (nächste Zelle).

</details>

````python
# Nach einer Fehlermeldung: offene Verbindung schließen
con.close()
````

### 2.2 Referentielle Integrität

Ein Fremdschlüssel verweist auf einen Datensatz in einer anderen Tabelle. Die **referentielle Integrität** verlangt, dass es diesen Datensatz auch wirklich gibt.

SQLite prüft Fremdschlüssel aber nur, wenn wir es mit `PRAGMA foreign_keys = ON` einschalten. Diese Einstellung gilt nur für die aktuelle Verbindung, wir müssen sie also nach jedem `connect()` neu setzen.

**Vorhersage 2.7** Die nächste Zelle schaltet die Prüfung ein und versucht dann, Dallmeyer (Kürzel DA) zu löschen. Was wird passieren? Sieh dir dazu die Tabelle *Kurse* an.

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("PRAGMA foreign_keys = ON")
cur.execute("DELETE FROM Lehrer WHERE Kuerzel = 'DA'")
con.commit()

con.close()
````

<details>
<summary><b>Erklärung</b></summary>

SQLite meldet `IntegrityError: FOREIGN KEY constraint failed` und löscht nichts. Der Kurs Bio1 verweist mit dem Kürzel DA noch auf Dallmeyer. Würde Dallmeyer gelöscht, zeigte dieser Fremdschlüssel ins Leere.

**Führe unbedingt die nächste Zelle aus.** Die abgebrochene Änderung hält die Datenbank sonst gesperrt, und jede weitere Änderung scheitert mit `database is locked`.

*PRAGMA* heißt auf Griechisch „Sache“ oder „Handlung“. Ein Pragma verarbeitet keine Daten, sondern stellt ein, wie das DBMS arbeiten soll. Es gehört nicht zum SQL-Standard, andere Datenbanksysteme haben dafür eigene Befehle.

</details>

## SQ 3 · Abfragen auf einer Tabelle

**Ziel:** Du kannst mit `SELECT` gezielt Spalten und Datensätze aus einer Tabelle abfragen, sortieren und Doppelte entfernen.

Ab hier verändern wir die Datenbank nicht mehr, wir fragen nur ab. `commit()` brauchen wir deshalb nicht.

### 3.1 Projektion und Selektion

Die Grundform einer Abfrage ist:

```
SELECT Spalten
FROM Tabelle
WHERE Bedingung
```

- Hinter `SELECT` stehen die Spalten, die wir sehen wollen. `*` steht für **alle** Spalten. Die Auswahl von Spalten heißt **Projektion**.
- Mit `WHERE` wählen wir die Datensätze aus, die eine Bedingung erfüllen. Die Auswahl von Zeilen heißt **Selektion**.

**Beispiel (Projektion):** Gib die Namen aller Schülerinnen und Schüler aus.

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT Name, Vorname
    FROM Schueler
""")

for datensatz in cur.fetchall():
    print(datensatz)

con.close()
````

**Beispiel (Selektion):** Gib die Daten des Schülers mit der Nummer 121 aus.

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT *
    FROM Schueler
    WHERE SNr = 121
""")

for datensatz in cur.fetchall():
    print(datensatz)

con.close()
````

### 3.2 Bedingungen in WHERE

| Operator | Bedeutung | Beispiel |
|:--|:--|:--|
| `=`, `<>` | gleich, ungleich | `Kuerzel = 'SC'` |
| `<`, `<=`, `>`, `>=` | Vergleiche | `SNr >= 121` |
| `AND`, `OR`, `NOT` | Bedingungen verknüpfen | `Kuerzel = 'SC' OR Kuerzel = 'AD'` |
| `LIKE` | Textmuster, `%` steht für beliebig viele Zeichen | `Name LIKE 'R%'` (beginnt mit R) |

Anders als in Python steht in SQL für „gleich“ ein **einfaches** `=`. Strings stehen in **einfachen Hochkommata**. Bei diesen Werten kommt es, anders als bei den Schlüsselwörtern, auf die **Groß- und Kleinschreibung** an: `'Schulte'` und `'schulte'` sind verschiedene Strings.

**Vorhersage 3.1** Welche Datensätze liefert die folgende Abfrage? Schau in die Tabelle *Schueler* oben und überlege, bevor du die Zelle ausführst.

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT *
    FROM Schueler
    WHERE SNr > 120 AND Name LIKE 'R%'
""")

for datensatz in cur.fetchall():
    print(datensatz)

con.close()
````

<details>
<summary><b>Erklärung</b></summary>

Nur `(122, 'Rödding', 'Maike')`. Beide Bedingungen müssen erfüllt sein. Riedberg beginnt zwar mit R, hat aber die SNr 120, und die ist nicht größer als 120. Dreis hat eine größere SNr, beginnt aber nicht mit R.

</details>

**Vorhersage 3.2** Liefern die beiden Abfragen in der Zelle unten ein Ergebnis? Achte auf die Schreibweise von *schulte*.

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("SELECT * FROM Lehrer WHERE Lehrer = 'schulte'")
print(cur.fetchall())

cur.execute("SELECT * FROM Lehrer WHERE Lehrer LIKE 'schulte'")
print(cur.fetchall())

con.close()
````

Die zweite Abfrage findet Schulte. `LIKE` ignoriert in SQLite die Groß- und Kleinschreibung, allerdings nur bei Buchstaben ohne Umlaute: `'%ö%'` findet Rödding, `'%Ö%'` nicht.

**Merke:** Liefert eine Abfrage nichts, obwohl die Daten da sein müssten, prüfe zuerst die Schreibweise der Werte.

</details>

### 3.3 Sortieren, Doppelte entfernen, Anzahl begrenzen

- `ORDER BY Spalte` **sortiert** das Ergebnis aufsteigend, `ORDER BY Spalte DESC` absteigend.
- `SELECT DISTINCT` entfernt **doppelte Zeilen** aus dem Ergebnis.
- `LIMIT n` begrenzt das Ergebnis auf die ersten n Datensätze. `LIMIT` steht immer am Ende, also nach `ORDER BY`.

**Beispiel:** Gib die Kürzel aller Lehrkräfte aus, die einen LK unterrichten. Führe beide Zellen aus und vergleiche.

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT Kuerzel
    FROM Kurse
""")

for datensatz in cur.fetchall():
    print(datensatz)

con.close()
````

Schulte unterrichtet zwei LKs, deshalb erscheint SC ohne `DISTINCT` doppelt.

## SQ 4 · Abfragen über mehrere Tabellen

**Ziel:** Du kannst Informationen aus mehreren Tabellen über Fremd- und Primärschlüssel wieder zusammenführen.

### 4.1 Das Problem

Durch die Normalisierung sind die Informationen auf mehrere Tabellen verteilt. Ein Beispiel: Die Frage „Welche Lehrkraft unterrichtet M1?“ lässt sich mit einer Tabelle allein nicht beantworten. In *Kurse* steht nur das Kürzel SC, der Name *Schulte* steht in *Lehrer*. Wir müssen beide Tabellen **verbinden**.

Dazu schreiben wir beide Tabellen hinter `FROM`.

**Vorhersage 4.1** *Kurse* hat 5 Datensätze, *Lehrer* hat 4. Wie viele Zeilen liefert die folgende Abfrage? Überlege, dann führe sie aus.

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT KID, Lehrer
    FROM Kurse, Lehrer
""")

for datensatz in cur.fetchall():
    print(datensatz)

con.close()
````

<details>
<summary><b>Erklärung</b></summary>

Man erhält **20 Zeilen**. `FROM Kurse, Lehrer` bildet **alle Kombinationen** aus einem Kurs und einer Lehrkraft: 5 · 4 = 20. Man nennt das das **Kreuzprodukt**. Die meisten Kombinationen sind sinnlos, z. B. `('M1', 'Adam')`, da M1 von Schulte unterrichtet wird.

</details>

### 4.2 Die Verknüpfungsbedingung

Von den 20 Kombinationen wollen wir nur die behalten, die **zusammengehören**. Das ist genau dann der Fall, wenn der **Fremdschlüssel** in *Kurse* mit dem **Primärschlüssel** in *Lehrer* übereinstimmt. Diese Verknüpfungsbedingung schreiben wir in `WHERE`.

Kommt ein Spaltenname in beiden Tabellen vor, schreiben wir den **Tabellennamen** mit Punkt davor, z. B. `Kurse.Kuerzel`.

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT KID, Lehrer
    FROM Kurse, Lehrer
    WHERE Kurse.Kuerzel = Lehrer.Kuerzel
""")

for datensatz in cur.fetchall():
    print(datensatz)

con.close()
````

Jetzt bleiben nur die 5 richtigen Zeilen übrig. Eine Verknüpfung von Tabellen nennt man **Verbund** (*Join*).

**Vorhersage 4.2** Wir wollen zusätzlich das Kürzel sehen und schreiben `SELECT KID, Kuerzel, Lehrer`. Was passiert?

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT KID, Kuerzel, Lehrer
    FROM Kurse, Lehrer
    WHERE Kurse.Kuerzel = Lehrer.Kuerzel
""")

for datensatz in cur.fetchall():
    print(datensatz)

con.close()
````

</details>

### 4.3 Weitere Bedingungen

Zur Verknüpfungsbedingung können wir mit `AND` weitere Bedingungen hinzufügen.

**Beispiel:** Gesucht sind alle Kurse von Schulte oder Adam, diesmal über den **Namen** der Lehrkraft.

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT KID, Lehrer
    FROM Kurse, Lehrer
    WHERE Kurse.Kuerzel = Lehrer.Kuerzel
      AND (Lehrer = 'Schulte' OR Lehrer = 'Adam')
""")

for datensatz in cur.fetchall():
    print(datensatz)

con.close()
````

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT KID, Lehrer
    FROM Kurse, Lehrer
    WHERE Kurse.Kuerzel = Lehrer.Kuerzel
      AND Lehrer = 'Schulte' OR Lehrer = 'Adam'
""")

for datensatz in cur.fetchall():
    print(datensatz)

con.close()
````


</details>

### 4.4 Mehr als zwei Tabellen

Für jede Verknüpfung zweier Tabellen brauchen wir eine Bedingung, bei vier Tabellen also drei. Du kannst dir das wie einen **Weg durch die Tabellen** vorstellen, der den Fremdschlüsseln folgt:

```
Schueler  ──SNr──  LKWahl  ──KID──  Kurse  ──Kuerzel──  Lehrer
```

**Beispiel:** Gib für jede LK-Wahl den Namen, die KursID und die Lehrkraft aus, sortiert nach Nachnamen.

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT Name, Vorname, LKWahl.KID, Lehrer
    FROM Schueler, LKWahl, Kurse, Lehrer
    WHERE Schueler.SNr = LKWahl.SNr
      AND LKWahl.KID = Kurse.KID
      AND Kurse.Kuerzel = Lehrer.Kuerzel
    ORDER BY Name
""")

for datensatz in cur.fetchall():
    print(datensatz)

con.close()
````

Für eine übersichtlichere Ausgabe **entpacken** wir jedes Tupel in einzelne Variablen und benutzen einen f-String. `{name:10}` reserviert 10 Zeichen für den Namen, so entstehen gerade Spalten.

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT Name, Vorname, LKWahl.KID, Lehrer
    FROM Schueler, LKWahl, Kurse, Lehrer
    WHERE Schueler.SNr = LKWahl.SNr
      AND LKWahl.KID = Kurse.KID
      AND Kurse.Kuerzel = Lehrer.Kuerzel
    ORDER BY Name
""")

for name, vorname, kid, lehrer in cur.fetchall():          # unpacking
    print(f'{vorname:10} {name:10} {kid:4} {lehrer:10}')

con.close()
````

## SQ 5 · Verbund mit JOIN

**Ziel:** Du kannst Abfragen über mehrere Tabellen mit `JOIN … ON` formulieren.

### 5.1 JOIN … ON

In SQ 4 haben wir die Tabellen hinter `FROM` aufgezählt und die Verknüpfungsbedingungen in `WHERE` geschrieben. Dieselben Abfragen lassen sich auch mit `JOIN … ON` formulieren:

```
SELECT Spalten
FROM Tabelle1
JOIN Tabelle2 ON Verknüpfungsbedingung
WHERE weitere Bedingungen
```

- Hinter `JOIN` steht die Tabelle, die dazukommt.
- Hinter `ON` steht, **wie** sie mit den bisherigen Tabellen verknüpft ist, also Fremdschlüssel = Primärschlüssel.
- In `WHERE` stehen nur noch die Bedingungen, mit denen wir **Datensätze auswählen**.

Das Ergebnis ist dasselbe. Der Vorteil: Verknüpfung und Auswahl sind sauber getrennt, und bei jeder Tabelle sieht man sofort, wie sie angebunden ist.

**Beispiel:** Alle Kurse von Schulte oder Adam (vgl. 4.3)

| mit WHERE | mit JOIN |
|:--|:--|
| `SELECT KID, Lehrer` | `SELECT KID, Lehrer` |
| `FROM Kurse, Lehrer` | `FROM Kurse` |
| `WHERE Kurse.Kuerzel = Lehrer.Kuerzel` | `JOIN Lehrer ON Kurse.Kuerzel = Lehrer.Kuerzel` |
| `  AND (Lehrer = 'Schulte' OR Lehrer = 'Adam')` | `WHERE Lehrer = 'Schulte' OR Lehrer = 'Adam'` |

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT KID, Lehrer
    FROM Kurse
    JOIN Lehrer ON Kurse.Kuerzel = Lehrer.Kuerzel
    WHERE Lehrer = 'Schulte' OR Lehrer = 'Adam'
""")

for datensatz in cur.fetchall():
    print(datensatz)

con.close()
````

Die Klammern um die `OR`-Bedingung sind hier nicht mehr nötig, weil in `WHERE` nur noch diese eine Bedingung steht.

**Beispiel:** Vier Tabellen (vgl. 4.4). Für jede weitere Tabelle kommt eine Zeile mit `JOIN … ON` dazu. Man liest die Abfrage von oben nach unten wie den Weg durch die Tabellen.

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT Name, Vorname, LKWahl.KID, Lehrer
    FROM Schueler
    JOIN LKWahl ON Schueler.SNr = LKWahl.SNr
    JOIN Kurse ON LKWahl.KID = Kurse.KID
    JOIN Lehrer ON Kurse.Kuerzel = Lehrer.Kuerzel
    ORDER BY Name
""")

for datensatz in cur.fetchall():
    print(datensatz)

con.close()
````

## SQ 6 · Zählen und Summieren

**Ziel:** Du kannst mit Aggregatfunktionen Datensätze zählen und Werte addieren und das Ergebnis in Python weiterverwenden.

### 6.1 Aggregatfunktionen

**Aggregatfunktionen** fassen viele Werte zu **einem einzigen** Wert zusammen.

| Funktion | Wirkung |
|:--|:--|
| `SUM(Spalte)` | addiert alle Werte einer Spalte |
| `COUNT(*)` | zählt die Datensätze im Ergebnis |
| `COUNT(DISTINCT Spalte)` | zählt die verschiedenen Werte einer Spalte |

Mit `WHERE` wählen wir vorher aus, welche Zeilen addiert oder gezählt werden.

**Beispiel:** Wie viele Wochenstunden LK-Unterricht gibt es insgesamt, und wie viele davon hat Schulte?

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT SUM(Stunden)
    FROM Kurse
""")

for datensatz in cur.fetchall():
    print(datensatz)

con.close()
````

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT SUM(Stunden)
    FROM Kurse
    WHERE Kuerzel = 'SC'
""")

for datensatz in cur.fetchall():
    print(datensatz)

con.close()
````

**Beispiel:** Wie viele Schülerinnen und Schüler haben den Kurs Ph1 gewählt?

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT COUNT(*)
    FROM LKWahl
    WHERE KID = 'Ph1'
""")

for datensatz in cur.fetchall():
    print(datensatz)

con.close()
````

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("SELECT COUNT(*) FROM LKWahl")
print(cur.fetchall())

cur.execute("SELECT COUNT(DISTINCT SNr) FROM LKWahl")
print(cur.fetchall())

con.close()
````

`[(6,)]` und `[(3,)]`. `COUNT(*)` zählt alle **LK-Wahlen**. `COUNT(DISTINCT SNr)` zählt jede Schülernummer nur einmal, also die Anzahl der **Schülerinnen und Schüler**, die gewählt haben. Jeder hat zwei LKs, deshalb ist die erste Zahl doppelt so groß.

</details>

### 6.2 Das Ergebnis weiterverwenden

Das Ergebnis einer Aggregatfunktion ist eine Liste mit einem Tupel mit einem einzigen Wert, z. B. `[(6,)]`. Mit `cur.fetchall()[0][0]` erhalten wir die Zahl selbst: `[0]` wählt das erste Tupel, das zweite `[0]` darin den ersten Wert. So können wir die Zahl in einen Satz einbauen oder damit rechnen.

````python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT COUNT(*)
    FROM Schueler
""")

anzahl = cur.fetchall()[0][0]
print(f'Es sind {anzahl} Schülerinnen und Schüler gespeichert.')

con.close()
````
