### Binäre Suche

[Video](https://youtu.be/HCfHYOmwcRs) - [AufgabenVideo](https://youtu.be/7WRzDeWWx54)

In einer **sortierten** Liste von Zahlen können wir mit der binären Suche schnell überprüfen, ob ein Element vorhanden ist.

![](binaereSuche1.png)

```python
def binaereSuche(a, x):
    '''
    a: sortierte Liste mit Zahlen
    x: gesuchte Zahl
    returns: Index von x in a, falls x in a
             -1              , falls x nicht in a

    >>> a = [2,4,7,8,12,15,17,18,21,22,25]
    >>> binaereSuche(a,12)
    4
    '''
    L = 0
    R = len(a)-1
    while L <= R:
        mid = (L+R)//2        # besser: L + (R-L)//2
        if a[mid] == x:
            return mid
        if a[mid] < x:
            L = mid + 1
        else:
            R = mid - 1
    return -1
```

```python
a = [2,4,7,8,12,15,17,18,21,22,25]
binaereSuche(a,12)
```

Bei der Berechnung von mid mit der Formel $(L+R)//2$ kann es in manchen Programmiersprachen zu einem overflow kommen.
Daher wird die Berechnung mit $L + (R-L)//2$ bevorzugt.

Mit Ausgaben in der Schleife können wir unsere Tabellen überprüfen. Zeile 0 enthält die Werte vor der Schleife.

```python
def binaereSuche_tabelle(a, x):
    L = 0
    R = len(a)-1
    schleife = 0
    print('Schleife', 'L', 'R', 'mid', 'return', sep='\t')
    print(schleife, L, R, sep='\t')
    while L <= R:
        schleife += 1
        mid = L + (R-L)//2
        if a[mid] == x:
            print(schleife, L, R, mid, mid, sep='\t')
            return mid
        if a[mid] < x:
            L = mid + 1
        else:
            R = mid - 1
        print(schleife, L, R, mid, sep='\t')
    print('', '', '', '', -1, sep='\t')
    return -1

a = [4, 7, 10, 11, 13, 17, 21, 24, 27, 29, 30, 32, 40, 45]
binaereSuche_tabelle(a, 29)
print()
binaereSuche_tabelle(a, 42);
```

#### Wie schnell ist die binäre Suche?

Bei jedem Schleifendurchlauf wird der Suchbereich von `L` bis `R` halbiert. Nach $k$ Durchläufen sind von $n$ Elementen
höchstens noch $n / 2^k$ übrig. Spätestens wenn kein Element mehr übrig ist, endet die Schleife.

Mit $2^{10} = 1024 \approx 10^3$ können wir abschätzen:

| Anzahl der Elemente | höchstens Schleifendurchläufe |
|---|---|
| 1000 $\approx 2^{10}$ | 10 |
| 1 Million $\approx 2^{20}$ | 20 |
| 1 Milliarde $\approx 2^{30}$ | 30 |

Die **lineare Suche** schaut sich ein Element nach dem anderen an. Bei einer Milliarde Elementen braucht sie im
schlechtesten Fall eine Milliarde Schritte. Auch `x in a` und `a.index(x)` suchen linear, denn Python weiß nicht,
dass die Liste sortiert ist.

```python
def lineareSuche_schritte(a, x):
    schritte = 0
    for i in range(len(a)):
        schritte += 1
        if a[i] == x:
            return schritte
    return schritte

def binaereSuche_schritte(a, x):
    L = 0
    R = len(a)-1
    schritte = 0
    while L <= R:
        schritte += 1
        mid = L + (R-L)//2
        if a[mid] == x:
            return schritte
        if a[mid] < x:
            L = mid + 1
        else:
            R = mid - 1
    return schritte

a = list(range(0, 2000000, 2))      # 1 Million gerade Zahlen
print('x', 'linear', 'binär', sep='\t')
for x in [10, 777776, 1999998, 5]:
    print(x, lineareSuche_schritte(a, x), binaereSuche_schritte(a, x), sep='\t')
```

#### Nur für sortierte Listen

Die binäre Suche verlässt sich darauf, dass links von `mid` nur kleinere und rechts nur größere Zahlen stehen.
Bei einer unsortierten Liste kann sie deshalb ein vorhandenes Element übersehen.

```python
a = [5, 1, 4, 2, 3]
print(binaereSuche(a, 1))      # 1 steht am Index 1, wird aber nicht gefunden
print(binaereSuche(a, 4))
```

#### Weitere Anwendungen der binären Suche

Die binäre Suche kann sehr schnell in einer Liste den Übergang von True nach False finden, wenn es einen solchen Übergang nur einmal gibt.

```python
def upperBound(a):
    '''
    Die Liste a enthält k1-mal den Wert True, gefolgt von k2-mal False.
    returns: k1-1 = Index des letzten Elements mit True
             -1, falls die Liste kein True enthält

    >>> a = [True, True, True, True, True, False, False, False, False]
    >>> upperBound(a)
    4
    '''
    L = 0
    R = len(a)-1
    ans = -1
    while L <= R:
        mid = L + (R-L)//2
        if a[mid]:
            ans = mid
            L = mid + 1
        else:
            R = mid - 1
    return ans
```

##### Anwendung: Wie viele Zahlen sind kleiner als x?

In einer sortierten Liste ordnen wir jedem Element den Wert True zu, wenn es kleiner als `x` ist.
Vorne stehen dann alle True, dahinter alle False. `ans` ist der Index des letzten True, also gibt es `ans + 1` Zahlen,
die kleiner als `x` sind. Das funktioniert auch, wenn Zahlen mehrfach vorkommen.

```python
def anzahl_kleiner(a, x):
    '''
    a: sortierte Liste mit Zahlen
    returns: Anzahl der Elemente von a, die kleiner als x sind

    >>> anzahl_kleiner([2, 4, 4, 4, 7, 9], 5)
    4
    >>> anzahl_kleiner([2, 4, 4, 4, 7, 9], 4)
    1
    >>> anzahl_kleiner([2, 4, 4, 4, 7, 9], 1)
    0
    '''
    L = 0
    R = len(a)-1
    ans = -1
    while L <= R:
        mid = L + (R-L)//2
        if a[mid] < x:
            ans = mid
            L = mid + 1
        else:
            R = mid - 1
    return ans + 1

a = [2, 4, 4, 4, 7, 9]
print(anzahl_kleiner(a, 5))
print(anzahl_kleiner(a, 5) - anzahl_kleiner(a, 4))    # so oft kommt 4 vor
```

##### Anwendung des UpperBound: RotatedList

    a = [15, 17, 22, 23, 33, 56, 72, 87, 4, 5, 6, 9, 13]

<img src=rotated1.png width='450'>

```python
def rotatedList(a):
    '''
    Eine sortierte Liste mit verschiedenen Zahlen wurde eventuell an einem unbekannten Punkt rotiert, d.h. die Anfangsliste bis
    zu diesem Punkt wurde ans Ende der Liste verschoben.
    returns: Index mit dem Beginn der sortierten Liste

    Idee: wir ordnen einem Element am Index i den Wert True zu, wenn gilt: a[0] <= a[i]

    >>> rotatedList([15,17,18,21,22,25,2,4,7,8,12])
    6
    >>> rotatedList([3,4,1,2])
    2
    >>> rotatedList([1,2,3,4])
    0
    '''
    L = 0
    R = len(a)-1
    ans = -1
    while L <= R:
        mid = L + (R-L)//2
        if a[0] <= a[mid] :
            ans = mid
            L = mid + 1
        else:
            R = mid - 1

    if ans == len(a)-1:    # es wurde nicht rotiert
        return 0
    return ans+1           # ans zeigt auf letztes Element x mit a[0] <= x

rotatedList([15,17,18,21,22,25,2,4,7,8,12])
```

##### Anwendung des UpperBound: FindMaximum

    a = [15, 17, 22, 23, 33, 56, 72, 33, 21, 8, 3]

![](binaereSuche3.png)

```python
def findMaximum(a):
    '''
    In einer Liste mit mindestens 3 Elementen, die steigt und dann wieder fällt
    sollen wir den Index mit dem Maximum finden.

    Idee: wir ordnen einem Element am Index i den Wert True zu, wenn gilt: a[i-1] < a[i]

    >>> a = [4,8,3]
    >>> findMaximum(a)
    1

    >>> a = [2,3,4,6,9,12,16,20,11,8,6,4,1]
    >>> findMaximum(a)
    7

    >>> findMaximum([1,9,8,7,6])
    1

    '''
    L = 0
    R = len(a)-1
    ans = -1
    while L <= R:
        mid = L + (R-L)//2
        if mid == 0 or a[mid] > a[mid-1]:
            ans = mid
            L = mid + 1        # rechts ein besseres True suchen
        else:
            R = mid - 1
    return ans
```

Achtung: Für `mid = 0` gibt es kein linkes Nachbarelement. `a[mid-1]` wäre dann `a[-1]`, also das **letzte** Element
der Liste. Ohne die Bedingung `mid == 0` liefert die Funktion z.B. für `[1, 9, 8, 7, 6]` den Wert -1 statt 1.
Das Element am Index 0 bekommt deshalb immer True, denn links vom Maximum steigt die Liste.

```python
a = [15, 17, 22, 23, 33, 56, 72, 33, 21, 8, 3]
print(findMaximum(a))
print(findMaximum([1, 9, 8, 7, 6]))
```

#### Zusammenfassung

| | |
|---|---|
| Voraussetzung | Die Liste ist sortiert. |
| Idee | `mid` in der Mitte von `L` und `R` ansehen und mit dem Suchbereich links oder rechts weitermachen |
| Ende | gefunden, oder `L > R`: Der Suchbereich ist leer, Rückgabe -1. |
| Schnelligkeit | Bei $n$ Elementen höchstens etwa $\log_2 n$ Durchläufe, bei 1 Million Elementen 20 |
| Übergang von True nach False | `ans` merkt sich das letzte True, danach rechts weitersuchen (upperBound) |
| Anwendungen | Anzahl der Elemente kleiner als x, rotierte Liste, Maximum einer Liste, die erst steigt und dann fällt |
