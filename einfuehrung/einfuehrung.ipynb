### Algorithmus und Programm

Wir beschäftigen uns mit Algorithmen und deren Programmierung in
Python. Ein **Algorithmus** ist eine Schritt-für-Schritt-Anleitung zur Lösung eines Problems oder zur Durchführung einer Aufgabe.

Ein in einer Computersprache formulierter Algorithmus heißt **Programm**.


**Pseudocode** ist eine vereinfachte Beschreibung eines Algorithmus.
Er verwendet eine leicht verständliche Sprache, die an Programmiersprachen erinnert,
aber nicht an die genaue Syntax einer bestimmten Programmiersprache gebunden ist.


Beispiel: Der Collatz-Algorithmus im Pseudocode

 
    lies x ein
    setze z auf 0
    solange x nicht gleich 1 tue:
        wenn x gerade, dann halbiere x
        sonst verdreifache x und erhöhe um 1
        erhöhe z um 1    
    gib z aus

Um zu prüfen, was ein Algorithmus macht, ist es manchmal hilfreich, ein **Ablaufprotokoll** zu erstellen. Dabei werden die Werte der (wichtigsten) beteiligten Variablen schrittweise mitverfolgt. Beim Collatz-Algorithmus reicht es, den Wert der eingegebenen Zahl zu verfolgen und anschließend die Anzahl der Durchgänge zu zählen:

    Eingabe: 17
    52 26 13 40 20 10 5 16 8 4 2 1
    Ausgabe: 12

Der Algorithmus wurde 1937 von Lothar Collatz formuliert. Es ist ein bis heute
ungelöstes mathematisches Problem, ob dieser Algorithmus fur jede Eingabe zu 
einem Ende kommt.

```python
# Collatz
x = int(input('Bitte eine positive ganze Zahl eingeben: '))
z = 0
while x != 1:
    if x % 2 == 0:
        x = x//2
    else:
        x = 3*x + 1
    z += 1
print(z)
```

### Entwicklungsumgebung Thonny

[Thonny](https://thonny.org/) ist eine Entwicklungsumgebung (IDE – *Integrated Development Environment*) für die Programmiersprache Python. Sie unterstützt  beim Schreiben, Ausführen und Testen von Programmen.

Thonny bietet verschiedene Werkzeuge, die das Programmieren erleichtern:

* Code-Editor: Hier wird der Programmcode geschrieben und bearbeitet.
* Ausführen von Programmen: Programme können direkt aus Thonny gestartet werden.
* Fehlermeldungen: Thonny zeigt an, wenn ein Programm syntaktische Fehler enthält oder während der Ausführung Probleme auftreten.
* Debugger: Mit dem Debugger kann ein Programm Schritt für Schritt ausgeführt werden. Dabei lässt sich beobachten, welche Anweisungen gerade ausgeführt werden und welche Werte die Variablen besitzen.
* Python-Konsole (Kommandozeile): Hier können wir Eingaben für unsere Programme machen und sehen Ausgaben.
