# Rekursive Zeichnungen mit der Turtle

Mit dem Modul `turtle` steuern wir eine Schildkröte (engl. *turtle*), die mit einem Stift über die Zeichenfläche läuft.
Viele schöne Bilder bestehen aus kleineren Kopien von sich selbst. Solche Bilder lassen sich mit Rekursion besonders
einfach zeichnen: Die Funktion zeichnet eine einfache **Grundfigur** und ruft sich an den passenden Stellen mit einer
kleineren Länge selbst auf.

**Hinweis:** Die Turtle zeichnet in einem eigenen Fenster. Das klappt in Thonny und in einem lokal installierten Jupyter,
aber nicht in Jupyter im Browser (JupyterLite). Wegen `t.mainloop()` am Ende bleibt das Fenster offen, bis wir es schließen.
Führt man die Zelle danach noch einmal aus, erscheint manchmal die Fehlermeldung `Terminator`. Dann die Zelle einfach ein
weiteres Mal ausführen.

#### Die wichtigsten Turtle-Befehle

Der Nullpunkt (0, 0) liegt in der Mitte des Fensters. Am Anfang steht die Turtle dort und schaut nach rechts.
Richtungen werden in Grad angegeben: 0 = rechts, 90 = oben, 180 = links, 270 = unten.

| Befehl | Wirkung |
|---|---|
| `t.forward(100)` oder `t.fd(100)` | 100 Einheiten vorwärts gehen |
| `t.back(100)` | 100 Einheiten rückwärts gehen, ohne sich umzudrehen |
| `t.left(90)`, `t.right(90)` | um 90 Grad nach links bzw. rechts drehen |
| `t.penup()` oder `t.up()` | Stift hochnehmen: Bewegungen zeichnen nichts |
| `t.pendown()` oder `t.down()` | Stift absetzen |
| `t.goto(x, y)` oder `t.setpos(x, y)` | zum Punkt (x, y) gehen |
| `t.position()` | aktuelle Position als Tupel |
| `t.heading()` | aktuelle Richtung in Grad |
| `t.setheading(90)` | Richtung festlegen, hier nach oben |
| `t.dot(10, 'red')` | roten Punkt mit Durchmesser 10 zeichnen |
| `t.pensize(3)`, `t.color('green')` | Strichstärke und Farbe festlegen |
| `t.speed(10)` | Geschwindigkeit: 1 = langsam, 10 = schnell, 0 = am schnellsten |
| `t.hideturtle()` | das Symbol der Turtle ausblenden |
| `t.mainloop()` | Fenster offen halten, bis es geschlossen wird |

```python
import turtle as t

t.pensize(3)
for i in range(4):          # Quadrat
    t.forward(200)
    t.left(90)
t.penup()
t.goto(100, 100)            # in die Mitte des Quadrats
t.pendown()
t.dot(20, 'red')
t.mainloop()
```

#### Rekursive Linien

<img src="linien.png" width="700">

Wir zeichnen einen Strich, lassen eine Lücke und machen dann dasselbe noch einmal, nur etwas kürzer.
Der Parameter `bremse` gibt an, wie viele Striche noch gezeichnet werden. Bei jedem rekursiven Aufruf wird er um 1 kleiner.
Ist er 0, ist der **Basisfall** erreicht: Die Funktion kehrt zurück, ohne sich noch einmal aufzurufen.

```python
import turtle as t

def zeichne(laenge, bremse):
    if bremse == 0:
        return
    t.pendown()
    t.forward(laenge)          # Strich
    t.penup()
    t.forward(laenge/2)        # Lücke
    zeichne(laenge*0.75, bremse-1)

t.hideturtle()
t.pensize(3)
t.penup()
t.goto(-340, 0)
zeichne(110, 20)
t.mainloop()
```

#### Quadrat

<img src="quadrat.png" width="500">

Die Funktion zeichnet einen Strich und dreht sich um 90 Grad nach links. Dann ruft sie sich mit einer etwas kürzeren
Länge selbst auf. Hier dient die Länge selbst als Bremse: Ist sie kleiner als 20, ist der Basisfall erreicht.

```python
import turtle as t

def zeichne(laenge):
    if laenge < 20:
        return
    t.fd(laenge)
    t.left(90)
    zeichne(laenge*0.98)

t.up()
t.goto(-350, -350)
t.down()
t.pensize(1)
t.hideturtle()
t.speed(10)
zeichne(700)
t.mainloop()
```

Bei den beiden ersten Beispielen ruft sich die Funktion nur **einmal** selbst auf, und zwar ganz am Ende.
Solche Bilder kann man genauso gut mit einer Schleife zeichnen:

```python
import turtle as t

t.up()
t.goto(-350, -350)
t.down()
t.hideturtle()
t.speed(10)
laenge = 700
while laenge >= 20:
    t.fd(laenge)
    t.left(90)
    laenge = laenge*0.98
t.mainloop()
```

Richtig nützlich wird die Rekursion, wenn sich die Funktion **mehrmals** selbst aufruft. Dann entstehen Verzweigungen.

#### Verzweigung

<img src="verzweigung.png" width="700">

Die Grundfigur ist ein **V**: ein Ast nach rechts oben und einer nach links oben. Am Ende jedes Asts soll dasselbe Bild
noch einmal entstehen, nur halb so groß. Die Funktion `zeichne` zeichnet deshalb **nur das V**. Am Ende jedes Asts
ruft sie sich selbst mit der halben Länge auf.

Dabei sind zwei Dinge wichtig:

- Beim rekursiven Aufruf muss die Turtle gegenüber dem kleinen V **genauso stehen** wie beim ersten Aufruf gegenüber dem
  großen V: am unteren Punkt des V und mit Blick in die Richtung, in die das V wachsen soll.
- Nach dem rechten Ast und dem rekursiven Aufruf steht die Turtle irgendwo oben im rechten Teilbild. Damit sie den linken
  Ast an der richtigen Stelle beginnt, merkt sie sich am Anfang **Position und Richtung** und kehrt vor dem linken Ast
  dorthin zurück.

Der Parameter `bremse` gibt die Anzahl der Rekursionsstufen an. `zeichne(200, 6)` zeichnet 6 Stufen.

```python
import turtle as t

def zeichne(laenge, bremse):
    if bremse == 0:
        return
    # Ausgangsposition merken
    pos = t.position()
    winkel = t.heading()
    # 1. Verzweigung
    t.right(45)
    t.forward(laenge)
    zeichne(laenge*0.5, bremse-1)
    # zurück auf Ausgangsposition
    t.penup()
    t.setpos(pos)
    t.setheading(winkel)
    t.pendown()
    # 2. Verzweigung
    t.left(45)
    t.forward(laenge)
    zeichne(laenge*0.5, bremse-1)

t.hideturtle()
t.left(90)
t.pensize(3)
t.speed(10)
zeichne(200, 6)
t.mainloop()
```

Mit `t.speed(1)` sieht man die Reihenfolge: Erst wird der rechte Ast mit allem, was an ihm hängt, fertig gezeichnet.
Erst danach beginnt der linke. Jeder Aufruf ist erst fertig, wenn alle Aufrufe, die er selbst gestartet hat, fertig sind.

Statt sich Position und Richtung zu merken, kann die Turtle auch auf demselben Weg zurücklaufen. Dazu muss sich jeder
Aufruf an eine Regel halten: **Am Ende steht die Turtle wieder an derselben Stelle und schaut in dieselbe Richtung wie am
Anfang.** Dann kann sich jeder Aufruf darauf verlassen, dass die Turtle nach einem rekursiven Aufruf wieder am Ende des
Asts steht.

```python
import turtle as t

def zeichne(laenge, bremse):
    if bremse == 0:
        return
    t.right(45)                   # rechter Ast
    t.forward(laenge)
    zeichne(laenge*0.5, bremse-1)
    t.back(laenge)                # zurück zum unteren Punkt des V
    t.left(90)                    # linker Ast
    t.forward(laenge)
    zeichne(laenge*0.5, bremse-1)
    t.back(laenge)
    t.right(45)                   # wieder in die Ausgangsrichtung drehen

t.hideturtle()
t.left(90)
t.pensize(3)
t.speed(10)
zeichne(200, 6)
t.mainloop()
```

#### Rekursiver Baum mit Äpfeln

<img src="turtle_02.png" width="700">

Der Baum entsteht genauso wie die Verzweigung. Er wirkt natürlicher, weil die beiden Äste verschieden sind: Der rechte
Ast ist um 60 Grad gedreht, der linke um 40 Grad. Der linke Ast ist etwas kürzer, und die beiden Teilbäume werden
verschieden stark verkleinert (0.7 und 0.6). Im Basisfall wird diesmal etwas gezeichnet: ein roter Apfel.

```python
import turtle as t

def zeichne(laenge, bremse):
    if bremse == 0:
        t.dot(10, 'red')
        return
    # Ausgangsposition merken
    pos = t.position()
    winkel = t.heading()
    # rechter Ast
    t.right(60)
    t.forward(laenge)
    zeichne(laenge*0.7, bremse-1)
    # zurück auf Ausgangsposition
    t.penup()
    t.setpos(pos)
    t.setheading(winkel)
    t.pendown()
    # linker Ast
    t.left(40)
    t.forward(laenge*0.8)
    zeichne(laenge*0.6, bremse-1)

t.speed(10)
t.hideturtle()
t.penup()
t.setpos(-50, -300)       # Startposition
t.setheading(90)
t.pendown()
t.forward(200)            # Stamm zeichnen
zeichne(200, 6)
t.mainloop()
```

#### Farben für die Rekursionsstufen

<img src="baum_farben.png" width="600">

Jede Rekursionsstufe kann eine eigene Farbe bekommen. Dazu nehmen wir `bremse` als Index in eine Liste mit Farben.
Mit `% len(farben)` bleibt der Index immer gültig, auch wenn es mehr Stufen als Farben gibt. Auch die Strichstärke
hängt von `bremse` ab, so werden die Äste nach außen dünner.

Achtung: Der rekursive Aufruf für den rechten Ast ändert Farbe und Strichstärke. Vor dem linken Ast müssen wir sie
deshalb noch einmal setzen.

```python
import turtle as t

farben = ['blue', 'green', 'orange', 'purple']

def zeichne(laenge, bremse):
    if bremse == 0:
        t.dot(10, 'red')
        return
    pos = t.position()
    winkel = t.heading()
    t.color(farben[bremse % len(farben)])
    t.pensize(bremse)
    t.right(60)
    t.forward(laenge)
    zeichne(laenge*0.7, bremse-1)
    t.penup()
    t.setpos(pos)
    t.setheading(winkel)
    t.pendown()
    t.color(farben[bremse % len(farben)])    # Farbe und Stärke neu setzen
    t.pensize(bremse)
    t.left(40)
    t.forward(laenge*0.8)
    zeichne(laenge*0.6, bremse-1)

t.speed(10)
t.hideturtle()
t.penup()
t.setpos(-50, -300)
t.setheading(90)
t.pendown()
t.pensize(7)
t.color('brown')
t.forward(200)
zeichne(200, 6)
t.mainloop()
```

#### So entwirft man ein rekursives Bild

1. **Grundfigur** festlegen, z.B. das V. Nur sie wird in der Funktion gezeichnet.
2. **Startlage** festlegen: An welchem Punkt beginnt die Grundfigur, und in welche Richtung schaut die Turtle dabei?
3. Festlegen, **wo die kleineren Kopien** der Figur hin sollen.
4. Vor jedem rekursiven Aufruf die Turtle so hinstellen, dass sie gegenüber der kleinen Kopie **in derselben Lage** ist
   wie beim ersten Aufruf gegenüber der Grundfigur.
5. Nach jedem rekursiven Aufruf wieder in eine bekannte Lage kommen: Position und Richtung merken und zurücksetzen oder
   auf demselben Weg zurücklaufen.
6. Die **Bremse** einbauen: `bremse` bei jedem rekursiven Aufruf um 1 verkleinern und bei 0 aufhören.

#### Sierpinski-Dreieck

<img src="turtle_04.png" width="700">

Bisher wurde außerhalb des Basisfalls gezeichnet. Beim Sierpinski-Dreieck ist es umgekehrt: Gezeichnet wird **nur im
Basisfall**, und zwar ein ganzes Dreieck. Die Rekursion sorgt nur dafür, dass die Turtle an die richtigen Stellen kommt.

`zeichne(laenge, bremse)` beginnt immer an der linken unteren Ecke des Dreiecks und schaut nach rechts. Ist `bremse`
noch nicht 0, wird das Dreieck nicht gezeichnet, sondern in drei halb so große Dreiecke aufgeteilt: links unten, rechts
unten und oben. Die Turtle läuft mit hochgenommenem Stift zur linken unteren Ecke jedes kleinen Dreiecks und ruft dort
`zeichne` auf. Am Ende läuft sie zurück zur Ausgangsecke. Auch hier hinterlässt also jeder Aufruf die Turtle so, wie er
sie vorgefunden hat. Bei Stufe $n$ werden $3^n$ Dreiecke gezeichnet.

Mit der linken Maustaste wird die Stufe um 1 erhöht, mit der rechten um 1 verringert. Die Funktion `sierpinski` löscht
dann das Fenster mit `t.reset()` und zeichnet neu. `t.tracer(0, 0)` schaltet das schrittweise Zeichnen ab, `t.update()`
zeigt dann das fertige Bild auf einmal.

```python
import turtle as t

def leftmouse(x, y):
    global stufe
    stufe += 1
    sierpinski()

def rightmouse(x, y):
    global stufe
    if stufe > 0:
        stufe -= 1
    sierpinski()

def zeichne(laenge, bremse):
    if bremse == 0:
        t.pendown()
        for i in range(3):    # Dreieck zeichnen
            t.forward(laenge)
            t.left(120)
        t.penup()
        return
    halb = laenge/2
    zeichne(halb, bremse-1)   # 1. Dreieck links unten
    t.forward(halb)
    zeichne(halb, bremse-1)   # 2. Dreieck rechts unten
    t.left(120)
    t.forward(halb)
    t.right(120)
    zeichne(halb, bremse-1)   # 3. Dreieck oben
    t.right(120)              # zurück in die Ausgangsposition
    t.forward(halb)
    t.left(120)

def sierpinski():
    t.reset()
    t.hideturtle()
    t.penup()                 # Ausgangsposition
    t.setpos(-350, -280)
    t.setheading(0)
    zeichne(650, stufe)       # Aufruf der rekursiven Funktion
    t.update()                # Bild zeigen

t.tracer(0, 0)
t.onscreenclick(leftmouse, btn=1)
t.onscreenclick(rightmouse, btn=3)
stufe = 2
sierpinski()
t.mainloop()
```

#### Kochsche Schneeflocke

<img src="turtle_05.png" width="700">

Auch hier wird nur im Basisfall gezeichnet, und zwar eine gerade Strecke. Im rekursiven Fall wird die Strecke
gedrittelt, und das mittlere Drittel wird durch zwei Seiten eines gleichseitigen Dreiecks ersetzt. Aus einer Strecke
werden so vier Strecken, jede ein Drittel so lang:

<img src="koch_stufen.png" width="600">

Jeder Aufruf endet am Ende seiner Strecke und schaut dann wieder in dieselbe Richtung wie am Anfang
(links 60, rechts 120, links 60 ergibt zusammen 0 Grad). Die Schneeflocke besteht aus drei solchen Kurven.

Bei Stufe $n$ besteht jede Seite aus $4^n$ Strecken. Weil 4 Strecken mit je einem Drittel der Länge zusammen
$\frac{4}{3}$-mal so lang sind wie die alte Strecke, wird die Kurve mit jeder Stufe um ein Drittel länger.

```python
import turtle as t

def leftmouse(x, y):
    global stufe
    stufe += 1
    koch()

def rightmouse(x, y):
    global stufe
    if stufe > 0:
        stufe -= 1
    koch()

def zeichne(laenge, bremse):
    if bremse == 0:
        t.pendown()
        t.forward(laenge)
        t.penup()
        return
    drittel = laenge/3
    zeichne(drittel, bremse-1)
    t.left(60)
    zeichne(drittel, bremse-1)
    t.right(120)
    zeichne(drittel, bremse-1)
    t.left(60)
    zeichne(drittel, bremse-1)

def koch():
    t.reset()
    t.hideturtle()
    t.penup()                 # Ausgangsposition
    t.setpos(-380, 180)
    t.setheading(0)
    for i in range(3):        # 3 Seiten der Schneeflocke
        zeichne(650, stufe)
        t.right(120)
    t.update()

t.tracer(0, 0)
t.onscreenclick(leftmouse, btn=1)
t.onscreenclick(rightmouse, btn=3)
stufe = 2
koch()
t.mainloop()
```

#### Fraktale

Bilder, die aus verkleinerten Kopien von sich selbst bestehen, nennt man **selbstähnlich**. Selbstähnliche Bilder wie
das Sierpinski-Dreieck oder die Kochsche Schneeflocke heißen auch **Fraktale**. Ähnliche Formen gibt es in der Natur,
z.B. bei Bäumen, Farnen, Blumenkohl oder Küstenlinien.
