### Endliche Automaten


#### Automaten im Alltag

Viele Geräte und Programme reagieren auf Eingaben je nachdem, in welchem *Zustand* sie gerade sind:

<ul>
  <li><em>Ampel:</em> Nach rot kommt rot-gelb, nach grün kommt gelb.</li>
  <li><em>Snack-, Getränke- und Ticketautomat:</em> Ob nach dem Einwurf einer Münze ein Snack kommt, hängt davon ab, wie viel Geld schon eingeworfen wurde.</li>
  <li><em>Leergutautomat:</em> Er nimmt Flaschen an, zählt das Pfand und druckt auf Knopfdruck einen Bon.</li>
  <li><em>Grafische Benutzeroberfläche:</em> Ein Button ist aktiv oder ausgegraut, ein Fenster offen oder geschlossen, ein Login nach drei Fehlversuchen gesperrt.</li>
  <li><em>Netzwerkprotokolle:</em> Ein Rechner muss sich merken, ob eine Verbindung gerade aufgebaut wird, besteht oder beendet wird.</li>
  <li><em>Textsuche:</em> Ein Suchprogramm merkt sich, wie viel vom Suchwort es schon gefunden hat.</li>
</ul>

All diese Abläufe kann man mit einem *Automatenmodell* beschreiben. Ein Automat besteht aus

<ul>
  <li>endlich vielen <em>Zuständen</em> (genau einer davon ist der <em>Startzustand</em>),</li>
  <li><em>Eingaben</em>, die der Automat nacheinander liest,</li>
  <li><em>Übergängen</em>, die festlegen, in welchen Zustand der Automat bei einer Eingabe wechselt,</li>
  <li>und eventuell <em>Ausgaben</em>.</li>
</ul>

#### Beispiel: Ampel

Die Ampel hat die Zustände rot, rotgelb, grün und gelb. Die einzige Eingabe ist das Signal $t$ des Taktgebers (Timer).

<p><img src="ampel.png" width="550"></p>

Das Bild heißt *Zustandsdiagramm*. Die Kreise sind die Zustände, die Pfeile die Übergänge, an den Pfeilen steht die Eingabe.
Das Dreieck (oder ein Startpfeil) markiert den Startzustand.

Dieselbe Information steht in der *Zustandsübergangstabelle*. In der ersten Spalte stehen die Zustände, in der Kopfzeile die Eingaben.
In den Zellen steht der Folgezustand:

| | t |
|---:|:---:|
| rot | rotgelb |
| rotgelb | grün |
| grün | gelb |
| gelb | rot |

#### Endliche Automaten ohne Ausgabe

Ein Automat ohne Ausgabe kann trotzdem eine Entscheidung treffen. Man zeichnet einige Zustände als *Endzustände* (Doppelkreis).
Der Automat liest ein Wort Zeichen für Zeichen. Ist er danach in einem Endzustand, *akzeptiert* er das Wort, sonst nicht.

Ein solcher Automat heißt *deterministischer endlicher Automat (DEA)*. Er besteht aus:

<ol>
  <li> \( Q \), einer endlichen Menge von Zuständen</li>
  <li> \( \Sigma \), einem endlichen Eingabe-Alphabet</li>
  <li> \( \delta \), einer Übergangsfunktion</li>
  <li> \( s \in Q \), einem Startzustand</li>
  <li> \( F \subseteq Q\), einer Menge von Endzuständen</li>
</ol>

$\delta(q,a)$ ist der Zustand, in den der DEA geht, wenn er im Zustand $q$ das Eingabesymbol $a$ liest.
*Deterministisch* heißt: Für jeden Zustand und jedes Eingabesymbol gibt es genau einen Folgezustand.

#### Beispiel

<p><img src="bild1.png" width="400"></p>

<ol>
  <li> \( Q  =  \{q_0, q_1, q_2\}\)</li>
  <li> \( \Sigma = \{0,1\}\)</li>
  <li> \( \delta \) = gegeben durch die Pfeile</li>
  <li> \( s = q_0 \)</li>
  <li> \( F = \{q_0,q_1\} \)</li>
</ol>

Der Automat akzeptiert alle Wörter über $\{0,1\}$, die keine zwei aufeinanderfolgenden Einsen haben,
z.B. $010010$, aber nicht $0110$.

#### Vom Zustandsdiagramm zur Zustandsübergangstabelle

Für jeden Zustand eine Zeile, für jedes Eingabesymbol eine Spalte. Den Startzustand markiert man mit $\rightarrow$,
Endzustände mit $*$:

| | 0 | 1 |
|---:|:---:|:---:|
| $\rightarrow * \, q_0$ | $q_0$ | $q_1$ |
| $* \, q_1$ | $q_0$ | $q_2$ |
| $q_2$ | $q_2$ | $q_2$ |

Umgekehrt zeichnet man aus der Tabelle für jede Zeile einen Kreis und für jede Zelle einen Pfeil
vom Zustand der Zeile zum Zustand in der Zelle, beschriftet mit dem Symbol der Spalte.

Aus dem Zustand $q_2$ kommt der Automat nie mehr heraus, und $q_2$ ist kein Endzustand. Hat der Automat $q_2$ erreicht, wird das Wort also nicht akzeptiert, egal was noch kommt.
Einen solchen Zustand nennt man auch *Fangzustand* (englisch *dead state* oder *trap state*).
In Diagrammen lässt man ihn manchmal weg. Fehlt ein Pfeil, bedeutet das dann: Das Wort wird nicht akzeptiert.

#### Simulation in Python

Die Zustandsübergangstabelle lässt sich direkt als Dictionary in Python übernehmen.

```python
delta = {
    ('q0', '0'): 'q0', ('q0', '1'): 'q1',
    ('q1', '0'): 'q0', ('q1', '1'): 'q2',
    ('q2', '0'): 'q2', ('q2', '1'): 'q2',
}
start = 'q0'
endzustaende = {'q0', 'q1'}


def dea_akzeptiert(wort, ausgabe=False):
    zustand = start
    for zeichen in wort:
        neu = delta[(zustand, zeichen)]
        if ausgabe:
            print(f'{zustand} --{zeichen}--> {neu}')
        zustand = neu
    return zustand in endzustaende


print(dea_akzeptiert('010010', ausgabe=True))
print(dea_akzeptiert('0110'))
```

#### Anwendung: Textsuche (Substringproblem)

Ein Suchprogramm soll prüfen, ob ein Text das Muster $010$ enthält. Der DEA merkt sich in seinem Zustand,
wie viel vom Muster er schon gelesen hat:

<ul>
  <li> \( A \): noch nichts vom Muster</li>
  <li> \( B \): \( 0 \) gelesen</li>
  <li> \( C \): \( 01 \) gelesen</li>
  <li> \( D \): \( 010 \) gefunden (Endzustand)</li>
</ul>

<p><img src="dea010.png" width="400"></p>

Interessant sind die Pfeile zurück: Kommt in $C$ eine 1, ist der Anfang $01$ verdorben und man muss wieder
von vorne beginnen. Kommt in $B$ eine weitere 0, bleibt man in $B$, denn die letzte 0 kann der Anfang des Musters sein.

Solche Automaten stecken in Suchfunktionen von Editoren, in Virenscannern und in der Prüfung von Eingaben (z.B. Postleitzahl, E-Mail-Adresse).

```python
delta = {
    ('A', '0'): 'B', ('A', '1'): 'A',
    ('B', '0'): 'B', ('B', '1'): 'C',
    ('C', '0'): 'D', ('C', '1'): 'A',
    ('D', '0'): 'D', ('D', '1'): 'D',
}


def enthaelt_010(text):
    zustand = 'A'
    for zeichen in text:
        zustand = delta[(zustand, zeichen)]
    return zustand == 'D'


for text in ['1101001', '1101101', '0010', '011']:
    print(text, enthaelt_010(text))
```

#### Automaten mit Ausgabe: Mealy-Automaten

Reale Automaten entscheiden nicht nur ja/nein, sie *tun* etwas: Sie geben einen Snack aus, Wechselgeld zurück oder schalten eine Lampe.
Beim **Mealy-Automaten** erzeugt jeder Übergang eine Ausgabe. Er besteht aus:

<ol>
  <li> \( Q \), einer endlichen Menge von Zuständen</li>
  <li> \( \Sigma \), einem endlichen Eingabe-Alphabet</li>
  <li> \( \Gamma \), einem endlichen Ausgabe-Alphabet</li>
  <li> \( \delta \), der Übergangsfunktion (Zustand und Eingabe &rarr; Folgezustand)</li>
  <li> \( \lambda \), der Ausgabefunktion (Zustand und Eingabe &rarr; Ausgabe)</li>
  <li> \( s \in Q \), einem Startzustand</li>
</ol>

Endzustände braucht man nicht. Im Zustandsdiagramm schreibt man an den Pfeil: `Eingabe/Ausgabe`.
Das Zeichen – bedeutet: keine Ausgabe.

#### Beispiel: Snackautomat

Ein Snack kostet 2 €. Der Automat nimmt 1-€- und 2-€-Münzen an. Er muss sich merken, wie viel Geld schon eingeworfen wurde:

<p><img src="snack.png" width="380"></p>

In der Zustandsübergangstabelle steht in jeder Zelle `Folgezustand / Ausgabe`:

| | 1 | 2 |
|---:|:---:|:---:|
| $\rightarrow$ 0 € | 1 € / – | 0 € / Snack |
| 1 € | 0 € / Snack | 0 € / Snack, 1 € zurück |

#### Beispiel: Ein Mealy-Automat für Bits

Der Automat gibt eine 1 aus, wenn das aktuelle Eingabesymbol eine 1 ist und das vorherige auch. Sonst gibt er 0 aus.

<p><img src="mealy1.png" width="250"></p>

Eingabe `0110111` &rarr; Ausgabe `0010011`.

| | 0 | 1 |
|---:|:---:|:---:|
| $\rightarrow q_0$ | $q_0$ / 0 | $q_1$ / 0 |
| $q_1$ | $q_0$ / 0 | $q_1$ / 1 |

#### Anwendung: Zustandsbasierte Protokolle

Ein *Protokoll* legt fest, wie zwei Rechner miteinander kommunizieren: Wer sendet wann welche Nachricht, und wie wird darauf geantwortet?
Das Protokoll TCP (Transmission Control Protocol) wird z.B. beim Aufruf einer Webseite verwendet. Bevor Daten fließen, bauen Browser (Client) und Webserver eine Verbindung auf,
ähnlich wie bei einem Telefonat: Erst klingelt es, dann meldet sich der andere, dann spricht man, am Ende verabschiedet man sich.
Der Client muss sich dabei merken, in welcher Phase er gerade ist. Das ist der Zustand des Automaten.

<p><img src="tcp.png" width="700"></p>

Die Beschriftung an den Pfeilen heißt wieder `Eingabe/Ausgabe`. Die Eingabe ist ein *Ereignis*: entweder ein Wunsch der Anwendung (z.B. des Browsers)
oder eine Nachricht, die vom Server ankommt. Die Ausgabe ist die Nachricht, die der Client daraufhin an den Server sendet.

| Zustand | Bedeutung |
|---:|:---|
| geschlossen | Es besteht keine Verbindung. |
| wartet | Der Client hat um eine Verbindung gebeten und wartet auf die Antwort des Servers. |
| verbunden | Die Verbindung steht, Daten können ausgetauscht werden. |
| trennt | Der Client hat den Wunsch zum Beenden gesendet und wartet auf die Bestätigung. |

| Übergang | Was passiert? |
|---:|:---|
| `verbinden/SYN` | Der Browser will eine Seite laden. Der Client sendet `SYN` ("Ich möchte eine Verbindung aufbauen") und wartet. |
| `SYN+ACK/ACK` | Der Server antwortet mit `SYN+ACK` ("Einverstanden, ich bin bereit"). Der Client bestätigt mit `ACK` ("Verstanden"). Jetzt ist die Verbindung aufgebaut. |
| `Daten/ACK` | Im Zustand *verbunden* kommen Daten vom Server (z.B. Teile der Webseite). Der Client bestätigt jedes Mal den Empfang mit `ACK`. Er bleibt im Zustand *verbunden*. |
| `trennen/FIN` | Der Browser schließt die Seite. Der Client sendet `FIN` ("Ich möchte die Verbindung beenden"). |
| `ACK/–` | Der Server bestätigt das Ende. Der Client sendet nichts mehr (–) und ist wieder *geschlossen*. |

Das Diagramm ist stark vereinfacht. Das echte TCP hat mehr Zustände, Nachrichten und Sonderfälle (z.B. wenn eine Nachricht verloren geht).
Es zeigt aber, worauf es ankommt: Dieselbe Nachricht kann je nach Zustand etwas ganz Verschiedenes bedeuten, oder sie ist gar nicht erlaubt.
Nachrichten, für die im aktuellen Zustand kein Pfeil existiert, werden ignoriert. Daten können z.B. nur im Zustand *verbunden* verarbeitet werden.

Weitere Beispiele für Übergänge, die durch Ereignisse ausgelöst werden: Ein Hotelkarten-Schloss wechselt zwischen *gesperrt* und *offen*,
ein Medienspieler zwischen *Pause*, *Wiedergabe* und *gestoppt*.
