## Funktionsweise eines Rechners


#### Das EVA-Prinzip

Jede Datenverarbeitung folgt dem **EVA-Prinzip**: **E**ingabe, **V**erarbeitung, **A**usgabe.

- **Eingabe:** Daten, Befehle oder Signale gelangen in das System, zum Beispiel über die Tastatur.
- **Verarbeitung:** Der Prozessor (CPU) berechnet und verarbeitet die Daten und speichert sie bei Bedarf.
- **Ausgabe:** Die Ergebnisse werden ausgegeben, zum Beispiel auf dem Bildschirm, oder an ein anderes System weitergegeben.

#### Die Von-Neumann-Architektur

Wie ein Rechner innen aufgebaut ist, hat der Mathematiker **John von Neumann** 1945 beschrieben. Fast alle heutigen Computer folgen noch diesem Bauplan, der **Von-Neumann-Architektur**.

<img src='bild2.png' width='250'>


| Komponente | Aufgabe |
|:--|:--|
| **Rechenwerk** (ALU, *Arithmetic Logic Unit*) | führt Rechenoperationen (z. B. Addition) und logische Verknüpfungen aus |
| **Steuerwerk** (*Control Unit*) | holt die Befehle des Programms der Reihe nach aus dem Speicher, entschlüsselt sie und steuert alle anderen Komponenten so, dass der Befehl ausgeführt wird |
| **Speicherwerk** (Arbeitsspeicher, RAM) | speichert **Programme und Daten** |
| **Ein-/Ausgabewerk** | verbindet den Rechner mit Tastatur, Bildschirm und anderen Geräten |
| **Bussystem** | verbindet die Komponenten miteinander |

Rechenwerk und Steuerwerk bilden zusammen den **Prozessor** (CPU, *Central Processing Unit*).

**Register** sind kleine, sehr schnelle Speicher **im Prozessor**. Sie halten die Werte, mit denen der Prozessor gerade arbeitet. Wichtige Register sind:

- der **Akkumulator** im Rechenwerk: Er enthält das aktuelle Rechenergebnis.
- das **Befehlsregister** im Steuerwerk: Es enthält den Befehl, der gerade ausgeführt wird.
- der **Befehlszähler** im Steuerwerk: Er enthält die Adresse des nächsten Befehls.

Das **Bussystem** besteht aus mehreren Leitungsbündeln:

- Über den **Adressbus** wird festgelegt, **welche** Speicherzelle gelesen oder beschrieben wird.
- Über den **Datenbus** wird der **Inhalt** übertragen, also die Daten oder Befehle.
- Über den **Steuerbus** wird festgelegt, **was** gerade passieren soll, zum Beispiel lesen oder schreiben.

##### Die Von-Neumann-Prinzipien

1. **Programm und Daten liegen im selben Speicher.** Einer Speicherzelle sieht man nicht an, ob sie einen Befehl oder eine Zahl enthält. Das entscheidet erst der Prozessor: Was der Befehlszähler als Nächstes holt, wird als Befehl ausgeführt.
2. Der Speicher besteht aus **durchnummerierten Zellen**. Die Nummer einer Zelle heißt **Adresse**.
3. Die Befehle werden **nacheinander** ausgeführt. Nur **Sprungbefehle** können die Reihenfolge ändern.
4. Der Prozessor arbeitet in einem ständigen **Befehlszyklus**: Befehl holen, entschlüsseln, ausführen, nächsten Befehl holen, …

Weil Programm und Daten im selben Speicher liegen, kann ein Rechner beliebige Programme ausführen, ohne dass man ihn umbauen muss. Man lädt einfach ein anderes Programm in den Speicher.

Ein Nachteil: Befehle und Daten müssen nacheinander über denselben Bus. Der Bus ist oft langsamer als der Prozessor, der deshalb warten muss. Man spricht vom **Von-Neumann-Flaschenhals**.

##### Die Assemblerbefehle

Programme für Johnny schreiben wir mit den folgenden **Assemblerbefehlen**. 

| Code | Befehl | Bedeutung |
|:--|:--|:--|
| 01 | `TAKE x` | kopiert den Wert aus `[x]` in den Akkumulator |
| 02 | `ADD x` | erhöht den Akkumulator um `[x]` |
| 03 | `SUB x` | erniedrigt den Akkumulator um `[x]` |
| 04 | `SAVE x` | speichert den Inhalt des Akkumulators in `[x]` |
| 05 | `JMP x` | springt zu Adresse `x`, d. h. der nächste Befehl steht in Zelle `x` |
| 06 | `TST x` | falls `[x] = 0`: überspringt den nächsten Befehl |
| 07 | `INC x` | erhöht `[x]` um 1 |
| 08 | `DEC x` | erniedrigt `[x]` um 1 |
| 09 | `NULL x` | setzt `[x]` auf 0 |
| 10 | `HLT x` | hält das Programm an, `x` spielt keine Rolle |

##### Die Mikrobefehle

Jeder Assemblerbefehl ist aus noch kleineren Schritten zusammengesetzt, den **Mikrobefehlen**. Jeder Mikrobefehl öffnet eine Verbindung zwischen zwei Bauteilen oder löst eine einfache Rechnung aus.

```text
db->ram    Inhalt des Datenbus im RAM speichern (an der Adresse auf dem Adressbus)
ram->db    Inhalt des RAM (an der Adresse auf dem Adressbus) auf den Datenbus legen
db->ins    Inhalt des Datenbus ins Befehlsregister übernehmen
db->acc    Inhalt des Datenbus in den Akkumulator übernehmen
acc->db    Inhalt des Akkumulators auf den Datenbus legen

plus       Akkumulator um den Inhalt des Datenbus erhöhen
minus      Akkumulator um den Inhalt des Datenbus erniedrigen
acc++      Akkumulator um 1 erhöhen
acc--      Akkumulator um 1 erniedrigen
acc=0      Akkumulator auf 0 setzen

pc->ab     Befehlszähler auf den Adressbus legen
pc++       Befehlszähler um 1 erhöhen
=0:pc++    falls der Akkumulator 0 ist: Befehlszähler um 1 erhöhen

ins->ab    Adressteil (YYY) des Befehlsregisters auf den Adressbus legen
ins->pc    Adressteil (YYY) des Befehlsregisters in den Befehlszähler schreiben
ins->mc    zum Mikroprogramm des Befehls springen, dessen Code (XX) im Befehlsregister steht

mc=0       zurück zum Anfang des Mikrocodes (FETCH)
stop       Programm beenden
```

Im echten Rechner sind Datenbus und Adressbus Leitungen und speichern nichts. Im Modell Johnny verhalten sie sich wie kleine Zwischenspeicher: Ein Wert bleibt auf dem Bus, bis ein anderer daraufgelegt wird.

Zum Kennenlernen gibt es in Johnny außerdem zwei Knöpfe, die nicht im Mikrocode verwendet werden können:

```text
ManuellAb   eine Zahl von Hand auf den Adressbus schreiben
ManuellDb   eine Zahl von Hand auf den Datenbus schreiben
```
