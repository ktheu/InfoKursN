### ADT Liste

Mit einer Python-Liste können wir Elemente speichern, ohne zu wissen, wie sie im Speicher liegen. Jetzt bauen wir eine
Liste selbst: aus einzelnen Einträgen, die über Referenzen aufeinander verweisen. Dabei lernen wir das Konzept
des **Abstrakten Datentyps** kennen.

Zur Unterscheidung: Die Python-Liste (`list`) ist im Hintergrund ein **Array**, die Elemente liegen im Speicher
nebeneinander. Unsere Klasse `Liste` ist eine **einfach verkettete Liste**.

## 1. Abstrakte Datentypen

**Abstrakter Datentyp (ADT)** = Datenstruktur + (abstrakte) Operationen.

Abstrakt heißt: nicht implementiert, es gibt nur Vorgaben.

Die Datenstruktur wird unabhängig von ihrer späteren Implementierung in einer Programmiersprache beschrieben.
Ein ADT legt die **Schnittstelle** nach außen fest: welche Operationen es gibt und was sie leisten. Wie die Daten
gespeichert sind, bleibt verborgen. Das kennen wir als Geheimnisprinzip: Wer den ADT benutzt, braucht nur die Schnittstelle.
Die Implementierung lässt sich austauschen, ohne dass sich am benutzenden Programm etwas ändert.

## 2. Der ADT Liste

Eine *Liste* ist eine (ggf. leere) Folge von Elementen zusammen mit einem so genannten
(ggf. undefinierten) *aktuellen Element*.

<img src='bild1.png' width='600'>

Das aktuelle Element ist die Stelle, an der wir gerade arbeiten, ähnlich wie der Cursor in einem Text. Undefiniert ist es, wenn die
Liste leer ist oder wenn wir hinter dem letzten Element angekommen sind.

#### Schnittstelle des ADT Liste

| Operation | Bedeutung |
|---|---|
| `empty` | liefert True, falls die Liste leer ist |
| `endpos` | liefert True, wenn die Liste abgearbeitet ist |
| `reset` | das erste Listenelement wird zum aktuellen Element |
| `advance` | der Nachfolger des aktuellen Elements wird zum aktuellen Element |
| `elem` | liefert das aktuelle Element |
| `insert(x)` | fügt x vor dem aktuellen Element ein, x wird zum aktuellen Element |
| `delete` | löscht das aktuelle Element, der Nachfolger wird zum aktuellen Element |

#### Umsetzung der Schnittstelle in eine Klasse

Die Methoden sind noch nicht implementiert.

```python
class Liste:
    '''  Eine Liste ist eine (ggf. leere) Folge von Elementen zusammen mit einem
    (ggf. undefinierten) aktuellen Element  '''

    def empty(self):
        ''' liefert True, falls die Liste leer ist '''
        pass

    def endpos(self):
        ''' liefert True, wenn die Liste abgearbeitet ist '''
        pass

    def reset(self):
        ''' das erste Listenelement wird aktuelles Element '''
        pass

    def advance(self):
        ''' der Nachfolger des aktuellen Elements wird aktuelles Element '''
        pass

    def elem(self):
        ''' liefert das aktuelle Element '''
        pass

    def insert(self, x):
        ''' fügt x vor dem aktuellen Element ein, x wird zum neuen aktuellen Element '''
        pass

    def delete(self):
        ''' löscht das aktuelle Element, der Nachfolger wird neues aktuelles Element '''
        pass
```

Welche dieser Methoden wären bei einer Implementation mittels einer normalen Python-Liste problematisch?

Das sind `insert` und `delete`. Die Elemente eines Arrays liegen im Speicher nebeneinander. Fügen wir in der Mitte ein oder
löschen dort, müssen alle Elemente dahinter verschoben werden. Das kostet $O(n)$.

## 3. Verkettete Einträge

Wir speichern die Elemente nicht nebeneinander, sondern in einzelnen **Einträgen**. Jeder Eintrag enthält den Inhalt und einen
Zeiger auf den nächsten Eintrag. Ein Zeiger ist eine Referenz: Das Attribut `next` verweist auf ein Objekt der Klasse `Eintrag`.
Beim letzten Eintrag steht in `next` der Wert `None`.

<img src='bild2.png'>

```python
class Eintrag:
    def __init__(self):
        self.inhalt = None
        self.next = None
```

Um uns durch die Einträge zu hangeln, nutzen wir eine `while`-Schleife wie bei der rekursiven Assoziation im
Abschnitt Klassen. Die Schleife endet, wenn wir bei `None` ankommen.

```python
e1 = Eintrag()
e1.inhalt = 'a'
e2 = Eintrag()
e2.inhalt = 'b'
e3 = Eintrag()
e3.inhalt = 'c'
e1.next = e2
e2.next = e3

e = e1
while e is not None:
    print(e.inhalt, end=' ')
    e = e.next
```

## 4. Die Zeiger `anf` und `pos`

Problem bei `insert`: Auf welches Element brauchen wir Zugriff, damit wir ein neues Element vor dem aktuellen Element
(roter Pfeil) einfügen können?

<img src='bild4.png'>

Wir müssen den Zeiger im Eintrag **vor** dem aktuellen Element ändern. Vom aktuellen Element aus kommen wir dort aber nicht hin,
denn die Zeiger gehen nur in eine Richtung.

Lösung für das `insert`-Problem:

<img src='bild5.png' width='600'>

`pos` zeigt auf den Listen-Eintrag vor dem aktuellen Element.

`anf` zeigt auf einen Dummy-Eintrag vor dem ersten Element. Der Dummy-Eintrag bleibt immer ohne Inhalt. Durch ihn
hat auch das erste Element einen Vorgänger, Einfügen und Löschen am Anfang brauchen keinen Sonderfall.

| | |
|---|---|
| aktuelles Element | `pos.next.inhalt` |
| Liste ist leer | `anf.next is None` |
| Liste ist abgearbeitet | `pos.next is None`, dann gibt es kein aktuelles Element |

#### Beispiel 1

Wir stellen den Ablauf mit Einträgen und den Zeigern `anf` und `pos` dar und bestimmen jeweils das aktuelle Element und das
Ergebnis von `endpos`.

    a = Liste()
    a.insert(42)
    a.advance()
    a.elem()
    a.delete()
    a.insert('c')
    a.elem()
    a.delete()

<img src='bild6.png' width='1001'>

#### Beispiel 2

Wir verfolgen den Zustand der Liste und die Position des aktuellen Elements. Das aktuelle Element ist unterstrichen, ein einzelner
Strich bedeutet: Es gibt kein aktuelles Element.

    a = Liste()
    a.insert(1)
    a.insert(2)
    a.advance()
    a.insert(3)
    a.advance()
    a.delete()
    a.insert(4)
    a.reset()
    a.advance()
    a.insert(6)
    a.reset()
    while not a.endpos():
        print(a.elem())
        a.advance()

## 5. Implementation der Liste

```python
class Liste:
    def __init__(self):
        self.anf = Eintrag()
        self.pos = self.anf

    def empty(self):
        return self.anf.next is None

    def endpos(self):
        return self.pos.next is None

    def reset(self):
        self.pos = self.anf

    def advance(self):
        if self.endpos(): raise RuntimeError("Fehler: Liste am Ende")
        self.pos = self.pos.next

    def elem(self):
        if self.endpos(): raise RuntimeError("Fehler: Liste am Ende")
        return self.pos.next.inhalt

    def insert(self, x):
        hilf = Eintrag()
        hilf.inhalt = x
        hilf.next = self.pos.next
        self.pos.next = hilf

    def delete(self):
        if self.endpos(): raise RuntimeError("Fehler: Liste am Ende")
        self.pos.next = self.pos.next.next
```

#### `insert`

Ein neuer Eintrag `hilf` wird zwischen `pos` und das bisherige aktuelle Element gehängt. Die Reihenfolge ist wichtig: Zuerst
übernimmt `hilf` den Nachfolger, erst dann zeigt `pos` auf `hilf`. Der Zeiger `pos` selbst bleibt stehen, dadurch ist der neue
Eintrag das aktuelle Element.

<img src='insert.png' width='700'>

#### `delete`

Der Zeiger im Eintrag `pos` wird am aktuellen Element vorbei auf dessen Nachfolger gesetzt. Damit ist das Element nicht mehr
erreichbar, und der Nachfolger ist das neue aktuelle Element.

<img src='delete.png' width='560'>

#### Laufzeit

Keine der Methoden enthält eine Schleife. Alle brauchen unabhängig von der Länge der Liste nur wenige Schritte, also $O(1)$.

#### Fehlersituationen

`advance`, `elem` und `delete` brauchen ein aktuelles Element. Gibt es keins, erzeugen wir mit `raise` selbst einen Fehler.

```python
a = Liste()
a.insert(42)
a.advance()
print(a.endpos())
a.elem()
```

`anf` und `pos` sind öffentliche Attribute, weil das den Code und die Zeichnungen übersichtlicher macht. Nach dem
Geheimnisprinzip müssten sie privat sein. Ein Programm, das die Liste benutzt, soll nur die Methoden der Schnittstelle aufrufen.

#### Das Beispiel 2 im Programm

Die Funktion `zeige` hangelt sich durch die Einträge und setzt das aktuelle Element in eckige Klammern. So können wir
jeden Schritt aus Beispiel 2 nachvollziehen. Die leeren Klammern bedeuten: Es gibt kein aktuelles Element.

```python
def zeige(a):
    '''
    a: Liste
    gibt die Elemente aus, das aktuelle Element steht in eckigen Klammern
    '''
    text = ''
    e = a.anf
    
    while e.next is not None:
        if e is a.pos:
            text += f'[{e.next.inhalt}] '
        else:
            text += f'{e.next.inhalt} '
        e = e.next
    if e is a.pos:
        text += '[ ]'
    print(text)

a = Liste()
zeige(a)
a.insert(1)
zeige(a)
a.insert(2)
zeige(a)
a.advance()
zeige(a)
a.insert(3)
zeige(a)
a.advance()
zeige(a)
a.delete()
zeige(a)
a.insert(4)
zeige(a)
a.reset()
zeige(a)
a.advance()
zeige(a)
a.insert(6)
zeige(a)
```

## 6. Die Liste benutzen

Ab jetzt benutzen wir die Liste nur noch über die Methoden der Schnittstelle. `zeige` bleibt unsere Kontrolle im Notebook.

#### Die typische Schleife

Mit `reset` beginnen wir vorn, mit `advance` gehen wir weiter, bis `endpos` True liefert.

```python
a.reset()
while not a.endpos():
    print(a.elem(), end=' ')
    a.advance()
```

#### Hinten anfügen

`insert` fügt vor dem aktuellen Element ein. Rufen wir danach `advance` auf, stehen wir wieder hinter dem letzten Element. So bleibt die
Reihenfolge erhalten.

```python
a = Liste()
for x in [4, 8, 15, 16, 23, 42]:
    a.insert(x)
    a.advance()

summe = 0
a.reset()
while not a.endpos():
    summe += a.elem()
    a.advance()
print(summe)
```

#### Lineare Suche

Die **lineare Suche** schaut sich ein Element nach dem anderen an. Auf einem Array können wir dafür die Position als Index benutzen,
in unserer Liste zählen wir die Position selbst mit. In beiden Fällen brauchen wir im schlechtesten Fall $n$ Schritte, also $O(n)$.

```python
def lineareSuche_array(a, x):
    '''
    a: Python-Liste
    x: gesuchter Wert
    returns: Position des ersten Vorkommens von x, -1 wenn x nicht vorkommt
    '''
    for i in range(len(a)):
        if a[i] == x:
            return i
    return -1

def lineareSuche(a, x):
    '''
    a: Liste
    x: gesuchter Wert
    verändert a: wenn x vorkommt, ist das erste Vorkommen danach das aktuelle Element
    returns: Position des ersten Vorkommens von x, -1 wenn x nicht vorkommt
    '''
    i = 0
    a.reset()
    while not a.endpos():
        if a.elem() == x:
            return i
        a.advance()
        i += 1
    return -1

print(lineareSuche_array([4, 8, 15], 15), lineareSuche(a, 15), lineareSuche(a, 99))
```

Nach einer erfolgreichen Suche steht die Liste auf dem gefundenen Element. Wir können es gleich löschen oder davor einfügen.

```python
if lineareSuche(a, 16) != -1:
    a.delete()
    a.insert(17)
zeige(a)
```

#### Beim Durchlaufen löschen

Nach `delete` ist der Nachfolger schon das aktuelle Element. Würden wir danach auch noch `advance` aufrufen, würden wir ein Element
überspringen. Deshalb gehört `advance` in den `else`-Zweig.

```python
def loesche_gerade(a):
    '''
    a: Liste aus ganzen Zahlen
    verändert a: alle geraden Zahlen werden gelöscht
    '''
    a.reset()
    while not a.endpos():
        if a.elem() % 2 == 0:
            a.delete()
        else:
            a.advance()

loesche_gerade(a)
zeige(a)
```

## 7. Eine andere Implementation

Dass ein ADT unabhängig von seiner Implementierung ist, können wir direkt ausprobieren. `ListeArray` hat dieselbe Schnittstelle wie
`Liste`, speichert die Elemente aber in einer Python-Liste. `pos` ist hier der Index des aktuellen Elements.

```python
class ListeArray:
    def __init__(self):
        self.daten = []
        self.pos = 0

    def empty(self):
        return len(self.daten) == 0

    def endpos(self):
        return self.pos == len(self.daten)

    def reset(self):
        self.pos = 0

    def advance(self):
        if self.endpos(): raise RuntimeError("Fehler: Liste am Ende")
        self.pos += 1

    def elem(self):
        if self.endpos(): raise RuntimeError("Fehler: Liste am Ende")
        return self.daten[self.pos]

    def insert(self, x):
        self.daten.insert(self.pos, x)

    def delete(self):
        if self.endpos(): raise RuntimeError("Fehler: Liste am Ende")
        self.daten.pop(self.pos)
```

Dasselbe Programm läuft mit beiden Klassen und liefert dasselbe Ergebnis. Es hängt nur von der Schnittstelle ab.

```python
def programm(a):
    '''
    a: leere Liste oder ListeArray
    '''
    for x in [3, 6, 9, 12]:
        a.insert(x)
        a.advance()
    print(lineareSuche(a, 9))
    a.delete()
    a.insert(0)
    a.reset()
    while not a.endpos():
        print(a.elem(), end=' ')
        a.advance()
    print()

programm(Liste())
programm(ListeArray())
```

Unterschiedlich ist die Laufzeit. Wir fügen $n$ Elemente immer am Anfang ein. Bei der verketteten Liste kostet das pro Element $O(1)$,
die Zeit bleibt klein. Beim Array werden jedes Mal alle Elemente verschoben. Verdoppeln wir $n$, vervierfacht sich die Zeit ungefähr.

```python
import time

def laufzeit(a, n):
    start = time.perf_counter()
    for i in range(n):
        a.insert(i)
    return time.perf_counter() - start

print('n', 'Liste', 'ListeArray', sep='\t')
for n in [20000, 40000, 80000]:
    print(n, round(laufzeit(Liste(), n), 3), round(laufzeit(ListeArray(), n), 3), sep='\t')
```

Dafür kann das Array etwas, das die verkettete Liste nicht kann: direkt auf die Position `i` zugreifen. Bei der verketteten Liste
müssen wir uns dorthin durchhangeln. Deshalb lohnt sich die binäre Suche nur auf einem Array.

| Operation | Array | verkettete Liste |
|---|---|---|
| Zugriff auf Position `i` | $O(1)$ | $O(n)$ |
| Einfügen, Löschen an der aktuellen Stelle | $O(n)$ | $O(1)$ |
| lineare Suche | $O(n)$ | $O(n)$ |
| binäre Suche (sortiert) | $O(\log n)$ | nicht sinnvoll |

## 8. Das Klassendiagramm

Die Klasse `Liste` kennt über `anf` und `pos` Objekte der Klasse `Eintrag`. Die Assoziation bei der Klasse `Eintrag` verweist auf sich
selbst. Dies macht die Liste zu einer *rekursiven Datenstruktur*.

<img src='liste_uml.png' width='600'>

#### Zusammenfassung

| | |
|---|---|
| ADT | Datenstruktur + Operationen, festgelegt ist nur die Schnittstelle |
| Schnittstelle | die Operationen eines ADT und was sie leisten |
| Array | Elemente liegen im Speicher nebeneinander, z. B. die Python-Liste |
| verkettete Liste | Einträge mit Inhalt und Zeiger auf den nächsten Eintrag |
| Eintrag | `inhalt` und `next`, beim letzten Eintrag ist `next` gleich `None` |
| aktuelles Element | die Stelle, an der wir arbeiten, `elem()` liefert es |
| `anf` | zeigt auf den Dummy-Eintrag vor dem ersten Element |
| `pos` | zeigt auf den Eintrag vor dem aktuellen Element |
| `endpos()` | True, wenn es kein aktuelles Element gibt |
| `insert(x)` | fügt x vor dem aktuellen Element ein, x wird aktuelles Element |
| `delete()` | löscht das aktuelle Element, der Nachfolger wird aktuelles Element |
| typische Schleife | `reset()`, `while not endpos():`, `elem()`, `advance()` |
| lineare Suche | ein Element nach dem anderen ansehen, $O(n)$ |
| rekursive Datenstruktur | die Klasse `Eintrag` verweist auf sich selbst |
