#### Binärer Baum

Ein *binärer Baum* ist entweder leer oder besteht aus einem Knoten, dem zwei binäre Bäume zugeordnet sind. Dieser heißt dann der *Elternknoten* des linken bzw. rechten Teilbaums. Ein Knoten ohne einen Elternknoten heißt *Wurzel*. Die Knoten, die x als Elternknoten haben, sind *Kinder* von x. Knoten ohne Kinder heißen *Blätter*.

#### Heap

Ein *Heap* ist ein binärer Baum für den gilt:  
 
- Jeder Knoten enthält einen Schlüssel  
- Der Schlüssel im Elternknoten ist kleiner oder gleich dem Schlüssel im Kind  
- Alle Ebenen sind vollständig gefüllt, bis auf die letzte, die muss von links beginnend gefüllt sein.  

Ein Heap aus der Entfernung betrachtet:

Wir übertragen die Zahlen in einem Heap ebenenweise in eine Liste:

<img src='heapsort3a.png' width='600'>

```python
a = [42,80,55,90,81,60,58,92,94,93,90]
```

Für die Indizes in der Liste gilt:

    linkes Kind von Knoten i: 2i+1 
    rechtes Kind von Knoten i: 2i+2 
    Elternknoten von Knoten i: (i-1)//2

Mit den Formeln für die Indizes können wir prüfen, ob eine Liste ein Heap ist: Jedes Element muss größer oder gleich
seinem Elternknoten sein.

```python
def ist_heap(a):
    for i in range(1, len(a)):
        if a[(i-1)//2] > a[i]:
            return False
    return True

print(ist_heap(a))
print(ist_heap([2, 5, 4, 6, 3]))     # 3 am Index 4 ist kleiner als sein Elternknoten 5 am Index 1
```

#### Idee für HeapSort: 

    Gegeben Liste a  
    Konstruiere daraus einen Heap.  
    for i in range(n):
        liefere Wurzel als aktuelles Minimum 
        entferne Wurzel 
        reorganisiere Heap

#### Reorganisation eines Heaps

<img src='heapsort4.png' width='1000'>

Der Aufwand der Reorganisation ist proportional zur Länge des Weges: $O(log(n))$

<img src='heapsort5.png' width='400'>

Die Funktion *sift* 
prüft, ob zwischen *start* und *end* ein ordentlicher Heap ist, oder ob *start* versickern muss.
Die Funktion kann sich darauf verlassen, dass unterhalb von *start* alles in Ordnung ist.

<img src='heapsort6.png' width='300'>

```python
def sift(a, start, end):
    i = start
    x = a[start]
    j = 2 * i + 1
    if j < end and a[j+1] < a[j]:
        j+=1
    while j <= end and a[j] < x:
        a[i] = a[j]
        i = j
        j = j * 2 + 1
        if j < end and a[j+1] < a[j]:
            j+=1
    a[i] = x
```

Beispiel: Nur die Wurzel 10 ist falsch, darunter ist alles in Ordnung. Die 10 versickert, bis ihre Kinder größer sind.
Dabei wird immer mit dem kleineren Kind getauscht.

```python
a = [10, 3, 4, 7, 5, 8, 9]
sift(a, 0, len(a)-1)
print(a)
```

#### Aufbau eines Heaps

    letzter Knoten: a[len(a)-1]  
    Elternknoten des letzten Knotens: a[(len(a)-2)//2]  

    Beginne beim Elternknoten des letzten Knotens. 
    Gehe von dort rückwärts Ebene für Ebene durch und rufe den Elementen jeweils sift zu.

<img src='heapsort8.png' width='420'>

Inhalt der Liste nach Aufbau des Heaps und nach jeder Reorganisation:

<img src='heapsort9.png' width='250'>

Mit einer Ausgabe nach jedem Aufruf von `sift` können wir den Aufbau des Heaps verfolgen.

```python
def mache_heap(a):
    n = len(a)
    for k in range((n-2)//2, -1, -1):
        sift(a, k, n-1)
        print(f'sift({k}):', *a)

a = [14, 42, 2, 13, 7, 24, 49, 5, 18]
mache_heap(a)
print(ist_heap(a))
```

#### HeapSort

<img src='heapsort10.png' width='400'>

```python
def sift(a, start , end):
    i = start
    x = a[start]
    j = 2 * i + 1
    if j < end and a[j+1] < a[j]:
        j+=1
    while j <= end and a[j] < x:
        a[i] = a[j]
        i = j
        j = j * 2 + 1
        if j < end and a[j+1] < a[j]:
            j+=1
    a[i] = x

def heap_sort(a):
    n = len(a)
    for k in range ((n-2)//2,-1,-1):
        sift(a,k,n-1)
    for k in range(n-1,0,-1):
        a[0],a[k]=a[k],a[0]
        sift(a,0,k-1)
```

```python
a = [14, 42, 2, 13, 7, 24, 49, 5, 18]
heap_sort(a)
a = a[::-1]
print(a)
```

Die Wurzel ist immer das kleinste Element. Sie wird mit dem letzten Element des Heaps getauscht, danach ist der Heap
um ein Element kürzer. Hinten sammeln sich so die kleinsten Elemente: Die Liste ist am Ende **absteigend** sortiert.
Deshalb drehen wir sie mit `a[::-1]` um.

Mit einer Ausgabe nach dem Aufbau des Heaps und nach jeder Reorganisation erhalten wir die Tabelle von oben.

```python
def heap_sort_protokoll(a):
    n = len(a)
    for k in range((n-2)//2, -1, -1):
        sift(a, k, n-1)
    print(*a)
    for k in range(n-1, 0, -1):
        a[0], a[k] = a[k], a[0]
        sift(a, 0, k-1)
        print(*a[:k], '|', *a[k:])

heap_sort_protokoll([14, 42, 2, 13, 7, 24, 49, 5, 18])
```

Rechts vom `|` steht der schon sortierte Teil.

#### Analyse von HeapSort

Aufwand für Aufbau des Heaps: $O(n)$   (ca. die Hälfte der Elemente des Heaps sind Blätter, nur jeweils ca. n/2 Elemente müssen reorganisiert werden, wobei nach oben hin die Anzahl der Elemente mit den längeren Sickerwegen abnimmt.)   

Aufwand für eine Reorganisation: $O(\log n)$

Insgesamt: $O(n \cdot \log n)$ im best, worst und average case.  

Zusätzlicher Platzbedarf:$O(1)$

#### Laufzeit und Platzbedarf der Sortieralgorithmen

|   | best | average | worst | zusätzlicher Platz |
|----------|----------|----------|--------|--------|
| SelectionSort  | $O(n^2)$   | $O(n^2)$   | $O(n^2)$   | $O(1)$   | 
| BubbleSort  | $O(n)$   | $O(n^2)$   | $O(n^2)$   | $O(1)$   | 
| InsertionSort  | $O(n)$   | $O(n^2)$   | $O(n^2)$   | $O(1)$   |
| MergeSort  | $O(n \cdot \log(n))$   | $O(n \cdot \log(n))$   | $O(n \cdot \log(n))$    | $O(n)$   |
| QuickSort  | $O(n \cdot \log(n))$   | $O(n \cdot \log(n))$   | $O(n^2)$    | $O(\log(n))$   |
| HeapSort  | $O(n \cdot \log(n))$   | $O(n \cdot \log(n))$   | $O(n \cdot \log(n))$   | $O(1)$   |

#### Heaps in Python: heapq

Das Modul `heapq` enthält fertige Funktionen für Heaps in einer Liste:

| Funktion | Bedeutung | Laufzeit |
|---|---|---|
| `heapify(a)` | macht aus der Liste `a` einen Heap | $O(n)$ |
| `heappush(a, x)` | fügt `x` in den Heap ein | $O(\log n)$ |
| `heappop(a)` | entfernt das kleinste Element und gibt es zurück | $O(\log n)$ |
| `a[0]` | das kleinste Element, ohne es zu entfernen | $O(1)$ |

`heapify` baut den Heap auf dieselbe Weise auf wie unsere Funktion `mache_heap`.

```python
from heapq import heapify, heappush, heappop

a = [14, 42, 2, 13, 7, 24, 49, 5, 18]
heapify(a)
print(a)
heappush(a, 1)
print(a)
print(heappop(a), heappop(a), heappop(a))
```

Ein Heap eignet sich als **Prioritätswarteschlange** (PriorityQueue): Man kann jederzeit neue Elemente einfügen
und bekommt immer das mit der höchsten Priorität (der kleinsten Zahl) zuerst. Tuples werden nach der ersten
Komponente verglichen.

```python
aufgaben = []
heappush(aufgaben, (3, 'Zimmer aufräumen'))
heappush(aufgaben, (1, 'Hausaufgaben'))
heappush(aufgaben, (2, 'Einkaufen'))
heappush(aufgaben, (1, 'Informatik lernen'))

while aufgaben:
    prioritaet, aufgabe = heappop(aufgaben)
    print(prioritaet, aufgabe)
```

#### Zusammenfassung

| | |
|---|---|
| Heap | binärer Baum, Elternknoten $\le$ Kinder, alle Ebenen bis auf die letzte voll |
| Heap als Liste | Kinder von `i`: `2*i+1` und `2*i+2`, Elternknoten: `(i-1)//2` |
| `sift` | lässt ein Element nach unten versickern, $O(\log n)$ |
| Heap aufbauen | `sift` für alle Elternknoten, von hinten nach vorne, $O(n)$ |
| HeapSort | Wurzel nach hinten tauschen, Heap reorganisieren, $O(n \cdot \log n)$ in allen Fällen, Platz $O(1)$ |
| `heapq` | `heapify`, `heappush`, `heappop`, z.B. für Prioritätswarteschlangen |
