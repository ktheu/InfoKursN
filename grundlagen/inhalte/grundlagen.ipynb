## GL 1 · Variablen und Zuweisungen

### Inhalte

Zunächst betrachten wir zwei Arten von Zahlen:

| Datentyp | In Python | Beispiele |
|:--|:--|:--|
| **Ganzzahl** | `int` | `8`, `-3`, `0` |
| **Dezimalzahl** (Gleitkommazahl, *floating point number*) | `float` | `3.2`, `-0.5`, `2.0` |

**Dezimalzahlen werden mit einem Punkt geschrieben**, nicht mit einem Komma.


Mit einer **Zuweisung** ordnen wir einer Variablen einen Wert zu. Das Gleichheitszeichen ist dabei keine mathematische Aussage: Rechts wird gerechnet, links wird gespeichert.

````python
k = 8          # int
x = 3.2        # float
````

Mit der Funktion `print` geben wir Werte aus. `print` kann – mit Komma getrennt – beliebig viele Werte ausgeben. In der Ausgabe stehen sie durch ein Leerzeichen getrennt.

````python
print(k)           # 8
print(x)           # 3.2
print(k, x, 10)    # 8 3.2 10
````

Durch eine **erneute Zuweisung** verändern wir den Wert einer Variablen. Der alte Wert ist danach verloren.

````python
x = 5
x = x + 1      # liest 5, rechnet 6, speichert 6
print(x)       # 6
````

Für besonders häufige Zuweisungen gibt es **Kurzformen**:

````python
x += 1    # entspricht  x = x + 1
x -= 1    # entspricht  x = x - 1
x *= 2    # entspricht  x = x * 2   … usw.
````

## GL 2 · Kommentare und Variablennamen

### Inhalte

**Kommentare** werden nicht ausgeführt. Sie dienen zur Dokumentation des Programms.

````python
'''
Dies ist ein
mehrzeiliger Kommentar
'''
x = 1

# dies ist ein einzeiliger Kommentar
y = 2
z = 3   # auch dies ist ein Kommentar
````

**Regeln für Variablennamen:**

- Variablennamen sind **case-sensitiv**: `anzahl` und `Anzahl` sind zwei verschiedene Variablen.
- Das erste Zeichen ist ein **Buchstabe oder ein Unterstrich**, keine Ziffer.
- Wir beschränken uns freiwillig auf den **Unterstrich, die Buchstaben des englischen Alphabets und die Ziffern**. Leerzeichen und Zeichen wie `-` sind nicht erlaubt.
- Schlüsselwörter der Sprache wie `if`, `for` oder `while` sind als Namen verboten.

Wenn es der Lesbarkeit nicht schadet, erlauben wir uns kurze Namen wie `x` oder `k`. Ansonsten nehmen wir **sprechende Namen**. **Variablennamen beginnen in der Regel mit einem kleinen Buchstaben.** Zusammengesetzte Namen schreiben wir entweder in *camelCase* oder mit Unterstrich verbunden (*snake_case*), aber nicht gemischt:

````python
zaehl = 1
radius = 2.2
anzahlTreffer = 5            # camelCase
entfernungZurWand = 3.2      # camelCase
anzahl_treffer = 5           # snake_case
entfernung_zur_wand = 3.2    # snake_case
````

## GL 3 · Arithmetische Operatoren

### Inhalte

Neben den Grundrechenarten kennt Python die Exponentation, den Modulo-Operator und die ganzzahlige Division.

| Operator | Bedeutung | Beispiel | Ergebnis |
|:--|:--|:--|:--|
| `+` | Addition | `11 + 3` | 14 |
| `-` | Subtraktion | `11 - 3` | 8 |
| `*` | Multiplikation | `11 * 3` | 33 |
| `/` | Division | `11 / 3` | 3.6666666666666665 |
| `**` | Exponentation (11 hoch 3) | `11 ** 3` | 1331 |
| `%` | Modulo: Rest bei der Division | `11 % 3` | 2 |
| `//` | ganzzahlige Division | `11 // 3` | 3 |

**`//` und `%` gehören zusammen:** 11 geteilt durch 3 ergibt 3 Rest 2, denn 11 = 3 · 3 + 2.

**Die Division `/` liefert immer einen `float`**, auch wenn sie aufgeht: `6 / 2` ergibt `3.0`.

**Auswertungsreihenfolge:** Exponentation vor Punktrechnung vor Strichrechnung. Im Zweifelsfall setzen wir Klammern.

````python
print(2 + 3 * 4)       # 14   Punkt vor Strich
print((2 + 3) * 4)     # 20   Klammern zuerst
print(2 * 3 ** 2)      # 18   Exponentation vor Punktrechnung
````

## GL 4 · Strings: Länge und Konkatenation

### Inhalte

**Strings** sind Zeichenketten. Sie werden in einfache `'…'` oder doppelte `"…"` Hochkommata eingeschlossen. Beide Schreibweisen sind gleichwertig.

````python
a = 'Hallo Welt'
b = "Guten Morgen"
print(a)          # Hallo Welt
print(b)          # Guten Morgen
````

Die **Länge** eines Strings ist die Anzahl seiner Zeichen. Wir erhalten sie mit `len()`. Leerzeichen zählen mit.

````python
s = 'Habicht'
print(len(s))     # 7
````

Mit dem Zeichen `+` können wir Strings aneinanderhängen (**konkatenieren**). Leerzeichen werden dabei nicht automatisch eingefügt, wir müssen sie selbst ergänzen.

````python
s1 = 'Hallo'
s2 = 'du'
s3 = 'Welt'
s4 = s1 + ' ' + s2 + ' ' + s3
print(s4)         # Hallo du Welt
````

**Vorsicht:** `'3' + '4'` ergibt den String `'34'`, denn Strings werden verkettet, nicht addiert. Einen String und eine Zahl kann man nicht mit `+` verbinden: `'Alter: ' + 15` führt zu einem *TypeError*.

- `'AdaLovelace'` hat 3 + 8 = 11 Zeichen.
- `'Hallo Welt'` hat 10 Zeichen, denn das Leerzeichen zählt mit.
- `'3' + '4'` verkettet zwei Strings zu `'34'`. Das Ergebnis ist keine Zahl.

## GL 5 · Strings: Indexing und Slicing

### Inhalte

Jedes Zeichen eines Strings hat **zwei Indizes**: von vorne gezählt positiv, beginnend bei `0`, und von hinten gezählt negativ, beginnend bei `-1`.

<img src="./habicht.png" width="300" alt="Das Wort Habicht mit den positiven Indizes 0 bis 6 und den negativen Indizes -7 bis -1">

```
  H   a   b   i   c   h   t
  0   1   2   3   4   5   6
 -7  -6  -5  -4  -3  -2  -1
```

Mit **Indexing** greifen wir auf ein einzelnes Zeichen zu:

````python
s = 'Habicht'
print(s[0])           # H   erstes Zeichen
print(s[1])           # a   zweites Zeichen
print(s[len(s)-1])    # t   letztes Zeichen
print(s[-1])          # t   letztes Zeichen, kürzer
print(s[-2])          # h   vorletztes Zeichen
````

Der größte positive Index ist `len(s) - 1`. Ein Zugriff wie `s[7]` führt zu einem *IndexError*.

Mit **Slicing** erzeugen wir Teilstrings: `s[von:bis]`. Fehlt die von-Zahl, beginnt der Teilstring am Anfang. Fehlt die bis-Zahl, reicht er bis zum Ende.

````python
s = 'Habicht'
print(s[:3])          # Hab     die ersten 3 Zeichen
print(s[-3:])         # cht     die letzten 3 Zeichen
print(s[2:])          # bicht   alles außer den ersten beiden Zeichen
print(s[:-2])         # Habic   alles außer den letzten beiden Zeichen
````

**Der Index der bis-Zahl wird nicht mitgenommen.** Die Differenz `bis - von` ist deshalb genau die Länge des Ergebnisses.

````python
s = 'Habicht'
print(s[1:5])         # abic    Index 1 bis 4, Länge 5 - 1 = 4
````

## GL 6 · Strings: Methoden und Vergleiche

### Inhalte

Mit den String-Methoden `lower()` und `upper()` verwandeln wir alle Buchstaben in Klein- bzw. Großbuchstaben. Sie liefern einen **neuen String**, der ursprüngliche String bleibt unverändert.

````python
s = 'Habicht'
print(s.lower())    # habicht
print(s.upper())    # HABICHT
print(s)            # Habicht   unverändert
````

Mit dem Zeichen `<` prüfen wir die **alphabetische Ordnung**. Verglichen wird Zeichen für Zeichen, bis sich zwei Zeichen unterscheiden. Dabei gilt: **Großbuchstaben < Kleinbuchstaben**.

````python
print('aab' < 'abb')    # True    das zweite Zeichen entscheidet
print('a' < 'A')        # False   Großbuchstaben kommen zuerst
````

Mit dem Schlüsselwort `in` prüfen wir, ob ein Zeichen oder ein Teilstring in einem anderen String enthalten ist.

````python
print('a' in 'Habicht')    # True
print('e' in 'Habicht')    # False
````

Mit der String-Methode `find()` finden wir den **Index des ersten Auftretens** eines Teilstrings. Das Ergebnis `-1` heißt: Es wurde nichts gefunden.

````python
s = 'Habicht'
print(s.find('ich'))    # 3
print(s.find('b'))      # 2
print(s.find('e'))      # -1
````

Spätere Auftreten finden wir, wenn wir als zweiten Wert mitgeben, **ab welchem Index** gesucht werden soll:

````python
s = 'aaabbaaaabbaaaa'
print(s.find('bb'))       # 3
print(s.find('bb', 4))    # 9    suche 'bb' ab Index 4
````

## GL 8 · Ausgabe mit f-Strings

### Inhalte

Für die Ausgabe von Text und Variablen nutzen wir einen **f-String**. Er beginnt mit einem `f` vor dem Hochkomma. Variablen oder ganze Ausdrücke schreiben wir in **geschweiften Klammern** direkt in den String. Eine Umwandlung mit `str()` ist nicht nötig.

````python
x = int(input('Bitte eine ganze Zahl eingeben: '))
print(f'Das Doppelte von {x} ist {2*x}.')
````

```
Bitte eine ganze Zahl eingeben: 12
Das Doppelte von 12 ist 24.
```

Bei Bedarf fügen wir nach einem Doppelpunkt eine **Formatierung** hinzu. Mit `:.2f` wird eine Dezimalzahl auf zwei Nachkommastellen gerundet:

````python
pi = 3.1415926535
print(f'Pi auf zwei Nachkommastellen: {pi:.2f}')
# Pi auf zwei Nachkommastellen: 3.14
````

Die Zahl vor dem `f` gibt die Anzahl der Nachkommastellen an: `{pi:.4f}` ergibt `3.1416`. Der Wert der Variablen selbst bleibt dabei unverändert, gerundet wird nur die Ausgabe.

### Aufgabe 1

1. Was gibt `print(f'{2/3:.2f}')` aus? Wird gerundet oder abgeschnitten?
2. Gegeben sind `name = 'Lea'`, `punkte = 17` und `maxPunkte = 20`. Gib mit einem f-String den Satz `Lea hat 17 von 20 Punkten erreicht (85.0 %).` aus. Der Prozentwert soll berechnet und mit einer Nachkommastelle ausgegeben werden.

#### Lösung

**1.** Ausgabe: `0.67`. `2/3` ist 0.6666…, die Formatierung **rundet** auf zwei Stellen. Abgeschnitten wäre es 0.66.

**2.**

````python
name = 'Lea'
punkte = 17
maxPunkte = 20
print(f'{name} hat {punkte} von {maxPunkte} Punkten erreicht ({punkte/maxPunkte*100:.1f} %).')
````

In den geschweiften Klammern darf ein ganzer Ausdruck stehen. Hier wird der Prozentwert 17 / 20 · 100 = 85.0 direkt im f-String berechnet und mit `:.1f` formatiert.

### Aufgabe 2

Schreibe einen Preisrechner: Das Programm liest einen Nettopreis ein und gibt die Mehrwertsteuer (19 %) und den Bruttopreis aus, jeweils mit zwei Nachkommastellen und dem Zusatz `Euro`.

#### Lösung

````python
netto = float(input('Nettopreis in Euro: '))
mwst = netto * 0.19
brutto = netto + mwst
print(f'Mehrwertsteuer: {mwst:.2f} Euro')
print(f'Bruttopreis:    {brutto:.2f} Euro')
````

```
Nettopreis in Euro: 49.99
Mehrwertsteuer: 9.50 Euro
Bruttopreis:    59.49 Euro
```

Ohne Formatierung erschiene `9.4981` bzw. `59.4881`. Bei Geldbeträgen sind aber zwei Nachkommastellen üblich. Mit `:.2f` erscheint außerdem `9.50` statt `9.5`, die Null am Ende bleibt also stehen.
